### 1.1 รหัสยูนิโค้ดและจำนวนไบต์ (ตัวอย่างที่ 7.1)

หัวข้อ 7.1.1 อธิบายว่าคอมพิวเตอร์เก็บข้อความเป็นลำดับ **รหัสยูนิโค้ด (Code Point)** และบันทึกลงไฟล์ด้วยการเข้ารหัส **UTF-8** ซึ่งใช้ 1 ไบต์ต่ออักษรอังกฤษพื้นฐาน และ 3 ไบต์ต่ออักษรไทย สระบน สระล่าง และวรรณยุกต์ของไทยเป็นรหัสแยกจากพยัญชนะ ใน Python ฟังก์ชัน `ord(ch)` ให้รหัสของอักขระ `len(text)` นับจำนวนรหัส และ `text.encode("utf-8")` แปลงเป็นไบต์

```python
def describe_code_points(text):
    """ตารางรหัสยูนิโค้ดของแต่ละอักขระ พร้อมจำนวนไบต์ใน UTF-8"""
    rows = []
    for ch in text:
        rows.append({
            "อักขระ": ch if ch != " " else "␠",
            "รหัส": f"U+{ord(ch):04X}",
            "ชื่อยูนิโค้ด": unicodedata.name(ch, "?"),
            "ไบต์ UTF-8": len(ch.encode("utf-8")),
        })
    return pd.DataFrame(rows)


for text in ["ปั๊มร้อน", "pump hot"]:
    print(f"{text!r}: {len(text)} รหัส, {len(text.encode('utf-8'))} ไบต์")
describe_code_points("ปั๊มร้อน")
```

```python
# ขนาดของใบสั่งงานภาษาไทย 50,000 ใบ ใบละ 120 อักขระ (อักษรไทยใบละ 3 ไบต์)
n_orders, chars_per_order, bytes_per_thai_char = 50_000, 120, 3
total_bytes = n_orders * chars_per_order * bytes_per_thai_char
print(f"{total_bytes:,} ไบต์ = {total_bytes / 1e6:.0f} MB  (หนังสือ: 18 MB)")
```

### 1.2 ลำดับสระและวรรณยุกต์ที่พิมพ์ต่างกัน

ขั้นที่ 3 ของตัวอย่างที่ 7.1 ชี้ว่าผู้พิมพ์บางคนพิมพ์ไม้ตรีก่อนไม้หันอากาศ ได้ลำดับรหัส ป ๊ ั ม ซึ่งหลายโปรแกรมแสดงผลเหมือน "ปั๊ม" ทุกประการ แต่ Python ถือว่าเป็นคนละสตริง ลองดูว่าการทำให้เป็นมาตรฐานแบบ NFC ของยูนิโค้ดช่วยได้หรือไม่ แล้วเทียบกับ `normalize` ของ PyThaiNLP และฟังก์ชันสั้น ๆ ที่เขียนเอง

```python
correct = "ปั๊ม"
swapped = "ป๊ัม"   # พิมพ์ไม้ตรี (U+0E4A) ก่อนไม้หันอากาศ (U+0E31)

THAI_TONE_MARKS = "่้๊๋"          # ่ ้ ๊ ๋
THAI_UPPER_VOWELS = "ัิีึื"  # ั ิ ี ึ ื


def fix_thai_mark_order(text):
    """สลับวรรณยุกต์ที่พิมพ์ก่อนสระบน และลบวรรณยุกต์ที่พิมพ์ซ้ำ (ฉบับย่อของ normalize)"""
    text = re.sub(f"([{THAI_TONE_MARKS}])([{THAI_UPPER_VOWELS}])", r"\2\1", text)
    return re.sub(f"([{THAI_TONE_MARKS}])\\1+", r"\1", text)


print("สตริงเท่ากันหรือไม่:", correct == swapped)
print("รหัสของสตริงที่พิมพ์สลับ:", [f"U+{ord(c):04X}" for c in swapped])
print("NFC แก้ได้หรือไม่:", unicodedata.normalize("NFC", swapped) == correct)
print("ฟังก์ชันที่เขียนเองแก้ได้หรือไม่:", fix_thai_mark_order(swapped) == correct)
if HAVE_PYTHAINLP:
    print("pythainlp.util.normalize แก้ได้หรือไม่:", thai_normalize(swapped) == correct)
print("ลบวรรณยุกต์ซ้ำ:", fix_thai_mark_order("ร้้อน"))
```

NFC ไม่ช่วย เพราะยูนิโค้ดไม่ได้กำหนดลำดับมาตรฐานของเครื่องหมายไทยกลุ่มนี้ จึงต้องใช้เครื่องมือเฉพาะภาษาไทยอย่าง `normalize` ของ PyThaiNLP ถ้าไม่ทำขั้นนี้ การค้นหาคำว่า "ปั๊ม" จะหาใบสั่งงานที่พิมพ์สลับไม่พบ

### 1.3 เตรียมข้อความใบสั่งงานซ่อม และกับดักของคำว่า "ไม่" (ตัวอย่างที่ 7.3)

ฟังก์ชัน `normalize_text` ด้านล่างทำตามรหัสเทียม PREPROCESS ใน หัวข้อ 7.2.1 ได้แก่ แก้ลำดับสระ/วรรณยุกต์ แปลงเลขไทยเป็นเลขอารบิก แปลงอักษรอังกฤษเป็นตัวพิมพ์เล็ก แปลงคำพ้องเฉพาะโดเมน และลบเครื่องหมายวรรคตอนซ้ำ ส่วนการตัดคำจะทำในส่วนที่ 2 ที่นี่ใช้รายการโทเคนที่ตัวอย่างที่ 7.3 กำหนดไว้ เพื่อเทียบผลของการตัดคำหยุดสองแบบ

```python
THAI_DIGITS = str.maketrans("๐๑๒๓๔๕๖๗๘๙", "0123456789")
# พจนานุกรมคำพ้องเฉพาะโดเมน (ระดับข้อความ ใช้ก่อนตัดคำ และระดับโทเคน ใช้หลังตัดคำ)
TEXT_SYNONYMS = {"คอนแทกเตอร์": "คอนแทคเตอร์", "เบรคเกอร์": "เบรกเกอร์", "ทริพ": "ทริป"}
TOKEN_SYNONYMS = {"trip": "ทริป", "tripped": "ทริป"}


def normalize_text(text):
    """ทำข้อความให้เป็นรูปแบบมาตรฐานตามหัวข้อ 7.2.1 (ยังไม่ตัดคำ)"""
    text = thai_normalize(text) if HAVE_PYTHAINLP else fix_thai_mark_order(text)
    text = text.translate(THAI_DIGITS).lower()
    for variant, canonical in TEXT_SYNONYMS.items():
        text = text.replace(variant, canonical)
    text = re.sub(r"([!?.,])\1+", " ", text)          # เครื่องหมายวรรคตอนซ้ำ เช่น "!!"
    return re.sub(r"\s+", " ", text).strip()


raw = "เบรกเกอร์ MCCB-3 Trip อีกแล้ว!!  มอเตอร์ปั๊ม ๒ ร้อนมาก ไม่มีเสียงผิดปกติ"
print("ข้อความดิบ:     ", raw)
print("หลังทำมาตรฐาน: ", normalize_text(raw))

# โทเคนตามขั้นที่ 2 ของตัวอย่างที่ 7.3 (สมมติว่าตัวตัดคำรู้จัก "mccb-3")
book_tokens = ["เบรกเกอร์", "mccb-3", "trip", "อีก", "แล้ว", "มอเตอร์", "ปั๊ม", "2", "ร้อน", "มาก",
               "ไม่", "มี", "เสียง", "ผิดปกติ"]
tokens = [TOKEN_SYNONYMS.get(t, t) for t in book_tokens]
stop_words = {"อีก", "แล้ว", "มาก", "มี"}
keep_words = {"ไม่"}                                     # คำที่ห้ามตัด

careful = [t for t in tokens if t not in stop_words or t in keep_words]
careless = [t for t in tokens if t not in stop_words | {"ไม่"}]
print(f"\nตัดคำแล้ว {len(tokens)} โทเคน: {' | '.join(tokens)}")
print(f"ตัดคำหยุดแบบระวัง {len(careful)} โทเคน:   {' | '.join(careful)}")
print(f"ตัดคำหยุดแบบไม่ระวัง {len(careless)} โทเคน: {' | '.join(careless)}")
```

### 2.1 เลือกคำยาวที่สุด และจำนวนคำน้อยที่สุด (ตัวอย่างที่ 7.5)

หัวข้อ 7.2.2 ให้รหัสเทียมของสองอัลกอริทึม เซลล์ถัดไปเขียนตามรหัสเทียมนั้นตรง ๆ

- `longest_matching` อ่านจากซ้ายไปขวา เลือกคำในพจนานุกรมที่ยาวที่สุดที่ขึ้นต้น ณ ตำแหน่งปัจจุบัน (ละโมบ) ถ้าไม่พบคำใดให้ตัดอักขระเดียว
- `maximal_matching` ใช้กำหนดการพลวัต `best[j]` = จำนวนคำน้อยที่สุดที่ตัด `s[0:j]` ได้ และ `back[j]` = จุดเริ่มของคำสุดท้าย แล้วย้อนรอยจาก `j = n` ฉบับนี้เพิ่มทางเลือก "อักขระที่ไม่รู้จัก" ให้มีต้นทุนสูง (100) เพื่อให้ตัดสตริงที่มีคำนอกพจนานุกรมได้เสมอ โดยไม่กระทบกรณีที่ตัดด้วยพจนานุกรมได้ครบ

```python
UNKNOWN_COST = 100   # ต้นทุนของการตัดอักขระที่ไม่รู้จักออกมาเดี่ยว ๆ (สูงกว่าคำในพจนานุกรมมาก)


def longest_matching(s, dictionary):
    """การเลือกคำยาวที่สุด (รหัสเทียม LONGEST-MATCHING ในหัวข้อ 7.2.2)"""
    i, words = 0, []
    while i < len(s):
        candidates = [w for w in dictionary if s.startswith(w, i)]
        w = max(candidates, key=len) if candidates else s[i]
        words.append(w)
        i += len(w)
    return words


def maximal_matching(s, dictionary, max_word_len=20, return_table=False):
    """การตัดให้ได้จำนวนคำน้อยที่สุดด้วยกำหนดการพลวัต (รหัสเทียม MAXIMAL-MATCHING)"""
    n = len(s)
    best = [0] + [math.inf] * n
    back = [None] * (n + 1)
    for j in range(1, n + 1):
        for i in range(max(0, j - max_word_len), j):
            cost = 1 if s[i:j] in dictionary else (UNKNOWN_COST if j - i == 1 else None)
            if cost is not None and best[i] + cost < best[j]:
                best[j], back[j] = best[i] + cost, i
    words, j = [], n
    while j > 0:
        words.append(s[back[j]:j])
        j = back[j]
    words.reverse()
    return (words, best, back) if return_table else words


def cut_positions(words):
    """ตำแหน่งรอยตัดภายในสตริง (นับอักขระตามรหัสยูนิโค้ด ไม่รวม 0 และ n)"""
    positions, pos = set(), 0
    for w in words[:-1]:
        pos += len(w)
        positions.add(pos)
    return positions


def precision_recall_f1(tp, fp, fn):
    precision = tp / (tp + fp) if tp + fp else 0.0
    recall = tp / (tp + fn) if tp + fn else 0.0
    f1 = 2 * tp / (2 * tp + fp + fn) if tp + fp + fn else 0.0
    return precision, recall, f1


def segmentation_scores(gold_words, pred_words):
    gold, pred = cut_positions(gold_words), cut_positions(pred_words)
    return precision_recall_f1(len(gold & pred), len(pred - gold), len(gold - pred))
```

```python
s = "ไปหามเหสี"
toy_dictionary = {"ไป", "หา", "หาม", "มเหสี", "เห", "สี"}

lm_words = longest_matching(s, toy_dictionary)
mm_words, best, back = maximal_matching(s, toy_dictionary, return_table=True)
print("เลือกคำยาวที่สุด:     ", " | ".join(lm_words), f"({len(lm_words)} คำ)")
print("จำนวนคำน้อยที่สุด:   ", " | ".join(mm_words), f"({len(mm_words)} คำ)")

# ตาราง best[j] เฉพาะตำแหน่งที่มีคำในพจนานุกรมมาสิ้นสุด (เทียบกับตารางในตัวอย่างที่ 7.5)
rows = [{"j": 0, "คำที่สิ้นสุดที่ j (เริ่มที่ i)": "—", "best[j]": 0, "back[j]": "—"}]
for j in range(1, len(s) + 1):
    ending = [f"{s[i:j]} ({i})" for i in range(j) if s[i:j] in toy_dictionary]
    if ending:
        rows.append({"j": j, "คำที่สิ้นสุดที่ j (เริ่มที่ i)": ", ".join(ending), "best[j]": best[j], "back[j]": back[j]})
display(pd.DataFrame(rows))

gold_words = ["ไป", "หา", "มเหสี"]
for name, words in [("เลือกคำยาวที่สุด", lm_words), ("จำนวนคำน้อยที่สุด", mm_words)]:
    p, r, f = segmentation_scores(gold_words, words)
    print(f"{name}: รอยตัด {sorted(cut_positions(words))} | P = {p:.3f}, R = {r:.3f}, F1 = {f:.3f}")
print("หนังสือ: เลือกคำยาวที่สุด P = 0.333, R = 0.500, F1 = 0.400 | จำนวนคำน้อยที่สุด F1 = 1.000")
```

```python
# กรณี "ตากลม": ทั้งสองแบบเป็นคำในพจนานุกรมและได้ 2 คำเท่ากัน
ta_dictionary = {"ตา", "ตาก", "กลม", "ลม"}
print("เลือกคำยาวที่สุด:   ", longest_matching("ตากลม", ta_dictionary))
print("จำนวนคำน้อยที่สุด: ", maximal_matching("ตากลม", ta_dictionary), "(อีกทางที่ได้ 2 คำเท่ากันคือ ตาก | ลม)")
```

### 2.2 ตัวตัดคำของ PyThaiNLP และพจนานุกรมผู้ใช้ (ตัวอย่างที่ 7.6)

`word_tokenize(text, engine=...)` ของ PyThaiNLP มีตัวตัดคำหลายแบบ ค่าเริ่มต้น `newmm` คือการตัดให้ได้จำนวนคำน้อยที่สุดบนกลุ่มอักขระไทย (TCC) ส่วน `longest` คือการเลือกคำยาวที่สุด พจนานุกรมที่ใช้คือพจนานุกรมคำไทยกว่า 60,000 คำ (`thai_words()`) และเปลี่ยนเป็นพจนานุกรมผู้ใช้ได้ด้วยอาร์กิวเมนต์ `custom_dict`

ฟังก์ชัน `tokenize_th` ด้านล่างรวมทุกขั้นของการเตรียมข้อความไว้ในที่เดียว และจะใช้ตลอด notebook

1. `normalize_text` (ส่วนที่ 1.3) รวมถึงกฎคำพ้อง "คอนแทกเตอร์" → "คอนแทคเตอร์"
2. ใช้นิพจน์ปกติแยกรหัสตู้ เช่น "ตู้MDB2" → `ตู้mdb2` คำอังกฤษ/รหัส เช่น `mccb-3` `ov2` และตัวเลข ออกมาก่อน
3. ตัดคำส่วนที่เป็นอักษรไทยด้วย `newmm` และพจนานุกรม `thai_words()` + `DOMAIN_WORDS` (ถ้าไม่มี PyThaiNLP ใช้ `maximal_matching` กับพจนานุกรมสำรอง)
4. แปลงคำพ้องระดับโทเคน เช่น `trip` → `ทริป`

```python
# คำศัพท์เฉพาะโดเมนของโรงงานจำลอง (ในงานจริงควรสร้างร่วมกับช่างผู้ชำนาญ)
DOMAIN_WORDS = {
    "เบรกเกอร์", "คอนแทคเตอร์", "ติดค้าง", "ฟิวส์", "สายไฟ", "หม้อแปลง", "มอเตอร์", "ไหม้", "ลัดวงจร", "ไฟรั่ว",
    "แรงดันตก", "ค่าฉนวน", "ลูกปืน", "สายพาน", "ปั๊ม", "พัดลม", "เกียร์", "ซีล", "เสียงดัง", "หลวม", "รั่ว",
    "สึก", "ร้อนจัด", "อุณหภูมิสูง", "เซนเซอร์", "อินเวอร์เตอร์", "รีเลย์", "ทรานสมิตเตอร์", "สื่อสาร",
    "ค่าเพี้ยน", "ค้าง", "ขึ้นรหัส", "ไลน์", "ทริป", "สั่น", "ขาด", "ผิดปกติ", "ฮีตซิงก์", "ตัวต้านทานเบรก",
    "พารามิเตอร์", "โอเวอร์โหลด", "แรงดันเกิน", "กระแสเกิน", "ลดความเร็ว", "เร่งความเร็ว", "บัส", "ตัวเก็บประจุ",
    "ขาดเฟส", "ลงดิน", "ร้อนเกิน", "ระบายความร้อน", "ขั้วต่อ", "แรงบิด", "ฉนวน", "ทดสอบ",
}
# พจนานุกรมสำรองเมื่อไม่มี PyThaiNLP: คำโดเมน + คำทั่วไปที่พบในข้อมูลจำลองและคำถามในส่วนที่ 8
FALLBACK_WORDS = DOMAIN_WORDS | {
    "ไม่", "มี", "เสียง", "ได้", "อีก", "แล้ว", "อีกแล้ว", "ด่วน", "ช่วย", "ดู", "ด้วย", "ตรวจ", "เปลี่ยน", "ร้อน",
    "มาก", "ที่", "ตัว", "น้ำ", "ตอน", "หยุด", "แก้", "อย่างไร", "คือ", "อะไร", "หมายถึง", "ต้อง", "ก่อน", "หลัง",
    "ปิด", "เปิด", "ไฟ", "รอ", "นาน", "เท่าไร", "ฝา", "กี่", "ปี", "ขนาด", "ต่อ", "ยังไง", "ตั้ง", "เวลา", "ไหน",
    "บ่อย", "เข้า", "กระแส", "แรงดัน", "เกิน", "ค่า", "ป้องกัน", "ตู้", "เครื่อง", "จ่าย", "ตลอด", "ทำงาน",
}
TRIE = dict_trie(set(thai_words()) | DOMAIN_WORDS) if HAVE_PYTHAINLP else None
TOKEN_PATTERN = re.compile(r"ตู้[a-z]+\d*|[a-z]+\d*(?:-[a-z0-9]+)*|\d+(?:\.\d+)?|[฀-๿]+")


def segment_thai(run, engine="newmm"):
    """ตัดคำสตริงที่เป็นอักษรไทยล้วน"""
    if HAVE_PYTHAINLP:
        return [w for w in word_tokenize(run, engine=engine, custom_dict=TRIE) if w.strip()]
    return maximal_matching(run, FALLBACK_WORDS)


def tokenize_th(text, engine="newmm"):
    """เตรียมข้อความ + ตัดคำ + แปลงคำพ้อง คืนรายการโทเคน"""
    text = re.sub(r"ตู้\s*([a-z]+\d*)", r" ตู้\1 ", normalize_text(text))
    tokens = []
    for run in TOKEN_PATTERN.findall(text):
        is_thai_word = "฀" <= run[0] <= "๿" and not re.search("[a-z]", run)
        tokens.extend(segment_thai(run, engine) if is_thai_word else [run])
    return [TOKEN_SYNONYMS.get(t, t) for t in tokens]


print("ตัวอย่างที่ 7.3:", " | ".join(tokenize_th(raw)))
```

```python
orders_76 = ["คอนแทคเตอร์ติดค้าง", "คอนแทกเตอร์ติดค้าง ปั๊ม 3", "เปลี่ยนคอนแทคเตอร์ตู้MDB2"]

if HAVE_PYTHAINLP:
    rows = []
    for text in orders_76:
        rows.append({
            "ใบสั่งงาน": text,
            "newmm (พจนานุกรมทั่วไป)": " | ".join(w for w in word_tokenize(text) if w.strip()),
            "longest (พจนานุกรมทั่วไป)": " | ".join(w for w in word_tokenize(text, engine="longest") if w.strip()),
            "tokenize_th (มาตรฐาน + พจนานุกรมผู้ใช้)": " | ".join(tokenize_th(text)),
        })
    display(pd.DataFrame(rows))
    # ตัวอย่างที่ 7.5 กับพจนานุกรมจริงของ PyThaiNLP
    print("newmm  'ไปหามเหสี':", word_tokenize("ไปหามเหสี"))
    print("longest 'ไปหามเหสี':", word_tokenize("ไปหามเหสี", engine="longest"))
else:
    for text in orders_76:
        print(text, "→", " | ".join(tokenize_th(text)))

counts = Counter(t for text in orders_76 for t in tokenize_th(text))
print("ความถี่ของ 'คอนแทคเตอร์' หลังแก้:", counts["คอนแทคเตอร์"], "(หนังสือ: 3)")
```

สังเกตว่าพจนานุกรมทั่วไปตัด "คอนแทคเตอร์" เป็นชิ้นที่ไม่มีความหมาย และตัด "คอนแทกเตอร์" อีกแบบหนึ่ง การนับคำจึงไม่พบว่าทั้งสามใบพูดถึงอุปกรณ์เดียวกัน หลังเพิ่มคำลงพจนานุกรมผู้ใช้และกฎคำพ้อง ความถี่ของ "คอนแทคเตอร์" เป็น 3 ตามตัวอย่างที่ 7.6

ส่วน "ไปหามเหสี" พจนานุกรมจริงของ PyThaiNLP มีคำว่า "ไปหา" อยู่แล้ว ตัวตัดคำทั้งสองแบบจึงได้ ไปหา | มเหสี (2 คำ) ซึ่งถูกความหมาย ตัวอย่างในหนังสือใช้พจนานุกรมขนาดเล็กเพื่อให้เห็นจุดอ่อนของการเลือกคำยาวที่สุด ผลของตัวตัดคำจริงจึงขึ้นกับพจนานุกรมมากพอ ๆ กับอัลกอริทึม

### 2.3 วัดผลงานติดป้ายรายโทเคน: NER ระดับชื่อเฉพาะ (ตัวอย่างที่ 7.2)

การตัดคำและการรู้จำชื่อเฉพาะ (NER) เป็นงานติดป้ายรายโทเคนเหมือนกัน ฟังก์ชัน `bio_spans` อ่านชื่อเฉพาะ (จุดเริ่ม, จุดจบ, ชนิด) จากป้าย BIO แล้วนับว่าถูกเมื่อ **ขอบเขตและชนิดตรงกันทุกประการ**

```python
def bio_spans(tags):
    """อ่านชื่อเฉพาะจากป้าย BIO คืนเซตของ (start, end, type)"""
    spans, start, kind = set(), None, None
    for i, tag in enumerate(list(tags) + ["O"]):
        if tag == "O" or tag.startswith("B-") or tag[2:] != kind:
            if start is not None:
                spans.add((start, i, kind))
            start, kind = (i, tag[2:]) if tag != "O" else (None, None)
    return spans


ner_tokens = ["หม้อแปลง", "TR-02", "ที่", "สถานี", "บางพลี", "มี", "น้ำมัน", "รั่ว"]
gold_tags = ["B-EQ", "I-EQ", "O", "B-LOC", "I-LOC", "O", "B-SYM", "I-SYM"]
pred_tags = ["B-EQ", "I-EQ", "O", "B-LOC", "O", "O", "B-SYM", "I-SYM"]

gold_spans, pred_spans = bio_spans(gold_tags), bio_spans(pred_tags)
tp, fp, fn = len(gold_spans & pred_spans), len(pred_spans - gold_spans), len(gold_spans - pred_spans)
p, r, f = precision_recall_f1(tp, fp, fn)
token_acc = np.mean([g == q for g, q in zip(gold_tags, pred_tags)])
for start, end, kind in sorted(pred_spans - gold_spans):
    print("ทำนายผิด:", " ".join(ner_tokens[start:end]), kind)
print(f"TP = {tp}, FP = {fp}, FN = {fn} | P = {p:.3f}, R = {r:.3f}, F1 = {f:.3f}  (หนังสือ 0.667 ทั้งสามค่า)")
print(f"ความถูกต้องรายโทเคน = {token_acc:.1%}  (หนังสือ 87.5%)")
```

## ส่วนที่ 3 Byte-Pair Encoding ที่เขียนเอง (ตัวอย่างที่ 7.4)

LLM ไม่ได้ใช้ "คำ" เป็นโทเคน แต่ใช้ **หน่วยย่อยของคำ (Subword)** ที่เรียนรู้ด้วย BPE (หัวข้อ 7.2.1) เริ่มจากอักขระเดี่ยวและสัญลักษณ์จบคำ `_` แล้วรวมคู่สัญลักษณ์ที่อยู่ติดกันบ่อยที่สุดซ้ำ ๆ เมื่อความถี่เท่ากันให้เลือกคู่ที่พบก่อนตามลำดับการอ่าน (ข้อตกลงของหนังสือ) ฟังก์ชันด้านล่างเขียนตามรหัสเทียม LEARN-BPE และ APPLY-BPE

```python
def merge_pair(symbols, pair):
    """แทนที่ทุกคู่ a b ที่ติดกันด้วยสัญลักษณ์ใหม่ ab"""
    out, i = [], 0
    while i < len(symbols):
        if i + 1 < len(symbols) and (symbols[i], symbols[i + 1]) == pair:
            out.append(symbols[i] + symbols[i + 1])
            i += 2
        else:
            out.append(symbols[i])
            i += 1
    return out


def count_pairs(words):
    """นับความถี่ของคู่สัญลักษณ์ที่ติดกัน ถ่วงด้วยความถี่ของคำ (Counter เก็บลำดับที่พบครั้งแรก)"""
    counts = Counter()
    for symbols, freq in words:
        for pair in zip(symbols, symbols[1:]):
            counts[pair] += freq
    return counts


def learn_bpe(corpus, num_merges, verbose=True):
    """corpus: รายการ (คำ, ความถี่) ตามลำดับการอ่าน คืนรายการกฎการรวม"""
    words = [(list(word) + ["_"], freq) for word, freq in corpus]
    rules = []
    for step in range(1, num_merges + 1):
        counts = count_pairs(words)
        top = max(counts.values())
        best = next(pair for pair in counts if counts[pair] == top)   # เสมอ → คู่ที่พบก่อน
        rules.append(best)
        words = [(merge_pair(symbols, best), freq) for symbols, freq in words]
        if verbose:
            ranked = ", ".join(f"({a},{b})={c}" for (a, b), c in counts.most_common(4))
            print(f"รวมครั้งที่ {step}: {best[0]} + {best[1]} → {best[0] + best[1]}   [ความถี่สูงสุด: {ranked}]")
            print("    ", "; ".join(f"{' '.join(s)} (×{f})" for s, f in words))
    return rules


def apply_bpe(word, rules):
    symbols = list(word) + ["_"]
    for pair in rules:
        symbols = merge_pair(symbols, pair)
    return symbols


fan_corpus = [("fan", 5), ("fans", 2), ("fault", 3)]
fan_rules = learn_bpe(fan_corpus, 3)
print("\nกฎการรวม:", fan_rules)
for word in ["fan", "fans", "fault", "faults"]:
    pieces = apply_bpe(word, fan_rules)
    print(f"{word:7s} → {pieces} ({len(pieces)} โทเคน)")
```

```python
# ข้อความไทยกับ BPE: เรียนรู้จากคำในใบสั่งงานจำลอง (BPE ทำงานกับอักขระ จึงไม่ต้องรู้ขอบเขตของคำย่อยภายในคำ)
BPE_THAI_MERGES = 22
thai_corpus = [("มอเตอร์", 10), ("เบรกเกอร์", 6), ("ร้อน", 6), ("ร้อนจัด", 2), ("สั่น", 4), ("ทริป", 5)]
thai_rules = learn_bpe(thai_corpus, BPE_THAI_MERGES, verbose=False)
print(f"เรียนรู้ {len(thai_rules)} กฎ")
for word in ["เบรกเกอร์", "ร้อนจัด", "ทริพ", "คอนแทคเตอร์", "มอเตอร์ร้อน"]:
    pieces = apply_bpe(word, thai_rules)
    print(f"{word} → {pieces} ({len(pieces)} โทเคน)")
```

สังเกตว่า "เบรกเกอร์" ที่พบบ่อยกลายเป็นโทเคนเดียว ส่วนคำที่พบน้อยหรือไม่เคยเห็น เช่น "ร้อนจัด" "ทริพ" (สะกดผิด) และ "คอนแทคเตอร์" ถูกแยกเป็นหลายชิ้น ซึ่งส่วนใหญ่เป็นอักขระเดี่ยวเพราะคลังข้อความเล็กมาก แต่ทุกคำยังแทนได้โดยไม่มีโทเคน "ไม่รู้จัก"

**คำถามชวนคิด** ถ้าคลังข้อความที่ใช้ฝึกตัวแบ่งโทเคนของ LLM มีภาษาไทยน้อยมาก ข้อความไทยจะถูกแบ่งเป็นโทเคนจำนวนมากหรือน้อย และมีผลต่อค่าใช้จ่ายและหน้าต่างบริบทอย่างไร (ข้อควรระวังในหัวข้อ 7.2.1)

## ส่วนที่ 4 Bag-of-Words, TF-IDF และความคล้ายโคไซน์

### 4.1 คำนวณตัวอย่างที่ 7.7 ด้วย NumPy

สูตรใน หัวข้อ 7.3.1 คือ $\text{tf-idf}(t, d) = \text{tf}(t, d) \times \log_{10}(N/\text{df}(t))$ และ $\cos(\mathbf{a}, \mathbf{b}) = \mathbf{a} \cdot \mathbf{b} / (\lVert \mathbf{a} \rVert \lVert \mathbf{b} \rVert)$ เซลล์ถัดไปคำนวณทีละขั้นตามตัวอย่าง

```python
docs_77 = {
    "D1": ["มอเตอร์", "ร้อน", "สั่น"],
    "D2": ["มอเตอร์", "สั่น", "สั่น", "ลูกปืน"],
    "D3": ["หม้อแปลง", "ร้อน", "น้ำมัน", "รั่ว"],
    "D4": ["เบรกเกอร์", "ทริป", "มอเตอร์"],
}
query_77 = ["มอเตอร์", "สั่น"]
vocab_77 = list(dict.fromkeys(t for d in docs_77.values() for t in d))   # เรียงตามลำดับที่พบ


def cosine(a, b):
    a, b = np.asarray(a, float), np.asarray(b, float)
    denom = np.linalg.norm(a) * np.linalg.norm(b)
    return float(a @ b / denom) if denom else 0.0


# ขั้นที่ 1 เมทริกซ์ BoW และ df
bow = np.array([[d.count(t) for t in vocab_77] for d in docs_77.values()], dtype=float)
df = (bow > 0).sum(axis=0)
N = len(docs_77)
display(pd.DataFrame(np.vstack([bow, df]).astype(int), index=list(docs_77) + ["df"], columns=vocab_77))

# ขั้นที่ 3–4 idf และ TF-IDF
idf = np.log10(N / df)
tfidf = bow * idf
q_bow = np.array([query_77.count(t) for t in vocab_77], dtype=float)
q_tfidf = q_bow * idf
print("idf:", dict(zip(vocab_77, idf.round(3))))

# ขั้นที่ 2 และ 5 ความคล้ายโคไซน์
result_77 = pd.DataFrame({
    "cos (BoW)": [cosine(q_bow, d) for d in bow],
    "cos (TF-IDF)": [cosine(q_tfidf, d) for d in tfidf],
    "หนังสือ BoW": [0.816, 0.866, 0, 0.408],
    "หนังสือ TF-IDF": [0.735, 0.702, 0, 0.056],
}, index=list(docs_77)).round(3)
result_77
```

```python
# เทียบกับ TfidfVectorizer ของ scikit-learn (ค่าเริ่มต้น: idf = ln((1+N)/(1+df)) + 1 และปรับความยาวเป็น 1)
texts_77 = [" ".join(d) for d in docs_77.values()]
vectorizer = TfidfVectorizer(tokenizer=str.split, token_pattern=None, lowercase=False)
X_77 = vectorizer.fit_transform(texts_77)
q_77 = vectorizer.transform([" ".join(query_77)])
sk_idf = dict(zip(vectorizer.get_feature_names_out(), vectorizer.idf_.round(3)))
print("idf ของ scikit-learn:", {t: sk_idf[t] for t in vocab_77})
result_77["cos (sklearn)"] = cosine_similarity(q_77, X_77).ravel().round(3)
result_77[["cos (BoW)", "cos (TF-IDF)", "cos (sklearn)"]]
```

ตัวเลขของ scikit-learn ต่างจากหนังสือ เพราะสูตร idf ต่างกัน (หมายเหตุท้ายตัวอย่างที่ 7.7) คำที่พบทุกเอกสารจะไม่ได้น้ำหนักเป็นศูนย์ และ "มอเตอร์" จึงยังทำให้ D4 ได้คะแนนสูงกว่าสูตรพื้นฐาน แต่ **ลำดับ** ของเอกสารเหมือนกัน คือ D1 > D2 > D4 > D3 ซึ่งเป็นสิ่งที่ระบบค้นหาใช้จริง

### 4.2 ใบสั่งงานซ่อมจำลอง

ส่วนที่เหลือของ notebook ใช้ใบสั่งงานซ่อมจำลอง 600 ใบในช่วงสองปี แต่ละใบมีหมวดจริง 3 หมวด คือ ไฟฟ้า (E) เครื่องกล (M) และระบบควบคุม (C) ข้อความสร้างจากอุปกรณ์และอาการของแต่ละหมวด แล้วใส่ความหลากหลายแบบที่พบในงานจริง ได้แก่ การสะกดหลายแบบ ("trip" "ทริพ" "คอนแทกเตอร์") เลขไทย การเว้นวรรคไม่สม่ำเสมอ อาการของหมวดอื่นปนมา (เช่น มอเตอร์ร้อนในงานเครื่องกล) และป้ายผิดราวร้อยละ 3

```python
CLASS_SPEC = {
    "E": {"weight": 0.4,
          "equipment": ["เบรกเกอร์", "ฟิวส์", "คอนแทคเตอร์", "สายไฟ", "หม้อแปลง", "ตู้MDB2", "มอเตอร์"],
          "symptoms": ["ทริป", "ขาด", "ไหม้", "ร้อน", "ลัดวงจร", "ไฟรั่ว", "แรงดันตก", "ค่าฉนวนต่ำ"]},
    "M": {"weight": 0.4,
          "equipment": ["ลูกปืน", "สายพาน", "ปั๊ม", "พัดลม", "เกียร์", "ซีล", "มอเตอร์"],
          "symptoms": ["สั่น", "เสียงดัง", "หลวม", "รั่ว", "สึก", "ร้อน"]},
    "C": {"weight": 0.2,
          "equipment": ["PLC", "HMI", "เซนเซอร์", "อินเวอร์เตอร์", "รีเลย์", "ทรานสมิตเตอร์"],
          "symptoms": ["error", "สื่อสารไม่ได้", "ค่าเพี้ยน", "ค้าง", "ขึ้นรหัส", "alarm"]},
}
# รูปเขียนต่าง ๆ ของคำเดียวกันที่พบในใบสั่งงานจริง
SURFACE_FORMS = {"ทริป": ["ทริป", "trip", "Trip", "ทริพ"], "คอนแทคเตอร์": ["คอนแทคเตอร์", "คอนแทกเตอร์"],
                 "เบรกเกอร์": ["เบรกเกอร์", "เบรคเกอร์"], "ร้อน": ["ร้อน", "ร้อนจัด", "อุณหภูมิสูง"]}
SUFFIXES = ["", "", "อีกแล้ว", "ด่วน", "!!", "ช่วยดูด้วย", "ตรวจแล้ว"]
THAI_DIGIT_CHARS = "๐๑๒๓๔๕๖๗๘๙"


def surface(word, rng):
    forms = SURFACE_FORMS.get(word, [word])
    return forms[rng.integers(len(forms))]


def make_work_order(label, rng):
    spec = CLASS_SPEC[label]
    parts = [surface(rng.choice(spec["equipment"]), rng)]
    if rng.random() < 0.4:                                   # ระบุไลน์ผลิต บางคนใช้เลขไทย
        digit = int(rng.integers(1, 6))
        parts.append("ไลน์" + (THAI_DIGIT_CHARS[digit] if rng.random() < 0.3 else str(digit)))
    parts.append(surface(rng.choice(spec["symptoms"]), rng))
    if rng.random() < 0.4:                                   # อาการที่สองจากหมวดเดียวกัน
        parts.append(surface(rng.choice(spec["symptoms"]), rng))
    if rng.random() < 0.15:                                  # อาการของหมวดอื่นปนมา (ทำให้กำกวม)
        other = rng.choice([c for c in CLASS_SPEC if c != label])
        parts.append(surface(rng.choice(CLASS_SPEC[other]["symptoms"]), rng))
    parts.append(rng.choice(SUFFIXES))
    sep = " " if rng.random() < 0.5 else ""                   # เว้นวรรคไม่สม่ำเสมอ
    text = parts[0]
    for part in (p for p in parts[1:] if p):
        # คำอังกฤษสองคำที่ติดกันต้องเว้นวรรค (ไม่เช่นนั้น "PLC" + "alarm" จะกลายเป็นคำเดียว)
        both_latin = re.search("[A-Za-z]$", text) and re.match("[A-Za-z]", part)
        text += (" " if both_latin else sep) + part
    return text


def make_work_orders(n, rng, label_noise=0.03):
    labels = list(CLASS_SPEC)
    weights = [CLASS_SPEC[c]["weight"] for c in labels]
    start = pd.Timestamp("2023-01-01")
    rows = []
    for _ in range(n):
        label = rng.choice(labels, p=weights)
        text = make_work_order(label, rng)
        if rng.random() < label_noise:                       # ป้ายผิดจากการติดป้ายด้วยมือ
            label = rng.choice([c for c in labels if c != label])
        rows.append({"date": start + pd.Timedelta(days=int(rng.integers(0, 730))), "text": text, "label": label})
    return pd.DataFrame(rows).sort_values("date").reset_index(drop=True)


orders = make_work_orders(600, np.random.default_rng(RANDOM_SEED))
orders["tokens"] = orders["text"].map(tokenize_th)
print(orders["label"].value_counts().to_dict())
orders.head(8)
```

### 4.3 ค้นหาใบสั่งงานเก่าที่คล้ายกัน

ระบบค้นหาแบบ TFIDF-SEARCH ในหนังสือใช้ได้กับใบสั่งงานทั้ง 600 ใบโดยตรง ใช้ `TfidfVectorizer` ร่วมกับ `tokenize_th` เป็นตัวตัดคำ แล้วเรียงตามความคล้ายโคไซน์

```python
search_vectorizer = TfidfVectorizer(tokenizer=tokenize_th, token_pattern=None, lowercase=False)
X_orders = search_vectorizer.fit_transform(orders["text"])
print("ขนาดเมทริกซ์เอกสาร-คำ:", X_orders.shape, f"| ค่าที่ไม่เป็นศูนย์ {X_orders.nnz / np.prod(X_orders.shape):.1%}")


def search_orders(query, top_k=5):
    scores = cosine_similarity(search_vectorizer.transform([query]), X_orders).ravel()
    top = np.argsort(-scores, kind="stable")[:top_k]
    return orders.loc[top, ["date", "text", "label"]].assign(cos=scores[top].round(3))


search_orders("มอเตอร์สั่น เสียงดัง")
```

```python
# คำที่ได้น้ำหนัก idf ต่ำสุดและสูงสุดในคลังใบสั่งงานจำลอง
idf_table = pd.Series(search_vectorizer.idf_, index=search_vectorizer.get_feature_names_out()).sort_values()
print("idf ต่ำสุด (พบบ่อย):", idf_table.head(6).round(2).to_dict())
print("idf สูงสุด (พบน้อย):", idf_table.tail(6).round(2).to_dict())
```

**คำถามชวนคิด** ลองค้นด้วย `search_orders("มอเตอร์อุณหภูมิสูง")` และ `search_orders("มอเตอร์ร้อน")` ผลต่างกันอย่างไร ทั้งที่ช่างหมายถึงอาการเดียวกัน ข้อนี้คือข้อจำกัดข้อที่ 2 ของการนับคำในหัวข้อ 7.3.1 (ไม่รู้ว่าคำต่างกันแต่ความหมายเหมือนกัน)

## ส่วนที่ 5 Skip-gram และเวกเตอร์ฝังคำ

### 5.1 คู่ฝึกของ Skip-gram (ตัวอย่างที่ 7.8)

Word2Vec แบบ Skip-gram (หัวข้อ 7.3.2) ใช้คำกลางทายคำรอบข้างในหน้าต่างขนาด $m$ คู่ฝึกจึงสร้างได้จากข้อความโดยไม่ต้องติดป้าย

```python
def skipgram_pairs(tokens, window):
    """รหัสเทียม SKIPGRAM-PAIRS (ดัชนีเริ่มที่ 0)"""
    pairs = []
    for i, center in enumerate(tokens):
        for j in range(max(0, i - window), min(len(tokens), i + window + 1)):
            if j != i:
                pairs.append((center, tokens[j]))
    return pairs


sentence = ["เบรกเกอร์", "ทริป", "เพราะ", "มอเตอร์", "ลัดวงจร"]
for m in (1, 2):
    pairs = skipgram_pairs(sentence, m)
    per_center = [sum(c == w for c, _ in pairs) for w in sentence]
    print(f"m = {m}: {len(pairs)} คู่ (ต่อคำกลาง {per_center})")
print("คู่ของ m = 1:", skipgram_pairs(sentence, 1))
```

### 5.2 ความคล้ายของเวกเตอร์ฝังคำ (ตัวอย่างที่ 7.9)

ใช้เวกเตอร์ 3 มิติที่สมมติในหนังสือ เวกเตอร์ของข้อความคือค่าเฉลี่ยของเวกเตอร์ทุกคำ

```python
embedding = {
    "ร้อน": np.array([0.8, 0.1, 0.3]),
    "อุณหภูมิสูง": np.array([0.7, 0.2, 0.4]),
    "สั่น": np.array([0.1, 0.9, 0.2]),
    "มอเตอร์": np.array([0.3, 0.4, 0.8]),
}


def sentence_vector(tokens):
    return np.mean([embedding[t] for t in tokens], axis=0)


for a, b in [("ร้อน", "อุณหภูมิสูง"), ("ร้อน", "สั่น"), ("อุณหภูมิสูง", "สั่น")]:
    print(f"cos({a}, {b}) = {cosine(embedding[a], embedding[b]):.3f}")

query_vec = sentence_vector(["มอเตอร์", "อุณหภูมิสูง"])
for name, toks in [("A มอเตอร์ ร้อน", ["มอเตอร์", "ร้อน"]), ("B มอเตอร์ สั่น", ["มอเตอร์", "สั่น"])]:
    vec = sentence_vector(toks)
    print(f"คำค้นกับ {name}: เวกเตอร์ {vec.round(3)} | cos = {cosine(query_vec, vec):.3f}")
print("หนังสือ: 0.980, 0.288, 0.428 | A 0.995, B 0.843 | BoW ให้ 0.5 เท่ากันทั้งคู่")
```

### 5.3 เวกเตอร์จากบริบทของใบสั่งงานจำลอง (สมมติฐานการกระจาย)

Word2Vec ต้องฝึกด้วยข้อความจำนวนมาก แต่แนวคิดเบื้องหลัง "คำที่อยู่ในบริบทคล้ายกันมีความหมายคล้ายกัน" ทดลองได้ทันทีด้วยวิธีนับ คือนับว่าแต่ละคำปรากฏใกล้คำอื่นกี่ครั้งในหน้าต่าง $m = 2$ (ใช้ `skipgram_pairs` ชุดเดียวกัน) แปลงเป็น **PPMI** (Positive Pointwise Mutual Information) ซึ่งลดน้ำหนักของคำที่พบบ่อยทุกที่ แล้วลดมิติด้วย SVD เหลือ 10 มิติ (Jurafsky & Martin, 2025) ผลคือเวกเตอร์หนาแน่นของแต่ละคำที่เรียนจากข้อมูล โดยไม่มีใครบอกว่า "ร้อน" กับ "อุณหภูมิสูง" เป็นคำพ้องความหมาย

```python
def ppmi_embeddings(token_lists, window=2, dims=10, min_count=5, seed=RANDOM_SEED):
    counts = Counter(t for toks in token_lists for t in toks)
    vocab = [t for t, c in counts.items() if c >= min_count]
    index = {t: i for i, t in enumerate(vocab)}
    co = np.zeros((len(vocab), len(vocab)))
    for toks in token_lists:
        for center, context in skipgram_pairs(toks, window):
            if center in index and context in index:
                co[index[center], index[context]] += 1
    total = co.sum()
    expected = co.sum(axis=1, keepdims=True) * co.sum(axis=0, keepdims=True) / total
    with np.errstate(divide="ignore", invalid="ignore"):
        pmi = np.log(co / expected)
    ppmi = np.where(co > 0, np.maximum(pmi, 0), 0.0)
    vectors = TruncatedSVD(n_components=dims, random_state=seed).fit_transform(ppmi)
    return {t: vectors[index[t]] for t in vocab}


learned = ppmi_embeddings(orders["tokens"])
for a, b in [("ร้อน", "อุณหภูมิสูง"), ("สั่น", "เสียงดัง"), ("ทริป", "ขาด"), ("plc", "hmi"),
             ("ทริป", "หลวม"), ("plc", "สั่น"), ("ฟิวส์", "ลูกปืน"), ("ร้อน", "สั่น")]:
    if a in learned and b in learned:
        print(f"cos({a}, {b}) = {cosine(learned[a], learned[b]):+.3f}")
```

คำที่ใช้แทนกันได้ในประโยคเดียวกัน (ร้อน กับ อุณหภูมิสูง) และอาการของหมวดเดียวกัน (สั่น กับ เสียงดัง, ทริป กับ ขาด) ได้ความคล้ายสูง ส่วนคำต่างหมวด (ทริป กับ หลวม, ฟิวส์ กับ ลูกปืน) ได้ค่าต่ำ แต่ "ร้อน" กับ "สั่น" ก็ได้ค่าค่อนข้างสูง เพราะในข้อมูลจำลอง "ร้อน" เป็นอาการของทั้งงานไฟฟ้าและงานเครื่องกล จึงมีบริบทร่วมกับ "สั่น" มาก นี่คือข้อเตือนว่าเวกเตอร์ฝังคำวัด "การใช้ในบริบทคล้ายกัน" ไม่ใช่ "ความหมายเหมือนกัน" โดยตรง ระบบจริงใช้คลังข้อความใหญ่กว่านี้มาก และมีเวกเตอร์ฝังคำภาษาไทยที่ฝึกไว้แล้วให้ดาวน์โหลด (เช่น ผ่าน PyThaiNLP) แต่ศัพท์เฉพาะโรงงานอาจไม่มีหรือมีน้อย

**คำถามชวนคิด** เวกเตอร์แบบนี้มีเวกเตอร์เดียวต่อคำ ใบสั่งงาน "ไม่ร้อน" จะได้เวกเตอร์ข้อความ (ค่าเฉลี่ย) ใกล้กับ "ร้อน" หรือไม่ และกลไก Attention ในส่วนที่ 6 แก้ปัญหานี้อย่างไร

## ส่วนที่ 6 Attention, Transformer และโมเดลภาษา

### 6.1 Scaled Dot-Product Attention (ตัวอย่างที่ 7.10)

$\text{Attention}(Q, K, V) = \text{softmax}(QK^\top / \sqrt{d_k})\,V$ (หัวข้อ 7.4.1) ซอฟต์แมกซ์คำนวณทีละแถว ฟังก์ชันด้านล่างลบค่าสูงสุดของแถวก่อนยกกำลังเพื่อป้องกันตัวเลขล้น ซึ่งไม่เปลี่ยนผลลัพธ์

```python
def softmax(scores, axis=-1, temperature=1.0):
    z = np.asarray(scores, dtype=float) / temperature
    z = z - z.max(axis=axis, keepdims=True)
    e = np.exp(z)
    return e / e.sum(axis=axis, keepdims=True)


def scaled_dot_product_attention(Q, K, V, scale=True):
    """คืน (Z, A) โดย A คือเมทริกซ์น้ำหนักความสนใจ n × n"""
    d_k = K.shape[-1]
    scores = Q @ K.T / (np.sqrt(d_k) if scale else 1.0)
    A = softmax(scores, axis=-1)
    return A @ V, A


tokens_710 = ["เบรกเกอร์", "ร้อน", "มัน"]
K_710 = np.array([[1.0, 0.0], [0.0, 1.0], [0.5, 0.5]])
V_710 = K_710.copy()
q3 = np.array([[2.0, 0.0]])          # Query ของโทเคนที่ 3 "มัน"

for scale in (True, False):
    z3, a3 = scaled_dot_product_attention(q3, K_710, V_710, scale=scale)
    label = "ปรับสเกล" if scale else "ไม่ปรับสเกล"
    print(f"{label:11s}: น้ำหนัก {a3.round(3).ravel()} | z3 = {z3.round(3).ravel()}")
print("หนังสือ: ปรับสเกล 0.576, 0.140, 0.284 และ z3 = (0.718, 0.282) | ไม่ปรับสเกล 0.665, 0.090, 0.245")
```

```python
# ทำไมต้องหารด้วย sqrt(d_k): คะแนนของเวกเตอร์สุ่มมิติสูงกระจายกว้าง ซอฟต์แมกซ์จึงอิ่มตัว
attn_rng = np.random.default_rng(RANDOM_SEED)
rows = []
for d_k in (2, 16, 64, 256):
    q = attn_rng.standard_normal((500, d_k))
    k = attn_rng.standard_normal((10, d_k))
    raw_scores = q @ k.T
    rows.append({
        "d_k": d_k,
        "ความแปรปรวนของ q·k": raw_scores.var().round(1),
        "น้ำหนักสูงสุดเฉลี่ย (ไม่ปรับสเกล)": softmax(raw_scores).max(axis=1).mean().round(3),
        "น้ำหนักสูงสุดเฉลี่ย (ปรับสเกล)": softmax(raw_scores / np.sqrt(d_k)).max(axis=1).mean().round(3),
    })
pd.DataFrame(rows)
```

เมื่อ $d_k$ เพิ่มขึ้น ความแปรปรวนของผลคูณจุดเพิ่มตาม $d_k$ และถ้าไม่ปรับสเกล น้ำหนักเกือบทั้งหมดจะไปอยู่ที่โทเคนเดียว (ค่าเข้าใกล้ 1) การหารด้วย $\sqrt{d_k}$ ทำให้การกระจายของน้ำหนักแทบไม่ขึ้นกับมิติ (ส่วน **มาสก์เชิงเหตุภาพ** ที่ห้ามมองโทเคนในอนาคตอยู่ในแบบฝึกปฏิบัติข้อ 3)

### 6.2 นับพารามิเตอร์ของบล็อก Transformer (ตัวอย่างที่ 7.11)

ใช้วิธีนับเดียวกับ หัวข้อ 7.4.2: $W_Q, W_K, W_V, W_O$ พร้อมไบแอส + FFN สองชั้น + Layer Normalization สองชุด

```python
def transformer_block_params(d_model, d_ff):
    attention = 4 * (d_model * d_model + d_model)
    ffn = (d_model * d_ff + d_ff) + (d_ff * d_model + d_model)
    layer_norm = 2 * (2 * d_model)
    return {"Attention": attention, "FFN": ffn, "LayerNorm": layer_norm, "รวม": attention + ffn + layer_norm}


base = transformer_block_params(512, 2048)
print({k: f"{v:,}" for k, v in base.items()})
print(f"ตัวเข้ารหัส 6 ชั้น: {6 * base['รวม']:,} (หนังสือ 18,914,304) | สัดส่วน FFN {base['FFN'] / base['รวม']:.1%}")
for n in (1_000, 2_000, 8_000):
    print(f"ข้อความ {n:,} โทเคน: เมทริกซ์ความสนใจ {n * n:,} ค่าต่อหัวต่อชั้น")
```

### 6.3 โมเดลภาษา อุณหภูมิ และความเสี่ยงสะสม (ตัวอย่างที่ 7.12–7.14)

โมเดลภาษาให้ลอจิตของโทเคนถัดไป แล้วแปลงเป็นความน่าจะเป็นด้วย $p_k = e^{z_k/T} / \sum_j e^{z_j/T}$ (หัวข้อ 7.4.3) เซลล์ถัดไปคำนวณตัวอย่างที่ 7.12 แล้วสุ่มโทเคน 1,000 ครั้งที่แต่ละอุณหภูมิ

```python
next_tokens = ["เกิน", "ไหล", "ต่ำ", "มอเตอร์"]
logits = np.array([3.0, 1.5, 0.5, -1.0])
temperatures = [0.5, 1.0, 2.0]

table = pd.DataFrame({f"T = {t}": softmax(logits, temperature=t) for t in temperatures}, index=next_tokens)
print(f"ครอสเอนโทรปีเมื่อโทเคนจริงคือ 'เกิน' (T = 1): {-np.log(table.loc['เกิน', 'T = 1.0']):.3f} (หนังสือ 0.280)")
sample_rng = np.random.default_rng(RANDOM_SEED)
for t in temperatures:
    draws = sample_rng.choice(len(next_tokens), size=1000, p=softmax(logits, temperature=t))
    table[f"สุ่ม 1,000 ครั้ง (T = {t})"] = np.bincount(draws, minlength=len(next_tokens))
display(table.round(4))

fig, ax = plt.subplots(figsize=(7, 3))
width = 0.25
for i, t in enumerate(temperatures):
    ax.bar(np.arange(4) + (i - 1) * width, softmax(logits, temperature=t), width, label=f"T = {t}")
ax.set_xticks(np.arange(4), ["kern (exceed)", "lai (flow)", "tam (low)", "motor"])
ax.set_ylabel("probability")
ax.set_title("Example 7.12: next-token probability vs temperature")
ax.legend()
plt.show()
```

```python
# ตัวอย่างที่ 7.14: ความน่าจะเป็นที่ทุกข้อเท็จจริงถูกต้อง เมื่อแต่ละข้อถูก 0.98 และเป็นอิสระต่อกัน
n_facts = np.arange(1, 51)
for n in (1, 10, 20, 50):
    print(f"{n:2d} ข้อเท็จจริง: ถูกครบ {0.98 ** n:.3f} | มีข้อผิดอย่างน้อยหนึ่งข้อ {1 - 0.98 ** n:.3f}")

# ตัวอย่างที่ 7.13 และ LoRA ในหัวข้อ 7.4.3
for name, params in [("7,000 ล้าน", 7e9), ("175,000 ล้าน", 175e9)]:
    print(f"โมเดล {name} พารามิเตอร์: 16 บิต {params * 2 / 1e9:,.1f} GB | 4 บิต {params * 0.5 / 1e9:,.1f} GB")
d, r = 4096, 8
print(f"LoRA r = {r}: ฝึก {d * r + r * d:,} จาก {d * d:,} ค่า ({(d * r + r * d) / (d * d):.2%})")
```

**คำถามชวนคิด** ที่ $T = 2$ มีโอกาสราวร้อยละ 7 ที่โมเดลจะต่อข้อความเป็น "กระแสมอเตอร์" ถ้าระบบของคุณใช้ LLM สกัดค่ากระแสจากรายงานผลทดสอบลงตาราง ควรตั้งอุณหภูมิเท่าใด และอุณหภูมิต่ำช่วยลด Hallucination ได้ทั้งหมดหรือไม่ (เปรียบเทียบกับตัวอย่างที่ 7.14 และ หัวข้อ 7.4.4)

`MultinomialNB(alpha=1.0)` ของ scikit-learn คือ Naive Bayes แบบพหุนามพร้อมการปรับเรียบแบบลาปลาซ (บวก 1) ตามสูตรใน หัวข้อ 7.5.1 `CountVectorizer` ให้จำนวนนับดิบ และคำที่ไม่อยู่ในคลังศัพท์ (เช่น "มอเตอร์") ถูกละทิ้งโดยอัตโนมัติ (การเขียน Naive Bayes เองอยู่ในแบบฝึกปฏิบัติข้อ 2)

### 7.2 ฝึกและวัดผลบนใบสั่งงานจำลอง 600 ใบ

หัวข้อ 7.5.1 เน้นให้ **แบ่งข้อมูลตามเวลา** เพื่อจำลองการใช้งานจริง (ฝึกด้วยอดีต ทดสอบกับใบที่เข้ามาภายหลัง) ใช้ใบสั่งงานก่อน 1 กรกฎาคม 2024 เป็นชุดฝึก ที่เหลือเป็นชุดทดสอบ แล้วเปรียบเทียบสองตัวจำแนกบน TF-IDF เดียวกัน

```python
split_date = pd.Timestamp("2024-07-01")
train, test = orders[orders["date"] < split_date], orders[orders["date"] >= split_date]
print(f"ชุดฝึก {len(train)} ใบ ({train['date'].min():%Y-%m} ถึง {train['date'].max():%Y-%m}) | "
      f"ชุดทดสอบ {len(test)} ใบ ({test['date'].min():%Y-%m} ถึง {test['date'].max():%Y-%m})")

clf_vectorizer = TfidfVectorizer(tokenizer=tokenize_th, token_pattern=None, lowercase=False)
X_train = clf_vectorizer.fit_transform(train["text"])
X_test = clf_vectorizer.transform(test["text"])
CLASS_ORDER = ["E", "M", "C"]

models = {
    "MultinomialNB": MultinomialNB(alpha=1.0),
    "LogisticRegression": LogisticRegression(max_iter=2000, random_state=RANDOM_SEED),
}
predictions = {}
for name, model in models.items():
    model.fit(X_train, train["label"])
    predictions[name] = model.predict(X_test)
    print(f"\n=== {name} ===")
    print(classification_report(test["label"], predictions[name], labels=CLASS_ORDER, digits=3, zero_division=0))
```

```python
def per_class_metrics(cm, labels):
    """ตัววัดรายคลาสจากเมทริกซ์ความสับสน (แถว = จริง, คอลัมน์ = ทำนาย) แบบตัวอย่างที่ 7.16"""
    cm = np.asarray(cm)
    tp = np.diag(cm)
    fp = cm.sum(axis=0) - tp
    fn = cm.sum(axis=1) - tp
    rows = [dict(zip(["ความแม่นยำ", "ความระลึก", "F1"], precision_recall_f1(*v))) for v in zip(tp, fp, fn)]
    table = pd.DataFrame(rows, index=labels).assign(TP=tp, FP=fp, FN=fn)
    summary = {"ความถูกต้อง": tp.sum() / cm.sum(), "F1 มหภาค": table["F1"].mean()}
    return table[["TP", "FP", "FN", "ความแม่นยำ", "ความระลึก", "F1"]].round(3), summary


# ตรวจด้วยเมทริกซ์ความสับสนของตัวอย่างที่ 7.16
cm_716 = [[17, 2, 1], [3, 16, 1], [2, 1, 7]]
table_716, summary_716 = per_class_metrics(cm_716, CLASS_ORDER)
display(table_716)
print({k: round(v, 3) for k, v in summary_716.items()}, "(หนังสือ: ความถูกต้อง 0.800, F1 มหภาค 0.789)")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(9, 3.6))
for ax, (name, pred) in zip(axes, predictions.items()):
    cm = confusion_matrix(test["label"], pred, labels=CLASS_ORDER)
    ConfusionMatrixDisplay(cm, display_labels=CLASS_ORDER).plot(ax=ax, colorbar=False, cmap="Blues")
    _, summary = per_class_metrics(cm, CLASS_ORDER)
    ax.set_title(f"{name}\nacc = {summary['ความถูกต้อง']:.3f}, macro-F1 = {summary['F1 มหภาค']:.3f}")
    ax.set_xlabel("predicted")
    ax.set_ylabel("true")
plt.tight_layout()
plt.show()
```

```python
# คำที่มีน้ำหนักสูงสุดของแต่ละคลาสใน Logistic Regression (อธิบายได้ว่าโมเดลตัดสินจากคำใด)
logreg = models["LogisticRegression"]
features = clf_vectorizer.get_feature_names_out()
for c, coef in zip(logreg.classes_, logreg.coef_):
    top = np.argsort(-coef)[:6]
    print(f"{c}: {', '.join(features[top])}")

# ส่งใบที่โมเดลไม่มั่นใจให้หัวหน้าช่างตรวจ (รูปที่ 7.11)
THRESHOLD = 0.6
proba_test = logreg.predict_proba(X_test)
confident = proba_test.max(axis=1) >= THRESHOLD
auto_acc = (logreg.classes_[proba_test.argmax(axis=1)][confident] == test["label"].to_numpy()[confident]).mean()
print(f"\nเกณฑ์ {THRESHOLD}: ส่งแผนกอัตโนมัติ {confident.mean():.1%} ของใบ ความถูกต้องของส่วนนี้ {auto_acc:.3f}"
      f" | ส่งคนตรวจ {(~confident).sum()} ใบ")
display(test.loc[~confident, ["text", "label"]].head(6))
```

### 8.2 ค้นคืนสามแบบ: คำสำคัญ ระดับอักขระ และแบบผสม (ตัวอย่างที่ 7.18)

- **คำสำคัญ** (`s_kw`): TF-IDF บนโทเคนจาก `tokenize_th` จับรหัสอย่าง OV2 ได้ตรงตัว แต่ไม่รู้คำพ้อง
- **ระดับอักขระ** (`s_vec`): TF-IDF บนชุดอักขระติดกัน 2–4 ตัว (`analyzer="char_wb"`) ไม่ต้องตัดคำ จึงทนต่อการสะกดต่างกันและการตัดคำผิด notebook นี้ใช้แทน "เวกเตอร์ฝังข้อความ" เพราะไม่ต้องดาวน์โหลดโมเดล (ส่วนทางเลือก 8.5 แสดงการใช้โมเดลฝังข้อความจริง)
- **แบบผสม**: $\text{score} = \alpha \cdot s_{\text{vec}} + (1 - \alpha) \cdot s_{\text{kw}}$

ตรวจตัวอย่างที่ 7.18 ก่อน แล้วสร้างฟังก์ชันค้นคืนบนคู่มือจำลอง

```python
def hybrid_score(s_vec, s_kw, alpha):
    return alpha * np.asarray(s_vec) + (1 - alpha) * np.asarray(s_kw)


s_vec_718, s_kw_718 = np.array([0.62, 0.80]), np.array([1.00, 0.0])   # ชิ้น A, B
for alpha in (1.0, 0.5):
    scores = hybrid_score(s_vec_718, s_kw_718, alpha)
    print(f"α = {alpha}: A = {scores[0]:.2f}, B = {scores[1]:.2f} → เลือก {'AB'[int(np.argmax(scores))]}")

kw_vectorizer = TfidfVectorizer(tokenizer=tokenize_th, token_pattern=None, lowercase=False)
char_vectorizer = TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 4), preprocessor=normalize_text)
X_kw = kw_vectorizer.fit_transform(manual["text"])
X_char = char_vectorizer.fit_transform(manual["text"])


def retrieval_scores(question, alpha=0.5):
    s_kw = cosine_similarity(kw_vectorizer.transform([question]), X_kw).ravel()
    s_vec = cosine_similarity(char_vectorizer.transform([question]), X_char).ravel()
    return {"keyword": s_kw, "char": s_vec, "hybrid": hybrid_score(s_vec, s_kw, alpha)}


def retrieve(question, k=3, method="hybrid", alpha=0.5):
    scores = retrieval_scores(question, alpha)[method]
    top = np.argsort(-scores, kind="stable")[:k]
    return manual.iloc[top][["section", "page", "title"]].assign(score=scores[top].round(3))


retrieve("OV2 คืออะไร")
```

### 8.3 ประกอบพรอมต์และตอบพร้อมแหล่งอ้างอิง

notebook นี้ไม่เรียก LLM ภายนอก จึงแสดงสองส่วนที่สำคัญที่สุดของ RAG แทน คือ (1) **พรอมต์** ที่จะส่งให้ LLM ตามรหัสเทียม RAG-ANSWER และ (2) **ตัวตอบแบบสกัด (Extractive)** ที่เลือกประโยคจากชิ้นที่ค้นได้มาแสดงตรง ๆ พร้อมเลขหัวข้อและหน้า ซึ่งไม่มีทางหลอน เพราะไม่สร้างข้อความใหม่ ตัวตอบมีกลไกป้องกันสองข้อตาม หัวข้อ 7.5.3 คือ ตอบว่า "ไม่พบในเอกสาร" เมื่อคำสำคัญของคำถามไม่ปรากฏในชิ้นที่ค้นได้อย่างน้อยครึ่งหนึ่ง และตอบด้วยข้อความมาตรฐานเมื่อคำถามเกี่ยวกับการข้ามระบบความปลอดภัย

```python
MIN_COVERAGE = 0.5   # สัดส่วนขั้นต่ำของคำสำคัญในคำถามที่ต้องพบในชิ้นที่ค้นได้
QUESTION_WORDS = {"คือ", "อะไร", "อย่างไร", "ยังไง", "ไหน", "เท่าไร", "กี่", "ต้อง", "หมายถึง", "ได้", "ไม่", "ที่",
                  "ของ", "ควร", "เป็น", "วิธี", "ทำ", "บ่อย", "รหัส", "แก้", "ขึ้น", "ตอน", "หลัง", "ก่อน", "นาน",
                  "จะ", "ไหม"}
SAFETY_PATTERNS = ["ปลดอินเทอร์ล็อก", "ข้ามอินเทอร์ล็อก", "bypass", "ทำงานขณะมีไฟ", "ไม่ต้องตัดไฟ"]
SAFETY_MESSAGE = ("คำถามนี้เกี่ยวกับการข้ามระบบความปลอดภัยหรือการทำงานขณะมีไฟ ระบบไม่ให้คำแนะนำ "
                  "โปรดปฏิบัติตาม SOP ที่ได้รับอนุมัติ และปรึกษาผู้ควบคุมงาน")


def build_prompt(question, chunks):
    context = "\n".join(f"[{r.section} หน้า {r.page}] {r.title}: {r.body}" for r in chunks.itertuples())
    return ("คุณเป็นผู้ช่วยค้นคู่มือการซ่อมบำรุง ตอบเป็นภาษาไทยจากข้อความอ้างอิงด้านล่างเท่านั้น "
            "ระบุเลขหัวข้อและหน้าทุกประโยค ถ้าไม่พบคำตอบให้ตอบว่า \"ไม่พบในเอกสาร\"\n\n"
            f"ข้อความอ้างอิง:\n{context}\n\nคำถาม: {question}\nคำตอบ:")


def extractive_answer(question, k=2, alpha=0.5):
    if any(p in normalize_text(question) for p in SAFETY_PATTERNS):
        return SAFETY_MESSAGE, None
    scores = retrieval_scores(question, alpha)["hybrid"]
    top = np.argsort(-scores, kind="stable")[:k]
    chunks = manual.iloc[top]
    # คะแนนโคไซน์อย่างเดียวแยกคำถามนอกขอบเขตไม่ได้ในคลังเอกสารเล็ก ๆ (คำทั่วไปอย่าง "กี่ ปี" ก็ทำให้ได้คะแนน)
    # จึงตรวจเพิ่มว่าคำสำคัญของคำถามปรากฏในชิ้นที่ค้นได้มากพอหรือไม่
    content = [t for t in tokenize_th(question) if t not in QUESTION_WORDS]
    chunk_words = {t for text in chunks["text"] for t in tokenize_th(text)}
    missing = [t for t in content if t not in chunk_words]
    coverage = 1 - len(missing) / len(content) if content else 0.0
    if coverage < MIN_COVERAGE:
        return f"ไม่พบในเอกสาร (คำสำคัญที่ไม่พบ: {', '.join(missing)})", None
    q_tokens = set(content)
    # เลือกชิ้นที่มีคำร่วมกับคำถามมากที่สุดจาก k ชิ้น แล้วแสดงข้อความต้นฉบับทั้งชิ้นพร้อมแหล่งอ้างอิง
    source = max(chunks.itertuples(), key=lambda r: len(q_tokens & set(tokenize_th(r.text))))
    answer = f"{source.body} [คู่มือ VF-200 หัวข้อ {source.section} หน้า {source.page}]"
    return answer, build_prompt(question, chunks)


question = "อินเวอร์เตอร์ขึ้นแรงดันเกินตอนหยุดมอเตอร์ แก้อย่างไร"
answer, prompt = extractive_answer(question)
print("คำถาม:", question)
print("คำตอบ:", answer)
print("\n--- พรอมต์ที่จะส่งให้ LLM ---\n" + prompt)
```

### 8.4 วัดผลการค้นคืนด้วย Recall@k และ MRR (ตัวอย่างที่ 7.19)

ชุดคำถามทดสอบ 14 ข้อด้านล่างเขียนแบบที่ช่างถามจริง แต่ละข้อระบุหัวข้อที่มีคำตอบ บางข้อใช้รหัสตรงตัว บางข้อใช้คำที่ต่างจากคู่มือ (เช่น "หยุดมอเตอร์" แทน "ลดความเร็ว")

```python
def recall_at_k_and_mrr(ranks, ks=(1, 3, 5)):
    """ranks: อันดับของชิ้นที่ถูกต้อง (เริ่มที่ 1) หรือ None ถ้าไม่พบ"""
    recalls = {f"Recall@{k}": np.mean([r is not None and r <= k for r in ranks]) for k in ks}
    mrr = np.mean([1 / r if r is not None else 0.0 for r in ranks])
    return {**recalls, "MRR": mrr}


# ตรวจกับตัวอย่างที่ 7.19
print({k: round(v, 3) for k, v in recall_at_k_and_mrr([1, 3, 1, None, 2], ks=(1, 3)).items()},
      "(หนังสือ 0.400, 0.800, 0.567)")

TEST_QUESTIONS = [
    ("OV2 คืออะไร", "6.2"),
    ("อินเวอร์เตอร์ขึ้นแรงดันเกินตอนหยุดมอเตอร์ แก้อย่างไร", "6.2"),
    ("ตั้งเวลาลดความเร็วที่พารามิเตอร์ไหน", "4.2"),
    ("ต่อตัวต้านทานเบรกยังไง", "2.5"),
    ("รหัส OC1 หมายถึงอะไร", "6.3"),
    ("ฮีตซิงก์ร้อนเกิน ต้องตรวจอะไร", "6.5"),
    ("มอเตอร์โอเวอร์โหลดบ่อย", "6.6"),
    ("หลังปิดไฟต้องรอนานเท่าไรก่อนเปิดฝา", "1.2"),
    ("เปลี่ยนพัดลมระบายความร้อนกี่ปี", "7.1"),
    ("สายไฟกำลังขนาดเท่าไร", "2.3"),
    ("กระแสรั่วลงดิน GF", "6.7"),
    ("ไฟเข้าขาดเฟส", "6.8"),
    ("modbus สื่อสารไม่ได้", "4.6"),
    ("เครื่องตัดตอนจ่ายไฟเข้า แรงดันต่ำ", "6.4"),
]


def rank_of_gold(question, gold_section, method, alpha=0.5):
    scores = retrieval_scores(question, alpha)[method]
    order = manual["section"].to_numpy()[np.argsort(-scores, kind="stable")]
    return int(np.where(order == gold_section)[0][0]) + 1


rows = {}
for method in ["keyword", "char", "hybrid"]:
    ranks = [rank_of_gold(q, gold, method) for q, gold in TEST_QUESTIONS]
    rows[method] = {**recall_at_k_and_mrr(ranks), "อันดับรายข้อ": ranks}
pd.DataFrame(rows).T
```

```python
# เลือก α ด้วยชุดคำถามทดสอบ (หัวข้อ 7.5.3 ขั้นที่ 3) แทนการตั้งตามความรู้สึก
alphas = np.round(np.linspace(0, 1, 11), 1)
mrr_by_alpha = [recall_at_k_and_mrr([rank_of_gold(q, g, "hybrid", a) for q, g in TEST_QUESTIONS])["MRR"]
                for a in alphas]
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(alphas, mrr_by_alpha, marker="o")
ax.set_xlabel("alpha (0 = keyword only, 1 = char vectors only)")
ax.set_ylabel("MRR")
ax.set_title("Hybrid search: MRR on 14 test questions")
plt.show()
print("α ที่ให้ MRR สูงสุด:", alphas[int(np.argmax(mrr_by_alpha))])
```

**คำถามชวนคิด** คำถามข้อใดที่การค้นด้วยคำสำคัญอย่างเดียวพลาด (อันดับมากกว่า 1) และเพราะอะไร ชุดคำถามทดสอบ 14 ข้อเล็กเกินไปที่จะเลือก $\alpha$ ได้อย่างมั่นใจ คุณจะเก็บชุดคำถามทดสอบที่ใหญ่และเป็นตัวแทนของคำถามจริงได้อย่างไร และควรระวังอะไรถ้าคนเขียนคำถามทดสอบเป็นคนเดียวกับคนที่ปรับระบบ

### 8.5 (ทางเลือก) โมเดลฝังข้อความและ LLM ขนาดเล็กจาก Hugging Face

### ข้อ 3 Self-attention แบบมีมาสก์เชิงเหตุภาพ

เขียน `causal_attention(Q, K, V)` ที่คืน `(Z, A)` เหมือน `scaled_dot_product_attention` แต่กำหนดคะแนน $S_{ij} = -\infty$ สำหรับทุก $j > i$ ก่อนซอฟต์แมกซ์ (รหัสเทียม SELF-ATTENTION) เพื่อไม่ให้โทเคนมองโทเคนในอนาคต
