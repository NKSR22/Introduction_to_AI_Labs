### 1.1 ผังทางวิ่งของ AGV

หุ่นยนต์ลำเลียงอัตโนมัติ (AGV) ต้องขนวัตถุดิบจากคลังวัตถุดิบ **S** ไปยังสถานีประกอบ **G** ผังทางวิ่งในรูปที่ 2.4 ของหนังสือมี 7 จุด ได้แก่ S, A, B, C, X, Y และ G ทางวิ่งวิ่งได้ทั้งสองทาง (กราฟไม่มีทิศทาง) และตัวเลขบนเส้นเชื่อมคือระยะทางเป็นเมตร

ในโค้ดเก็บผังเป็น **dict ซ้อน dict** คือ `FACTORY_MAP[จุด][จุดข้างเคียง] = ระยะทาง` ซึ่งตรงกับตารางที่ 2.2 และเก็บพิกัด $(x, y)$ ของแต่ละจุดไว้ใช้วาดรูป (พิกัดชุดเดียวกับตัวอย่างที่ 2.12 ซึ่งจะใช้คำนวณฮิวริสติกในปฏิบัติการที่ 4)

```python
# ผังโรงงาน (ตารางที่ 2.2): จุด -> {จุดข้างเคียง: ระยะทาง (m)} เรียงจุดข้างเคียงตามตัวอักษร
FACTORY_MAP = {
    "S": {"A": 20, "C": 25, "X": 20},
    "A": {"B": 20, "S": 20},
    "B": {"A": 20, "X": 20},
    "C": {"S": 25, "Y": 30},
    "X": {"B": 20, "G": 40, "S": 20, "Y": 15},
    "Y": {"C": 30, "G": 10, "X": 15},
    "G": {"X": 40, "Y": 10},
}

# พิกัดของแต่ละจุด (m) ตามตัวอย่างที่ 2.12 ใช้วาดผังและคำนวณฮิวริสติก
COORDS = {"S": (5, 27), "A": (4, 42), "B": (19, 43), "C": (5, 3),
          "X": (24, 27), "Y": (30, 15), "G": (40, 15)}

START, GOAL = "S", "G"


# ตรวจว่าผังเป็นกราฟไม่มีทิศทางจริง: ถ้า u ไป v ได้ระยะ w แล้ว v ต้องไป u ได้ระยะ w เช่นกัน
for u, nbrs in FACTORY_MAP.items():
    for v, w in nbrs.items():
        assert FACTORY_MAP[v][u] == w, f"ทางวิ่ง {u}-{v} ไม่สมมาตร"

# แสดงเป็นตารางเดียวกับตารางที่ 2.2 ของหนังสือ
pd.DataFrame({
    "จุด": list(FACTORY_MAP),
    "จุดข้างเคียง (ระยะทาง m)": [", ".join(f"{v} ({w})" for v, w in nbrs.items())
                                 for nbrs in FACTORY_MAP.values()],
}).set_index("จุด")
```

### 1.2 วาดผังด้วย networkx

ฟังก์ชัน `draw_factory_map` สร้างกราฟ `networkx` จาก `FACTORY_MAP` แล้ววาดโดยวางแต่ละจุดตามพิกัดจริง ฟังก์ชันนี้จะใช้ซ้ำตลอดปฏิบัติการ โดยระบายจุดที่อัลกอริทึม **ขยาย** เป็นสีน้ำเงินทึบ และวาด **เส้นทางที่ได้** เป็นเส้นหนาสีส้ม

```python
def draw_factory_map(graph=None, expanded=(), path=(), title="", ax=None, node_text=None):
    """วาดผังโรงงาน: วงกลมทึบสีน้ำเงิน = จุดที่ถูกขยาย, เส้นหนาสีส้ม = เส้นทางที่ได้
    node_text: dict ข้อความเล็กใต้แต่ละจุด เช่น ค่าฮิวริสติก"""
    graph = FACTORY_MAP if graph is None else graph
    G = nx.Graph()
    G.add_nodes_from(graph)
    for u, nbrs in graph.items():
        for v, w in nbrs.items():
            G.add_edge(u, v, weight=w)
    if ax is None:
        _, ax = plt.subplots(figsize=(5.5, 5))

    on_path = {frozenset(e) for e in zip(path, path[1:])}
    edges = list(G.edges())
    nx.draw_networkx_edges(
        G, COORDS, ax=ax, edgelist=edges,
        edge_color=[COLORS["orange"] if frozenset(e) in on_path else COLORS["gray"] for e in edges],
        width=[4 if frozenset(e) in on_path else 1.5 for e in edges])
    nx.draw_networkx_edge_labels(G, COORDS, ax=ax, edge_labels=nx.get_edge_attributes(G, "weight"),
                                 font_size=8, font_color=COLORS["gray"])
    nodes = list(G.nodes())
    nx.draw_networkx_nodes(G, COORDS, ax=ax, nodelist=nodes, node_size=650,
                           node_color=[COLORS["blue"] if n in expanded else "white" for n in nodes],
                           edgecolors=COLORS["blue"], linewidths=2)
    for n in nodes:
        x, y = COORDS[n]
        ax.text(x, y, n, ha="center", va="center", fontsize=12, fontweight="bold",
                color="white" if n in expanded else COLORS["blue"])
        if node_text and n in node_text:
            ax.text(x, y - 3.5, f"({node_text[n]})", ha="center", va="top", fontsize=9, color=COLORS["gray"])
    ax.set_title(title)
    ax.set_aspect("equal")
    ax.margins(0.15)
    ax.axis("off")
    return ax


draw_factory_map(title="Factory AGV map (distances in m)")
plt.show()
```

### 1.3 ปัญหา โหนด และการขยายโหนด

ตามหัวข้อ 2.1.1 ปัญหาการค้นหาประกอบด้วยสถานะเริ่มต้น การกระทำ แบบจำลองการเปลี่ยนสถานะ การทดสอบเป้าหมาย และต้นทุนการกระทำ คลาส `RouteProblem` เขียนองค์ประกอบเหล่านี้เป็นเมธอด ข้อดีคืออัลกอริทึมการค้นหาเรียกใช้เพียงเมธอดเหล่านี้ จึงนำไปใช้กับปัญหาอื่นได้ทันทีถ้าปัญหานั้นมีเมธอดชื่อเดียวกัน (จะได้ลองในส่วนที่ 6)

โหนดของต้นไม้ค้นหา **ไม่ใช่สถานะ** แต่เป็นโครงสร้างข้อมูลที่เก็บ `state`, `parent`, `action`, `path_cost` ($g$) และ `depth` (หัวข้อ 2.2.1) สถานะเดียวกันจึงอาจมีหลายโหนดถ้าไปถึงได้หลายเส้นทาง ฟังก์ชัน `expand` สร้างโหนดลูกตามรหัสเทียม EXPAND และ `path_states` ย้อนตาม `parent` เพื่อหาเส้นทางคำตอบ

ผลลัพธ์ของทุกอัลกอริทึมเก็บใน `SearchResult` ซึ่งบันทึกโหนดที่ถูกขยาย สถานะที่ถูกทดสอบเป้าหมาย และโหนดที่ถูกสร้าง เพื่อใช้เปรียบเทียบในส่วนที่ 5 ส่วนฟังก์ชันตัวช่วย `log` และ `show_trace` ใช้บันทึกและแสดงตารางไล่อัลกอริทึม

```python
class RouteProblem:
    """ปัญหาหาเส้นทางบนกราฟ: สถานะ = ชื่อจุด, การกระทำ = วิ่งไปจุดข้างเคียง, ต้นทุน = ระยะทาง"""

    def __init__(self, graph, initial, goal):
        self.graph = graph
        self.initial = initial
        self.goal = goal

    def actions(self, state):
        return sorted(self.graph[state])            # ข้อตกลง: สร้างลูกตามลำดับตัวอักษร

    def result(self, state, action):
        return action                                # การกระทำ "ไป X" ทำให้อยู่ที่ X

    def action_cost(self, state, action, next_state):
        return self.graph[state][next_state]

    def is_goal(self, state):
        return state == self.goal


@dataclass
class Node:
    """โหนดในต้นไม้ค้นหา (หัวข้อ 2.2.1) เก็บสถานะ โหนดแม่ การกระทำ ต้นทุนเส้นทาง g และความลึก"""
    state: Any
    parent: Optional["Node"] = field(default=None, repr=False)
    action: Any = None
    path_cost: float = 0
    depth: int = 0


def expand(problem, node):
    """สร้างโหนดลูกทีละโหนดตามรหัสเทียม EXPAND"""
    for action in problem.actions(node.state):
        next_state = problem.result(node.state, action)
        cost = node.path_cost + problem.action_cost(node.state, action, next_state)
        yield Node(next_state, node, action, cost, node.depth + 1)


def path_states(node):
    """ย้อนตาม parent จากโหนดเป้าหมายกลับไปถึงราก แล้วกลับลำดับ"""
    states = []
    while node is not None:
        states.append(node.state)
        node = node.parent
    return states[::-1]


@dataclass
class SearchResult:
    """ผลการค้นหา: โหนดคำตอบ (หรือ None) และสถิติที่ใช้เปรียบเทียบอัลกอริทึม"""
    algorithm: str
    node: Optional[Node]
    expanded_nodes: list      # โหนดที่ถูกขยายตามลำดับ
    tested: list              # สถานะที่ถูกทดสอบเป้าหมายตามลำดับ
    generated: list           # โหนดที่ถูกสร้างและเก็บไว้ (รวมราก)
    status: str               # "found", "failure" หรือ "cutoff"

    @property
    def expanded(self):
        return [n.state for n in self.expanded_nodes]

    @property
    def path(self):
        return path_states(self.node) if self.node else []

    @property
    def cost(self):
        return self.node.path_cost if self.node else None

    def summary(self, show_path=True):
        row = {"อัลกอริทึม": self.algorithm}
        if show_path:
            row["เส้นทางที่ได้"] = " → ".join(map(str, self.path)) or "-"
        row["ต้นทุน"] = num(self.cost)
        row["จำนวนการกระทำ"] = len(self.path) - 1 if self.path else None
        row["จำนวนโหนดที่ขยาย"] = len(self.expanded_nodes)
        return row


# ---------- ตัวช่วยแสดงผล ----------
def num(x):
    """แสดงจำนวนเต็มเป็นจำนวนเต็ม และทศนิยมไม่เกิน 2 ตำแหน่ง"""
    if x is None:
        return None
    return int(x) if float(x).is_integer() else round(float(x), 2)


def fmt_states(items, brackets="[]"):
    return brackets[0] + ", ".join(str(s) for s in items) + brackets[1]


def log(trace, row):
    """บันทึกหนึ่งแถวของตารางไล่อัลกอริทึม (ถ้าผู้เรียกส่งรายการ trace มา)"""
    if trace is not None:
        trace.append(row)


def show_trace(trace):
    """แสดงตารางไล่อัลกอริทึมโดยใช้คอลัมน์แรกเป็นดัชนี"""
    table = pd.DataFrame(trace)
    display(table.set_index(table.columns[0]))


# ทดลองขยายโหนดราก S
factory_problem = RouteProblem(FACTORY_MAP, START, GOAL)
root = Node(factory_problem.initial)
for child in expand(factory_problem, root):
    print(f"ลูก {child.state}: action = ไป {child.action}, g = {child.path_cost} m, depth = {child.depth}, "
          f"เส้นทาง = {'-'.join(path_states(child))}")
```

### 1.4 ต้นไม้ค้นหาใหญ่กว่าปริภูมิสถานะ (ตัวอย่างที่ 2.5)

ถ้าขยายทีละชั้นโดย **ไม่ตัดสถานะซ้ำ** ต้นไม้ค้นหาจะมีเส้นทางวนกลับ (เช่น S-A-S) และเส้นทางซ้ำซ้อน (เช่น B ที่ไปถึงได้ทั้งจาก S-A-B และ S-X-B) เซลล์ต่อไปนี้สร้างโหนดทุกโหนดถึงความลึก 2 ซึ่งควรได้ 1 + 3 + 8 = 12 โหนด ตรงกับตารางในตัวอย่างที่ 2.5

```python
level = [root]
for depth in range(3):
    paths = ["-".join(path_states(node)) for node in level]
    print(f"ความลึก {depth}: {len(level)} โหนด -> {', '.join(paths)}")
    level = [child for node in level for child in expand(factory_problem, node)]
```

**คำถามชวนคิด 1:** ที่ความลึก 2 มีโหนดเป้าหมาย S-X-G แล้ว แต่ทำไมเส้นทางนี้ยังไม่ใช่คำตอบที่ดีที่สุด และอัลกอริทึมแบบใดบ้างที่จะ "หยุดเร็วเกินไป" ที่โหนดนี้

## ส่วนที่ 2 Breadth-First Search (BFS)

BFS ขยายโหนดทีละชั้น เหมือนคลื่นที่แผ่ออกจากจุดที่โยนก้อนหิน กลไกคือใช้ **คิว FIFO** เป็น frontier (`deque`: ต่อท้ายด้วย `append` นำออกจากหัวด้วย `popleft`) และมีรายละเอียดสองข้อตามหัวข้อ 2.2.2

1. **ทดสอบเป้าหมายทันทีที่สร้างโหนดลูก** (Early Goal Test)
2. **reached เป็นเพียงเซตของสถานะ** เพราะเส้นทางแรกที่พบสถานะหนึ่งมีจำนวนการกระทำน้อยที่สุดอยู่แล้ว

ฟังก์ชันรับพารามิเตอร์ `trace` เป็นรายการว่าง ถ้าส่งมา ฟังก์ชันจะบันทึกตารางไล่อัลกอริทึมทีละขั้นให้ด้วย

```python
def breadth_first_search(problem, trace=None):
    """BFS แบบกราฟ: frontier เป็นคิว FIFO, reached เป็นเซตของสถานะ
    ทดสอบเป้าหมายทันทีที่สร้างโหนดลูก (Early Goal Test)"""
    node = Node(problem.initial)
    expanded, tested, generated = [], [node.state], [node]
    if problem.is_goal(node.state):
        return SearchResult("BFS", node, expanded, tested, generated, "found")
    frontier = deque([node])
    reached = {problem.initial: True}       # dict ใช้แทนเซตที่จำลำดับการเพิ่มได้

    def row(step, expanded_state, notes, done=False):
        if trace is None:                           # ไม่ต้องสร้างข้อความถ้าไม่ได้ขอตาราง
            return None
        return {"ขั้น": step, "โหนดที่ขยาย": expanded_state, "ลูกที่สร้าง (ผลการตรวจ)": notes,
                "frontier หลังขั้นนี้": "—" if done else fmt_states(n.state for n in frontier),
                "reached หลังขั้นนี้": "—" if done else fmt_states(reached, "{}")}

    log(trace, row(0, "—", f"(ตรวจ {node.state}: ไม่ใช่เป้าหมาย)"))
    while frontier:
        node = frontier.popleft()                   # นำโหนดที่หัวคิวออก
        expanded.append(node)
        notes = []
        for child in expand(problem, node):
            tested.append(child.state)
            if problem.is_goal(child.state):        # ทดสอบตั้งแต่ตอนสร้าง
                generated.append(child)
                notes.append(f"{child.state} (เป้าหมาย!)")
                log(trace, row(len(expanded), node.state, ", ".join(notes), done=True))
                return SearchResult("BFS", child, expanded, tested, generated, "found")
            if child.state in reached:
                notes.append(f"{child.state} (พบแล้ว ข้าม)")
            else:
                reached[child.state] = True
                frontier.append(child)              # ต่อท้ายคิว
                generated.append(child)
                notes.append(f"{child.state} (ใหม่)")
        log(trace, row(len(expanded), node.state, ", ".join(notes)))
    return SearchResult("BFS", None, expanded, tested, generated, "failure")
```

รัน BFS บนผังโรงงานแล้วเทียบตารางกับตัวอย่างที่ 2.6 ควรได้เส้นทาง S → X → G (60 m) โดยขยาย 4 โหนด (S, A, C, X)

```python
bfs_trace = []
bfs_result = breadth_first_search(factory_problem, trace=bfs_trace)
show_trace(bfs_trace)
print("เส้นทาง:", " → ".join(bfs_result.path), "| ระยะทาง", bfs_result.cost, "m",
      "| ขยาย", len(bfs_result.expanded), "โหนด:", bfs_result.expanded)
```

### 2.1 แสดงต้นไม้ค้นหา

ฟังก์ชัน `draw_search_tree` วาดเฉพาะโหนดที่ BFS สร้างและเก็บไว้ (ไม่วาดโหนดที่ถูกข้ามเพราะพบแล้ว) วงกลมทึบคือโหนดที่ถูกขยาย ตัวเลขสีส้มคือลำดับการขยาย วงกลมขอบเขียวคือโหนดเป้าหมายที่พบ และวงกลมขาวคือโหนดที่ยังรออยู่ใน frontier ตอนจบ เทียบกับรูปที่ 2.6 (ก) ในหนังสือ

```python
def draw_search_tree(result, title="", ax=None):
    """วาดต้นไม้ค้นหาจาก result.generated: ทึบน้ำเงิน = ขยายแล้ว (เลขสีส้ม = ลำดับการขยาย),
    ขอบเขียว = โหนดเป้าหมายที่พบ, ขาวขอบเทา = ยังรออยู่ใน frontier"""
    nodes = result.generated
    children = {id(n): [] for n in nodes}
    for n in nodes:
        if n.parent is not None and id(n.parent) in children:
            children[id(n.parent)].append(n)
    order = {id(n): i + 1 for i, n in enumerate(result.expanded_nodes)}

    # วางตำแหน่ง: ใบไม้เรียงซ้ายไปขวา โหนดแม่อยู่กึ่งกลางของลูก ความลึกเป็นแกนตั้ง
    pos, next_x = {}, [0]
    def place(n):
        kids = children[id(n)]
        if not kids:
            pos[id(n)] = (next_x[0], -n.depth)
            next_x[0] += 1
        else:
            for k in kids:
                place(k)
            pos[id(n)] = (sum(pos[id(k)][0] for k in kids) / len(kids), -n.depth)
    place(nodes[0])

    if ax is None:
        _, ax = plt.subplots(figsize=(6, 4))
    for n in nodes:
        if n.parent is not None:
            (x1, y1), (x2, y2) = pos[id(n.parent)], pos[id(n)]
            ax.plot([x1, x2], [y1, y2], color=COLORS["gray"], linewidth=1.2, zorder=1)
            ax.text((x1 + x2) / 2, (y1 + y2) / 2, str(num(n.path_cost - n.parent.path_cost)),
                    fontsize=7, color=COLORS["gray"], ha="center", va="center",
                    bbox=dict(fc="white", ec="none", pad=0.5))
    for n in nodes:
        x, y = pos[id(n)]
        expanded = id(n) in order
        edge = COLORS["green"] if result.node is n else (COLORS["blue"] if expanded else COLORS["gray"])
        ax.scatter(x, y, s=520, zorder=2, color=COLORS["blue"] if expanded else "white",
                   edgecolors=edge, linewidths=3 if result.node is n else 1.5)
        ax.text(x, y, str(n.state), ha="center", va="center", fontsize=10, fontweight="bold",
                color="white" if expanded else COLORS["blue"], zorder=3)
        if expanded:
            ax.annotate(str(order[id(n)]), (x, y), xytext=(9, 9), textcoords="offset points",
                        color=COLORS["orange"], fontsize=9, fontweight="bold", zorder=3)
    ax.set_title(title)
    ax.margins(0.15)
    ax.axis("off")
    return ax


fig, axes = plt.subplots(1, 2, figsize=(11, 4))
draw_search_tree(bfs_result, title="BFS search tree (orange = expansion order)", ax=axes[0])
draw_factory_map(expanded=bfs_result.expanded, path=bfs_result.path,
                 title=f"BFS: {' -> '.join(bfs_result.path)}, {bfs_result.cost} m", ax=axes[1])
plt.tight_layout()
plt.show()
```

**คำถามชวนคิด 2:** BFS พบคำตอบที่ "ตื้นที่สุด" (2 การกระทำ) แต่ไม่ใช่คำตอบที่ "สั้นที่สุดตามระยะทาง" (45 m) เพราะอะไร ถ้าเปลี่ยนต้นทุนทุกทางวิ่งให้เท่ากันหมด (เช่น ทุกเส้นยาว 1 หน่วย) BFS จะได้คำตอบที่ดีที่สุดหรือไม่

**คำถามชวนคิด 3:** ถ้าย้ายการทดสอบเป้าหมายไปทำตอนนำโหนดออกจากคิว (แทนตอนสร้าง) BFS จะขยายโหนดเพิ่มขึ้นกี่โหนดในผังนี้ ลองแก้โค้ดแล้วนับดู

## ส่วนที่ 3 Depth-First และ Depth-Limited Search

DFS ลงลึกตามกิ่งเดียวจนสุดก่อนแล้วจึงถอยกลับ กลไกคือใช้ **สแตก LIFO** เป็น frontier ใน Python ใช้ `list` โดย `append` ใส่ไว้บนสุดและ `pop()` นำโหนดบนสุดออก หนังสือใช้ข้อตกลงสองข้อ (หัวข้อ 2.2.3)

1. **ใส่ลูกลงสแตกในลำดับกลับกัน** ลูกตัวแรกตามตัวอักษรจึงอยู่บนสุดและถูกขยายก่อน
2. **DFS แบบต้นไม้ที่ตรวจวงวน** ไม่เก็บ reached แต่ไม่สร้างลูกที่ซ้ำกับสถานะบนเส้นทางจากรากถึงโหนดปัจจุบัน

ฟังก์ชันเดียวกันนี้ใช้เป็น **DLS** ได้ด้วยการกำหนด `limit` = $\ell$ โหนดที่ความลึก $\ell$ ถูกทดสอบเป้าหมายได้แต่ไม่ถูกขยาย (หนังสือใช้เงื่อนไข DEPTH ≥ ℓ) และคืนสถานะ `"cutoff"` ถ้ามีโหนดถูกตัดเพราะถึงขีดจำกัด หรือ `"failure"` ถ้าสำรวจครบแล้วไม่พบคำตอบเลย

ในคอลัมน์ frontier บนสุดของสแตกอยู่ซ้าย ถ้าสถานะเดียวกันอยู่ในสแตกหลายโหนด จะแสดงเส้นทางในวงเล็บ เช่น `X(S-A-B-X)`

```python
def fmt_stack(stack):
    """แสดงสแตกจากบนสุด (ซ้าย) ลงล่าง ถ้าสถานะซ้ำกันให้แสดงเส้นทางกำกับ"""
    top_first = list(reversed(stack))
    counts = Counter(n.state for n in top_first)
    return fmt_states(n.state if counts[n.state] == 1 else f"{n.state}({'-'.join(map(str, path_states(n)))})"
                      for n in top_first)


def depth_first_search(problem, limit=None, trace=None):
    """DFS แบบต้นไม้ที่ตรวจวงวน ถ้ากำหนด limit จะเป็น DLS:
    โหนดที่ความลึก >= limit ถูกทดสอบเป้าหมายได้ แต่ไม่ถูกขยายต่อ"""
    name = "DFS" if limit is None else f"DLS (ℓ = {limit})"
    root = Node(problem.initial)
    frontier = [root]                       # สแตก: ท้ายรายการคือบนสุด
    expanded, tested, generated = [], [], [root]
    status = "failure"
    log(trace, {"ขั้น": 0, "โหนดที่นำออก (เส้นทาง)": "—", "ความลึก": "", "เป้าหมาย?": "—",
                "การดำเนินการ": "—", "frontier (บนสุดอยู่ซ้าย)": fmt_stack(frontier)})
    while frontier:
        node = frontier.pop()               # นำโหนดบนสุดออก
        tested.append(node.state)
        row = {"ขั้น": len(tested), "โหนดที่นำออก (เส้นทาง)": "-".join(map(str, path_states(node))),
               "ความลึก": node.depth}
        if problem.is_goal(node.state):
            row.update({"เป้าหมาย?": "✓", "การดำเนินการ": "คืนคำตอบ", "frontier (บนสุดอยู่ซ้าย)": "—"})
            log(trace, row)
            return SearchResult(name, node, expanded, tested, generated, "found")
        row["เป้าหมาย?"] = "✗"
        if limit is not None and node.depth >= limit:
            status = "cutoff"               # ถึงกำแพงความลึก ไม่ขยายต่อ
            row["การดำเนินการ"] = "ถึงขีดจำกัด → cutoff"
        else:
            expanded.append(node)
            on_path = set(path_states(node))
            children, skipped = [], []
            for child in expand(problem, node):
                (skipped if child.state in on_path else children).append(child)
            frontier.extend(reversed(children))  # ใส่กลับลำดับ ลูกตัวแรกจึงอยู่บนสุด
            generated.extend(children)
            text = "ขยาย ได้ " + (", ".join(str(c.state) for c in children) if children else "ไม่มีลูกใหม่")
            if skipped:
                text += " (ข้าม " + ", ".join(str(c.state) for c in skipped) + ")"
            row["การดำเนินการ"] = text
        row["frontier (บนสุดอยู่ซ้าย)"] = fmt_stack(frontier)
        log(trace, row)
    return SearchResult(name, None, expanded, tested, generated, status)
```

## ส่วนที่ 4 Uniform-Cost และ Iterative Deepening Search

### 4.1 Uniform-Cost Search (ตัวอย่างที่ 2.10)

UCS ขยายโหนดที่มีต้นทุนเส้นทาง $g(n)$ ต่ำที่สุดก่อนเสมอ ซึ่งคือ BEST-FIRST-SEARCH ที่ใช้ $f(n) = g(n)$ frontier เป็น **คิวลำดับความสำคัญ** (`heapq`) และต่างจาก BFS สองข้อ คือ (1) ทดสอบเป้าหมายตอน **นำโหนดออก** และ (2) ยอมให้เส้นทางที่ถูกกว่า **แทนที่** เส้นทางเดิม

ฟังก์ชัน `best_first_search` เขียนแบบทั่วไปโดยรับฟังก์ชันประเมิน `f` เป็นพารามิเตอร์ ปฏิบัติการที่ 4 จะใช้ฟังก์ชันเดียวกันนี้กับ Greedy และ A* เพียงเปลี่ยน `f`

> **หมายเหตุเรื่องการเขียนโปรแกรมจริง** เมื่อพบเส้นทางที่ถูกกว่า โค้ดใส่โหนดใหม่ลง heap โดยไม่ลบโหนดเก่า (การลบกลาง heap ทำได้ยาก) แต่จดไว้ใน `in_frontier` ว่าโหนดล่าสุดของสถานะนั้นคือโหนดใด เมื่อโหนดเก่าถูกนำออกมาภายหลังจะถูกข้ามไป ตารางไล่อัลกอริทึมแสดงเฉพาะโหนดล่าสุดของแต่ละสถานะ จึงตรงกับตารางในหนังสือ

```python
def best_first_search(problem, f, name, describe, trace=None):
    """BEST-FIRST-SEARCH (หัวข้อ 2.2.1): frontier เป็นคิวลำดับความสำคัญเรียงตาม f(node)
    - ถ้า f เท่ากัน เลือกตามลำดับของสถานะ (ตัวอักษร) ตามข้อตกลงของหนังสือ
    - ทดสอบเป้าหมายตอน "นำออก" จาก frontier
    - ถ้าพบสถานะเดิมด้วยต้นทุน g ที่ต่ำกว่า ให้เส้นทางใหม่แทนที่เส้นทางเดิม
    describe(node) คืนข้อความของโหนดสำหรับตารางไล่อัลกอริทึม"""
    root = Node(problem.initial)
    counter = itertools.count()                 # ใช้กันกรณีค่า f และสถานะเท่ากันทุกอย่าง
    heap = [(f(root), root.state, next(counter), root)]
    in_frontier = {root.state: root}            # โหนดล่าสุดของแต่ละสถานะที่ยังรออยู่ใน frontier
    reached = {root.state: root}
    expanded, tested, generated = [], [], [root]

    def frontier_text():
        waiting = sorted(in_frontier.values(), key=lambda n: (f(n), n.state))
        return fmt_states(f"{n.state}:{num(f(n))}" for n in waiting)

    if trace is not None:
        log(trace, {"ขั้น": 0, "นำออก": "—", "เป้าหมาย?": "—", "ลูกที่สร้าง (ผล)": describe(root),
                    "frontier หลังขั้นนี้": frontier_text(), "ขยายแล้ว": "{ }"})
    while heap:
        _, _, _, node = heapq.heappop(heap)
        if in_frontier.get(node.state) is not node:
            continue                            # โหนดเก่าที่ถูกเส้นทางที่ถูกกว่าแทนที่ไปแล้ว ไม่ใช้
        del in_frontier[node.state]
        tested.append(node.state)
        if problem.is_goal(node.state):
            if trace is not None:
                log(trace, {"ขั้น": len(expanded) + 1, "นำออก": describe(node), "เป้าหมาย?": "✓",
                            "ลูกที่สร้าง (ผล)": "—", "frontier หลังขั้นนี้": "—", "ขยายแล้ว": "—"})
            return SearchResult(name, node, expanded, tested, generated, "found")
        expanded.append(node)
        notes = []
        for child in expand(problem, node):
            old = reached.get(child.state)
            if old is not None and child.path_cost >= old.path_cost:
                if trace is not None:
                    notes.append(f"{describe(child)} (เดิม g={num(old.path_cost)} ข้าม)")
                continue
            if trace is not None:
                notes.append(f"{describe(child)} (ใหม่)" if old is None
                             else f"{describe(child)} (เดิม g={num(old.path_cost)} → ปรับ)")
            reached[child.state] = child
            in_frontier[child.state] = child
            generated.append(child)
            heapq.heappush(heap, (f(child), child.state, next(counter), child))
        if trace is not None:
            log(trace, {"ขั้น": len(expanded), "นำออก": describe(node), "เป้าหมาย?": "✗",
                        "ลูกที่สร้าง (ผล)": "; ".join(notes), "frontier หลังขั้นนี้": frontier_text(),
                        "ขยายแล้ว": fmt_states((n.state for n in expanded), "{}")})
    return SearchResult(name, None, expanded, tested, generated, "failure")


def uniform_cost_search(problem, trace=None):
    """UCS = BEST-FIRST-SEARCH ที่ f(n) = g(n)"""
    return best_first_search(problem, f=lambda n: n.path_cost, name="UCS",
                             describe=lambda n: f"{n.state}:{num(n.path_cost)}", trace=trace)


ucs_trace = []
ucs_result = uniform_cost_search(factory_problem, trace=ucs_trace)
show_trace(ucs_trace)
print("เส้นทาง:", " → ".join(ucs_result.path), "| ระยะทาง", ucs_result.cost, "m",
      "| ขยาย", len(ucs_result.expanded), "โหนด:", ucs_result.expanded)
```

สังเกตขั้นที่ 3 ซึ่ง UCS สร้าง G ด้วย $g = 60$ แล้ว แต่ยังไม่หยุด จนขั้นที่ 5 พบเส้นทางผ่าน Y ที่ถูกกว่า ($g = 45$) จึงปรับเส้นทาง และขั้นที่ 6 ซึ่ง UCS ขยาย B ($g = 40$) ก่อน G ($g = 45$) ทั้งที่ B อยู่คนละทิศกับเป้าหมาย เพราะ UCS ไม่รู้ทิศทางของเป้าหมาย

### 4.2 Iterative Deepening Search (ตัวอย่างที่ 2.11)

IDS เรียก DLS ซ้ำโดยเพิ่ม $\ell$ ทีละ 1 จนกว่าผลจะไม่ใช่ `cutoff` ควรได้การทดสอบเป้าหมายรวม 13 ครั้ง ขยายรวม 5 โหนด และได้เส้นทาง S → X → G (60 m) เท่ากับ BFS

```python
def iterative_deepening_search(problem, max_limit=50, trace=None):
    """เรียก DLS ด้วย ℓ = 0, 1, 2, ... จนกว่าจะพบคำตอบหรือได้ failure"""
    expanded, tested, generated = [], [], []
    for limit in range(max_limit + 1):
        result = depth_first_search(problem, limit=limit)
        expanded += result.expanded_nodes
        tested += result.tested
        generated += result.generated
        log(trace, {"รอบ": f"ℓ = {limit}",
                    "โหนดที่ทดสอบเป้าหมาย (ตามลำดับ)": ", ".join(map(str, result.tested)),
                    "จำนวนโหนดที่ขยาย": f"{len(result.expanded)} ({', '.join(map(str, result.expanded))})"
                                       if result.expanded else "0",
                    "ผล": "พบ " + str(result.node.state) if result.node else result.status})
        if result.status != "cutoff":
            return SearchResult("IDS", result.node, expanded, tested, generated, result.status)
    return SearchResult("IDS", None, expanded, tested, generated, "cutoff")


ids_trace = []
ids_result = iterative_deepening_search(factory_problem, trace=ids_trace)
show_trace(ids_trace)
print(f"รวม: ทดสอบเป้าหมาย {len(ids_result.tested)} ครั้ง | ขยาย {len(ids_result.expanded)} โหนด",
      f"| เส้นทาง {' → '.join(ids_result.path)} ({ids_result.cost} m)")
print("S ถูกทดสอบเป้าหมาย", ids_result.tested.count("S"), "ครั้ง")
```

**คำถามชวนคิด 6:** IDS ทดสอบเป้าหมายซ้ำโหนดชั้นบนหลายรอบ (S ถูกทดสอบ 3 ครั้ง) ทำไม Russell และ Norvig จึงยังแนะนำ IDS เมื่อปริภูมิสถานะใหญ่และไม่รู้ความลึกของคำตอบ (ใบ้: เทียบหน่วยความจำ $O(bd)$ กับ $O(b^d)$ และดูสูตร $N(\text{IDS})$ ในหัวข้อ 2.2.4)

### 5.2 ต้นไม้ค้นหาโตแบบเลขชี้กำลัง

ผังโรงงานมีเพียง 7 สถานะ แต่ถ้าค้นหาแบบต้นไม้โดย **ไม่ตัดสถานะซ้ำเลย** จำนวนโหนดในแต่ละชั้นจะโตแบบเลขชี้กำลัง (ตัวอย่างที่ 2.5 นับไว้ถึงความลึก 2) ฟังก์ชันด้านล่างนับต่อไปจนถึงความลึก 12 แล้ววาดกราฟแกนตั้งแบบลอการิทึม ซึ่งเลขชี้กำลังจะปรากฏเป็นเส้นตรง

```python
def count_tree_nodes(graph, start, max_depth):
    """นับจำนวนโหนดในแต่ละชั้นของต้นไม้ค้นหา เมื่อไม่ตัดสถานะซ้ำและไม่ตรวจวงวน"""
    counts, level = [1], [start]
    for _ in range(max_depth):
        level = [nbr for state in level for nbr in graph[state]]
        counts.append(len(level))
    return counts


tree_counts = count_tree_nodes(FACTORY_MAP, START, 12)
growth = pd.DataFrame({"ความลึก": range(len(tree_counts)), "จำนวนโหนดในชั้น": tree_counts,
                       "จำนวนโหนดสะสม": np.cumsum(tree_counts)})
growth["อัตราส่วนต่อชั้นก่อน"] = (growth["จำนวนโหนดในชั้น"] / growth["จำนวนโหนดในชั้น"].shift(1)).round(2)
display(growth.set_index("ความลึก"))

fig, ax = plt.subplots(figsize=(6, 3.5))
ax.semilogy(growth["ความลึก"], growth["จำนวนโหนดสะสม"], "o-", color=COLORS["orange"],
            label="tree search (no duplicate check)")
ax.axhline(len(FACTORY_MAP), color=COLORS["blue"], linestyle="--", label="number of states = 7")
ax.set_xlabel("depth")
ax.set_ylabel("nodes (log scale)")
ax.legend()
ax.set_title("Search tree vs state space of the factory map")
plt.show()
```

อัตราส่วนจำนวนโหนดระหว่างชั้นติดกันแกว่งอยู่ราว 2.7–2.8 ซึ่งคือตัวประกอบการแตกกิ่งยังผลของต้นไม้นี้ (จุดในผังมีจุดข้างเคียง 2–4 จุด เฉลี่ย $18/7 \approx 2.6$) ต้นไม้ลึก 12 ชั้นจึงมีเกือบ 3 แสนโหนด การค้นหาแบบกราฟที่จำสถานะที่พบแล้ว (เช่น BFS ในส่วนที่ 2) จึงไม่เคยเก็บเกิน 7 สถานะ

ในปัญหาจริง ปริภูมิสถานะเองก็ใหญ่มาก ตารางที่ 2.4 ของหนังสือประเมินเวลาและหน่วยความจำของ BFS เมื่อ $b = 10$ สร้างโหนดได้ $10^6$ โหนดต่อวินาที และใช้หน่วยความจำ 1 KB ต่อโหนด เซลล์ต่อไปคำนวณตารางนั้นซ้ำ

```python
def readable_time(seconds):
    for unit, size in [("ปี", 365 * 24 * 3600), ("วัน", 24 * 3600), ("ชั่วโมง", 3600), ("นาที", 60)]:
        if seconds >= size * 2:
            return f"{seconds / size:.1f} {unit}"
    return f"{seconds:g} วินาที"


def readable_bytes(n_bytes):
    for unit, size in [("PB", 1e15), ("TB", 1e12), ("GB", 1e9), ("MB", 1e6)]:
        if n_bytes >= size:
            return f"{n_bytes / size:g} {unit}"
    return f"{n_bytes:g} B"


branching, nodes_per_second, bytes_per_node = 10, 1e6, 1e3
rows = []
for depth in [6, 8, 10, 12, 14]:
    n_nodes = branching ** depth
    rows.append({"ความลึก d": depth, "จำนวนโหนด ≈ b^d": f"10^{depth}",
                 "เวลา": readable_time(n_nodes / nodes_per_second),
                 "หน่วยความจำ": readable_bytes(n_nodes * bytes_per_node)})
pd.DataFrame(rows).set_index("ความลึก d")
```

**คำถามชวนคิด 7:** จากตารางนี้ ปัจจัยใดเป็นข้อจำกัดของ BFS ก่อนกัน ระหว่างเวลากับหน่วยความจำ และ DFS หรือ IDS ช่วยแก้ข้อจำกัดใดได้

**คำถามชวนคิด 8:** สมมติระบบจำหน่ายไฟฟ้าขนาดเล็กมีสวิตช์ 12 ตัว แต่ละตัวเปิดหรือปิดได้ และต้องการหาลำดับการสับสวิตช์ที่น้อยที่สุดเพื่อถ่ายโอนโหลด ปริภูมิสถานะมีกี่สถานะ BFS แบบกราฟยังใช้ได้หรือไม่ ถ้าสวิตช์เพิ่มเป็น 60 ตัวล่ะ

```python
def close_edge(graph, u, v):
    """คืนผังใหม่ที่ไม่มีทางวิ่ง u–v (ไม่แก้ผังเดิม)"""
    new_graph = {node: dict(nbrs) for node, nbrs in graph.items()}   # สำเนาสองชั้น
    # TODO: ลบ v ออกจากจุดข้างเคียงของ u และลบ u ออกจากจุดข้างเคียงของ v ใน new_graph
    return new_graph
```

### ข้อ 2 ถังตวงน้ำมันหม้อแปลง (Water Jug)

ช่างบำรุงรักษาต้องตวงน้ำมันฉนวนจากถังใหญ่ใส่ภาชนะตวงให้ได้ **2 ลิตรพอดี** เพื่อส่งทดสอบค่าความคงทนต่อแรงดันไฟฟ้า (Breakdown Voltage) แต่มีภาชนะตวงเพียง 2 ใบ คือใบใหญ่ 4 ลิตรและใบเล็ก 3 ลิตร ซึ่ง **ไม่มีขีดบอกปริมาตร** การกระทำที่ทำได้มี 6 แบบ

| การกระทำ | ผล |
|---|---|
| เติมใบใหญ่จนเต็ม / เติมใบเล็กจนเต็ม | ภาชนะใบนั้นมี 4 หรือ 3 ลิตร |
| เทใบใหญ่ทิ้ง / เทใบเล็กทิ้ง (เทกลับถัง) | ภาชนะใบนั้นว่าง |
| เทใบใหญ่ลงใบเล็ก / เทใบเล็กลงใบใหญ่ | เทจนใบต้นทางหมดหรือใบปลายทางเต็ม แล้วแต่อย่างใดเกิดก่อน |

นิยามปัญหา: สถานะ = `(ปริมาตรในใบใหญ่, ปริมาตรในใบเล็ก)` สถานะเริ่มต้น `(0, 0)` เป้าหมาย = ใบใหญ่มี 2 ลิตร และทุกการกระทำมีต้นทุน 1 (ต้องการจำนวนการกระทำน้อยที่สุด จึงเหมาะกับ BFS)

จงเติมฟังก์ชัน `jug_successors(state)` ให้คืน dict `{ชื่อการกระทำ: สถานะใหม่}` ของทุกการกระทำที่ทำให้สถานะเปลี่ยน คลาส `JugProblem` ในเซลล์ตรวจจะเรียกฟังก์ชันนี้ แล้วใช้ `breadth_first_search` ตัวเดิมจากส่วนที่ 2 แก้ปัญหาโดยไม่ต้องแก้ BFS เลย

```python
JUG_CAPACITY = (4, 3)    # (ใบใหญ่, ใบเล็ก) หน่วยลิตร


def jug_successors(state):
    """คืน dict {ชื่อการกระทำ: สถานะใหม่} จากสถานะ (big, small) เฉพาะการกระทำที่ทำให้สถานะเปลี่ยน"""
    big, small = state
    cap_big, cap_small = JUG_CAPACITY
    candidates = {}
    # TODO: เติมการกระทำ 6 แบบลงใน candidates เช่น
    #   candidates["เติมใบใหญ่จนเต็ม"] = (cap_big, small)
    #   "เติมใบเล็กจนเต็ม", "เทใบใหญ่ทิ้ง", "เทใบเล็กทิ้ง"
    #   "เทใบใหญ่ลงใบเล็ก": ปริมาตรที่เทได้ = min(big, cap_small - small)
    #   "เทใบเล็กลงใบใหญ่": ปริมาตรที่เทได้ = min(small, cap_big - big)
    return {action: new_state for action, new_state in candidates.items() if new_state != state}
```

```python
# เซลล์ตรวจคำตอบข้อ 2 (ไม่ต้องแก้)
class JugProblem:
    """ปัญหาถังตวง: สถานะ (big, small) เป้าหมายคือใบใหญ่มีปริมาตร target ลิตร"""

    def __init__(self, initial=(0, 0), target=2):
        self.initial = initial
        self.target = target

    def actions(self, state):
        return list(jug_successors(state))

    def result(self, state, action):
        return jug_successors(state)[action]

    def action_cost(self, state, action, next_state):
        return 1

    def is_goal(self, state):
        return state[0] == self.target


def path_actions(node):
    """รายการ (การกระทำ, สถานะหลังทำ) ตามเส้นทางจากรากถึง node"""
    steps = []
    while node.parent is not None:
        steps.append((node.action, node.state))
        node = node.parent
    return steps[::-1]


spot_checks = {((4, 1), "เทใบใหญ่ลงใบเล็ก"): (2, 3), ((1, 3), "เทใบเล็กลงใบใหญ่"): (4, 0),
               ((0, 0), "เติมใบเล็กจนเต็ม"): (0, 3), ((4, 3), "เทใบใหญ่ทิ้ง"): (0, 3)}
if not jug_successors((0, 0)):
    print("ข้อ 2: ยังไม่ได้ทำ")
else:
    wrong = [f"{s} --{a}--> ควรได้ {e} แต่ได้ {jug_successors(s).get(a)}"
             for (s, a), e in spot_checks.items() if jug_successors(s).get(a) != e]
    jug_result = breadth_first_search(JugProblem())
    if wrong:
        print("ข้อ 2: ยังไม่ถูก\n  " + "\n  ".join(wrong))
    elif jug_result.node is None:
        print("ข้อ 2: BFS หาคำตอบไม่พบ ลองตรวจการกระทำอีกครั้ง")
    else:
        steps = path_actions(jug_result.node)
        display(pd.DataFrame([{"ขั้น": i, "การกระทำ": a, "ใบใหญ่ (L)": s[0], "ใบเล็ก (L)": s[1]}
                              for i, (a, s) in enumerate(steps, start=1)]).set_index("ขั้น"))
        print(f"BFS ขยาย {len(jug_result.expanded)} โหนด พบสถานะทั้งหมด {len(jug_result.generated)} สถานะ")
        if len(steps) == 6:
            print("ข้อ 2: ถูกต้อง — ใช้ 6 การกระทำ ซึ่งน้อยที่สุด")
        else:
            print(f"ข้อ 2: ได้ {len(steps)} การกระทำ แต่คำตอบที่ดีที่สุดใช้ 6 การกระทำ")
```

**คำถามข้อ 2:** ทำไม BFS จึงรับประกันว่าได้จำนวนการกระทำน้อยที่สุดในปัญหานี้ ถ้าเปลี่ยนต้นทุนให้ "เติมจากถังใหญ่" มีต้นทุนเท่ากับปริมาตรที่เติม (คิดเป็นปริมาณน้ำมันที่ใช้) ควรเปลี่ยนไปใช้อัลกอริทึมใด

### ข้อ 3 เขาวงกตในคลังสินค้า (Maze)

AGV อีกคันวิ่งในคลังสินค้าที่แบ่งพื้นเป็นช่องตาราง `#` คือชั้นวาง `S` คือจุดเริ่ม และ `G` คือจุดรับสินค้า AGV วิ่งได้เฉพาะ 4 ทิศ ช่องละ 1 การกระทำ (สถานะ = `(แถว, คอลัมน์)` โดยแถว 0 อยู่บนสุด)

จงเติมฟังก์ชัน `maze_neighbors(grid, cell)` ให้คืนรายการช่องข้างเคียงที่ AGV วิ่งไปได้ โดยเรียงลำดับ **ขึ้น ลง ซ้าย ขวา** และตัดช่องที่อยู่นอกตารางหรือเป็นชั้นวางออก คลาส `MazeProblem` จะใช้ฟังก์ชันนี้ เพื่อให้ BFS และ DFS ชุดเดิมหาเส้นทางได้

```python
WAREHOUSE_MAZE = [
    "S...#.........",
    ".##.#.######..",
    ".#..#......#..",
    ".#.##.##.#.#..",
    ".#....#..#....",
    ".####.#.##.##.",
    "......#......G",
]
def draw_grid(grid, path=(), expanded=(), title="", ax=None):
    """วาดแผนที่ตาราง: สีเทา = ชั้นวาง/ผนัง (#), สีฟ้าอ่อน = ช่องที่ถูกขยาย, เส้นสีส้ม = เส้นทางที่ได้"""
    n_rows, n_cols = len(grid), len(grid[0])
    image = np.ones((n_rows, n_cols, 3))
    for (r, c) in expanded:
        image[r, c] = (0.78, 0.87, 0.96)
    for r in range(n_rows):
        for c in range(n_cols):
            if grid[r][c] == "#":
                image[r, c] = (0.42, 0.42, 0.42)
    if ax is None:
        _, ax = plt.subplots(figsize=(0.4 * n_cols + 1, 0.4 * n_rows + 1))
    ax.imshow(image, interpolation="nearest")
    if path:
        ax.plot([c for r, c in path], [r for r, c in path], color=COLORS["orange"], linewidth=3)
    for r in range(n_rows):
        for c in range(n_cols):
            if grid[r][c] in "SG":
                ax.text(c, r, grid[r][c], ha="center", va="center", fontsize=10, fontweight="bold",
                        color="white", bbox=dict(boxstyle="circle,pad=0.2", ec="none",
                        fc=COLORS["blue"] if grid[r][c] == "S" else COLORS["green"]))
    ax.set_xticks(np.arange(-0.5, n_cols, 1), minor=True)
    ax.set_yticks(np.arange(-0.5, n_rows, 1), minor=True)
    ax.grid(which="minor", color="#DDDDDD", linewidth=0.5)
    ax.tick_params(which="both", length=0, labelsize=7)
    ax.set_title(title, fontsize=10)
    return ax


draw_grid(WAREHOUSE_MAZE, title="Warehouse maze (# = rack)")
plt.show()
```

```python
def maze_neighbors(grid, cell):
    """คืนรายการช่องข้างเคียงที่วิ่งไปได้ เรียงลำดับ ขึ้น ลง ซ้าย ขวา"""
    row, col = cell
    neighbors = []
    for d_row, d_col in [(-1, 0), (1, 0), (0, -1), (0, 1)]:   # ขึ้น ลง ซ้าย ขวา
        new_row, new_col = row + d_row, col + d_col
        # TODO: ถ้า (new_row, new_col) อยู่ในตาราง และ grid[new_row][new_col] ไม่ใช่ "#"
        #       ให้เพิ่ม (new_row, new_col) ลงใน neighbors
    return neighbors
```

```python
# เซลล์ตรวจคำตอบข้อ 3 (ไม่ต้องแก้)
class MazeProblem:
    """ปัญหาเขาวงกต: สถานะ = (แถว, คอลัมน์), การกระทำ = ช่องข้างเคียงที่จะไป, ต้นทุนช่องละ 1"""

    def __init__(self, grid):
        self.grid = grid
        for r, line in enumerate(grid):
            for c, ch in enumerate(line):
                if ch == "S":
                    self.initial = (r, c)
                elif ch == "G":
                    self.goal = (r, c)

    def actions(self, state):
        return maze_neighbors(self.grid, state)

    def result(self, state, action):
        return action

    def action_cost(self, state, action, next_state):
        return 1

    def is_goal(self, state):
        return state == self.goal


expected_neighbors = {(0, 0): [(1, 0), (0, 1)], (4, 2): [(3, 2), (4, 3)], (6, 13): [(5, 13), (6, 12)]}
if maze_neighbors(WAREHOUSE_MAZE, (0, 0)) == []:
    print("ข้อ 3: ยังไม่ได้ทำ")
else:
    wrong = {c: maze_neighbors(WAREHOUSE_MAZE, c) for c, e in expected_neighbors.items()
             if maze_neighbors(WAREHOUSE_MAZE, c) != e}
    if wrong:
        print("ข้อ 3: ยังไม่ถูก ช่องข้างเคียงที่ได้:", wrong, "\n  ควรได้:", expected_neighbors)
    else:
        maze_problem = MazeProblem(WAREHOUSE_MAZE)
        maze_results = [breadth_first_search(maze_problem), depth_first_search(maze_problem)]
        display(pd.DataFrame([r.summary(show_path=False) for r in maze_results]).set_index("อัลกอริทึม"))
        bfs_steps = len(maze_results[0].path) - 1
        print("ข้อ 3: ถูกต้อง — BFS ได้เส้นทาง", bfs_steps, "ช่อง ซึ่งสั้นที่สุด" if bfs_steps == 25
              else "ช่อง (ควรได้ 25 ช่อง)")
        fig, axes = plt.subplots(1, 2, figsize=(13, 3.8))
        for ax, r in zip(axes, maze_results):
            draw_grid(WAREHOUSE_MAZE, path=r.path, expanded=r.expanded, ax=ax,
                      title=f"{r.algorithm}: {len(r.path) - 1} steps, expanded {len(r.expanded)} cells")
        plt.tight_layout()
        plt.show()
```

**คำถามข้อ 3:** เส้นทางของ DFS ยาวกว่าของ BFS เท่าใด แต่ DFS ขยายช่องน้อยกว่าหรือมากกว่า ถ้าคลังสินค้าใหญ่มาก (เช่น 1,000 × 1,000 ช่อง) และหน่วยความจำจำกัด ควรเลือกอัลกอริทึมใด เพราะอะไร

- **การนิยามปัญหาให้อยู่ในรูปมาตรฐาน** (สถานะเริ่มต้น การกระทำ ผลของการกระทำ ต้นทุน และการทดสอบเป้าหมาย) ทำให้อัลกอริทึมชุดเดียวแก้ได้ทั้งผังโรงงาน ถังตวงน้ำมัน และเขาวงกต โดยไม่ต้องแก้โค้ดของอัลกอริทึม (ส่วนที่ 1 และ 6)
- **อัลกอริทึมต่างกันที่โครงสร้างของ frontier**: คิว FIFO ให้ BFS สแตก LIFO ให้ DFS/DLS/IDS และคิวลำดับความสำคัญตาม $g(n)$ ให้ UCS ผลบนผังโรงงานตรงกับตารางที่ 2.6 คือ BFS และ IDS ได้ S → X → G (60 m) DFS ได้ S → A → B → X → G (100 m) และ UCS ได้ S → X → Y → G (45 m) ซึ่งดีที่สุด (ส่วนที่ 2–5)
- **จุดทดสอบเป้าหมายสำคัญ**: BFS ทดสอบตอนสร้างโหนดเพราะนับเพียงจำนวนการกระทำ ส่วน UCS ต้องทดสอบตอนนำออก มิฉะนั้นจะได้คำตอบ 60 m แทน 45 m (ส่วนที่ 4)
- **ต้นไม้ค้นหาโตแบบเลขชี้กำลัง** แม้ผังมีเพียง 7 สถานะ การจดจำสถานะที่พบแล้ว (การค้นหาแบบกราฟ) และการเลือกอัลกอริทึมให้เหมาะกับหน่วยความจำ (IDS) จึงสำคัญ (ส่วนที่ 5)
- **UCS ได้คำตอบที่ดีที่สุดแต่ขยายโหนดมากที่สุด** เพราะไม่รู้ทิศทางของเป้าหมาย
