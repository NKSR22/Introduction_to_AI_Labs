## ส่วนที่ 1 การพยากรณ์โหลดและพลังงานหมุนเวียน

เนื้อหาในหนังสือ: 8.1.1 การพยากรณ์โหลดและพลังงานหมุนเวียน

### 1.1 แบบจำลองทางกายภาพของกำลังผลิตโซลาร์เซลล์ (ตัวอย่างที่ 8.1)

กำลังผลิตของแผงแปรผันตามความเข้มแสง $G$ และลดลงเมื่อเซลล์ร้อน อุณหภูมิเซลล์ประมาณจากค่า NOCT บนแผ่นข้อมูล

$$P = P_{\text{STC}} \times \frac{G}{1000} \times \big[1 + \gamma\,(T_c - 25)\big] \times \eta_{\text{ระบบ}}, \qquad T_c \approx T_a + \frac{\text{NOCT} - 20}{800}\,G$$

ฟังก์ชัน `pv_power` รับค่าเป็นสเกลาร์หรืออาร์เรย์ก็ได้ จึงใช้ทั้งตรวจตัวอย่างในหนังสือและสร้างข้อมูลทั้งปีในหัวข้อ 1.2

```python
def pv_power(p_stc, irradiance, t_air, gamma=-0.004, noct=45.0, eta_system=0.9):
    """กำลังผลิตของโซลาร์เซลล์ (หน่วยเดียวกับ p_stc) คืน (อุณหภูมิเซลล์, ตัวประกอบอุณหภูมิ, กำลัง)"""
    t_cell = t_air + (noct - 20) / 800 * irradiance
    temp_factor = 1 + gamma * (t_cell - 25)
    power = p_stc * irradiance / 1000 * temp_factor * eta_system
    return t_cell, temp_factor, power


# ตัวอย่างที่ 8.1 ฟาร์ม 1,000 kW พยากรณ์ G = 800 W/m², Ta = 32 °C
tc, factor, p_forecast = pv_power(1000, 800, 32)
compare("อุณหภูมิเซลล์ (°C)", tc, 57, 1e-9)
compare("ตัวประกอบอุณหภูมิ", factor, 0.872, 1e-9)
compare("กำลังผลิตที่พยากรณ์ (kW)", p_forecast, 627.84, 1e-6)
# ขั้นที่ 4 เมฆมาและแสงจริงเหลือ 500 W/m²
tc2, factor2, p_actual = pv_power(1000, 500, 32)
compare("กำลังผลิตจริงเมื่อเมฆมา (kW)", p_actual, 409.3, 0.05)
compare("พยากรณ์สูงเกินจริง (kW)", p_forecast - p_actual, 219, 0.5)
print(f"แสงผิดไป {(800 - 500) / 800:.1%} → กำลังผลิตผิดไป {(p_forecast - p_actual) / p_forecast:.1%}")

# ความไวของกำลังผลิตต่อแสงและอุณหภูมิอากาศ
g_grid = np.linspace(0, 1100, 111)
fig, ax = plt.subplots(figsize=(6.5, 3.6))
for t_air, color in [(24, COLORS["blue"]), (32, COLORS["orange"]), (40, COLORS["gray"])]:
    ax.plot(g_grid, pv_power(1000, g_grid, t_air)[2], color=color, label=f"Ta = {t_air} °C")
ax.scatter([800, 500], [p_forecast, p_actual], color="black", zorder=3)
ax.annotate("forecast 628 kW", (800, p_forecast), xytext=(520, 700), arrowprops=dict(arrowstyle="->"))
ax.annotate("actual 409 kW", (500, p_actual), xytext=(560, 250), arrowprops=dict(arrowstyle="->"))
ax.set_xlabel("irradiance G (W/m²)")
ax.set_ylabel("AC power (kW)")
ax.set_title("1,000 kW PV plant (Example 8.1 parameters)")
ax.legend()
plt.show()
```

**คำถามชวนคิด:** เส้นสามเส้นในกราฟแทบขนานกันและห่างกันไม่มาก แต่การเปลี่ยนแสงจาก 800 เป็น 500 W/m² ทำให้กำลังผลิตเปลี่ยนเกือบ 220 kW ข้อนี้บอกอะไรเกี่ยวกับตัวแปรที่ควรลงทุนพยากรณ์ให้แม่นที่สุด

### 1.2 สร้างข้อมูลโหลดและโซลาร์เซลล์รายชั่วโมงจำลองหนึ่งปี

ระบบไฟฟ้าขนาดเล็กสมมติมีโหลดสูงสุดราว 650 MW และฟาร์มโซลาร์เซลล์ 200 MW ข้อมูลจำลองสร้างจากกลไกที่หนังสืออธิบาย

- **อุณหภูมิอากาศ** = ฤดูกาล (ร้อนสุดราวปลายเดือนเมษายน) + รอบวัน (สูงสุดบ่ายสามโมง) + ความผิดปกติรายวันแบบสุ่มที่ต่อเนื่องกันหลายวัน
- **โหลด** = รูปแบบรายชั่วโมงของวันทำการ/วันหยุด + โหลดเครื่องปรับอากาศที่เพิ่มตามอุณหภูมิเหนือ 26 °C + สัญญาณรบกวน
- **โซลาร์เซลล์** = แสงในวันฟ้าใส (Clear-sky) × สัดส่วนที่เมฆยอมให้ผ่าน แล้วแปลงเป็นกำลังด้วย `pv_power` ฤดูฝน (กลางเดือนพฤษภาคมถึงตุลาคม) มีเมฆมากกว่า
- **ค่าพยากรณ์อากาศล่วงหน้าหนึ่งวัน** = ค่าจริงบวกความคลาดเคลื่อน (อุณหภูมิ σ = 1 °C และเมฆรายวัน σ = 0.15) เพราะตอนพยากรณ์จริงเราไม่รู้ค่าวัดของวันพรุ่งนี้ (ทบทวนการรั่วของข้อมูลในหัวข้อ 4.2)

```python
def simulate_grid_year(seed=RANDOM_SEED, pv_capacity_mw=200.0):
    """ข้อมูลรายชั่วโมงจำลองหนึ่งปีของโหลด อุณหภูมิ โซลาร์เซลล์ และค่าพยากรณ์อากาศล่วงหน้าหนึ่งวัน"""
    rng = np.random.default_rng(seed)
    time_index = pd.date_range("2025-01-01", periods=365 * 24, freq="h")
    n = len(time_index)
    hour = time_index.hour.to_numpy()
    day = time_index.dayofyear.to_numpy() - 1
    workday = (time_index.dayofweek.to_numpy() < 5).astype(int)

    # อุณหภูมิ: ความผิดปกติรายวันแบบ AR(1) ทำให้วันร้อนมักเกิดติดกันหลายวัน
    anomaly = np.zeros(365)
    for d in range(1, 365):
        anomaly[d] = 0.8 * anomaly[d - 1] + rng.normal(0, 0.8)
    temp = (29 + 3 * np.sin(2 * np.pi * (day - 30) / 365) + 4 * np.sin(2 * np.pi * (hour - 9) / 24)
            + anomaly[day] + rng.normal(0, 0.4, n))

    # โหลด (MW): รูปแบบรายชั่วโมง วันหยุดลดลงร้อยละ 12 และโหลดเครื่องปรับอากาศ
    profile = 0.72 + 0.25 * np.exp(-((hour - 14) / 4.5) ** 2) + 0.12 * np.exp(-((hour - 20) / 2.0) ** 2)
    load = 500 * profile * (1 - 0.12 * (1 - workday)) + 12 * np.maximum(temp - 26, 0) + rng.normal(0, 8, n)

    # โซลาร์เซลล์: แสงฟ้าใส × สัดส่วนที่เมฆยอมให้ผ่าน
    clear_sky = 1000 * np.clip(np.sin(np.pi * (hour + 0.5 - 6) / 12), 0, None) ** 1.3
    rainy_season = (np.arange(365) >= 135) & (np.arange(365) <= 290)        # กลางพฤษภาคม–ตุลาคม
    cloud_day = np.clip(rng.beta(2, 3, 365) + 0.25 * rainy_season, 0, 0.95)  # สัดส่วนเมฆรายวัน
    clearness = np.clip(1 - cloud_day[day] * rng.uniform(0.3, 1.3, n), 0.05, 1.0)
    irradiance = clear_sky * clearness
    pv = pv_power(pv_capacity_mw, irradiance, temp)[2]

    data = pd.DataFrame({
        "load": load, "temp": temp, "pv": pv, "irradiance": irradiance, "clear_sky": clear_sky,
        "hour": hour, "dow": time_index.dayofweek, "workday": workday, "doy": day + 1,
        "temp_fc": temp + rng.normal(0, 1.0, n),                                   # พยากรณ์อุณหภูมิ
        "cloud_fc": np.clip(cloud_day[day] + rng.normal(0, 0.15, 365)[day], 0, 1),  # พยากรณ์เมฆรายวัน
    }, index=time_index)
    data["net_load"] = data["load"] - data["pv"]
    return data


grid = simulate_grid_year()
print(grid.shape)
print(grid[["load", "pv", "net_load", "temp"]].describe().round(1).T[["mean", "min", "max"]])

# วันทำการหนึ่งวันในเดือนเมษายน: โหลด โซลาร์ และโหลดสุทธิ (เส้นโค้งเป็ด รูปที่ 8.2 ข)
day_slice = grid.loc["2025-04-23"]
fig, ax = plt.subplots(figsize=(7, 3.6))
ax.plot(day_slice["hour"], day_slice["load"], color=COLORS["blue"], label="load")
ax.fill_between(day_slice["hour"], 0, day_slice["pv"], color=COLORS["orange"], alpha=0.4, label="PV")
ax.plot(day_slice["hour"], day_slice["net_load"], color=COLORS["green"], lw=2, label="net load = load − PV")
ax.set_xlabel("hour")
ax.set_ylabel("MW")
ax.set_title("Simulated workday, 23 April 2025")
ax.legend(loc="upper left")
plt.show()
```

**คำถามชวนคิด:** โหลดสุทธิช่วงเที่ยงต่ำกว่าโหลดจริงมาก แต่ช่วง 17:00–19:00 น. ชันขึ้นเร็ว โรงไฟฟ้าแบบใดเหมาะจะรับช่วงชันนี้ และถ้ากำลังติดตั้งของโซลาร์เพิ่มเป็นสองเท่า รูปร่างของกราฟจะเปลี่ยนอย่างไร

### 1.3 เปรียบเทียบวิธีพยากรณ์โหลดล่วงหน้าหนึ่งวันด้วยการตรวจสอบแบบเลื่อนจุดเริ่มต้น

ตามตารางที่ 8.2 ในหนังสือ เราเริ่มจากค่าฐานแล้วค่อยเพิ่มความซับซ้อน

1. **การพยากรณ์แบบคงเดิม (Persistence)** ทายว่าโหลดชั่วโมงนี้ของพรุ่งนี้เท่ากับชั่วโมงเดียวกันของวันนี้ (`lag24`)
2. **การถดถอยเชิงเส้นพหุคูณ** ใช้ชั่วโมงแบบ one-hot แยกวันทำการ/วันหยุด อุณหภูมิพยากรณ์ และกำลังสองของอุณหภูมิ (แนวทางของ Hong et al., 2010)
3. **ต้นไม้แบบ Ensemble** ใช้ `HistGradientBoostingRegressor` กับชั่วโมง วันในสัปดาห์ วันทำการ อุณหภูมิพยากรณ์ และโหลดย้อนหลัง 24 และ 168 ชั่วโมง

การตรวจสอบแบบเลื่อนจุดเริ่มต้น (Rolling-origin) ใช้ `TimeSeriesSplit` ฝึกด้วยข้อมูลก่อนหน้าทั้งหมดแล้วทดสอบกับ 4 สัปดาห์ถัดไป เลื่อนไป 5 ครั้ง ไม่มีการสุ่มสลับลำดับเวลา แล้วรายงาน **คะแนนทักษะ** $\text{SS} = 1 - \text{RMSE}_{\text{โมเดล}}/\text{RMSE}_{\text{คงเดิม}}$

```python
def add_lag_features(data):
    out = data.copy()
    out["lag24"] = out["load"].shift(24)
    out["lag168"] = out["load"].shift(168)
    return out.dropna()


def linear_design(data):
    """เมทริกซ์ออกแบบของการถดถอยเชิงเส้นพหุคูณ: ชั่วโมง × วันทำการ (one-hot), T, T²"""
    hour_day = pd.get_dummies(data["hour"].astype(str) + "_" + data["workday"].astype(str), dtype=float)
    hour_day["temp"] = data["temp_fc"]
    hour_day["temp2"] = data["temp_fc"] ** 2
    return hour_day


TREE_FEATURES = ["hour", "dow", "workday", "temp_fc", "lag24", "lag168"]   # ไม่ใส่ doy ดูหมายเหตุในหัวข้อ 1.4
load_data = add_lag_features(grid)
x_linear = linear_design(load_data)
splitter = TimeSeriesSplit(n_splits=5, test_size=28 * 24)

fold_rows, oof = [], []
for fold, (train_idx, test_idx) in enumerate(splitter.split(load_data), start=1):
    train, test = load_data.iloc[train_idx], load_data.iloc[test_idx]
    pred_persist = test["lag24"].to_numpy()
    linear = LinearRegression().fit(x_linear.iloc[train_idx], train["load"])
    pred_linear = linear.predict(x_linear.iloc[test_idx])
    tree = HistGradientBoostingRegressor(max_iter=300, learning_rate=0.05, random_state=RANDOM_SEED)
    tree.fit(train[TREE_FEATURES], train["load"])
    pred_tree = tree.predict(test[TREE_FEATURES])
    fold_rows.append({"fold": fold, "test start": test.index[0].date(), "persistence": rmse(test["load"], pred_persist),
                      "linear": rmse(test["load"], pred_linear), "gradient boosting": rmse(test["load"], pred_tree)})
    oof.append(pd.DataFrame({"actual": test["load"], "persistence": pred_persist, "linear": pred_linear,
                             "gradient boosting": pred_tree}, index=test.index))

fold_table = pd.DataFrame(fold_rows).set_index("fold")
load_oof = pd.concat(oof)
summary = fold_table[["persistence", "linear", "gradient boosting"]].mean().to_frame("mean RMSE (MW)")
summary["skill score vs persistence"] = 1 - summary["mean RMSE (MW)"] / summary.loc["persistence", "mean RMSE (MW)"]
print("RMSE (MW) ของแต่ละรอบการเลื่อนจุดเริ่มต้น")
print(fold_table.round(2))
print()
print(summary.round(3))

week = load_oof.loc["2025-12-08":"2025-12-14"]
fig, ax = plt.subplots(figsize=(9, 3.4))
ax.plot(week.index, week["actual"], color="black", lw=1.5, label="actual")
ax.plot(week.index, week["persistence"], color=COLORS["gray"], ls="--", label="persistence (lag 24 h)")
ax.plot(week.index, week["linear"], color=COLORS["blue"], label="linear regression")
ax.plot(week.index, week["gradient boosting"], color=COLORS["orange"], label="gradient boosting")
ax.set_ylabel("load (MW)")
ax.set_title("Day-ahead load forecast, one test week (Mon–Sun)")
ax.legend(ncol=4, fontsize=8, loc="lower center")
plt.show()

# คะแนนทักษะของตัวอย่างที่ 4.13 ที่หนังสืออ้างในหัวข้อ 8.1.1
compare("คะแนนทักษะของตัวอย่างที่ 4.13 (เทียบค่าเฉลี่ย)", 1 - 13.33 / 85.9, 0.845, 5e-4)
```

**คำถามชวนคิด**

1. การพยากรณ์แบบคงเดิมพลาดมากที่สุดในวันใดของสัปดาห์ (ดูวันจันทร์และวันเสาร์ในกราฟ) เพราะเหตุใด
2. ต้นไม้แบบ Ensemble ได้ RMSE ต่ำสุด แต่ตารางที่ 8.2 ระบุว่ามัน "ประมาณค่านอกช่วงไม่ได้" ถ้าปีหน้ามีวันที่ร้อนกว่าทุกวันในข้อมูลฝึก โมเดลใดน่าจะพลาดน้อยกว่า ลองตรวจด้วยการป้อน `temp_fc` ที่สูงกว่าค่าสูงสุดในข้อมูลฝึก 3 °C

### 1.4 พยากรณ์ควอนไทล์ของโซลาร์เซลล์และตรวจความครอบคลุม (ตัวอย่างที่ 8.2)

ความไม่แน่นอนของโซลาร์เซลล์มาจากเมฆเป็นหลัก เราฝึก `HistGradientBoostingRegressor(loss="quantile")` สามโมเดลสำหรับ P10, P50 และ P90 ซึ่งภายในใช้ **Pinball Loss** เป็นฟังก์ชันสูญเสีย คุณลักษณะคือชั่วโมง แสงฟ้าใส (คำนวณได้ล่วงหน้า) เมฆพยากรณ์ และอุณหภูมิพยากรณ์ ฝึกด้วย 9 เดือนแรกและทดสอบกับ 3 เดือนสุดท้าย ใช้เฉพาะชั่วโมงที่มีแสง

> **หมายเหตุเรื่องคุณลักษณะ "วันของปี"** ข้อมูลทดสอบ (ตุลาคม–ธันวาคม) มีวันของปีสูงกว่าทุกวันในชุดฝึก ต้นไม้จึงต้องประมาณค่านอกช่วง ซึ่งเป็นจุดอ่อนในตารางที่ 8.2 ในการทดลองระหว่างจัดทำ notebook นี้ การใส่ `doy` ทำให้สัดส่วนที่ค่าจริงต่ำกว่า P90 เหลือเพียงราว 0.54 แทนที่จะใกล้ 0.90 เราจึงให้ฤดูกาลเข้ามาผ่านแสงฟ้าใส เมฆพยากรณ์ และอุณหภูมิแทน ลองเพิ่ม `"doy"` กลับเข้าไปใน `PV_FEATURES` แล้วดูผลด้วยตนเอง

ตัววัดที่ใช้

- `mean_pinball_loss` ของ scikit-learn (แบบฝึกปฏิบัติข้อ 1 จะให้เขียนเอง)
- **ความครอบคลุม (Coverage)** ของช่วง P10–P90 ควรใกล้ 0.80 และสัดส่วนที่ค่าจริงเกิน P90 ควรใกล้ 0.10

```python
# ตัวอย่างที่ 8.2 ด้วยฟังก์ชันของ scikit-learn (ค่าเดียวจึงเท่ากับ Pinball Loss ของจุดนั้น)
compare("L_0.9 เมื่อค่าจริง 945, P90 = 960", mean_pinball_loss([945], [960], alpha=0.9), 1.5, 1e-9)
compare("L_0.9 เมื่อค่าจริง 975, P90 = 960", mean_pinball_loss([975], [960], alpha=0.9), 13.5, 1e-9)

PV_FEATURES = ["hour", "clear_sky", "cloud_fc", "temp_fc"]
daylight = grid[grid["clear_sky"] > 0]
pv_train, pv_test = daylight.loc[:"2025-09-30"], daylight.loc["2025-10-01":]
quantiles = [0.1, 0.5, 0.9]
pv_pred = {}
for q in quantiles:
    model = HistGradientBoostingRegressor(loss="quantile", quantile=q, max_iter=300, learning_rate=0.05,
                                          random_state=RANDOM_SEED)
    pv_pred[q] = model.fit(pv_train[PV_FEATURES], pv_train["pv"]).predict(pv_test[PV_FEATURES])

y_pv = pv_test["pv"].to_numpy()
rows = []
for q in quantiles:
    rows.append({"quantile": f"P{int(q * 100)}", "pinball loss (MW)": mean_pinball_loss(y_pv, pv_pred[q], alpha=q),
                 "fraction of actual below": np.mean(y_pv < pv_pred[q])})
print(pd.DataFrame(rows).round(3).to_string(index=False))
coverage_80 = np.mean((y_pv >= pv_pred[0.1]) & (y_pv <= pv_pred[0.9]))
print(f"ความครอบคลุมของช่วง P10–P90: {coverage_80:.3f} (ควรใกล้ 0.80)")

# ช่วงที่ "มั่นใจเกินไป": หดช่วงเหลือครึ่งหนึ่งรอบ P50 แบบที่ตัวอย่างที่ 8.2 เตือนไว้
narrow_lo = pv_pred[0.5] - 0.5 * (pv_pred[0.5] - pv_pred[0.1])
narrow_hi = pv_pred[0.5] + 0.5 * (pv_pred[0.9] - pv_pred[0.5])
print(f"ถ้าหดช่วงเหลือครึ่งหนึ่ง ความครอบคลุมเหลือ {np.mean((y_pv >= narrow_lo) & (y_pv <= narrow_hi)):.3f} "
      f"และค่าจริงเกินขอบบน {np.mean(y_pv > narrow_hi):.1%} ของเวลา")

# P50 เทียบการพยากรณ์แบบคงเดิม (ชั่วโมงเดียวกันของเมื่อวาน)
persist_pv = grid["pv"].shift(24).loc[pv_test.index].to_numpy()
ss_pv = 1 - rmse(y_pv, pv_pred[0.5]) / rmse(y_pv, persist_pv)
print(f"RMSE ของ P50 = {rmse(y_pv, pv_pred[0.5]):.1f} MW, คงเดิม = {rmse(y_pv, persist_pv):.1f} MW, คะแนนทักษะ = {ss_pv:.3f}")
print(f"RMSE หารด้วยกำลังติดตั้ง = {rmse(y_pv, pv_pred[0.5]) / 200:.1%} (ใช้แทน MAPE ซึ่งใช้ไม่ได้ตอนเช้า/เย็น)")

show = pv_test.loc["2025-11-03":"2025-11-07"].index
mask = pv_test.index.isin(show)
fig, ax = plt.subplots(figsize=(9, 3.4))
x_axis = np.arange(mask.sum())
ax.fill_between(x_axis, pv_pred[0.1][mask], pv_pred[0.9][mask], color=COLORS["orange"], alpha=0.3, label="P10–P90")
ax.plot(x_axis, pv_pred[0.5][mask], color=COLORS["orange"], label="P50")
ax.plot(x_axis, y_pv[mask], color="black", lw=1.2, label="actual")
ax.set_xlabel("daylight hour index (5 days)")
ax.set_ylabel("PV power (MW)")
ax.set_title("Day-ahead quantile forecast of a 200 MW PV plant")
ax.legend(ncol=3, fontsize=8)
plt.show()
```

### 1.5 จากความคลาดเคลื่อนสู่กำลังสำรอง (ตัวอย่างที่ 8.3)

สูตรในหนังสือสมมติว่าความคลาดเคลื่อนของโหลดและของโซลาร์เซลล์แจกแจงปกติและเป็นอิสระกัน $R = z\sqrt{\sigma_{\text{โหลด}}^2 + \sigma_{\text{หมุนเวียน}}^2}$ เซลล์ถัดไปตรวจตัวอย่างในหนังสือ แล้วใช้ความคลาดเคลื่อนจริงของโมเดลในหัวข้อ 1.3–1.4 (ช่วงเวลาทดสอบที่ซ้อนกัน) เทียบกำลังสำรองจากสูตรกับควอนไทล์ 95 จากข้อมูลโดยตรง

```python
def reserve_mw(sigma_load, sigma_renewable, z=1.645):
    """กำลังสำรองจากความคลาดเคลื่อนที่เป็นอิสระกัน (หัวข้อ 8.1.1)"""
    return z * math.hypot(sigma_load, sigma_renewable)


compare("ตัวอย่างที่ 8.3: σ_สุทธิ (MW)", math.hypot(30, 40), 50, 1e-9)
compare("ตัวอย่างที่ 8.3: กำลังสำรอง (MW)", reserve_mw(30, 40), 82.25, 1e-9)
compare("ตัวอย่างที่ 8.3: ลดลงเมื่อ σ_หมุนเวียน = 30 (MW)", reserve_mw(30, 40) - reserve_mw(30, 30), 12.5, 0.05)
compare("z ด้านเดียวที่ 95% จาก scipy", stats.norm.ppf(0.95), 1.645, 5e-4)

# ความคลาดเคลื่อนจริงในช่วงเวลาที่ทั้งสองการพยากรณ์มีค่า (ทั้งวัน: กลางคืน PV = 0 พยากรณ์ได้แน่นอน)
pv_fc_all = pd.Series(0.0, index=grid.loc["2025-10-01":].index)
pv_fc_all.loc[pv_test.index] = pv_pred[0.5]
common = load_oof.index.intersection(pv_fc_all.index)
load_err = load_oof.loc[common, "actual"] - load_oof.loc[common, "gradient boosting"]
pv_err = grid.loc[common, "pv"] - pv_fc_all.loc[common]
net_err = load_err - pv_err                       # โหลดสุทธิจริง − โหลดสุทธิที่พยากรณ์
r_formula = reserve_mw(load_err.std(), pv_err.std())
r_empirical = np.quantile(net_err, 0.95)
print(f"σ_โหลด = {load_err.std():.1f} MW, σ_โซลาร์ = {pv_err.std():.1f} MW, สหสัมพันธ์ = {np.corrcoef(load_err, pv_err)[0, 1]:.2f}")
print(f"กำลังสำรองจากสูตร (ปกติ + อิสระ) = {r_formula:.1f} MW")
print(f"ควอนไทล์ 95 ของความคลาดเคลื่อนโหลดสุทธิจากข้อมูล = {r_empirical:.1f} MW")
print(f"ความเบ้ของความคลาดเคลื่อนโซลาร์ = {stats.skew(pv_err):.2f} (0 คือสมมาตร)")
```

**คำถามชวนคิด:** ถ้ากำลังสำรองจากสูตรต่างจากควอนไทล์จากข้อมูล ความต่างมาจากสมมติฐานข้อใดในหนังสือ ("ปกติ" หรือ "อิสระ") ลองดูความเบ้และสหสัมพันธ์ที่พิมพ์ออกมา และถ้าต้องเลือกตัวเลขหนึ่งตัวไปใช้กำหนดกำลังสำรองจริง คุณจะเลือกตัวใด และต้องตรวจอะไรเพิ่ม

### 2.1 การจัดสรรกำลังผลิตอย่างประหยัด (ตัวอย่างที่ 8.4)

เครื่องกำเนิดไฟฟ้าสามเครื่องมีต้นทุน $C_i(P_i) = a_iP_i^2 + b_iP_i + c_i$ และกำลังผลิตที่ระดับ $\lambda$ คือ $P_i(\lambda) = (\lambda - b_i)/(2a_i)$ ตรึงไว้ในช่วง $[P_i^{\min}, P_i^{\max}]$ เซลล์ถัดไปใช้ `scipy.optimize.minimize` (วิธี SLSQP) ซึ่งเป็นตัวหาค่าเหมาะที่สุดทั่วไปที่ไม่รู้เรื่องเงื่อนไขต้นทุนส่วนเพิ่มเท่ากันเลย เพื่อยืนยันคำตอบเชิงวิเคราะห์ของหนังสือ ส่วนการทำซ้ำแลมบ์ดาตามรหัสเทียมในหนังสือเป็นแบบฝึกปฏิบัติข้อ 2

```python
GENS = pd.DataFrame({
    "a": [0.010, 0.020, 0.025], "b": [2.0, 1.6, 1.75], "c": [100.0, 120.0, 80.0],
    "pmin": [50.0, 30.0, 20.0], "pmax": [200.0, 120.0, 100.0],
}, index=["G1", "G2", "G3"])


def output_at_lambda(gens, lam):
    """กำลังผลิตของทุกเครื่องที่ราคาส่วนเพิ่ม λ ตรึงที่ขีดจำกัด"""
    return np.clip((lam - gens["b"]) / (2 * gens["a"]), gens["pmin"], gens["pmax"]).to_numpy()


def generation_cost(gens, p):
    p = np.asarray(p, dtype=float)
    return (gens["a"] * p ** 2 + gens["b"] * p + gens["c"]).to_numpy()


def dispatch_slsqp(gens, demand):
    """จัดสรรกำลังผลิตด้วยตัวหาค่าเหมาะที่สุดทั่วไป (ใช้ตรวจคำตอบ)"""
    result = optimize.minimize(lambda p: generation_cost(gens, p).sum(), x0=gens["pmin"].to_numpy() + 1,
                               bounds=list(zip(gens["pmin"], gens["pmax"])), method="SLSQP",
                               constraints=[{"type": "eq", "fun": lambda p: p.sum() - demand}],
                               options={"ftol": 1e-12, "maxiter": 500})
    return result.x


# ตารางการแบ่งครึ่งช่วงของกรณี ก (D = 300 MW)
table = pd.DataFrame([{"λ": lam, **dict(zip(GENS.index, output_at_lambda(GENS, lam)))} for lam in (4, 6, 5)])
table["sum"] = table[["G1", "G2", "G3"]].sum(axis=1)
table["D − sum"] = 300 - table["sum"]
print(table.to_string(index=False))

for demand in (300, 400):
    p = dispatch_slsqp(GENS, demand)
    marginal = GENS["b"] + 2 * GENS["a"] * p
    print(f"\nD = {demand} MW: P = {np.round(p, 2)} MW, ต้นทุนส่วนเพิ่ม = {np.round(marginal.to_numpy(), 3)}")
    print(f"  ต้นทุนรวม = {generation_cost(GENS, p).sum():.3f} หน่วยเงินต่อชั่วโมง")
p300 = dispatch_slsqp(GENS, 300)
compare("กรณี ก: ต้นทุนรวม", generation_cost(GENS, p300).sum(), 1324.875, 1e-3)
compare("กรณี ก: ประหยัดเทียบการแบ่งเท่ากัน", generation_cost(GENS, [100, 100, 100]).sum()
        - generation_cost(GENS, p300).sum(), 60.125, 1e-3)
p400 = dispatch_slsqp(GENS, 400)
compare("กรณี ข: P2 (MW)", p400[1], 112.78, 0.01)
compare("กรณี ข: λ จากต้นทุนส่วนเพิ่มของ G2", GENS.loc["G2", "b"] + 2 * GENS.loc["G2", "a"] * p400[1], 6.111, 5e-4)

# ผลรวมกำลังผลิตเป็นฟังก์ชันเพิ่มของ λ (รูปที่ 8.4 ข) มีจุดหักเมื่อเครื่องติดขีดจำกัด
lams = np.linspace(3.5, 7, 200)
fig, ax = plt.subplots(figsize=(6.5, 3.6))
ax.plot(lams, [output_at_lambda(GENS, lam).sum() for lam in lams], color=COLORS["blue"], label="ΣP(λ) with limits")
ax.plot(lams, 95 * lams - 175, color=COLORS["gray"], ls=":", label="95λ − 175 (no limits)")
for demand, color in [(300, COLORS["green"]), (400, COLORS["orange"])]:
    ax.axhline(demand, color=color, ls="--", lw=1, label=f"D = {demand} MW")
ax.set_xlabel("λ (money/MWh)")
ax.set_ylabel("total generation (MW)")
ax.legend(fontsize=8)
plt.show()
```

**คำถามชวนคิด:** ที่ $D = 400$ MW เส้นสีน้ำเงินชันน้อยลงหลัง $\lambda = 6$ เพราะเหตุใด และถ้าความต้องการสูงถึง 430 MW (เท่ากับผลรวมกำลังผลิตสูงสุด) $\lambda$ จะเป็นเท่าใด ระบบยังจัดสรรได้หรือไม่

### 2.2 ประมาณสถานะด้วย WLS ตรวจข้อมูลผิดพลาด และการฉีดข้อมูลเท็จ (ตัวอย่างที่ 8.6)

ฟังก์ชัน `wls` คำนวณ $\hat{\mathbf{x}} = (\mathbf{H}^\top\mathbf{W}\mathbf{H})^{-1}\mathbf{H}^\top\mathbf{W}\mathbf{z}$ ส่วนเหลือ และ $J$ ค่าวิกฤตของไคกำลังสองได้จาก `scipy.stats.chi2.ppf`

```python
def wls(h, z, sigma):
    """กำลังสองน้อยที่สุดแบบถ่วงน้ำหนัก คืน (สถานะ, ส่วนเหลือ, J)"""
    h, z, sigma = np.atleast_2d(h), np.asarray(z, float), np.asarray(sigma, float)
    w = np.diag(1 / sigma ** 2)
    x_hat = np.linalg.solve(h.T @ w @ h, h.T @ w @ z)
    residual = z - h @ x_hat
    return x_hat, residual, float(np.sum((residual / sigma) ** 2))


H_2BUS = np.array([[-10.0], [10.0], [10.0]])      # M1: P12, M2: P21, M3: กำลังฉีดที่บัส 2
SIGMA_2BUS = np.array([0.01, 0.02, 0.02])
z_normal = np.array([0.62, -0.58, -0.60])
critical = stats.chi2.ppf(0.95, df=3 - 1)
print(f"ค่าวิกฤต χ²(0.95, 2) = {critical:.2f}")

for name, z in [("ปกติ", z_normal), ("M2 เสีย (อ่าน −0.40)", np.array([0.62, -0.40, -0.60]))]:
    theta, r, j = wls(H_2BUS, z, SIGMA_2BUS)
    print(f"{name}: θ2 = {theta[0]:.4f} rad, P12 = {-10 * theta[0]:.3f} pu, |r|/σ = {np.round(np.abs(r) / SIGMA_2BUS, 2)}, "
          f"J = {j:.2f} → {'พบข้อมูลผิดพลาด' if j > critical else 'ผ่าน'}")
theta_cut, _, _ = wls(H_2BUS[[0, 2]], np.array([0.62, -0.60]), SIGMA_2BUS[[0, 2]])
compare("ตัด M2 แล้ว P12 (pu)", -10 * theta_cut[0], 0.616, 1e-9)

# การฉีดข้อมูลเท็จ a = Hc เลื่อนสถานะโดยที่ J ไม่เปลี่ยน
attack = (H_2BUS * -0.002).ravel()
theta_fdi, _, j_fdi = wls(H_2BUS, z_normal + attack, SIGMA_2BUS)
print(f"ถูกฉีด {attack}: P12 = {-10 * theta_fdi[0]:.3f} pu, J = {j_fdi:.2f} (เท่าเดิม การทดสอบมองไม่เห็น)")

# ตรวจว่าเมื่อค่าวัดปกติจริง J แจกแจงไคกำลังสอง 2 องศาเสรี จึงแจ้งเตือนผิดราวร้อยละ 5
rng = np.random.default_rng(RANDOM_SEED)
true_z = (H_2BUS * -0.061).ravel()
j_values = np.array([wls(H_2BUS, true_z + rng.normal(0, SIGMA_2BUS), SIGMA_2BUS)[2] for _ in range(5000)])
print(f"จำลอง 5,000 ครั้ง: ค่าเฉลี่ยของ J = {j_values.mean():.2f} (ทฤษฎี 2), แจ้งเตือนผิด {np.mean(j_values > critical):.1%} (ทฤษฎี 5%)")
```

**คำถามชวนคิด:** การจำลองแสดงว่าแม้ค่าวัดทุกตัวปกติ การทดสอบก็แจ้งเตือนผิดราวร้อยละ 5 ถ้าศูนย์ควบคุมประมาณสถานะทุก 5 วินาที จะเกิดการแจ้งเตือนผิดกี่ครั้งต่อชั่วโมง และควรออกแบบระบบแจ้งเตือนอย่างไร (ทบทวนข้อควรระวังเรื่องความล้าจากสัญญาณเตือนในหัวข้อ 8.2.2)

### 2.3 จำแนกชนิดของฟอลต์ด้วยกฎ (ตัวอย่างที่ 8.7)

กฎสองข้อในหนังสือเขียนเป็นฟังก์ชันได้ไม่กี่บรรทัด ซึ่งเหมาะกับฟอลต์กระแสสูง แต่แยกเหตุการณ์ E5 (โหลดไม่สมดุล หรือฟอลต์อิมพีแดนซ์สูง) ไม่ได้

```python
def classify_fault(i_a, i_b, i_c, i_0x3, i_pre=0.40):
    """กฎในตัวอย่างที่ 8.7: เฟสที่เกิน 2 เท่าของกระแสก่อนเกิดเหตุ และ 3I0 เกินร้อยละ 10 ของกระแสเฟสสูงสุด"""
    phases = "".join(name for name, i in zip("ABC", (i_a, i_b, i_c)) if i > 2 * i_pre)
    ground = i_0x3 > 0.10 * max(i_a, i_b, i_c)
    if not phases:
        return "no fault by rule 1" + (" (ground current present)" if ground else "")
    return "-".join(phases) + ("-G" if ground else "")


events = pd.DataFrame({"I_A": [4.80, 0.40, 0.42, 5.10, 0.46], "I_B": [0.41, 3.90, 4.50, 5.00, 0.40],
                       "I_C": [0.39, 3.90, 4.30, 5.20, 0.40], "3I0": [4.40, 0.02, 2.60, 0.05, 0.06]},
                      index=["E1", "E2", "E3", "E4", "E5"])
events["rule result"] = [classify_fault(*row) for row in events[["I_A", "I_B", "I_C", "3I0"]].to_numpy()]
print(events.to_string())
```

### 3.1 ดัชนีสุขภาพและความคุ้มค่าของนโยบายบำรุงรักษา (ตัวอย่างที่ 8.9–8.10)

```python
def health_index(weights, scores, s_max=4):
    """ดัชนีสุขภาพแบบค่าเฉลี่ยถ่วงน้ำหนัก (ร้อยละ)"""
    weights, scores = np.asarray(weights, float), np.asarray(scores, float)
    return float((weights * scores).sum() / (s_max * weights.sum()) * 100)


HI_WEIGHTS = [10, 6, 8, 4, 3]          # DGA, น้ำมัน, PD, ประวัติโหลด, อายุ
compare("ตัวอย่างที่ 8.9: HI (%)", health_index(HI_WEIGHTS, [1, 2, 3, 3, 2]), 51.6, 0.05)
compare("ตัวอย่างที่ 8.9: HI เมื่อ DGA = 0 (%)", health_index(HI_WEIGHTS, [0, 2, 3, 3, 2]), 43.5, 0.05)


def maintenance_cost_predictive(recall, precision=0.8, failures=12.5, fixed=450_000, planned=30_000,
                                unplanned=260_000, false_alarm=5_000):
    """ต้นทุนรายปีของนโยบาย ค ในตัวอย่างที่ 8.10 (บาท)"""
    detected = recall * failures
    false_alarms = detected / precision - detected
    return fixed + detected * planned + (failures - detected) * unplanned + false_alarms * false_alarm


cost_run_to_failure = 12.5 * 260_000
cost_time_based = 25 * 20_000 + 0.3 * 12.5 * 260_000
compare("นโยบาย ก (บาท)", cost_run_to_failure, 3_250_000, 1e-6)
compare("นโยบาย ข (บาท)", cost_time_based, 1_475_000, 1e-6)
compare("นโยบาย ค r = 0.8 (บาท)", maintenance_cost_predictive(0.8), 1_412_500, 1e-6)
compare("นโยบาย ค r = 0.6 (บาท)", maintenance_cost_predictive(0.6), 1_984_375, 1e-6)
r_break_even = optimize.brentq(lambda r: maintenance_cost_predictive(r) - cost_time_based, 0, 1)
compare("จุดคุ้มทุนของความระลึก", r_break_even, 0.778, 5e-4)

recalls = np.linspace(0.5, 1.0, 51)
fig, ax = plt.subplots(figsize=(6.5, 3.4))
ax.plot(recalls, [maintenance_cost_predictive(r) / 1e6 for r in recalls], color=COLORS["blue"], label="predictive (precision 0.8)")
ax.plot(recalls, [maintenance_cost_predictive(r, precision=0.3) / 1e6 for r in recalls], color=COLORS["blue"], ls=":",
        label="predictive (precision 0.3)")
ax.axhline(cost_time_based / 1e6, color=COLORS["orange"], ls="--", label="time-based every 2 years")
ax.axvline(r_break_even, color=COLORS["gray"], lw=0.8)
ax.set_xlabel("recall r")
ax.set_ylabel("annual cost (million THB)")
ax.legend(fontsize=8)
plt.show()
```

**คำถามชวนคิด:** เส้นประ (ความแม่นยำ 0.3) อยู่สูงกว่าเส้นทึบเพียงเล็กน้อย เพราะการตรวจสัญญาณแจ้งเตือนผิดในตัวอย่างนี้ถูก (5,000 บาท) ถ้าทุกการแจ้งเตือนต้องหยุดสายการผลิตเพื่อตรวจ (ครั้งละ 50,000 บาท) ข้อสรุปเรื่องการตั้งเกณฑ์ให้ความระลึกสูงจะยังใช้ได้หรือไม่ ลองเปลี่ยนพารามิเตอร์ `false_alarm`

### 3.2 คุณลักษณะในโดเมนเวลา (ตัวอย่างที่ 8.11)

RMS อัตราส่วนยอดต่อ RMS (Crest Factor) และความโด่ง (Kurtosis) หารด้วย $N$ ตามข้อตกลงของหนังสือ ความโด่งตรวจซ้ำได้ด้วย `scipy.stats.kurtosis(x, fisher=False)`

```python
def time_features(x):
    """ค่าเฉลี่ย RMS ค่ายอด อัตราส่วนยอดต่อ RMS และความโด่งของหน้าต่างสัญญาณ (หารด้วย N)"""
    x = np.asarray(x, dtype=float)
    centered = x - x.mean()
    rms_value = math.sqrt(np.mean(x ** 2))
    peak = np.max(np.abs(x))
    kurt = np.mean(centered ** 4) / np.mean(centered ** 2) ** 2
    return {"mean": x.mean(), "RMS": rms_value, "peak": peak, "crest": peak / rms_value, "kurtosis": kurt}


window_normal = [2, -1, 2, -2, 1, -2, 1, -1]
window_fault = [4, -1, 0, -1, 0, -1, 0, -1]
feature_table = pd.DataFrame({"normal": time_features(window_normal), "early fault": time_features(window_fault)}).T
print(feature_table.round(3))
compare("ความโด่ง (เริ่มเสียหาย)", feature_table.loc["early fault", "kurtosis"], 5.2, 1e-9)
compare("ความโด่งจาก scipy", stats.kurtosis(window_fault, fisher=False), 5.2, 1e-9)
```

### 3.3 สัญญาณการสั่นจำลอง ความถี่ของความเสียหาย และการวิเคราะห์เอนเวโลป (ตัวอย่างที่ 8.12)

สร้างสัญญาณ 1 วินาทีที่ $f_s$ = 12,000 Hz ของมอเตอร์ในตัวอย่างที่ 8.12 (1,470 rpm, ตลับลูกปืน 9 เม็ด)

- **ปกติ** = การสั่นที่ความถี่รอบหมุน $1\times$ และ $2\times$ + สัญญาณรบกวน
- **รางนอกเสียหาย** = ปกติ + แรงกระแทกสั้น ๆ ทุกครั้งที่เม็ดลูกปืนกลิ้งผ่านรอย (อัตรา BPFO และมีการลื่นไถลเล็กน้อย) แต่ละครั้งกระตุ้นการสั่นพ้องของโครงสร้างที่ 3 kHz ซึ่งหน่วงหายภายในราว 1 มิลลิวินาที

การวิเคราะห์เอนเวโลปทำสามขั้น: กรองแถบผ่าน 2–4 kHz ด้วย `scipy.signal.butter` + `filtfilt` → หาขนาดเอนเวโลปด้วย `scipy.signal.hilbert` → FFT ของเอนเวโลป

```python
def bearing_frequencies(rpm, n_balls, ball_d, pitch_d, contact_deg=0.0):
    """ความถี่รอบหมุน BPFO และ BPFI (Hz)"""
    fr = rpm / 60
    ratio = ball_d / pitch_d * math.cos(math.radians(contact_deg))
    return fr, n_balls * fr / 2 * (1 - ratio), n_balls * fr / 2 * (1 + ratio)


def rotor_bar_sidebands(rpm, poles=4, f_supply=50.0):
    """สลิปและแถบข้าง (1 ± 2s) f ของซี่โรเตอร์หัก (ตัวอย่างที่ 8.13)"""
    n_sync = 120 * f_supply / poles
    slip = (n_sync - rpm) / n_sync
    return slip, f_supply * (1 - 2 * slip), f_supply * (1 + 2 * slip)


FR, BPFO, BPFI = bearing_frequencies(1470, 9, 7.94, 39.04)
compare("BPFO (Hz)", BPFO, 87.8, 0.05)
compare("BPFI (Hz)", BPFI, 132.7, 0.05)
slip, sb_low, sb_high = rotor_bar_sidebands(1470)
print(f"ตัวอย่างที่ 8.13: สลิป = {slip:.3f}, แถบข้าง = {sb_low:.1f} และ {sb_high:.1f} Hz, "
      f"0.3% = {20 * math.log10(0.003):.1f} dB, 1.5% = {20 * math.log10(0.015):.1f} dB")

FS, DURATION = 12_000, 1.0


def simulate_vibration(fault, seed=RANDOM_SEED, impact_amp=6.0):
    rng = np.random.default_rng(seed)
    t = np.arange(int(FS * DURATION)) / FS
    x = 1.0 * np.sin(2 * np.pi * FR * t) + 0.3 * np.sin(2 * np.pi * 2 * FR * t + 0.5) + rng.normal(0, 0.5, t.size)
    if fault:
        period = 1 / BPFO
        impact_times = np.arange(0, DURATION, period) + rng.normal(0, 0.01 * period, int(np.ceil(DURATION / period)))
        ring = np.arange(int(0.004 * FS)) / FS
        burst = np.exp(-ring / 0.001) * np.sin(2 * np.pi * 3000 * ring)
        for t0 in impact_times:
            start = int(round(t0 * FS))
            if 0 <= start < t.size:
                end = min(t.size, start + burst.size)
                x[start:end] += impact_amp * rng.uniform(0.7, 1.0) * burst[:end - start]
    return t, x


def amplitude_spectrum(x):
    spectrum = np.abs(np.fft.rfft(x - np.mean(x))) * 2 / len(x)
    return np.fft.rfftfreq(len(x), 1 / FS), spectrum


def envelope_spectrum(x, band=(2000, 4000)):
    b, a = signal.butter(4, band, btype="bandpass", fs=FS)
    envelope = np.abs(signal.hilbert(signal.filtfilt(b, a, x)))
    return amplitude_spectrum(envelope)


t_vib, x_normal = simulate_vibration(fault=False)
_, x_fault = simulate_vibration(fault=True)
print(pd.DataFrame({"normal": time_features(x_normal), "outer-race fault": time_features(x_fault)}).T.round(3))

fig, axes = plt.subplots(3, 1, figsize=(9, 8))
axes[0].plot(t_vib[:1200] * 1000, x_fault[:1200], color=COLORS["orange"], lw=0.7, label="outer-race fault")
axes[0].plot(t_vib[:1200] * 1000, x_normal[:1200], color=COLORS["blue"], lw=0.7, alpha=0.7, label="normal")
axes[0].set_xlabel("time (ms)")
axes[0].set_ylabel("acc. (m/s²)")
axes[0].legend(fontsize=8)
f_raw, s_raw = amplitude_spectrum(x_fault)
axes[1].plot(f_raw, s_raw, color=COLORS["gray"], lw=0.7)
axes[1].set_xlim(0, 300)
axes[1].set_title("raw spectrum of faulty bearing: BPFO hidden near the noise floor")
f_env, s_env = envelope_spectrum(x_fault)
f_env_n, s_env_n = envelope_spectrum(x_normal)
axes[2].plot(f_env_n, s_env_n, color=COLORS["blue"], lw=0.8, label="normal")
axes[2].plot(f_env, s_env, color=COLORS["orange"], lw=0.8, label="outer-race fault")
for k in (1, 2, 3):
    axes[2].axvline(k * BPFO, color=COLORS["orange"], ls=":", lw=1)
axes[2].axvline(BPFI, color=COLORS["green"], ls=":", lw=1, label="BPFI")
axes[2].set_xlim(0, 300)
axes[2].set_xlabel("frequency (Hz)")
axes[2].set_title("envelope spectrum (2–4 kHz band): peaks at BPFO, 2×BPFO, 3×BPFO")
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

top = f_env[1:400][np.argsort(s_env[1:400])[-3:]]
print("ความถี่ของยอดสูงสุด 3 ยอดในสเปกตรัมเอนเวโลป (Hz):", np.sort(np.round(top, 1)))
```

**คำถามชวนคิด:** ในสเปกตรัมดิบ ยอดที่ BPFO แทบมองไม่เห็นเพราะพลังงานของแรงกระแทกกระจายอยู่รอบ 3 kHz แต่ในสเปกตรัมเอนเวโลปกลับชัดมาก อธิบายด้วยคำว่า "จังหวะ" ของแรงกระแทก และถ้ามอเตอร์ถูกขับด้วยอินเวอร์เตอร์ที่ความเร็ว 900 rpm ต้องแก้โค้ดส่วนใดก่อนอ่านสเปกตรัม

### 3.4 ตรวจหาความผิดปกติด้วยระยะทางมาฮาลาโนบิส (ตัวอย่างที่ 8.14)

สร้างข้อมูลปกติ 500 จุดของการสั่น RMS และอุณหภูมิตัวถังจากค่าเฉลี่ย ส่วนเบี่ยงเบนมาตรฐาน และสหสัมพันธ์ 0.8 ในตัวอย่าง แล้ว

1. ตรวจจุด R และ Q ในหนังสือด้วยพารามิเตอร์ที่กำหนด
2. ประมาณ $\boldsymbol{\mu}$ และ $\boldsymbol{\Sigma}$ จากข้อมูลจำลอง (สิ่งที่ทำได้จริงในโรงงาน) แล้วเทียบกฎ $\mu \pm 3\sigma$ กับเกณฑ์ $D_M^2 > 13.82$ บนข้อมูลปกติชุดใหม่และข้อมูลความหลวมทางกลจำลอง (การสั่นสูงขึ้นแต่อุณหภูมิไม่ขึ้นตาม)

```python
def mahalanobis_sq(points, mean, cov):
    """D_M² ของทุกจุด (แถว) เทียบกับการแจกแจงปกติที่มีค่าเฉลี่ย mean และเมทริกซ์ความแปรปรวนร่วม cov"""
    diff = np.atleast_2d(points) - np.asarray(mean)
    return np.einsum("ij,ij->i", diff @ np.linalg.inv(cov), diff)


MU_TRUE = np.array([2.0, 60.0])
STD_TRUE = np.array([0.5, 5.0])
RHO = 0.8
COV_TRUE = np.array([[STD_TRUE[0] ** 2, RHO * STD_TRUE[0] * STD_TRUE[1]],
                     [RHO * STD_TRUE[0] * STD_TRUE[1], STD_TRUE[1] ** 2]])
THRESHOLD_D2 = stats.chi2.ppf(0.999, df=2)
point_r, point_q = np.array([3.0, 68.0]), np.array([2.8, 56.0])
compare("D_M² ของจุด R", mahalanobis_sq(point_r, MU_TRUE, COV_TRUE)[0], 4.0, 1e-9)
compare("D_M² ของจุด Q", mahalanobis_sq(point_q, MU_TRUE, COV_TRUE)[0], 14.58, 0.005)
compare("เกณฑ์ χ²(0.999, 2)", THRESHOLD_D2, 13.82, 0.005)

rng = np.random.default_rng(RANDOM_SEED)
normal_train = rng.multivariate_normal(MU_TRUE, COV_TRUE, 500)
normal_new = rng.multivariate_normal(MU_TRUE, COV_TRUE, 2000)
looseness = rng.multivariate_normal(MU_TRUE, COV_TRUE, 200) + np.array([1.0, -4.0])   # สั่นขึ้น เย็นลง

mu_hat, cov_hat = normal_train.mean(axis=0), np.cov(normal_train, rowvar=False)
sd_hat = np.sqrt(np.diag(cov_hat))
print(f"ประมาณจากข้อมูล: μ = {np.round(mu_hat, 2)}, σ = {np.round(sd_hat, 2)}, ρ = {cov_hat[0, 1] / sd_hat.prod():.3f}")


def flag_3sigma(points):
    return np.any(np.abs((points - mu_hat) / sd_hat) > 3, axis=1)


def flag_mahalanobis(points):
    return mahalanobis_sq(points, mu_hat, cov_hat) > THRESHOLD_D2


rows = []
for name, rule in [("μ ± 3σ per feature", flag_3sigma), ("Mahalanobis D² > 13.82", flag_mahalanobis)]:
    rows.append({"rule": name, "false alarm on new normal data": rule(normal_new).mean(),
                 "detection of looseness": rule(looseness).mean(),
                 "flags R": bool(rule(point_r[None])[0]), "flags Q": bool(rule(point_q[None])[0])})
print(pd.DataFrame(rows).round(3).to_string(index=False))

# วงรีระดับ D² = 13.82 เทียบกรอบ μ ± 3σ (รูปที่ 8.9)
grid_v, grid_t = np.meshgrid(np.linspace(0.3, 4.2, 200), np.linspace(38, 82, 200))
d2_grid = mahalanobis_sq(np.c_[grid_v.ravel(), grid_t.ravel()], mu_hat, cov_hat).reshape(grid_v.shape)
fig, ax = plt.subplots(figsize=(6.5, 4.5))
ax.scatter(*normal_train.T, s=6, color=COLORS["gray"], alpha=0.5, label="normal (training)")
ax.scatter(*looseness.T, s=8, color=COLORS["orange"], alpha=0.5, label="simulated looseness")
ax.contour(grid_v, grid_t, d2_grid, levels=[THRESHOLD_D2], colors=COLORS["blue"])
lo, hi = mu_hat - 3 * sd_hat, mu_hat + 3 * sd_hat
ax.add_patch(plt.Rectangle(lo, *(hi - lo), fill=False, ls="--", color=COLORS["gray"]))
for label, pt, color in [("R", point_r, COLORS["green"]), ("Q", point_q, "red")]:
    ax.scatter(*pt, color=color, s=60, zorder=3)
    ax.annotate(label, pt, xytext=(5, 5), textcoords="offset points")
ax.set_xlabel("vibration RMS (mm/s)")
ax.set_ylabel("frame temperature (°C)")
ax.set_title("ellipse: D² = 13.82, dashed box: μ ± 3σ")
ax.legend(fontsize=8, loc="upper left")
plt.show()
```

**คำถามชวนคิด:** กฎ $\mu \pm 3\sigma$ แจ้งเตือนผิดบนข้อมูลปกติน้อย แต่จับความหลวมได้น้อยมาก ส่วนมาฮาลาโนบิสจับได้มากกว่ามากโดยแจ้งเตือนผิดใกล้เคียงกัน ถ้ามอเตอร์มีสองโหมด (โหลดเต็มกลางวัน โหลดเบากลางคืน) แต่เราประมาณ $\boldsymbol{\Sigma}$ จากข้อมูลรวมทั้งสองโหมด จะเกิดอะไรขึ้นกับวงรี

### 3.5 ประเมิน RUL พร้อมความไม่แน่นอน และเลือกเวลาซ่อม (ตัวอย่างที่ 8.15–8.16)

ฟิตแบบจำลองเชิงเส้นด้วย `np.polyfit` และแบบเอกซ์โพเนนเชียลด้วยการฟิตเส้นตรงกับ $\ln y$ ช่วงความเชื่อมั่นของความชันใช้ค่า $t$ จาก `scipy.stats.t` แล้วหาเวลาซ่อมที่ต้นทุนคาดหมายต่ำสุด

```python
def fit_linear_rul(t, y, threshold, t_now, confidence=0.95):
    """RUL จากเส้นตรง และช่วงของ RUL จากความไม่แน่นอนของความชัน (เริ่มจากค่าที่ฟิตได้ ณ ปัจจุบัน)

    ใช้ได้ทั้งข้อมูลที่เพิ่มขึ้นสู่เกณฑ์ (เช่น การสั่น) และลดลงสู่เกณฑ์ (เช่น ดัชนีสุขภาพ)
    "RUL low" ใช้อัตราที่เร็วที่สุดในช่วง "RUL high" ใช้อัตราที่ช้าที่สุด
    ถ้าช่วงของความชันคร่อมศูนย์ (|slope| ≤ half) ขอบไกลเป็น math.inf
    """
    t, y = np.asarray(t, float), np.asarray(y, float)
    slope, intercept = np.polyfit(t, y, 1)
    sse = np.sum((y - (intercept + slope * t)) ** 2)
    se = math.sqrt(sse / (len(t) - 2) / np.sum((t - t.mean()) ** 2))
    half = stats.t.ppf(0.5 + confidence / 2, len(t) - 2) * se
    y_now = intercept + slope * t_now
    rul = (threshold - intercept) / slope - t_now
    remaining = abs(threshold - y_now)
    rate_fast, rate_slow = abs(slope) + half, abs(slope) - half
    rul_low = remaining / rate_fast
    rul_high = remaining / rate_slow if rate_slow > 0 else math.inf
    return {"slope": slope, "intercept": intercept, "SSE": sse, "slope SE": se, "slope half-width": half,
            "RUL": rul, "RUL low": rul_low, "RUL high": rul_high}


def fit_exponential_rul(t, y, threshold, t_now):
    """RUL จากแบบจำลอง y = a e^{bt} ที่ฟิตในสเกลลอการิทึม"""
    t, y = np.asarray(t, float), np.asarray(y, float)
    b, ln_a = np.polyfit(t, np.log(y), 1)
    a = math.exp(ln_a)
    return {"a": a, "b": b, "SSE": float(np.sum((y - a * np.exp(b * t)) ** 2)),
            "RUL": math.log(threshold / a) / b - t_now}


weeks = np.arange(5)
vib_rms = np.array([2.0, 2.3, 2.5, 2.9, 3.1])
lin = fit_linear_rul(weeks, vib_rms, 4.5, t_now=4)
expo = fit_exponential_rul(weeks, vib_rms, 4.5, t_now=4)
compare("ความชันเชิงเส้น", lin["slope"], 0.28, 1e-9)
compare("RUL เชิงเส้น (สัปดาห์)", lin["RUL"], 4.9, 0.05)
compare("ขอบล่างของ RUL", lin["RUL low"], 4.2, 0.05)
compare("ขอบบนของ RUL", lin["RUL high"], 6.1, 0.05)
compare("b ของเอกซ์โพเนนเชียล", expo["b"], 0.1108, 5e-5)
compare("RUL เอกซ์โพเนนเชียล (สัปดาห์)", expo["RUL"], 3.2, 0.05)
print(f"SSE เชิงเส้น = {lin['SSE']:.4f}, SSE เอกซ์โพเนนเชียล = {expo['SSE']:.4f} (ฟิตดีพอกัน แต่ RUL ต่างกัน)")

t_plot = np.linspace(0, 10.5, 200)
fig, ax = plt.subplots(figsize=(7, 3.8))
ax.scatter(weeks, vib_rms, color="black", zorder=3, label="measured")
ax.plot(t_plot, lin["intercept"] + lin["slope"] * t_plot, color=COLORS["blue"], label=f"linear → RUL {lin['RUL']:.1f} wk")
ax.plot(t_plot, expo["a"] * np.exp(expo["b"] * t_plot), color=COLORS["orange"], label=f"exponential → RUL {expo['RUL']:.1f} wk")
y_now = lin["intercept"] + lin["slope"] * 4
half = lin["slope half-width"]
t_future = np.linspace(4, 10.5, 50)
ax.fill_between(t_future, y_now + (lin["slope"] - half) * (t_future - 4), y_now + (lin["slope"] + half) * (t_future - 4),
                color=COLORS["blue"], alpha=0.15, label="95% slope band")
ax.axhline(4.5, color="darkred", ls="--", label="threshold 4.5 mm/s")
ax.axvline(4, color=COLORS["gray"], lw=0.8)
ax.set_ylim(1.5, 5.2)
ax.set_xlabel("week")
ax.set_ylabel("vibration RMS (mm/s)")
ax.legend(fontsize=8, loc="upper left")
plt.show()


def expected_repair_cost(tau, rul_mean=4.0, rul_sd=1.0, planned=30_000, extra_if_failed=230_000, waste_per_period=5_000):
    """ต้นทุนคาดหมายเมื่อนัดซ่อมในอีก tau หน่วยเวลา (สัปดาห์ในตัวอย่างที่ 8.16 หรือเดือนในแบบฝึกหัดข้อ 13)
    waste_per_period คือมูลค่าอายุที่ทิ้งต่อหนึ่งหน่วยเวลาเดียวกับ tau"""
    return planned + extra_if_failed * stats.norm.cdf(tau, rul_mean, rul_sd) + waste_per_period * (rul_mean - tau)


cost_table = pd.DataFrame({"tau (weeks)": [1, 2, 3, 4]})
cost_table["P(RUL < tau)"] = stats.norm.cdf(cost_table["tau (weeks)"], 4, 1)
cost_table["E[C] (THB)"] = expected_repair_cost(cost_table["tau (weeks)"])
print(cost_table.round(5).to_string(index=False))
best = optimize.minimize_scalar(expected_repair_cost, bounds=(0, 4), method="bounded")
print(f"เมื่อ τ เป็นค่าต่อเนื่อง ต้นทุนต่ำสุดที่ τ = {best.x:.2f} สัปดาห์ (E[C] = {best.fun:,.0f} บาท)")
for sd in (0.5, 1.0, 1.5):
    b = optimize.minimize_scalar(lambda tau: expected_repair_cost(tau, rul_sd=sd), bounds=(0, 4), method="bounded")
    print(f"  ถ้าส่วนเบี่ยงเบนมาตรฐานของ RUL = {sd} สัปดาห์ → τ ที่ดีที่สุด = {b.x:.2f} สัปดาห์")
```

**คำถามชวนคิด:** ผลรอบสุดท้ายแสดงว่าเมื่อความไม่แน่นอนของ RUL ลดลง เวลาซ่อมที่ดีที่สุดขยับเข้าใกล้ค่าเฉลี่ย 4 สัปดาห์ ข้อนี้แปลงเป็น "มูลค่าของการพยากรณ์ที่แม่นยำขึ้น" เป็นเงินได้อย่างไร (คำแนะนำ: เทียบ E[C] ที่ดีที่สุดของแต่ละกรณี)

### 4.1 MDP ของพัดลมระบายความร้อนและการทำซ้ำค่า (ตัวอย่างที่ 8.17)

เซลล์ถัดไปนิยาม MDP ตามตารางในตัวอย่างที่ 8.17 (สถานะ C, N, H การกระทำ OFF, ON) และฟังก์ชัน `value_iteration` ตามรหัสเทียมในหนังสือ ซึ่งบันทึกค่า $V_k$ และนโยบายทุกรอบไว้ในตาราง

```python
FAN_STATES = ["C", "N", "H"]      # เย็น ปกติ ร้อน
FAN_ACTIONS = ["OFF", "ON"]
# FAN_P[s][a] = รายการ (สถานะถัดไป, ความน่าจะเป็น) ตามตารางในตัวอย่างที่ 8.17
FAN_P = {
    "C": {"OFF": [("N", 1.0)], "ON": [("C", 1.0)]},
    "N": {"OFF": [("N", 0.5), ("H", 0.5)], "ON": [("C", 0.5), ("N", 0.5)]},
    "H": {"OFF": [("H", 1.0)], "ON": [("N", 1.0)]},
}
GAMMA = 0.9


def fan_reward(state, action, fan_cost=1.0, hot_cost=10.0):
    """รางวัล R(s, a): เปิดพัดลมเสีย fan_cost ต่อก้าว อยู่ที่ H เสีย hot_cost ต่อก้าว"""
    return 0.0 - (fan_cost if action == "ON" else 0.0) - (hot_cost if state == "H" else 0.0)


def q_from_v(v, fan_cost=1.0, gamma=GAMMA):
    """แทนค่าในด้านขวาของสมการเบลล์แมน: Q(s, a) = R(s, a) + γ Σ P(s'|s, a) V(s')"""
    return {(s, a): fan_reward(s, a, fan_cost) + gamma * sum(p * v[s2] for s2, p in FAN_P[s][a])
            for s in FAN_STATES for a in FAN_ACTIONS}


def greedy(q, state):
    """เลือกการกระทำที่ Q สูงสุด เมื่อเท่ากันเลือก OFF (ข้อตกลงของหนังสือ)"""
    return "OFF" if q[(state, "OFF")] >= q[(state, "ON")] else "ON"


def value_iteration(fan_cost=1.0, gamma=GAMMA, eps=1e-10, max_iter=10_000):
    """การทำซ้ำค่าตามรหัสเทียม VALUE-ITERATION คืน V*, Q* และตารางประวัติทุกรอบ"""
    v = {s: 0.0 for s in FAN_STATES}
    history = []
    for k in range(1, max_iter + 1):
        q = q_from_v(v, fan_cost, gamma)
        v_new = {s: max(q[(s, a)] for a in FAN_ACTIONS) for s in FAN_STATES}
        delta = max(abs(v_new[s] - v[s]) for s in FAN_STATES)
        v = v_new
        history.append({"round": k, **{f"V({s})": v[s] for s in FAN_STATES},
                        "policy": tuple(greedy(q, s) for s in FAN_STATES), "delta": delta})
        if delta < eps:
            break
    return v, q_from_v(v, fan_cost, gamma), pd.DataFrame(history)


v_star, q_star, vi_history = value_iteration()
print(vi_history.head(3).round(4).to_string(index=False))
print(f"... ลู่เข้า (Δ < 1e-10) หลัง {len(vi_history)} รอบ")
for s in FAN_STATES:
    print(f"V*({s}) = {v_star[s]:.3f}   Q*({s}, OFF) = {q_star[(s, 'OFF')]:.3f}   Q*({s}, ON) = {q_star[(s, 'ON')]:.3f}"
          f"   π*({s}) = {greedy(q_star, s)}")
compare("V*(C)", v_star["C"], -6.207, 5e-4)
compare("V*(N)", v_star["N"], -6.897, 5e-4)
compare("V*(H)", v_star["H"], -17.207, 5e-4)
compare("Q*(N, OFF)", q_star[("N", "OFF")], -10.847, 5e-4)

# จำนวนรอบ: ขอบเขตจากการหดตัว เทียบกับจำนวนรอบจริงที่ความคลาดเคลื่อน ≤ 0.01
error = np.max(np.abs(vi_history[[f"V({s})" for s in FAN_STATES]].to_numpy() - np.array([v_star[s] for s in FAN_STATES])), axis=1)
bound = math.log(0.01 / max(abs(v) for v in v_star.values())) / math.log(GAMMA)
actual_rounds = int(np.argmax(error <= 0.01)) + 1
print(f"ขอบเขตจากการหดตัว: k ≥ {bound:.1f} รอบ | จำนวนรอบจริงที่ max|V_k − V*| ≤ 0.01: {actual_rounds} รอบ")
print("นโยบายนิ่งตั้งแต่รอบที่", int(vi_history.index[vi_history["policy"] == vi_history["policy"].iloc[-1]][0]) + 1)

# รูปที่ 8.12 (ก): การลู่เข้าของการทำซ้ำค่า
fig, axes = plt.subplots(1, 2, figsize=(10, 3.6))
rounds = np.r_[0, vi_history["round"].to_numpy()][:81]
for s, color in zip(FAN_STATES, [COLORS["blue"], COLORS["green"], COLORS["orange"]]):
    values = np.r_[0, vi_history[f"V({s})"].to_numpy()][:81]
    axes[0].plot(rounds, values, color=color, label=f"V({s})")
    axes[0].axhline(v_star[s], color=color, ls="--", lw=0.8)
axes[0].set_xlabel("round k")
axes[0].set_ylabel("V_k(s)")
axes[0].legend()
axes[1].semilogy(vi_history["round"][:80], error[:80], color=COLORS["blue"], label="max |V_k − V*|")
axes[1].semilogy(vi_history["round"][:80], max(abs(v) for v in v_star.values()) * GAMMA ** vi_history["round"][:80],
                 color=COLORS["gray"], ls=":", label="17.2 × 0.9^k (bound)")
axes[1].set_xlabel("round k")
axes[1].legend()
plt.tight_layout()
plt.show()
```

**คำถามชวนคิด:** จำนวนรอบจริงน้อยกว่าขอบเขต $\ln(\varepsilon/E_0)/\ln\gamma$ เพราะเหตุใด (ดูกราฟด้านขวา เส้นจริงลดลงเร็วกว่าเส้นขอบเขตหรือไม่) และลองเรียก `value_iteration(fan_cost=3)` (แบบฝึกหัดท้ายบทข้อ 15) นโยบายเปลี่ยนไปอย่างไร

### 4.2 ไล่ Q-Learning 8 ก้าวแรกของหนังสือ (ตัวอย่างที่ 8.18)

เอเจนต์ไม่รู้ตาราง `FAN_P` เรียนจากประสบการณ์ $(s, a, r, s')$ เท่านั้น ฟังก์ชัน `q_learning_replay` เล่นประสบการณ์ที่กำหนดซ้ำ ก้าวที่ระบุการกระทำไว้คือการสำรวจ ก้าวที่เหลือเลือกแบบละโมบ (เท่ากันเลือก OFF)

```python
def q_learning_replay(experience, alpha=0.5, gamma=GAMMA, q=None):
    """เล่นประสบการณ์ (s, การกระทำที่บังคับหรือ None, s') ซ้ำ คืนตาราง Q และบันทึกทุกก้าว"""
    q = {(s, a): 0.0 for s in FAN_STATES for a in FAN_ACTIONS} if q is None else dict(q)
    log = []
    for step, (state, forced_action, next_state) in enumerate(experience, start=1):
        action = forced_action if forced_action is not None else greedy(q, state)
        r = fan_reward(state, action)
        max_next = max(q[(next_state, b)] for b in FAN_ACTIONS)
        q[(state, action)] += alpha * (r + gamma * max_next - q[(state, action)])
        log.append({"step": step, "s": state, "a": action, "r": r, "s'": next_state,
                    "max Q(s',·)": max_next, "new Q(s,a)": q[(state, action)]})
    return q, pd.DataFrame(log)


BOOK_EXPERIENCE = [("N", "OFF", "H"), ("H", None, "H"), ("H", None, "N"), ("N", None, "H"),
                   ("H", None, "H"), ("H", None, "N"), ("N", None, "C"), ("C", None, "N")]
q_after8, log8 = q_learning_replay(BOOK_EXPERIENCE)
print(log8.to_string(index=False))
q_compare = pd.DataFrame({
    "Q(·,OFF)": [q_after8[(s, "OFF")] for s in FAN_STATES], "Q(·,ON)": [q_after8[(s, "ON")] for s in FAN_STATES],
    "policy now": [greedy(q_after8, s) for s in FAN_STATES],
    "Q*(·,OFF)": [q_star[(s, "OFF")] for s in FAN_STATES], "Q*(·,ON)": [q_star[(s, "ON")] for s in FAN_STATES],
    "optimal": [greedy(q_star, s) for s in FAN_STATES]}, index=FAN_STATES)
print()
print(q_compare.round(3))
compare("Q(C, OFF) หลัง 8 ก้าว", q_after8[("C", "OFF")], -0.225, 1e-12)
```

### 4.3 ฝึก Q-Learning ต่อ 20,000 ก้าวและวาดกราฟการลู่เข้า (รูปที่ 8.12 ข)

ใช้ ε-greedy ที่ $\varepsilon = 0.1$ และอัตราการเรียนรู้ที่ **ลดลงตามจำนวนครั้งที่เยี่ยม** คู่ $(s, a)$ นั้น $\alpha = 1/N(s, a)^{0.7}$ เงื่อนไขการลู่เข้าของ Watkins & Dayan (1992) ต้องการให้ $\sum \alpha = \infty$ และ $\sum \alpha^2 < \infty$ ซึ่งเลขชี้กำลังระหว่าง 0.5 ถึง 1 ให้ทั้งสองข้อ เลขชี้กำลัง 1 (คือ $\alpha = 1/N$ ซึ่งเท่ากับการหาค่าเฉลี่ยตรง ๆ) ก็เข้าเงื่อนไข แต่ลู่เข้าช้ามากเมื่อ $\gamma$ ใกล้ 1 เซลล์ถัดไปจะแสดงให้เห็น

> ผลจากการจำลองชุดนี้ (seed = 42) บันทึกไว้ที่ `notebooks/src/data/ch08_qlearning_curve.csv` ด้วยโค้ดชุดเดียวกัน เพื่อใช้วาดรูปที่ 8.12 (ข) ในหนังสือ

```python
def fan_step(state, action, rng, fan_cost=1.0):
    """จำลองหนึ่งก้าวจากแบบจำลองจริง (เอเจนต์ไม่เห็นตาราง FAN_P เห็นเพียง r และ s')"""
    outcomes = FAN_P[state][action]
    index = rng.choice(len(outcomes), p=[p for _, p in outcomes])
    return fan_reward(state, action, fan_cost), outcomes[index][0]


def run_q_learning(n_steps=20_000, epsilon=0.1, alpha_power=0.7, fixed_alpha=None, seed=42,
                   record_every=50, fan_cost=1.0, start="N", q_init=0.0):
    """Q-Learning แบบตารางด้วย ε-greedy บน MDP ของพัดลม (งานต่อเนื่อง ไม่มีการจบตอน)

    อัตราการเรียนรู้ลดลงตามจำนวนครั้งที่เยี่ยมคู่ (s, a): α = 1 / N(s, a)^alpha_power
    ถ้ากำหนด fixed_alpha จะใช้ค่าคงที่แทน คืน (Q สุดท้าย, จำนวนครั้งที่เยี่ยม, ตารางค่า Q ทุก record_every ก้าว)
    """
    rng = np.random.default_rng(seed)
    q = {(s, a): q_init for s in FAN_STATES for a in FAN_ACTIONS}
    visits = {key: 0 for key in q}
    state = start

    def snapshot(step):
        return {"step": step, **{f"Q_{s}_{a}": q[(s, a)] for s in FAN_STATES for a in FAN_ACTIONS}}

    records = [snapshot(0)]
    for step in range(1, n_steps + 1):
        if rng.random() < epsilon:                      # สำรวจ
            action = FAN_ACTIONS[int(rng.integers(len(FAN_ACTIONS)))]
        else:                                           # ใช้ประโยชน์
            action = greedy(q, state)
        r, next_state = fan_step(state, action, rng, fan_cost)
        visits[(state, action)] += 1
        alpha = fixed_alpha if fixed_alpha is not None else 1 / visits[(state, action)] ** alpha_power
        td_target = r + GAMMA * max(q[(next_state, b)] for b in FAN_ACTIONS)
        q[(state, action)] += alpha * (td_target - q[(state, action)])
        state = next_state
        if step % record_every == 0:
            records.append(snapshot(step))
    return q, visits, pd.DataFrame(records)


start_time = time.time()
q_learned, visits, q_curve = run_q_learning(n_steps=20_000, epsilon=0.1, alpha_power=0.7, seed=RANDOM_SEED, record_every=50)
print(f"ใช้เวลา {time.time() - start_time:.1f} วินาที")
summary_q = pd.DataFrame({"learned Q": {k: q_learned[k] for k in q_learned}, "Q*": {k: q_star[k] for k in q_star},
                          "visits": visits})
summary_q["|error|"] = (summary_q["learned Q"] - summary_q["Q*"]).abs()
print(summary_q.round(3))
print("นโยบายที่เรียนได้:", {s: greedy(q_learned, s) for s in FAN_STATES}, "| นโยบายที่ดีที่สุด:", {s: greedy(q_star, s) for s in FAN_STATES})

# ก้าวสุดท้ายที่นโยบายของแต่ละสถานะยังผิด (คู่ที่ Q* ต่างกันน้อยต้องใช้ตัวอย่างมาก)
for s in FAN_STATES:
    wrong = q_curve[np.where(q_curve[f"Q_{s}_OFF"] >= q_curve[f"Q_{s}_ON"], "OFF", "ON") != greedy(q_star, s)]
    print(f"สถานะ {s}: นโยบายตาม Q ผิดครั้งสุดท้ายที่ก้าว {int(wrong['step'].max()) if len(wrong) else 0} "
          f"(|Q*(OFF) − Q*(ON)| = {abs(q_star[(s, 'OFF')] - q_star[(s, 'ON')]):.2f})")

fig, axes = plt.subplots(1, 3, figsize=(12, 3.6), sharex=True)
for ax, s in zip(axes, FAN_STATES):
    for a, color in [("OFF", COLORS["blue"]), ("ON", COLORS["orange"])]:
        ax.plot(q_curve["step"], q_curve[f"Q_{s}_{a}"], color=color, label=f"Q({s},{a})")
        ax.axhline(q_star[(s, a)], color=color, ls="--", lw=0.8)
    ax.set_title(f"state {s}")
    ax.set_xlabel("step")
    ax.legend(fontsize=8)
axes[0].set_ylabel("Q value (dashed = Q*)")
plt.suptitle("Q-learning on the fan MDP (ε = 0.1, α = 1/N^0.7, simulation, seed 42)")
plt.tight_layout()
plt.show()
```

### 4.4 ทดลองเปลี่ยน ε, α และต้นทุนของพัดลม

ตารางถัดไปรันการจำลองหลายแบบด้วย seed 5 ค่า แล้วรายงานค่าเฉลี่ยของความคลาดเคลื่อนสูงสุด $\max|Q - Q^*|$ และสัดส่วนของการรันที่ได้นโยบายที่ดีที่สุดครบทุกสถานะ สำหรับต้นทุนพัดลม 3 ต่อก้าว ต้องเทียบกับ $Q^*$ ของ MDP ใหม่ซึ่งหาด้วย `value_iteration(fan_cost=3)`

```python
def evaluate_setting(label, n_steps=20_000, seeds=range(5), fan_cost=1.0, **kwargs):
    _, q_ref, _ = value_iteration(fan_cost=fan_cost)
    errors, correct = [], []
    for seed in seeds:
        q, _, _ = run_q_learning(n_steps=n_steps, seed=seed, fan_cost=fan_cost, record_every=n_steps, **kwargs)
        errors.append(max(abs(q[k] - q_ref[k]) for k in q))
        correct.append(all(greedy(q, s) == greedy(q_ref, s) for s in FAN_STATES))
    return {"setting": label, "mean max|Q − Q*|": np.mean(errors), "optimal policy (share of runs)": np.mean(correct),
            "optimal policy (C, N, H)": "".join(greedy(q_ref, s)[1] for s in FAN_STATES)}


start_time = time.time()
experiments = pd.DataFrame([
    evaluate_setting("ε = 0.1, α = 1/N^0.7 (baseline)"),
    evaluate_setting("ε = 0.1, α = 1/N (sample average)", alpha_power=1.0),
    evaluate_setting("ε = 0.1, α = 0.5 fixed", fixed_alpha=0.5),
    evaluate_setting("ε = 0.1, α = 0.05 fixed", fixed_alpha=0.05),
    evaluate_setting("ε = 0.01, α = 1/N^0.7", epsilon=0.01),
    evaluate_setting("ε = 0.3, α = 1/N^0.7", epsilon=0.3),
    evaluate_setting("fan cost = 3, ε = 0.1, α = 1/N^0.7", fan_cost=3.0),
])
print(experiments.round(3).to_string(index=False))
print(f"(ใช้เวลา {time.time() - start_time:.1f} วินาที, ตัวอักษรท้ายตาราง F = OFF, N = ON)")
```

**คำถามชวนคิด**

1. อัตราการเรียนรู้คงที่ 0.5 ปรับค่าเร็วแต่ค่า Q แกว่งตามความสุ่มของแต่ละก้าวตลอด ส่วน 0.05 นิ่งกว่าแต่ช้ากว่า ทำไม $\alpha = 1/N$ ซึ่งดูเป็น "ค่าเฉลี่ยที่ถูกต้อง" จึงลู่เข้าช้าที่สุด (คำแนะนำ: เป้าหมาย TD ในช่วงแรกยังผิดเพราะค่า Q ของสถานะถัดไปยังไม่ถูก การให้น้ำหนักเท่ากันทุกตัวอย่างจึงจำค่าผิดในอดีตไว้นาน)
2. $\varepsilon$ = 0.01 สำรวจน้อยมาก คู่ใดถูกเยี่ยมน้อยที่สุด และเหตุใดจึงเป็นปัญหาแม้นโยบายจะยังถูก ถ้าต้องฝึกบนระบบจริง $\varepsilon$ สูงมีต้นทุนอะไร (ทบทวนข้อควรระวังในหัวข้อ 8.3.3)

### 4.5 ตัวควบคุม P และ PI เทียบกับนโยบายที่ RL เรียนได้ (ตัวอย่างที่ 8.19)

ระบบทำความร้อน $y[k+1] = 0.8\,y[k] + 0.2\,u[k]$ ค่าเป้าหมาย $r = 1$ ฟังก์ชัน `simulate_controller` รับพารามิเตอร์ของระบบ ($a$, $b$) จึงใช้ตรวจแบบฝึกหัดท้ายบทข้อ 17 ได้ด้วย

จากนั้นฝึก Q-Learning บนระบบเดียวกัน: สถานะคือ $y$ ที่แบ่งเป็นช่วงละ 0.05 การกระทำคือ $u \in \{0, 0.5, 1, 1.5, 2\}$ รางวัลคือ $-|r - y[k+1]|$ ฝึก 3,000 ตอน ตอนละ 30 ก้าวจากค่าเริ่มต้นสุ่ม แล้วนำนโยบายแบบละโมบมาเทียบกับ P และ PI จากนั้นทดสอบ **ช่องว่างระหว่างแบบจำลองกับของจริง** โดยให้ของจริงมีอัตราขยายต่ำกว่าแบบจำลองที่ใช้ฝึก ($b$ = 0.15 แทน 0.2)

```python
def simulate_controller(controller, steps=30, a=0.8, b=0.2, setpoint=1.0, y0=0.0):
    """จำลองวงปิด y[k+1] = a y[k] + b u[k] คืนตาราง k, y, e, Σe, u, y[k+1]"""
    y, error_sum, rows = y0, 0.0, []
    for k in range(steps):
        e = setpoint - y
        error_sum += e
        u = controller(e, error_sum, y)
        y_next = a * y + b * u
        rows.append({"k": k, "y[k]": y, "e[k]": e, "sum e": error_sum, "u[k]": u, "y[k+1]": y_next})
        y = y_next
    return pd.DataFrame(rows)


def p_controller(kp):
    return lambda e, error_sum, y: kp * e


def pi_controller(kp, ki):
    return lambda e, error_sum, y: kp * e + ki * error_sum


p_run = simulate_controller(p_controller(2), steps=60)
pi_run = simulate_controller(pi_controller(1, 0.5), steps=60)
print("ตัวควบคุม PI (Kp = 1, Ki = 0.5) หกก้าวแรก")
print(pi_run.head(6).round(6).to_string(index=False))
compare("P: y สภาวะคงตัว", p_run["y[k+1]"].iloc[-1], 0.667, 5e-4)
compare("PI: y สูงสุด", pi_run["y[k+1]"].max(), 1.087, 5e-4)
closed_loop = np.array([[0.8 - 0.2 * 1 - 0.2 * 0.5, 0.2 * 0.5], [-1, 1]])   # สถานะ (y[k], Σe ถึง k−1)
poles = np.linalg.eigvals(closed_loop)
print(f"สมการลักษณะเฉพาะ: {np.round(np.poly(closed_loop), 3)}, ขั้ว = {np.round(poles, 3)}, |z| = {np.abs(poles[0]):.3f}")

# ---------- RL แบบตารางบนระบบเดียวกัน
U_LEVELS = np.array([0.0, 0.5, 1.0, 1.5, 2.0])
Y_EDGES = np.linspace(0, 1.5, 31)


def y_state(y):
    return int(np.clip(np.digitize(y, Y_EDGES) - 1, 0, len(Y_EDGES) - 2))


def train_thermal_q(n_episodes=3000, steps=30, alpha=0.2, gamma=0.9, a=0.8, b=0.2, seed=RANDOM_SEED):
    rng = np.random.default_rng(seed)
    q = np.zeros((len(Y_EDGES) - 1, len(U_LEVELS)))
    for episode in range(n_episodes):
        epsilon = max(0.05, 1 - episode / (0.7 * n_episodes))     # ลด ε จาก 1 เหลือ 0.05
        y = rng.uniform(0, 1.3)
        for _ in range(steps):
            s = y_state(y)
            act = int(rng.integers(len(U_LEVELS))) if rng.random() < epsilon else int(np.argmax(q[s]))
            y_next = a * y + b * U_LEVELS[act]
            reward_value = -abs(1 - y_next)
            q[s, act] += alpha * (reward_value + gamma * q[y_state(y_next)].max() - q[s, act])
            y = y_next
    return q


start_time = time.time()
q_thermal = train_thermal_q()
print(f"ฝึก RL {3000 * 30:,} ก้าวใน {time.time() - start_time:.1f} วินาที")
rl_controller = lambda e, error_sum, y: U_LEVELS[int(np.argmax(q_thermal[y_state(y)]))]

rows, runs = [], {}
for plant, b_value in [("nominal plant (b = 0.2)", 0.2), ("real plant, lower gain (b = 0.15)", 0.15)]:
    for name, ctrl in [("P (Kp = 2)", p_controller(2)), ("PI (Kp = 1, Ki = 0.5)", pi_controller(1, 0.5)), ("RL policy", rl_controller)]:
        run = simulate_controller(ctrl, steps=60, b=b_value)
        runs[(plant, name)] = run
        rows.append({"plant": plant, "controller": name, "IAE (60 steps)": np.abs(1 - run["y[k+1]"]).sum(),
                     "max y": run["y[k+1]"].max(), "final y": run["y[k+1]"].iloc[-1]})
print(pd.DataFrame(rows).round(3).to_string(index=False))

fig, axes = plt.subplots(1, 2, figsize=(11, 3.6), sharey=True)
for ax, plant in zip(axes, ["nominal plant (b = 0.2)", "real plant, lower gain (b = 0.15)"]):
    for name, color in [("P (Kp = 2)", COLORS["gray"]), ("PI (Kp = 1, Ki = 0.5)", COLORS["blue"]), ("RL policy", COLORS["orange"])]:
        run = runs[(plant, name)]
        ax.step(np.r_[0, run["k"] + 1][:31], np.r_[0, run["y[k+1]"]][:31], where="post", color=color, label=name)
    ax.axhline(1, color="black", lw=0.8, ls=":")
    ax.set_title(plant)
    ax.set_xlabel("k")
axes[0].set_ylabel("y")
axes[0].legend(fontsize=8)
plt.tight_layout()
plt.show()
print("นโยบาย RL ที่เรียนได้ (u ต่อช่วงของ y):",
      {f"{Y_EDGES[i]:.2f}-{Y_EDGES[i + 1]:.2f}": float(U_LEVELS[int(np.argmax(q_thermal[i]))]) for i in range(0, 30, 3)})
```

**คำถามชวนคิด**

1. บนแบบจำลองที่ใช้ฝึก RL ขึ้นถึงค่าเป้าหมายเร็วกว่า PI และไม่พุ่งเกิน เพราะเรียนรู้ "เร่งเต็มที่แล้วถือค่า $u = 1$" แต่เมื่ออัตราขยายของของจริงต่ำลง RL เหลือความคลาดเคลื่อนสถานะคงตัวขณะที่ PI ยังเข้าเป้า เพราะอะไร (คำแนะนำ: นโยบายแบบตารางไม่มีพจน์ปริพันธ์ และสถานะ $y$ อย่างเดียวไม่บอกว่าระบบเปลี่ยนไป)
2. ผลนี้สอดคล้องกับข้อสรุปในหนังสือว่า "ไม่มีเหตุผลที่จะใช้ RL ในที่ที่ PID ทำได้ดีแล้ว" อย่างไร และถ้าจะใช้ RL ในระบบนี้ ควรให้ RL ทำหน้าที่ใดในสถาปัตยกรรมแบบชั้นของรูปที่ 8.13

### 4.6 ตัวกันการกระทำของแบตเตอรี่และขนาดตาราง Q (ตัวอย่างที่ 8.20)

ตัวกันการกระทำอย่างง่ายตรวจคำสั่งจ่ายไฟสองขั้น คือพิกัดกำลังและพลังงานที่เหลือก่อนถึงขีดล่างของสถานะประจุ (แบบฝึกปฏิบัติข้อ 4 จะขยายให้รองรับการชาร์จและประสิทธิภาพ) เซลล์ถัดไปทดสอบกับคำสั่งสุ่มของ "เอเจนต์ที่ยังฝึกไม่ดี" ตลอดหนึ่งวัน เพื่อดูว่าสถานะประจุไม่หลุดขีดจำกัดเลยไม่ว่าเอเจนต์จะสั่งอะไร

```python
def discharge_shield(command_kw, soc, capacity_kwh=200.0, p_max_kw=100.0, soc_min=0.10, dt_h=0.25):
    """ตัวกันการกระทำของตัวอย่างที่ 8.20 (เฉพาะการจ่ายไฟ ไม่คิดประสิทธิภาพ)"""
    allowed = min(command_kw, p_max_kw)                                    # ขั้นที่ 1 พิกัดกำลัง
    allowed = min(allowed, max(0.0, (soc - soc_min) * capacity_kwh / dt_h))  # ขั้นที่ 2 พลังงานที่เหลือ
    return max(0.0, allowed)


compare("ตัวอย่างที่ 8.20: คำสั่ง 120 kW ที่ SOC 12% (kW)", discharge_shield(120, 0.12), 16, 1e-9)

rng = np.random.default_rng(RANDOM_SEED)
soc, soc_trace, clipped = 0.5, [], 0
for step in range(96):                                    # 96 รอบ × 15 นาที = 1 วัน
    command = rng.uniform(0, 150)                         # เอเจนต์สั่งจ่ายไฟแบบสุ่ม
    allowed = discharge_shield(command, soc)
    clipped += allowed < command
    soc -= allowed * 0.25 / 200
    soc_trace.append(soc)
    if step % 8 == 7:
        soc = min(0.9, soc + 0.25)                        # ชาร์จคืนเป็นระยะ (สมมติ)
print(f"คำสั่งถูกปรับ {clipped} จาก 96 ครั้ง, SOC ต่ำสุด = {min(soc_trace):.3f} (ขีดล่าง 0.10)")

states_battery = 21 * 3 * 24
print(f"ตาราง Q ของแบตเตอรี่ในหัวข้อ 8.3.2: {states_battery:,} สถานะ × 3 การกระทำ = {states_battery * 3:,} ค่า")
```

### 5.2 ฝึก MLP 30-64-32-4 แล้วแจงหน่วยค่าน้ำหนักเป็น 8 บิต (ตัวอย่างที่ 8.23)

สร้างข้อมูลคุณลักษณะจำลองของมอเตอร์ 4 สภาพ (ปกติ ไม่สมดุล เยื้องศูนย์ ตลับลูกปืนเสียหาย) แต่ละสภาพทำให้คุณลักษณะกลุ่มต่างกันสูงขึ้น ฝึก `MLPClassifier(hidden_layer_sizes=(64, 32))` แล้ว

1. นับพารามิเตอร์จาก `coefs_` และ `intercepts_` เทียบกับการนับในหนังสือ
2. แจงหน่วยค่าน้ำหนักทุกชั้นแบบอะฟฟีน **ทั้งเทนเซอร์ (Per-tensor)** ด้วย `quantize_tensor` ที่เขียนเอง แปลงกลับ แล้วทำนายด้วยการคูณเมทริกซ์ใน NumPy (เหมือนโค้ดที่รันบนไมโครคอนโทรลเลอร์) เทียบความถูกต้องกับโมเดลทศนิยม 32 บิต
3. สร้างค่าโดดในชั้นแรกโดย **ไม่เปลี่ยนการทำนายของโมเดลทศนิยมเลย**: คูณค่าน้ำหนักขาเข้า (และไบแอส) ของหน่วยซ่อนหนึ่งหน่วยด้วย $f$ แล้วหารค่าน้ำหนักขาออกของหน่วยนั้นด้วย $f$ ซึ่งได้ผลเท่าเดิมเพราะ $	ext{ReLU}(fz) = f\,	ext{ReLU}(z)$ เมื่อ $f > 0$ สถานการณ์นี้เกิดได้จริงในโมเดลที่ฝึกแล้ว แล้วดูว่าช่วงที่กว้างขึ้นทำให้ค่าส่วนใหญ่ถูกปัดหยาบเพียงใด

```python
def quantize_tensor(weights, n_bits=8):
    """แจงหน่วยแบบอะฟฟีนทั้งเทนเซอร์เป็นจำนวนเต็มไม่มีเครื่องหมาย คืน (q, S, Z) ช่วงต้องครอบคลุม 0 เสมอ"""
    w = np.asarray(weights, dtype=float)
    q_max = 2 ** n_bits - 1
    x_min, x_max = min(w.min(), 0.0), max(w.max(), 0.0)
    scale = (x_max - x_min) / q_max
    zero = int(round(-x_min / scale))
    q = np.clip(np.round(w / scale) + zero, 0, q_max).astype(np.uint8 if n_bits <= 8 else np.uint16)
    return q, scale, zero


def dequantize(q, scale, zero):
    return scale * (q.astype(float) - zero)


# ตัวอย่างที่ 8.23 ช่วง [−0.8, 1.2]
q_ex, s_ex, z_ex = quantize_tensor([-0.8, 0.0, 0.5, 1.2])
print(f"S = {s_ex:.6f}, Z = {z_ex}, q = {q_ex}, แปลงกลับ = {np.round(dequantize(q_ex, s_ex, z_ex), 5)}")
compare("S", s_ex, 0.007843, 5e-7)
compare("q(0.5)", float(q_ex[2]), 166, 0)


def simulate_motor_features(n_per_class=600, n_features=30, seed=RANDOM_SEED):
    """คุณลักษณะจำลอง 30 ค่า: 0–9 แถบความถี่ต่ำ (1×, 2×), 10–19 แถบ BPFO/BPFI, 20–29 เวลา/อุณหภูมิ"""
    rng = np.random.default_rng(seed)
    shifts = {0: {}, 1: {0: 3.0, 1: 2.0, 20: 1.25}, 2: {2: 2.5, 3: 3.0, 21: 1.5}, 3: {10: 2.5, 11: 2.25, 12: 1.75, 22: 2.0}}
    x_parts, y_parts = [], []
    mixing = rng.normal(0, 0.25, (n_features, n_features)) + np.eye(n_features)   # คุณลักษณะสัมพันธ์กัน
    for label, shift in shifts.items():
        base = rng.normal(0, 1, (n_per_class, n_features))
        for j, delta in shift.items():
            base[:, j] += delta
        x_parts.append(base @ mixing)
        y_parts.append(np.full(n_per_class, label))
    return np.vstack(x_parts), np.concatenate(y_parts)


x_motor, y_motor = simulate_motor_features()
x_tr, x_te, y_tr, y_te = train_test_split(x_motor, y_motor, test_size=0.3, stratify=y_motor, random_state=RANDOM_SEED)
scaler = StandardScaler().fit(x_tr)
x_tr_s, x_te_s = scaler.transform(x_tr), scaler.transform(x_te)
with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    mlp = MLPClassifier(hidden_layer_sizes=(64, 32), alpha=1e-3, max_iter=400, random_state=RANDOM_SEED).fit(x_tr_s, y_tr)

n_params = sum(w.size for w in mlp.coefs_) + sum(b.size for b in mlp.intercepts_)
compare("จำนวนพารามิเตอร์", n_params, 4196, 0)
print(f"ขนาด float32 = {n_params * 4:,} ไบต์ ({n_params * 4 / 1024:.1f} KiB), int8 = {n_params:,} ไบต์ ({n_params / 1024:.1f} KiB)")


def mlp_forward(x, weights, biases):
    """ทำนายด้วยการคูณเมทริกซ์ + ReLU เหมือนโค้ดบนอุปกรณ์ขอบ คืนคลาสที่คะแนนสูงสุด"""
    h = x
    for i, (w, b) in enumerate(zip(weights, biases)):
        h = h @ w + b
        if i < len(weights) - 1:
            h = np.maximum(h, 0)
    return np.argmax(h, axis=1)


def quantized_copy(weights, quantizer):
    """แจงหน่วยแล้วแปลงกลับทุกชั้น (ค่าไบแอสเก็บเป็นทศนิยมตามแนวปฏิบัติทั่วไป)"""
    return [dequantize(*quantizer(w)) for w in weights]


acc_float = accuracy_score(y_te, mlp_forward(x_te_s, mlp.coefs_, mlp.intercepts_))
acc_int8 = accuracy_score(y_te, mlp_forward(x_te_s, quantized_copy(mlp.coefs_, quantize_tensor), mlp.intercepts_))
print(f"ความถูกต้องบนชุดทดสอบ: float32 = {acc_float:.3f}, int8 ทั้งเทนเซอร์ = {acc_int8:.3f}")
compare("ตรวจ forward ของเราเท่ากับ mlp.predict", accuracy_score(mlp.predict(x_te_s), mlp_forward(x_te_s, mlp.coefs_, mlp.intercepts_)), 1.0, 0)

# สร้างค่าโดดแบบไม่เปลี่ยนฟังก์ชัน: หน่วยซ่อน j ของชั้นแรกขยายด้วย f ขาออกหดด้วย f
j_unit = int(np.argmax(np.abs(mlp.coefs_[0]).max(axis=0)))
f_scale = 100.0 / np.abs(mlp.coefs_[0][:, j_unit]).max()       # ให้ค่าน้ำหนักที่ใหญ่สุดเป็น 100
weights_outlier = [w.copy() for w in mlp.coefs_]
biases_outlier = [b.copy() for b in mlp.intercepts_]
weights_outlier[0][:, j_unit] *= f_scale
biases_outlier[0][j_unit] *= f_scale
weights_outlier[1][j_unit, :] /= f_scale
for label, ws, bs in [("original", mlp.coefs_, mlp.intercepts_), ("rescaled unit (outlier 100)", weights_outlier, biases_outlier)]:
    q0, s0, z0 = quantize_tensor(ws[0])
    err = np.abs(dequantize(q0, s0, z0) - ws[0])
    acc_f = accuracy_score(y_te, mlp_forward(x_te_s, ws, bs))
    acc_q = accuracy_score(y_te, mlp_forward(x_te_s, quantized_copy(ws, quantize_tensor), bs))
    print(f"{label:28s}: S ชั้นแรก = {s0:.5f}, คลาดเคลื่อนเฉลี่ย = {err.mean():.5f}, "
          f"float32 = {acc_f:.3f}, int8 ทั้งเทนเซอร์ = {acc_q:.3f}")
```

### 5.3 แฝดดิจิทัลความร้อนของมอเตอร์และผลของสัญญาณรบกวน (ตัวอย่างที่ 8.24)

ทำซ้ำตารางในหนังสือก่อน แล้วจำลองค่าวัดที่มีสัญญาณรบกวน σ = 0.3 K ของมอเตอร์ปกติและมอเตอร์ที่แผ่นกรองอุดตัน รอบละ 5 ชั่วโมง (60 ก้าว) 300 รอบ เพื่อดูว่าเกณฑ์ส่วนต่างแต่ละค่าให้ **เวลาที่ตรวจพบ** และ **อัตราการแจ้งเตือนผิด** เท่าใด ทั้งแบบแจ้งเมื่อเกินเกณฑ์ครั้งเดียว และแบบยืนยันเมื่อเกินติดกัน 3 ก้าว

```python
def thermal_twin(theta0, load_pu, steps, theta_rated=60.0, tau_min=40.0, dt_min=5.0, resistance_factor=1.0):
    """แบบจำลองความร้อนอันดับหนึ่ง คืนอุณหภูมิเหนืออากาศ Δθ[1..steps] (K)"""
    theta_ss = theta_rated * load_pu ** 2 * resistance_factor
    ratio = dt_min / (tau_min * resistance_factor)
    out, theta = [], theta0
    for _ in range(steps):
        theta = theta + ratio * (theta_ss - theta)
        out.append(theta)
    return np.array(out)


AMBIENT = 35.0
predicted = AMBIENT + thermal_twin(20, 0.9, 6)
measured = AMBIENT + thermal_twin(20, 0.9, 6, resistance_factor=1.2)
twin_table = pd.DataFrame({"minute": np.arange(1, 7) * 5, "twin (°C)": predicted, "measured (°C)": measured,
                           "residual (K)": measured - predicted})
print(twin_table.round(2).to_string(index=False))
compare("ส่วนต่างที่นาที 25 (K)", twin_table["residual (K)"].iloc[4], 2.28, 0.005)
print(f"ถามว่าถ้า...: โหลดสูงสุดที่ไม่เกิน 100 °C เมื่อระบายความร้อนบกพร่อง = {math.sqrt(65 / 72):.3f} pu")


def first_alarm(residual, threshold, confirm=1):
    """ก้าวแรก (นับจาก 1) ที่ส่วนต่างเกินเกณฑ์ติดกัน confirm ก้าว คืน None ถ้าไม่แจ้งเตือน"""
    run = 0
    for k, value in enumerate(residual, start=1):
        run = run + 1 if value > threshold else 0
        if run >= confirm:
            return k
    return None


rng = np.random.default_rng(RANDOM_SEED)
N_RUNS, STEPS, NOISE = 300, 60, 0.3
twin_long = thermal_twin(20, 0.9, STEPS)
healthy_long = thermal_twin(20, 0.9, STEPS)
faulty_long = thermal_twin(20, 0.9, STEPS, resistance_factor=1.2)
rows = []
for threshold in (0.5, 1.0, 1.5, 2.0, 3.0):
    for confirm in (1, 3):
        false_alarm = np.mean([first_alarm(healthy_long + rng.normal(0, NOISE, STEPS) - twin_long, threshold, confirm) is not None
                               for _ in range(N_RUNS)])
        detect = [first_alarm(faulty_long + rng.normal(0, NOISE, STEPS) - twin_long, threshold, confirm) for _ in range(N_RUNS)]
        detect = [d for d in detect if d is not None]
        rows.append({"threshold (K)": threshold, "confirm steps": confirm, "false alarm (healthy, 5 h)": false_alarm,
                     "median detection time (min)": np.median(detect) * 5 if detect else np.nan})
alarm_table = pd.DataFrame(rows)
print(alarm_table.round(3).to_string(index=False))

fig, ax = plt.subplots(figsize=(7, 3.6))
minutes = np.arange(1, STEPS + 1) * 5
example_noise = rng.normal(0, NOISE, STEPS)
ax.plot(minutes, AMBIENT + twin_long, color=COLORS["blue"], label="twin prediction (healthy model)")
ax.plot(minutes, AMBIENT + faulty_long + example_noise, color=COLORS["orange"], lw=0.9, label="measured, clogged filter (+ noise)")
ax2 = ax.twinx()
ax2.bar(minutes, faulty_long + example_noise - twin_long, width=4, color=COLORS["gray"], alpha=0.3)
ax2.axhline(2.0, color="darkred", ls="--", lw=0.8)
ax2.set_ylabel("residual (K), dashed = 2 K")
ax.set_xlabel("minute")
ax.set_ylabel("temperature (°C)")
ax.legend(fontsize=8, loc="lower right")
plt.show()
```
