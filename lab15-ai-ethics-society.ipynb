## ส่วนที่ 1 ตัววัดความเป็นธรรมจากเมทริกซ์ความสับสน

อ้างอิง: หัวข้อ 9.1.1 ตารางที่ 9.3 และตัวอย่างที่ 9.2–9.3

ตัววัดความเป็นธรรมเชิงสถิติไม่ใช่ตัววัดใหม่ แต่คือตัววัดเดิมจากหัวข้อ 4.3.1 ที่ **คำนวณแยกตามกลุ่ม** แล้วนำมาเทียบกัน ถ้าให้ $A$ เป็นกลุ่ม $Y$ เป็นค่าจริง (1 = ลักลอบใช้ไฟ) และ $\hat{Y}$ เป็นผลทำนาย (1 = เตือนให้ตรวจ)

| นิยาม | ตัววัดที่ต้องเท่ากันทุกกลุ่ม | สูตร |
|---|---|---|
| Demographic Parity | อัตราการเตือน | $(TP+FP)/N$ |
| Equal Opportunity | ความระลึก (TPR) | $TP/(TP+FN)$ |
| Equalized Odds | ความระลึก และอัตราบวกลวง (FPR) | $TP/(TP+FN)$ และ $FP/(FP+TN)$ |
| Predictive Parity | ความแม่นยำ (PPV) | $TP/(TP+FP)$ |

เซลล์ถัดไปเขียนฟังก์ชัน `metrics_from_counts` ที่รับจำนวน TP, FN, FP, TN ของกลุ่มหนึ่ง และ `compare_groups` ที่รายงาน **ผลต่าง** (กลุ่ม B ลบกลุ่ม A) และ **อัตราส่วน** (ค่าต่ำหารค่าสูง ยิ่งใกล้ 1 ยิ่งเท่าเทียม) แล้วทดสอบกับตัวอย่างที่ 9.2

```python
def metrics_from_counts(tp, fn, fp, tn):
    """ตัววัดของกลุ่มหนึ่งจากเมทริกซ์ความสับสน (ตารางที่ 9.3)"""
    n = tp + fn + fp + tn
    return {
        "n": n,
        "base_rate": (tp + fn) / n,                       # อัตราฐาน: สัดส่วนผู้ลักลอบจริง
        "alert_rate": (tp + fp) / n,                      # Demographic Parity
        "tpr": tp / (tp + fn) if tp + fn else np.nan,     # Equal Opportunity
        "fpr": fp / (fp + tn) if fp + tn else np.nan,     # Equalized Odds (คู่กับ TPR)
        "ppv": tp / (tp + fp) if tp + fp else np.nan,     # Predictive Parity
        "accuracy": (tp + tn) / n,
    }


def compare_groups(table, metrics=("alert_rate", "tpr", "fpr", "ppv", "accuracy")):
    """ผลต่าง (B − A) และอัตราส่วน (ต่ำ/สูง) ของตัววัดแต่ละตัวระหว่างสองกลุ่มในตาราง"""
    a, b = table.loc["A"], table.loc["B"]
    rows = []
    for m in metrics:
        low, high = sorted([a[m], b[m]])
        rows.append({"metric": m, "A": a[m], "B": b[m], "diff_B_minus_A": b[m] - a[m],
                     "ratio_low_high": low / high if high > 0 else np.nan})
    return pd.DataFrame(rows).set_index("metric")


# ตัวอย่างที่ 9.2: (TP, FN, FP, TN) ของบ้าน 500 หลังต่อพื้นที่
BOOK_92 = {"A": (16, 4, 24, 456), "B": (10, 10, 60, 420)}
table_92 = pd.DataFrame({g: metrics_from_counts(*cm) for g, cm in BOOK_92.items()}).T
display(table_92)
gaps_92 = compare_groups(table_92)
display(gaps_92)

overall_92 = metrics_from_counts(*[a + b for a, b in zip(BOOK_92["A"], BOOK_92["B"])])
compare("ความถูกต้องรวม", overall_92["accuracy"], 0.902, 5e-4)
compare("ความระลึกรวม", overall_92["tpr"], 0.650, 5e-4)
compare("อัตราส่วนอัตราการเตือน (Disparate Impact)", gaps_92.loc["alert_rate", "ratio_low_high"], 0.571, 5e-4)
compare("อัตราส่วนความระลึก", gaps_92.loc["tpr", "ratio_low_high"], 0.625, 5e-4)
compare("อัตราส่วนอัตราบวกลวง", gaps_92.loc["fpr", "ratio_low_high"], 0.400, 5e-4)
compare("อัตราส่วนความแม่นยำ", gaps_92.loc["ppv", "ratio_low_high"], 0.357, 5e-4)
print("ผ่านกฎสี่ในห้าหรือไม่:", "ผ่าน" if gaps_92.loc["alert_rate", "ratio_low_high"] >= 0.8 else "ไม่ผ่าน (< 0.8)")
```

ตัววัดรวมดูดี (ความถูกต้อง 0.902) แต่เมื่อแยกกลุ่ม พื้นที่ B เสียเปรียบสองต่อ คือถูกเตือนบ่อยกว่า (อัตราบวกลวง 0.125 เทียบกับ 0.050) และจับผู้ลักลอบได้น้อยกว่า (ความระลึก 0.500 เทียบกับ 0.800) ทั้งที่อัตราฐานเท่ากันที่ 0.04 กราฟถัดไปวาดตัววัดทั้งสี่แบบเดียวกับรูปที่ 9.3 ในหนังสือ

```python
def plot_group_metrics(table, title, metrics=("alert_rate", "tpr", "fpr", "ppv"), ax=None):
    """กราฟแท่งจัดกลุ่มของตัววัดความเป็นธรรม (แบบรูปที่ 9.3)"""
    if ax is None:
        _, ax = plt.subplots(figsize=(7, 3.6))
    x = np.arange(len(metrics))
    for k, group in enumerate(table.index):
        values = [table.loc[group, m] for m in metrics]
        bars = ax.bar(x + (k - 0.5) * 0.38, values, width=0.38, color=COLORS.get(group, "gray"),
                      label=f"area {group}")
        ax.bar_label(bars, fmt="%.3f", fontsize=8)
    ax.set_xticks(x, ["alert rate", "TPR (recall)", "FPR", "PPV (precision)"][:len(metrics)])
    ax.set_ylim(0, 1.05)
    ax.set_title(title)
    ax.legend()
    return ax


plot_group_metrics(table_92, "Example 9.2: equal base rate 0.04, unequal errors")
plt.tight_layout()
plt.show()
```

### 1.2 เมื่ออัตราฐานต่างกัน (ตัวอย่างที่ 9.3)

หนังสือแสดงความสัมพันธ์ที่ได้จากนิยามโดยตรง

$$
\text{FPR} = \frac{p}{1-p} \cdot \frac{1-\text{PPV}}{\text{PPV}} \cdot \text{TPR}
$$

ถ้าสองกลุ่มมี TPR และ PPV เท่ากันแต่อัตราฐาน $p$ ต่างกัน FPR **ต้อง** ต่างกัน เซลล์ถัดไปสร้างเมทริกซ์ความสับสนของเขต C ($p = 0.10$) และเขต D ($p = 0.02$) ที่มี TPR = 0.80 และ FPR = 0.05 เท่ากัน แล้วตรวจสมการ

```python
def counts_from_rates(n, base_rate, tpr, fpr):
    """สร้าง (TP, FN, FP, TN) จากขนาดกลุ่ม อัตราฐาน TPR และ FPR (ปัดเป็นจำนวนเต็ม)"""
    positives = round(n * base_rate)
    tp = round(tpr * positives)
    fp = round(fpr * (n - positives))
    return tp, positives - tp, fp, n - positives - fp


def fpr_identity(p, ppv, tpr):
    return p / (1 - p) * (1 - ppv) / ppv * tpr


table_93 = pd.DataFrame({zone: metrics_from_counts(*counts_from_rates(1000, p, 0.80, 0.05))
                         for zone, p in {"C": 0.10, "D": 0.02}.items()}).T
display(table_93[["base_rate", "alert_rate", "tpr", "fpr", "ppv"]])
compare("ความแม่นยำเขต C", table_93.loc["C", "ppv"], 0.640, 5e-4)
compare("ความแม่นยำเขต D", table_93.loc["D", "ppv"], 0.246, 5e-4)
for zone in ["C", "D"]:
    row = table_93.loc[zone]
    compare(f"FPR จากสมการ เขต {zone}", fpr_identity(row["base_rate"], row["ppv"], row["tpr"]), 0.05, 1e-12)
```

**คำถามชวนคิด**

1. ตัวอย่างที่ 9.3 ผ่าน Equalized Odds แต่ไม่ผ่าน Predictive Parity และ Demographic Parity ถ้าต้องการให้ความแม่นยำของเขต D เท่ากับเขต C ต้องปรับเกณฑ์ตัดสินของเขต D ไปทางใด และอะไรจะเสียไป
2. ในงานตรวจต้นไม้ใกล้สายป้อน ความต่างของอัตราการเตือนระหว่างเขตสะท้อน "ความเสี่ยงจริง" ส่วนในงานตรวจการลักลอบใช้ไฟของตัวอย่างที่ 9.2 อัตราฐานเท่ากัน ความต่างของอัตราการเตือนจึงสะท้อนอะไร

## ส่วนที่ 2 ตรวจความลำเอียงของโมเดลตรวจการลักลอบใช้ไฟ

อ้างอิง: หัวข้อ 9.1.1 ตารางที่ 9.2 (ความลำเอียงจากการวัด) และตัวอย่างที่ 9.2

ส่วนนี้สร้างสถานการณ์ในบทนำของบทที่ 9 ขึ้นมาจริงด้วยข้อมูลจำลอง บ้าน 10,000 หลังแบ่งเป็นสองพื้นที่ พื้นที่ละ 5,000 หลัง **อัตราการลักลอบจริงเท่ากันที่ร้อยละ 4** ความต่างอยู่ที่ **วิธีวัด** เท่านั้น

| คุณลักษณะ | ความหมาย | พื้นที่ A (มิเตอร์อัจฉริยะ ทุก 15 นาที) | พื้นที่ B (มิเตอร์จานหมุน จดเดือนละครั้ง) |
|---|---|---|---|
| `drop_pct` | การใช้ไฟลดลงเทียบกับปีก่อน (ร้อยละ) | วัดแม่นยำ (คลาดเคลื่อน SD 4) | คลาดเคลื่อนมาก (SD 14) จากการจดค่าและรอบบิลที่ไม่ตรงกัน |
| `tamper_alerts` | สัญญาณงัดแงะที่มิเตอร์ส่งใน 3 เดือน | ผู้ลักลอบมีสัญญาณเฉลี่ย 1.5 ครั้ง บ้านสุจริต 0.05 ครั้ง | **เป็น 0 เสมอ** เพราะมิเตอร์ส่งสัญญาณไม่ได้ |
| `old_meter` | มิเตอร์รุ่นเก่า (1 = ใช่) | 0 | 1 (จึงเป็นตัวแปรแทนของพื้นที่) |
| `late_payments` | จำนวนเดือนที่ชำระค่าไฟล่าช้าในปีที่ผ่านมา | เหมือนกันทั้งสองพื้นที่ | เหมือนกันทั้งสองพื้นที่ |

ป้ายกำกับ `theft` มาจากการสมมติว่า **ลงตรวจทุกหลัง** แบบเดียวกับตัวอย่างที่ 9.2 (ในงานจริงมักมีป้ายกำกับเฉพาะบ้านที่เคยถูกเลือกตรวจ ซึ่งยากกว่านี้มาก) ค่าทุกค่าเป็นค่าสมมติเพื่อการเรียนรู้

```python
FEATURES = ["drop_pct", "tamper_alerts", "old_meter", "late_payments"]


def make_households(n_per_area=5000, theft_rate=0.04, seed=RANDOM_SEED, upgrade_rural_meters=False):
    """จำลองบ้านสองพื้นที่ที่อัตราการลักลอบเท่ากัน แต่คุณภาพการวัดต่างกัน

    upgrade_rural_meters=True จำลองการเปลี่ยนมิเตอร์ในพื้นที่ B เป็นมิเตอร์อัจฉริยะ (ส่วนที่ 3.4)
    """
    rng = np.random.default_rng(seed)
    frames = []
    for area in ["A", "B"]:
        n = n_per_area
        theft = (rng.random(n) < theft_rate).astype(int)
        smart = area == "A" or upgrade_rural_meters
        # การใช้ไฟที่ลดลงจริง: ผู้ลักลอบลดลงมาก บ้านสุจริตลดลงหรือเพิ่มขึ้นเล็กน้อยตามเหตุปกติ
        true_drop = np.where(theft == 1, rng.normal(35, 12, n), rng.normal(5, 10, n))
        measurement_noise = 4 if smart else 14          # ความลำเอียงจากการวัด
        drop_pct = true_drop + rng.normal(0, measurement_noise, n)
        if smart:
            tamper = rng.poisson(np.where(theft == 1, 1.5, 0.05))
        else:
            tamper = np.zeros(n, dtype=int)              # มิเตอร์จานหมุนส่งสัญญาณงัดแงะไม่ได้
        late = rng.poisson(np.where(theft == 1, 1.2, 0.8))
        frames.append(pd.DataFrame({
            "area": area,
            "drop_pct": drop_pct.round(1),
            "tamper_alerts": tamper,
            "old_meter": int(not smart),
            "late_payments": late,
            "theft": theft,
        }))
    return pd.concat(frames, ignore_index=True)


households = make_households()
display(households.head())
display(households.groupby("area")[FEATURES + ["theft"]].mean())
```

ตารางค่าเฉลี่ยยืนยันว่าอัตราการลักลอบ (`theft`) ของสองพื้นที่ใกล้ 0.04 เท่ากัน แต่พื้นที่ B ไม่มีสัญญาณงัดแงะเลย กราฟถัดไปแสดงการกระจายของ `drop_pct` แยกตามพื้นที่และสถานะจริง สังเกตว่าในพื้นที่ B สองกลุ่มซ้อนทับกันมากกว่า

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 3.4), sharey=True)
bins = np.linspace(-40, 90, 53)
for ax, area in zip(axes, ["A", "B"]):
    part = households[households["area"] == area]
    ax.hist(part.loc[part["theft"] == 0, "drop_pct"], bins=bins, density=True, alpha=0.6,
            color=COLORS["gray"], label="honest")
    ax.hist(part.loc[part["theft"] == 1, "drop_pct"], bins=bins, density=True, alpha=0.7,
            color=COLORS[area], label="theft")
    ax.set_title(f"area {area}: consumption drop (%)")
    ax.set_xlabel("drop_pct")
    ax.legend()
plt.tight_layout()
plt.show()
```

### 2.1 ฝึกโมเดลและดูตัววัดรวม

แบ่งข้อมูลเป็นชุดฝึกและชุดทดสอบครึ่งต่อครึ่ง โดยรักษาสัดส่วนของพื้นที่และป้ายกำกับ (stratify) แล้วฝึกการถดถอยลอจิสติกบนคุณลักษณะดิบ (ไม่ปรับมาตรฐาน เพื่อให้อ่านค่าน้ำหนักเป็นหน่วยจริงได้แบบตัวอย่างที่ 9.5)

การไฟฟ้ามีเจ้าหน้าที่จำกัด จึงใช้โมเดล **จัดรายการบ้านที่ควรตรวจ** โดยเตือนเมื่อความน่าจะเป็น $\ge$ `ALERT_THRESHOLD` = 0.10 (ราว 2.5 เท่าของอัตราฐาน) แทนเกณฑ์ 0.5 ที่จะเตือนน้อยเกินไปเมื่อข้อมูลไม่สมดุล (หัวข้อ 4.3.1)

```python
ALERT_THRESHOLD = 0.10


def split_households(df, seed=RANDOM_SEED):
    strata = df["area"] + "_" + df["theft"].astype(str)
    return train_test_split(df, test_size=0.5, stratify=strata, random_state=seed)


def fit_detector(train_df, features=FEATURES):
    model = LogisticRegression(max_iter=5000)
    model.fit(train_df[features], train_df["theft"])
    return model


train_df, test_df = split_households(households)
model = fit_detector(train_df)
test_scores = model.predict_proba(test_df[FEATURES])[:, 1]
test_pred = (test_scores >= ALERT_THRESHOLD).astype(int)

weights = pd.Series(model.coef_[0], index=FEATURES)
print("ค่าน้ำหนัก (หน่วยลอการิทึมของอัตราต่อรองต่อหน่วยคุณลักษณะ):")
print(weights.round(3).to_string())
print(f"จุดตัดแกน w0 = {model.intercept_[0]:.3f}")

overall = metrics_from_counts(*confusion_matrix(test_df["theft"], test_pred, labels=[1, 0]).ravel())
print(f"\nชุดทดสอบ {overall['n']} หลัง | ความถูกต้องรวม {overall['accuracy']:.3f} | "
      f"ความระลึกรวม {overall['tpr']:.3f} | ความแม่นยำรวม {overall['ppv']:.3f}")
```

ค่าน้ำหนักของ `old_meter` เป็นบวก ทั้งที่ชนิดมิเตอร์ไม่ได้ทำให้ใครลักลอบมากขึ้น (อัตราฐานเท่ากัน) โมเดลเรียนรู้ค่านี้เพื่อ **ชดเชย** ที่พื้นที่ B ไม่มีสัญญาณงัดแงะ ซึ่งในพื้นที่ A การไม่มีสัญญาณงัดแงะเป็นหลักฐานว่าบ้านสุจริต นี่คือกลไกที่ทำให้ชนิดมิเตอร์กลายเป็นตัวแปรแทนของพื้นที่

### 2.2 ตัววัดแยกกลุ่ม

ฟังก์ชัน `group_report` ใช้ `confusion_matrix` ของ scikit-learn นับ TP, FN, FP, TN ของแต่ละพื้นที่ แล้วส่งให้ `metrics_from_counts` (แบบฝึกปฏิบัติข้อ 1 จะให้เขียนฟังก์ชันนี้เองด้วย NumPy)

```python
def group_report(y_true, y_pred, groups):
    """ตารางตัววัดแยกกลุ่มจากอาร์เรย์ค่าจริง ผลทำนาย และกลุ่ม"""
    y_true, y_pred, groups = map(np.asarray, (y_true, y_pred, groups))
    rows = {}
    for g in sorted(set(groups)):
        mask = groups == g
        tp, fn, fp, tn = confusion_matrix(y_true[mask], y_pred[mask], labels=[1, 0]).ravel()
        rows[g] = metrics_from_counts(tp, fn, fp, tn)
    return pd.DataFrame(rows).T


baseline_table = group_report(test_df["theft"], test_pred, test_df["area"])
display(baseline_table)
baseline_gaps = compare_groups(baseline_table)
display(baseline_gaps)
plot_group_metrics(baseline_table, f"Simulated detector, threshold {ALERT_THRESHOLD}")
plt.tight_layout()
plt.show()
```

ผลจากข้อมูลจำลองมีรูปแบบเดียวกับตัวอย่างที่ 9.2 ทุกประการ ความถูกต้องรวมสูงกว่าร้อยละ 90 แต่พื้นที่ B มีอัตราบวกลวงสูงกว่าหลายเท่า (บ้านสุจริตถูกตรวจโดยไม่จำเป็น) และความระลึกต่ำกว่ามาก (ผู้ลักลอบรอดไป) อัตราส่วนอัตราการเตือนต่ำกว่า 0.8 จึงไม่ผ่านกฎสี่ในห้าด้วย

**คำถามชวนคิด**

1. ถ้าทีมรายงานผู้บริหารเพียง "ความถูกต้องรวม" ผู้บริหารจะตัดสินใจอย่างไร และใครรับผลเสีย (ความลำเอียงจากการประเมิน ตารางที่ 9.2)
2. ความลำเอียงในข้อมูลจำลองนี้เกิดจากแหล่งใดในตารางที่ 9.2 และการปรับอัลกอริทึมอย่างเดียวแก้ต้นเหตุได้หรือไม่

## ส่วนที่ 3 เกณฑ์ตัดสินแยกกลุ่ม การตัดตัวแปรแทน และการแก้ที่ต้นเหตุ

อ้างอิง: หัวข้อ 9.1.1 "วิธีลดความลำเอียง" (ก่อนฝึก ระหว่างฝึก หลังฝึก และนอกโมเดล)

### 3.1 TPR และ FPR ของแต่ละพื้นที่เมื่อเปลี่ยนเกณฑ์ตัดสิน

เกณฑ์ตัดสินเดียวกันให้ผลต่างกันในแต่ละพื้นที่ เพราะการกระจายของคะแนนต่างกัน กราฟถัดไปวาด TPR และ FPR ของแต่ละพื้นที่เทียบกับเกณฑ์ตัดสิน

```python
THRESHOLD_GRID = np.round(np.arange(0.01, 0.61, 0.01), 2)


def rates_at(scores, y_true, threshold):
    """(TPR, FPR, อัตราการเตือน) เมื่อเตือนที่คะแนน >= threshold"""
    pred = scores >= threshold
    y_true = np.asarray(y_true)
    return pred[y_true == 1].mean(), pred[y_true == 0].mean(), pred.mean()


area_test = test_df["area"].to_numpy()
y_test = test_df["theft"].to_numpy()
curves = {}
for area in ["A", "B"]:
    mask = area_test == area
    curves[area] = np.array([rates_at(test_scores[mask], y_test[mask], t) for t in THRESHOLD_GRID])

fig, axes = plt.subplots(1, 2, figsize=(10, 3.4))
for area in ["A", "B"]:
    axes[0].plot(THRESHOLD_GRID, curves[area][:, 0], color=COLORS[area], label=f"area {area}")
    axes[1].plot(THRESHOLD_GRID, curves[area][:, 1], color=COLORS[area], label=f"area {area}")
for ax, name in zip(axes, ["TPR (recall)", "FPR"]):
    ax.axvline(ALERT_THRESHOLD, color=COLORS["gray"], ls="--", lw=1)
    ax.set_xlabel("decision threshold")
    ax.set_title(name)
    ax.legend()
plt.tight_layout()
plt.show()
```

### 3.2 ปรับเกณฑ์ตัดสินแยกกลุ่ม (หลังฝึก)

วิธีหลังฝึกของ Hardt et al. (2016) คือเลือกเกณฑ์ของแต่ละกลุ่มให้ตัววัดที่ต้องการใกล้กัน ทดลองเลือกเกณฑ์ของแต่ละพื้นที่ให้ **TPR ≥ 0.80** (Equal Opportunity) โดยใช้เกณฑ์ที่สูงที่สุดในตารางที่ยังได้ TPR ถึงเป้า แล้วดูว่าตัววัดอื่นเป็นอย่างไร

```python
TARGET_TPR = 0.80


def grid_threshold_for_tpr(curve, target):
    """เกณฑ์สูงสุดใน THRESHOLD_GRID ที่ยังได้ TPR >= target (ใช้ตารางหยาบ แบบฝึกปฏิบัติข้อ 2 จะหาค่าแม่นยำ)"""
    ok = THRESHOLD_GRID[curve[:, 0] >= target]
    return ok.max() if len(ok) else THRESHOLD_GRID.min()


group_thresholds = {area: grid_threshold_for_tpr(curves[area], TARGET_TPR) for area in ["A", "B"]}
print("เกณฑ์ตัดสินแยกพื้นที่:", group_thresholds)
per_group_pred = np.array([s >= group_thresholds[a] for s, a in zip(test_scores, area_test)]).astype(int)
per_group_table = group_report(y_test, per_group_pred, area_test)
display(compare_groups(per_group_table))
overall_pg = metrics_from_counts(*confusion_matrix(y_test, per_group_pred, labels=[1, 0]).ravel())
print(f"ความถูกต้องรวม {overall_pg['accuracy']:.3f} (เดิม {overall['accuracy']:.3f}) | "
      f"จำนวนบ้านที่ต้องตรวจ {per_group_pred.sum()} (เดิม {test_pred.sum()})")
```

ความระลึกของสองพื้นที่ใกล้กันแล้ว แต่ต้องแลกด้วยการลดเกณฑ์ของพื้นที่ B จนบ้านสุจริตในพื้นที่ B ถูกตรวจมากขึ้นอีก (FPR สูงขึ้น) Equalized Odds จึงแย่ลง จำนวนบ้านที่ต้องตรวจเพิ่มขึ้น และความถูกต้องรวมลดลง

### 3.3 กราฟการแลกเปลี่ยน

ตรึงเกณฑ์ของพื้นที่ A ไว้ที่ `ALERT_THRESHOLD` แล้วเลื่อนเกณฑ์ของพื้นที่ B ตลอดช่วง กราฟแสดงผลต่างของ TPR ผลต่างของ FPR อัตราส่วนอัตราการเตือน และความถูกต้องรวม ไม่มีเกณฑ์ใดที่ทำให้ทุกเส้นดีพร้อมกัน

```python
mask_a, mask_b = area_test == "A", area_test == "B"
tpr_a, fpr_a, alert_a = rates_at(test_scores[mask_a], y_test[mask_a], ALERT_THRESHOLD)
trade = []
for t_b in THRESHOLD_GRID:
    tpr_b, fpr_b, alert_b = rates_at(test_scores[mask_b], y_test[mask_b], t_b)
    pred = np.where(mask_a, test_scores >= ALERT_THRESHOLD, test_scores >= t_b)
    trade.append({"t_B": t_b, "TPR gap |B-A|": abs(tpr_b - tpr_a), "FPR gap |B-A|": abs(fpr_b - fpr_a),
                  "alert-rate ratio": min(alert_a, alert_b) / max(alert_a, alert_b),
                  "overall accuracy": (pred == y_test).mean()})
trade = pd.DataFrame(trade).set_index("t_B")

ax = trade.plot(figsize=(8, 3.8), color=[COLORS["B"], COLORS["A"], COLORS["green"], COLORS["gray"]])
ax.axhline(0.8, color="black", ls=":", lw=1)
ax.text(0.45, 0.82, "four-fifths rule (ratio)", fontsize=8)
ax.set_xlabel("threshold for area B (area A fixed)")
ax.set_title("No single threshold for area B satisfies every fairness definition")
plt.tight_layout()
plt.show()
print("เกณฑ์ของ B ที่ทำให้ผลต่าง TPR น้อยที่สุด:", trade["TPR gap |B-A|"].idxmin(),
      "| ที่ทำให้ผลต่าง FPR น้อยที่สุด:", trade["FPR gap |B-A|"].idxmin())
```

### 3.4 ตัดชนิดมิเตอร์ออก หรือแก้ที่ต้นเหตุ

ทดลองสองทางเลือกแล้วเทียบกับโมเดลเดิม

- **ความเป็นธรรมโดยการไม่รับรู้ (Fairness through Unawareness)** — ฝึกใหม่โดยไม่ใช้ `old_meter`
- **แก้ที่ต้นเหตุทางวิศวกรรม** — จำลองว่าการไฟฟ้าเปลี่ยนมิเตอร์ในพื้นที่ B เป็นมิเตอร์อัจฉริยะ (`upgrade_rural_meters=True`) แล้วเก็บข้อมูลและฝึกใหม่ โดยยังใช้เกณฑ์ 0.10 เดียวกันทุกพื้นที่

```python
def scenario_row(name, y_true, y_pred, groups):
    table = group_report(y_true, y_pred, groups)
    total = metrics_from_counts(*confusion_matrix(y_true, y_pred, labels=[1, 0]).ravel())
    return {"scenario": name, "TPR A": table.loc["A", "tpr"], "TPR B": table.loc["B", "tpr"],
            "FPR A": table.loc["A", "fpr"], "FPR B": table.loc["B", "fpr"],
            "alert ratio": min(table["alert_rate"]) / max(table["alert_rate"]),
            "accuracy": total["accuracy"], "inspections": int(np.sum(y_pred))}


# ทางเลือก 1: ตัด old_meter
features_no_meter = [f for f in FEATURES if f != "old_meter"]
model_no_meter = fit_detector(train_df, features_no_meter)
pred_no_meter = (model_no_meter.predict_proba(test_df[features_no_meter])[:, 1] >= ALERT_THRESHOLD).astype(int)

# ทางเลือก 2: เปลี่ยนมิเตอร์ในพื้นที่ B แล้วเก็บข้อมูลใหม่
upgraded = make_households(upgrade_rural_meters=True)
train_up, test_up = split_households(upgraded)
model_up = fit_detector(train_up)
pred_up = (model_up.predict_proba(test_up[FEATURES])[:, 1] >= ALERT_THRESHOLD).astype(int)

scenarios = pd.DataFrame([
    scenario_row("baseline (threshold 0.10)", y_test, test_pred, area_test),
    scenario_row("per-group thresholds (TPR >= 0.80)", y_test, per_group_pred, area_test),
    scenario_row("drop old_meter feature", y_test, pred_no_meter, area_test),
    scenario_row("upgrade rural meters", test_up["theft"], pred_up, test_up["area"]),
]).set_index("scenario")
display(scenarios)
```

ผลที่ควรสังเกต

- การตัด `old_meter` **ไม่ได้ทำให้ความลำเอียงหายไป** เพราะต้นเหตุคือข้อมูลของพื้นที่ B มีคุณภาพต่ำกว่า (drop_pct คลาดเคลื่อนมาก และไม่มีสัญญาณงัดแงะ) ซึ่งยังอยู่ครบ ตรงกับข้อความในหนังสือว่า "การตัดคุณลักษณะอ่อนไหวทิ้งไม่พอ"
- การปรับเกณฑ์แยกกลุ่มย้ายปัญหาจากความระลึกไปเป็นอัตราบวกลวงและภาระการตรวจ
- การเปลี่ยนมิเตอร์ทำให้ตัววัดของสองพื้นที่ใกล้กันมากโดยไม่ต้องเสียความถูกต้อง ซึ่งสอดคล้องกับขั้นที่ 4 ของตัวอย่างที่ 9.2 ว่าทางแก้ที่ตรงจุดที่สุดอาจเป็นทางวิศวกรรม แต่ทางนี้มีต้นทุนและใช้เวลา ระหว่างรอจึงอาจต้องใช้มาตรการกระบวนการ เช่น ให้เจ้าหน้าที่ตรวจสอบก่อนดำเนินการกับผู้ใช้ไฟทุกราย

**คำถามชวนคิด** ถ้าคุณเป็นวิศวกรผู้รับผิดชอบ จะเสนอใช้สถานการณ์ใดในตารางระหว่างรองบประมาณเปลี่ยนมิเตอร์ และจะใช้ตัววัดใดติดตามผลหลังใช้งาน

อ้างอิง: หัวข้อ 9.1.2 รหัสเทียม PermutationImportance ตัวอย่างที่ 9.4–9.5 และรูปที่ 9.4

### 4.1 Permutation Feature Importance

เขียนตามรหัสเทียมในหนังสือ: วัดสมรรถนะบนชุดทดสอบ สลับค่าของคุณลักษณะหนึ่งแบบสุ่ม แล้ววัดใหม่ ผลที่ลดลงคือความสำคัญ เนื่องจากข้อมูลไม่สมดุลมาก (ผู้ลักลอบร้อยละ 4) ความถูกต้องแทบไม่เปลี่ยนเมื่อสลับค่า จึงใช้ **Average Precision** (พื้นที่ใต้เส้นโค้งความแม่นยำ–ความระลึก) ซึ่งคำนวณจากคะแนน `predict_proba` เป็นตัววัดแทน

```python
def permutation_importance_manual(model, X, y, features, metric, n_repeats=10, seed=RANDOM_SEED):
    """Permutation Feature Importance ตามรหัสเทียมในหัวข้อ 9.1.2 (metric รับ y_true กับคะแนน)"""
    rng = np.random.default_rng(seed)
    base = metric(y, model.predict_proba(X[features])[:, 1])
    result = {}
    for feature in features:
        scores = []
        for _ in range(n_repeats):
            X_perm = X[features].copy()
            X_perm[feature] = rng.permutation(X_perm[feature].to_numpy())
            scores.append(metric(y, model.predict_proba(X_perm)[:, 1]))
        result[feature] = base - np.mean(scores)
    return pd.Series(result), base


imp_manual, base_ap = permutation_importance_manual(model, test_df, test_df["theft"], FEATURES,
                                                    average_precision_score)
imp_sklearn = permutation_importance(model, test_df[FEATURES], test_df["theft"], scoring="average_precision",
                                     n_repeats=10, random_state=RANDOM_SEED).importances_mean
importance = pd.DataFrame({"manual": imp_manual, "sklearn": imp_sklearn}, index=FEATURES)
print(f"Average Precision ก่อนสลับ = {base_ap:.3f}")
display(importance)
```

ค่าจากฟังก์ชันที่เขียนเองกับของ scikit-learn ใกล้กันแต่ไม่เท่ากันพอดี เพราะใช้ลำดับการสุ่มต่างกัน ความสำคัญรวมของ `tamper_alerts` สูงมาก แต่ความสำคัญ **รวม** ซ่อนอีกเรื่องไว้ ลองคำนวณแยกตามพื้นที่

```python
per_area_importance = {}
for area in ["A", "B"]:
    part = test_df[test_df["area"] == area]
    per_area_importance[f"area {area}"], _ = permutation_importance_manual(
        model, part, part["theft"], FEATURES, average_precision_score)
per_area_importance = pd.DataFrame(per_area_importance)
display(per_area_importance)

ax = per_area_importance.plot.barh(figsize=(7, 3), color=[COLORS["A"], COLORS["B"]])
ax.set_xlabel("drop in average precision when shuffled")
ax.set_title("Permutation importance by area")
plt.tight_layout()
plt.show()
```

ในพื้นที่ B ความสำคัญของ `tamper_alerts` เป็น 0 พอดี (ค่าเป็น 0 ทุกหลัง สลับอย่างไรก็ไม่เปลี่ยน) และ `old_meter` ก็เป็น 0 ในแต่ละพื้นที่ เพราะค่าคงที่ภายในพื้นที่ การดูความสำคัญแยกกลุ่มจึงช่วยให้เห็นว่าโมเดลมี "ข้อมูลน้อยกว่า" สำหรับพื้นที่ B ซึ่งเป็นหลักฐานของความลำเอียงจากการวัด

### 4.2 ค่าแชปลีย์ด้วยการไล่ทุกลำดับ (ตัวอย่างที่ 9.4)

ค่าแชปลีย์ของคุณลักษณะ $i$ คือค่าเฉลี่ยของส่วนเพิ่มที่ $i$ ทำให้เกิดเมื่อเข้าร่วม เฉลี่ยทุกลำดับการเข้าร่วม ฟังก์ชัน `shapley_by_orders` ทำตามนิยามนี้ตรงตัว โดยรับ `value(S)` ซึ่งคืนผลทำนายเมื่อรู้ค่าจริงเฉพาะคุณลักษณะในเซต `S`

```python
def shapley_by_orders(players, value):
    """ค่าแชปลีย์ด้วยการไล่ทุกลำดับการเข้าร่วม (d! ลำดับ)"""
    phi = {p: 0.0 for p in players}
    orders = list(itertools.permutations(players))
    for order in orders:
        joined = frozenset()
        for p in order:
            phi[p] += value(joined | {p}) - value(joined)
            joined = joined | {p}
    return {p: total / len(orders) for p, total in phi.items()}


# ตัวอย่างที่ 9.4: v(S) ของโมเดลเตือนมอเตอร์
MOTOR_V = {frozenset(): 10, frozenset({"vibration"}): 30, frozenset({"temperature"}): 20,
           frozenset({"vibration", "temperature"}): 90}
phi_motor = shapley_by_orders(["vibration", "temperature"], MOTOR_V.__getitem__)
compare("φ การสั่น", phi_motor["vibration"], 45, 1e-9)
compare("φ อุณหภูมิ", phi_motor["temperature"], 35, 1e-9)
compare("ผลรวม = v(F) − v(∅)", sum(phi_motor.values()), 90 - 10, 1e-9)
```

### 4.3 SHAP ของโมเดลเชิงเส้นและแผนภาพน้ำตก (ตัวอย่างที่ 9.5)

สำหรับโมเดลเชิงเส้น $z = w_0 + \sum_i w_i x_i$ เมื่อถือว่าคุณลักษณะเป็นอิสระต่อกันและใช้ค่าเฉลี่ยเป็นข้อมูลพื้นหลัง ค่าแชปลีย์ในหน่วยของ $z$ คือ $\phi_i = w_i(x_i - \bar{x}_i)$ เซลล์ถัดไปคำนวณบ้าน H ของตัวอย่างที่ 9.5 สองวิธี คือด้วยสูตร และด้วยการไล่ทุกลำดับโดยให้ $v(S)$ คือ $z$ เมื่อคุณลักษณะนอก $S$ ใช้ค่าเฉลี่ย แล้ววาดแผนภาพน้ำตกแบบรูปที่ 9.4

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


def linear_shap(weights, x, x_mean):
    """ค่าแชปลีย์ของโมเดลเชิงเส้นในหน่วยของ z"""
    return np.asarray(weights) * (np.asarray(x, dtype=float) - np.asarray(x_mean, dtype=float))


def masked_linear_value(w0, weights, x, x_mean):
    """คืนฟังก์ชัน v(S): z เมื่อรู้ค่าจริงเฉพาะคุณลักษณะในเซต S ที่เหลือใช้ค่าเฉลี่ย"""
    def value(subset):
        mixed = np.array(x_mean, dtype=float)
        for j in subset:
            mixed[j] = x[j]
        return w0 + float(np.dot(weights, mixed))
    return value


def plot_waterfall(base, phis, labels, title, threshold=0.0):
    """แผนภาพน้ำตกแนวนอน: เริ่มจากค่าพื้นฐาน บวกค่าแชปลีย์ทีละตัว จบที่ผลทำนาย"""
    fig, ax = plt.subplots(figsize=(7, 0.6 * len(phis) + 1.6))
    level = base
    for k, (phi, label) in enumerate(zip(phis, labels)):
        color = COLORS["B"] if phi > 0 else COLORS["A"]
        ax.barh(k, phi, left=level, color=color)
        ax.text(level + phi + (0.03 if phi >= 0 else -0.03), k, f"{phi:+.2f}", va="center",
                ha="left" if phi >= 0 else "right", fontsize=9)
        level += phi
    levels = base + np.concatenate([[0], np.cumsum(phis)])
    low, high = min(levels.min(), threshold), max(levels.max(), threshold)
    ax.set_xlim(low - 0.15 * (high - low) - 0.3, high + 0.15 * (high - low) + 0.3)
    ax.axvline(base, color=COLORS["gray"], lw=1)
    ax.axvline(threshold, color="black", ls="--", lw=1)
    ax.set_yticks(range(len(labels)), labels)
    ax.invert_yaxis()
    ax.set_xlabel(f"z (log-odds): base {base:.2f} -> prediction {level:.2f}   (dashed = alert threshold)")
    ax.set_title(title)
    plt.tight_layout()
    plt.show()


BOOK_W0 = -3.0
BOOK_W = np.array([0.05, 0.8, 1.2])          # การใช้ไฟลดลง, สัญญาณงัดแงะ, มิเตอร์รุ่นเก่า
BOOK_MEAN = np.array([10.0, 0.5, 0.5])
HOUSE_H = np.array([40.0, 0.0, 1.0])

z_h = BOOK_W0 + BOOK_W @ HOUSE_H
z_bar = BOOK_W0 + BOOK_W @ BOOK_MEAN
phi_h = linear_shap(BOOK_W, HOUSE_H, BOOK_MEAN)
phi_h_enum = shapley_by_orders([0, 1, 2], masked_linear_value(BOOK_W0, BOOK_W, HOUSE_H, BOOK_MEAN))
compare("z_H", z_h, 0.2, 1e-9)
compare("σ(z_H)", sigmoid(z_h), 0.550, 5e-4)
compare("z̄", z_bar, -1.5, 1e-9)
compare("σ(z̄)", sigmoid(z_bar), 0.182, 5e-4)
for j, book in enumerate([1.5, -0.4, 0.6]):
    compare(f"φ{j + 1} สูตร / ไล่ลำดับ", phi_h_enum[j], book, 1e-9)
    assert abs(phi_h[j] - phi_h_enum[j]) < 1e-9
compare("z̄ + Σφ", z_bar + phi_h.sum(), z_h, 1e-9)
plot_waterfall(z_bar, phi_h, ["drop = 40%", "tamper alerts = 0", "old meter = 1"],
               "Example 9.5: why house H is flagged")
```

**คำอธิบายเชิงสมมติ (Counterfactual)** ถ้าเปลี่ยนเฉพาะชนิดมิเตอร์เป็นรุ่นใหม่ บ้าน H จะได้ $z = -1.0$ และไม่ถูกเตือน และถ้ามิเตอร์ยังเป็นรุ่นเก่า บ้าน H จะไม่ถูกเตือนเมื่อการใช้ไฟลดลงน้อยกว่าค่าที่แก้สมการ $-3 + 0.05x_1 + 1.2 = 0$ ได้

> **ข้อควรระวัง** การ "ตัดแท่งมิเตอร์รุ่นเก่า" ออกจากแผนภาพน้ำตกไม่ใช่คำอธิบายเชิงสมมติแบบเดียวกัน การตัดแท่งเท่ากับตั้ง $x_3$ เป็นค่าเฉลี่ย 0.5 ได้ $z = 0.2 - 0.6 = -0.4$ ส่วนคำอธิบายเชิงสมมติตั้ง $x_3 = 0$ ได้ $z = -1.0$ ทั้งสองแบบต่ำกว่าเกณฑ์ จึงให้ข้อสรุปเดียวกันว่าชนิดมิเตอร์พลิกผลการตัดสิน

```python
house_new_meter = HOUSE_H.copy()
house_new_meter[2] = 0
z_cf = BOOK_W0 + BOOK_W @ house_new_meter
compare("z เมื่อมิเตอร์เป็นรุ่นใหม่", z_cf, -1.0, 1e-9)
compare("σ(−1.0)", sigmoid(z_cf), 0.269, 5e-4)
x1_limit = -(BOOK_W0 + BOOK_W[2] * 1) / BOOK_W[0]
compare("ไม่ถูกเตือนเมื่อการใช้ไฟลดลงน้อยกว่า (%)", x1_limit, 36, 1e-9)
print(f"ตัดแท่งมิเตอร์ออกจากแผนภาพน้ำตก: z = {z_bar + phi_h[0] + phi_h[1]:.2f}")
```

### 4.4 อธิบายบ้านที่ถูกเตือนผิดในข้อมูลจำลอง

นำวิธีเดียวกันมาใช้กับโมเดลในส่วนที่ 2 เลือกบ้านสุจริตในพื้นที่ B ที่ถูกเตือน (FP) และจะไม่ถูกเตือนถ้าใช้มิเตอร์รุ่นใหม่ (เลือกหลังที่คะแนนสูงสุดในกลุ่มนี้) แล้วแจกแจงคะแนนด้วยค่าแชปลีย์ในหน่วยลอการิทึมของอัตราต่อรอง โดยใช้ค่าเฉลี่ยของชุดฝึกเป็นข้อมูลพื้นหลัง

```python
w0_model = model.intercept_[0]
w_model = model.coef_[0]
background_mean = train_df[FEATURES].mean().to_numpy()

fp_b = test_df[(test_df["area"] == "B") & (test_df["theft"] == 0) & (test_pred == 1)]
# เลือกบ้านที่ "ถ้าเป็นมิเตอร์รุ่นใหม่จะไม่ถูกเตือน" หลังที่มีคะแนนสูงสุด (ถ้าไม่มีให้ใช้หลังที่คะแนนสูงสุด)
fp_b_new_meter = fp_b[FEATURES].assign(old_meter=0)
flips = fp_b[model.predict_proba(fp_b_new_meter)[:, 1] < ALERT_THRESHOLD]
candidates = flips if len(flips) else fp_b
house_idx = candidates.index[np.argmax(model.predict_proba(candidates[FEATURES])[:, 1])]
print(f"บ้านสุจริตในพื้นที่ B ที่ถูกเตือน {len(fp_b)} หลัง | ในจำนวนนี้ {len(flips)} หลังจะไม่ถูกเตือนถ้าเป็นมิเตอร์รุ่นใหม่")
house = test_df.loc[house_idx, FEATURES].to_numpy(dtype=float)
print("บ้านที่เลือก:", test_df.loc[house_idx, FEATURES].to_dict())

phi_house = linear_shap(w_model, house, background_mean)
phi_house_enum = shapley_by_orders(list(range(len(FEATURES))),
                                   masked_linear_value(w0_model, w_model, house, background_mean))
z_house = w0_model + w_model @ house
z_base = w0_model + w_model @ background_mean
print(f"z = {z_house:.3f} (p = {sigmoid(z_house):.3f}) | ค่าพื้นฐาน z̄ = {z_base:.3f} (p = {sigmoid(z_base):.3f})")
print("สูตรตรงกับการไล่ลำดับ:", np.allclose(phi_house, [phi_house_enum[j] for j in range(len(FEATURES))]))
z_threshold = np.log(ALERT_THRESHOLD / (1 - ALERT_THRESHOLD))   # เกณฑ์ 0.10 ในหน่วย z
plot_waterfall(z_base, phi_house, [f"{f} = {v:g}" for f, v in zip(FEATURES, house)],
               "A false alarm in area B", threshold=z_threshold)

# คำอธิบายเชิงสมมติ: ถ้าบ้านหลังนี้ใช้มิเตอร์รุ่นใหม่ (เปลี่ยนเฉพาะ old_meter)
house_cf = house.copy()
house_cf[FEATURES.index("old_meter")] = 0
p_cf = sigmoid(w0_model + w_model @ house_cf)
print(f"ถ้าเปลี่ยนเฉพาะ old_meter เป็น 0: p = {p_cf:.4f} (เกณฑ์ {ALERT_THRESHOLD}) → {'ยังถูกเตือน' if p_cf >= ALERT_THRESHOLD else 'ไม่ถูกเตือน'}")
```

แผนภาพน้ำตกบอกว่าคะแนนของบ้านหลังนี้มาจาก `drop_pct` เป็นหลัก (ซึ่งในพื้นที่ B ส่วนหนึ่งเป็นความคลาดเคลื่อนของการวัด) และคำอธิบายเชิงสมมติแสดงว่า `old_meter` เพียงอย่างเดียวพลิกผลการเตือนได้ เช่นเดียวกับบ้าน H ในตัวอย่างที่ 9.5 จำนวนบ้านที่พิมพ์ไว้ด้านบนบอกว่าเรื่องนี้เกิดกับบ้านสุจริตกี่หลัง คำอธิบายบอกว่า **โมเดล** ใช้อะไร แต่ไม่ได้บอกว่าบ้านนี้ลักลอบจริงหรือไม่ เจ้าหน้าที่จึงควรได้รับคำอธิบายนี้พร้อมคำเตือนว่าข้อมูลมิเตอร์จานหมุนคลาดเคลื่อนสูง

## ส่วนที่ 5 ความเป็นส่วนตัวของข้อมูลมิเตอร์

อ้างอิง: หัวข้อ 9.3.1 ตารางที่ 9.8 ตัวอย่างที่ 9.13–9.14 และรูปที่ 9.9

### 5.1 การโจมตีด้วยการหาผลต่าง (ตัวอย่างที่ 9.13)

การเปิดเผยค่าเฉลี่ยสองครั้งที่ต่างกันเพียงหนึ่งหลังทำให้รู้ค่าของหลังนั้นได้ตรงตัว เซลล์ถัดไปคำนวณตัวเลขในหนังสือ แล้วจำลองหมู่บ้าน 100 หลังที่บ้าน X ใช้ไฟ 448 kWh

```python
def differencing_attack(mean_all, n_all, mean_without, n_without):
    """หาค่าของรายที่หายไปจากค่าเฉลี่ยสองครั้ง"""
    return mean_all * n_all - mean_without * n_without


compare("บ้าน X จากตัวเลขในหนังสือ (kWh)", differencing_attack(250.0, 100, 248.0, 99), 448, 1e-9)

rng = np.random.default_rng(RANDOM_SEED)
village = rng.gamma(shape=6, scale=40, size=100)      # การใช้ไฟรายเดือนของ 100 หลัง (kWh)
village[-1] = 448.0                                    # บ้าน X อยู่แถวสุดท้าย
recovered = differencing_attack(village.mean(), 100, village[:-1].mean(), 99)
print(f"ค่าเฉลี่ย 100 หลัง {village.mean():.1f} | 99 หลัง {village[:-1].mean():.1f} | "
      f"ผู้โจมตีได้ค่าบ้าน X = {recovered:.1f} kWh")
```

### 5.2 กลไกลาปลาซ (ตัวอย่างที่ 9.14)

ตัดค่าของแต่ละบ้านให้อยู่ในช่วง 0–1,000 kWh ก่อนรวม ความไวของผลรวมจึงเป็น $\Delta f = 1{,}000$ kWh แล้วเติมสัญญาณรบกวน $\text{Lap}(b)$ ที่ $b = \Delta f/\varepsilon$ ลงในผลรวมก่อนเปิดเผย ส่วนเบี่ยงเบนมาตรฐานของสัญญาณรบกวนคือ $\sqrt{2}\,b$ และ $P(|\text{noise}| > t) = e^{-t/b}$

```python
CLIP_KWH = 1000.0


def laplace_sum(values, epsilon, rng, clip=CLIP_KWH):
    """ผลรวมที่เติมสัญญาณรบกวนลาปลาซ (ตัดค่าแต่ละรายไว้ที่ 0..clip ความไว = clip)"""
    clipped = np.clip(values, 0, clip)
    return clipped.sum() + rng.laplace(0.0, clip / epsilon)


for eps, book_sd in [(1.0, 1414), (0.1, 14142)]:
    compare(f"SD ของสัญญาณรบกวนที่ ε = {eps}", math.sqrt(2) * CLIP_KWH / eps, book_sd, 0.5)
compare("P(|noise| > 448) ที่ ε = 1 (สูตร)", math.exp(-448 / CLIP_KWH), 0.64, 5e-3)

# ผู้โจมตีพยายามหาผลต่างจากผลรวมที่เติมสัญญาณรบกวนสองครั้ง ทำซ้ำ 10,000 ครั้ง
rng = np.random.default_rng(RANDOM_SEED)
trials = 10_000
guesses = np.array([laplace_sum(village, 1.0, rng) - laplace_sum(village[:-1], 1.0, rng)
                    for _ in range(trials)])
single_noise = rng.laplace(0, CLIP_KWH / 1.0, trials)
compare("P(|noise| > 448) จากการสุ่ม", np.mean(np.abs(single_noise) > 448), 0.64, 0.02)
print(f"สัดส่วนครั้งที่ผู้โจมตีเดาค่าบ้าน X ได้ใกล้เคียง ±50 kWh: {np.mean(np.abs(guesses - 448) <= 50):.3f}")

plt.figure(figsize=(7, 3))
plt.hist(guesses, bins=80, range=(-6000, 7000), color=COLORS["A"], alpha=0.8)
plt.axvline(448, color=COLORS["B"], lw=2, label="true value of house X (448 kWh)")
plt.xlabel("attacker's estimate of house X (kWh), epsilon = 1 per release")
plt.title("Differencing attack against Laplace-noised sums")
plt.legend()
plt.tight_layout()
plt.show()
```

ค่าที่ผู้โจมตีเดาได้กระจายกว้างหลายพัน kWh ผลต่างของผลรวมสองครั้งจึงไม่บอกค่าบ้าน X อีกต่อไป (การเปิดเผยสองครั้งใช้งบรวม $\varepsilon = 2$)

### 5.3 ความคลาดเคลื่อนของค่าเฉลี่ยเทียบกับขนาดกลุ่ม (รูปที่ 9.9)

ส่วนเบี่ยงเบนมาตรฐานของค่าเฉลี่ยรายบ้าน = $\sqrt{2}\,b / n$ เทียบกับค่าเฉลี่ยจริง 250 kWh เป็นร้อยละ กราฟวาดเส้นจากสูตร และจุดจากการจำลองจริงที่ $n$ = 100 และ 10,000

```python
def relative_error_pct(n, epsilon, mean_kwh=250.0, clip=CLIP_KWH):
    return math.sqrt(2) * clip / epsilon / n / mean_kwh * 100


# (n, ε): (ร้อยละในหนังสือ, ความคลาดเคลื่อนจากการปัดของหนังสือ)
book_914 = {(100, 1.0): (5.7, 0.05), (100, 0.1): (57, 0.5), (10_000, 1.0): (0.06, 0.005),
            (10_000, 0.1): (0.57, 0.005)}
for (n, eps), (book_pct, tol) in book_914.items():
    compare(f"n = {n:>6}, ε = {eps}: ความคลาดเคลื่อนสัมพัทธ์ (%)", relative_error_pct(n, eps), book_pct, tol)

n_values = np.logspace(1, 5, 100)
plt.figure(figsize=(7, 4))
rng = np.random.default_rng(RANDOM_SEED)
for eps, color in [(0.1, COLORS["B"]), (1.0, COLORS["A"]), (10.0, COLORS["green"])]:
    plt.loglog(n_values, [relative_error_pct(n, eps) for n in n_values], color=color, label=f"epsilon = {eps}")
    if eps in (0.1, 1.0):
        for n in (100, 10_000):
            houses = rng.gamma(shape=6, scale=250 / 6, size=n)   # ค่าเฉลี่ยราว 250 kWh
            means = np.array([laplace_sum(houses, eps, rng) / n for _ in range(2000)])
            plt.scatter(n, means.std() / 250 * 100, color=color, edgecolor="black", zorder=3)
plt.axhline(1, color=COLORS["gray"], ls="--", lw=1)
plt.text(12, 1.2, "1% (example acceptable error)", fontsize=8)
plt.xlabel("number of households n")
plt.ylabel("relative error of the mean (%)")
plt.title("Differential privacy works best for large groups (dots = simulation)")
plt.legend()
plt.tight_layout()
plt.show()
```

### 5.4 ตรวจ k-Anonymity

ตารางข้อมูลผู้ใช้ไฟที่ตัดชื่อและเลขที่ผู้ใช้ไฟออกแล้ว ยังอาจระบุตัวบุคคลได้จาก **ตัวแปรที่ใช้ระบุตัวได้ทางอ้อม (Quasi-identifier)** เช่น ตำบล ชนิดบ้าน และจำนวนผู้อยู่อาศัย ตารางมีสมบัติ k-Anonymity เมื่อทุกระเบียนมีอย่างน้อย $k$ ระเบียนที่มีค่าตัวแปรเหล่านี้เหมือนกัน (Sweeney, 2002)

ตารางจำลองนี้มีคอลัมน์ `medical_device` (บ้านที่มีเครื่องมือแพทย์ใช้ไฟ เช่น เครื่องผลิตออกซิเจน) ซึ่งเผยข้อมูลสุขภาพ จึงเป็นข้อมูลอ่อนไหวตามมาตรา 26 ของ PDPA

```python
def make_customer_table(n=60, seed=RANDOM_SEED):
    """ตารางผู้ใช้ไฟจำลองที่ตัดชื่อออกแล้ว (ค่าสมมติ)"""
    rng = np.random.default_rng(seed)
    subdistricts = {"Nai Mueang": "Mueang", "Ban Pet": "Mueang", "Sila": "Mueang",
                    "Tha Phra": "Rural", "Don Chang": "Rural", "Ban Kho": "Rural"}
    names = list(subdistricts)
    sub = rng.choice(names, n)
    return pd.DataFrame({
        "subdistrict": sub,
        "district": [subdistricts[s] for s in sub],
        "house_type": rng.choice(["detached", "townhouse", "shophouse"], n, p=[0.5, 0.3, 0.2]),
        "occupants": rng.integers(1, 7, n),
        "monthly_kwh": rng.gamma(6, 40, n).round(0),
        "medical_device": rng.random(n) < 0.08,
    })


def equivalence_class_sizes(df, quasi_identifiers):
    """ขนาดของกลุ่มที่มีค่าตัวแปรระบุตัวทางอ้อมเหมือนกัน (คืน Series ต่อแถว)"""
    return df.groupby(quasi_identifiers)[quasi_identifiers[0]].transform("size")


customers = make_customer_table()
QI_DETAILED = ["subdistrict", "house_type", "occupants"]
sizes = equivalence_class_sizes(customers, QI_DETAILED)
print(f"k ของตารางละเอียด = {sizes.min()} | ระเบียนที่ไม่ซ้ำใครเลย (k = 1): {(sizes == 1).sum()} จาก {len(customers)}")
display(customers.head(8))
```

ตารางละเอียดมี k = 1 แปลว่ามีบ้านที่ระบุตัวได้เฉพาะเจาะจง ใครที่รู้ว่าเพื่อนบ้านอยู่ตำบลใด บ้านแบบใด อยู่กันกี่คน ก็หาแถวของบ้านนั้นพบ และรู้ว่าบ้านนั้นมีเครื่องมือแพทย์หรือไม่ วิธีแก้คือ **ทำให้หยาบลง (Generalization)** เช่น ใช้อำเภอแทนตำบล และใช้ช่วงจำนวนผู้อยู่อาศัยแทนจำนวนจริง (แบบฝึกปฏิบัติข้อ 4 จะให้เขียนฟังก์ชันตรวจ k และเลือกการทำให้หยาบลงเอง)

```python
coarse = customers.assign(occupants_band=pd.cut(customers["occupants"], [0, 2, 4, 6], labels=["1-2", "3-4", "5-6"]))
for qi in (["district", "house_type", "occupants"], ["district", "house_type", "occupants_band"],
           ["district", "occupants_band"]):
    print(f"{str(qi):45s} k = {equivalence_class_sizes(coarse, qi).min()}")
```

ยิ่งหยาบ k ยิ่งสูง แต่ข้อมูลก็มีประโยชน์น้อยลง และ k-Anonymity ยังไม่ป้องกันกรณีที่ **ทุกคนในกลุ่มมีค่าอ่อนไหวเหมือนกัน** (ตารางที่ 9.8) ในข้อมูลมิเตอร์จริง รูปแบบโหลดรายวันเองก็เป็นตัวระบุตัวทางอ้อมที่ทรงพลังมาก การรวมข้อมูลระดับหม้อแปลงตั้งแต่ต้น (Privacy by Design ตัวอย่างที่ 9.12) จึงมักได้ผลกว่า

## ส่วนที่ 6 พลังงานและคาร์บอนของการฝึกและการอนุมาน

อ้างอิง: หัวข้อ 9.2.3 ตัวอย่างที่ 9.10–9.11

สูตรประมาณคือ $E = P_{\text{IT}} \times t \times \text{PUE}$ และ $\text{CO}_2\text{e} = E \times EF$ ใช้ค่าสมมติ $EF$ = 0.5 kgCO₂e/kWh ตามหนังสือ (ค่าจริงต้องใช้ค่าล่าสุดที่ อบก. ประกาศ) เริ่มจากตรวจตัวอย่างในหนังสือ

```python
EMISSION_FACTOR = 0.5   # kgCO2e/kWh ค่าสมมติของหนังสือ


def energy_kwh(power_w, hours, pue=1.0):
    return power_w / 1000 * hours * pue


# ตัวอย่างที่ 9.10: GPU 300 W 10 ชั่วโมง PUE 1.5 ค้นหาไฮเปอร์พารามิเตอร์ 50 รอบ
e_910 = energy_kwh(300, 10, 1.5)
compare("9.10 พลังงานการฝึก 1 รอบ (kWh)", e_910, 4.5, 1e-9)
compare("9.10 CO2e 1 รอบ (kg)", e_910 * EMISSION_FACTOR, 2.25, 1e-9)
compare("9.10 CO2e 50 รอบ (kg)", 50 * e_910 * EMISSION_FACTOR, 112.5, 1e-9)
# ตัวอย่างที่ 9.11
compare("9.11 กล้อง Edge AI 15 W ทั้งปี (kWh)", energy_kwh(15, 24 * 365), 131.4, 1e-9)
saved, used = 2_000_000 * 0.05, energy_kwh(250, 24 * 365)
compare("9.11 ระบบน้ำเย็น: ประหยัดได้กี่เท่าของที่ใช้", saved / used, 46, 0.5)
```

### 6.1 จับเวลาโมเดลของเราเอง

จับเวลาการฝึกโมเดลในส่วนที่ 2 การค้นหาไฮเปอร์พารามิเตอร์ด้วย `GridSearchCV` (ค่า `C` 10 ค่า × 5 fold = 50 รอบ ตรงกับตัวอย่างที่ 9.10) และการอนุมานต่อบ้านหนึ่งหลัง แล้วประมาณพลังงาน

ความเสี่ยง = โอกาสเกิด (L) × ความรุนแรง (S) ระดับละ 1–5 แบ่งเป็น ต่ำ (1–4) ปานกลาง (5–9) สูง (10–14) และสูงมาก (15–25) ตามเกณฑ์สมมติในหนังสือ เซลล์ถัดไปคำนวณตารางของตัวอย่างที่ 9.17 และวาดตำแหน่งก่อนและหลังใช้มาตรการแบบรูปที่ 9.12

```python
def risk_level(score):
    return "low" if score <= 4 else "medium" if score <= 9 else "high" if score <= 14 else "very high"


risks = pd.DataFrame({
    "hazard": ["1 drift after rooftop PV", "2 shed critical load", "3 same plants shed repeatedly",
               "4 spoofed shed command", "5 operator over-trust"],
    "L": [4, 2, 4, 2, 3], "S": [3, 5, 2, 5, 3],
    "L_after": [2, 1, 2, 1, 2], "S_after": [3, 5, 2, 5, 3],
}).set_index("hazard")
risks["score"] = risks["L"] * risks["S"]
risks["level"] = risks["score"].map(risk_level)
risks["score_after"] = risks["L_after"] * risks["S_after"]
risks["level_after"] = risks["score_after"].map(risk_level)
display(risks)
print("ระดับสูงขึ้นไป ก่อน/หลังใช้มาตรการ:",
      int(risks["level"].isin(["high", "very high"]).sum()), "/",
      int(risks["level_after"].isin(["high", "very high"]).sum()))

grid = np.array([[l * s for s in range(1, 6)] for l in range(1, 6)])
level_index = np.vectorize(lambda v: 0 if v <= 4 else 1 if v <= 9 else 2 if v <= 14 else 3)(grid)
fig, ax = plt.subplots(figsize=(5.5, 4.5))
cmap = plt.matplotlib.colors.ListedColormap(["#cfe8d2", "#fff3b0", "#f6c9a8", "#f4a3a3"])
ax.imshow(level_index, origin="lower", cmap=cmap, extent=(0.5, 5.5, 0.5, 5.5))
for k, (name, row) in enumerate(risks.iterrows()):
    jitter = {1: -0.12, 3: 0.12}.get(k, 0.0)   # เยื้องเฉพาะข้อ 2 และ 4 ที่ซ้อนกัน
    ax.annotate("", xy=(row["S_after"] + jitter, row["L_after"]), xytext=(row["S"] + jitter, row["L"]),
                arrowprops=dict(arrowstyle="->", color=COLORS["gray"]))
    ax.scatter(row["S"] + jitter, row["L"], s=170, color=COLORS["gray"], zorder=3)
    ax.scatter(row["S_after"] + jitter, row["L_after"], s=170, color=COLORS["A"], zorder=3)
    ax.text(row["S"] + jitter, row["L"], name[0], color="white", ha="center", va="center", fontsize=8, zorder=4)
    ax.text(row["S_after"] + jitter, row["L_after"], name[0], color="white", ha="center", va="center",
            fontsize=8, zorder=4)
ax.set_xlabel("severity S")
ax.set_ylabel("likelihood L")
ax.set_title("Risk matrix: before (gray) -> after (blue)")
plt.tight_layout()
plt.show()
```

### ข้อ 1 ตัววัดความเป็นธรรมแยกกลุ่มด้วย NumPy

เขียน `fairness_by_group(y_true, y_pred, groups)` ที่คืน DataFrame หนึ่งแถวต่อกลุ่ม (index คือชื่อกลุ่ม เรียงตามตัวอักษร) มีคอลัมน์ `n` `base_rate` `alert_rate` `tpr` `fpr` `ppv` `accuracy` ตามตารางที่ 9.3 **ห้ามใช้** `confusion_matrix` หรือ `metrics_from_counts` ให้นับ TP, FN, FP, TN เองด้วยหน้ากากบูลีน เช่น `((y == 1) & (p == 1)).sum()`

```python
# เซลล์ตรวจคำตอบข้อ 1 (ไม่ต้องแก้)
def expand_counts(counts_by_group):
    """แปลง (TP, FN, FP, TN) ของแต่ละกลุ่มเป็นอาร์เรย์ y_true, y_pred, groups"""
    y_true, y_pred, groups = [], [], []
    for g, (tp, fn, fp, tn) in counts_by_group.items():
        for yt, yp, k in [(1, 1, tp), (1, 0, fn), (0, 1, fp), (0, 0, tn)]:
            y_true += [yt] * k
            y_pred += [yp] * k
            groups += [g] * k
    return np.array(y_true), np.array(y_pred), np.array(groups)


result = fairness_by_group(*expand_counts(BOOK_92))
if result is None:
    print("ข้อ 1: ยังไม่ได้ทำ")
else:
    cols = ["n", "base_rate", "alert_rate", "tpr", "fpr", "ppv", "accuracy"]
    ok_book = np.allclose(result[cols].astype(float).to_numpy(), table_92[cols].astype(float).to_numpy())
    mine_sim = fairness_by_group(y_test, test_pred, area_test)
    ok_sim = np.allclose(mine_sim[cols].astype(float).to_numpy(), baseline_table[cols].astype(float).to_numpy())
    print("ตัวอย่างที่ 9.2:", "ตรง" if ok_book else "ไม่ตรง", "| ข้อมูลจำลองส่วนที่ 2:", "ตรง" if ok_sim else "ไม่ตรง")
    print("ถูกต้อง" if ok_book and ok_sim else "ยังไม่ถูก ลองตรวจอีกครั้ง")
```
