## ส่วนที่ 1 เซลล์ประสาทเทียม Perceptron และปัญหา XOR

### 1.1 เซลล์ประสาทเทียมเป็นวงจรลอจิก (ตัวอย่างที่ 5.1)

เซลล์ประสาทเทียมคำนวณ $z = \mathbf{w}^\top \mathbf{x} + b$ แล้วส่งผ่านฟังก์ชันกระตุ้น $a = f(z)$ ถ้าใช้ฟังก์ชันขั้นบันได เซลล์จะ "ยิง" ($a = 1$) เมื่อ $z \ge 0$ ไบแอส $b$ จึงทำหน้าที่เป็นเกณฑ์ว่าต้องมีสัญญาณเข้ากี่เส้นจึงจะยิง เซลล์ถัดไปใช้ค่าน้ำหนัก $w_1 = w_2 = 1$ เท่ากัน แต่เปลี่ยนไบแอสเพื่อสร้างลอจิก AND (สั่งปลดวงจรเมื่อกระแสเกินและอุณหภูมิขดลวดสูงพร้อมกัน) และ OR (ส่งสัญญาณเตือนเมื่อมีอย่างน้อยหนึ่งเงื่อนไข)

```python
def neuron(x, w, b, activation=step):
    """เซลล์ประสาทเทียมหนึ่งตัว: คืนค่า (z, a)"""
    z = float(np.dot(w, x) + b)
    return z, int(activation(z))


INPUTS_2 = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
rows = []
for x in INPUTS_2:
    z_and, a_and = neuron(x, w=[1, 1], b=-1.5)
    z_or, a_or = neuron(x, w=[1, 1], b=-0.5)
    rows.append({"x1": x[0], "x2": x[1], "z (AND, b=-1.5)": z_and, "a AND": a_and,
                 "z (OR, b=-0.5)": z_or, "a OR": a_or})
display(pd.DataFrame(rows))

# 2-out-of-3 Voting: เซนเซอร์สามตัว ปลดวงจรเมื่อร้อนเกินอย่างน้อยสองตัว
voting = pd.DataFrame([{"k (sensors hot)": k, "z = k - 1.5": neuron(np.r_[np.ones(k), np.zeros(3 - k)], [1, 1, 1], -1.5)[0],
                        "trip a": neuron(np.r_[np.ones(k), np.zeros(3 - k)], [1, 1, 1], -1.5)[1]} for k in range(4)])
display(voting.set_index("k (sensors hot)").T)
```

ผลตรงกับตารางในตัวอย่างที่ 5.1 ทุกแถว การเปลี่ยนไบแอสจาก −1.5 เป็น −0.5 อย่างเดียวเปลี่ยน AND เป็น OR

### 1.2 ฝึก Perceptron ให้เรียนรู้ลอจิก OR (ตัวอย่างที่ 5.2)

กฎการเรียนรู้ของ Perceptron ปรับค่าน้ำหนักเฉพาะเมื่อทายผิด

$$w_i \leftarrow w_i + \eta\,(y - \hat{y})\,x_i, \qquad b \leftarrow b + \eta\,(y - \hat{y})$$

ฟังก์ชัน `perceptron_train` ด้านล่างเขียนตามรหัสเทียม PERCEPTRON-TRAIN ในหนังสือทุกบรรทัด และเก็บทุกแถวของการคำนวณไว้ในตาราง เพื่อเทียบกับตารางในตัวอย่างที่ 5.2

```python
def perceptron_train(data, eta=1.0, max_epochs=20):
    """PERCEPTRON-TRAIN ในหนังสือ: data คือรายการ ((x1, x2, ...), y)
    คืนค่า (w, b, ตารางทุกแถว, จำนวนที่ทายผิดในแต่ละรอบ)"""
    w = np.zeros(len(data[0][0]))
    b = 0.0
    records, errors_per_epoch = [], []
    for epoch in range(1, max_epochs + 1):
        n_wrong = 0
        for x, y in data:                          # ป้อนทีละตัวอย่างตามลำดับ
            x = np.asarray(x, dtype=float)
            z = w @ x + b
            y_hat = 1 if z >= 0 else 0
            error = y - y_hat
            if error != 0:                         # ทายผิด → ขยับค่าน้ำหนัก
                w = w + eta * error * x
                b = b + eta * error
                n_wrong += 1
            records.append({"epoch": epoch, "x": tuple(int(v) for v in x), "y": y, "z": z,
                            "y_hat": y_hat, "y - y_hat": error, "w1": w[0], "w2": w[1], "b": b})
        errors_per_epoch.append(n_wrong)
        if n_wrong == 0:                           # ทายถูกทุกตัวในรอบนี้ → หยุด
            break
    return w, b, pd.DataFrame(records), errors_per_epoch


or_data = list(zip(map(tuple, INPUTS_2), [0, 1, 1, 1]))
w_or, b_or, table_or, errors_or = perceptron_train(or_data, eta=1.0)
display(table_or)
print("จำนวนที่ทายผิดในแต่ละรอบ:", errors_or, "(หนังสือ: [2, 2, 1, 0])")
print(f"ค่าน้ำหนักสุดท้าย w1 = {w_or[0]:g}, w2 = {w_or[1]:g}, b = {b_or:g} (หนังสือ: 1, 1, −1)")
```

ตารางทั้ง 16 แถวตรงกับตัวอย่างที่ 5.2 Perceptron หยุดในรอบที่ 4 ได้เส้นแบ่ง $x_1 + x_2 - 1 = 0$ ซึ่งผ่านจุด (0,1) และ (1,0) พอดี

### 1.3 Perceptron กับปัญหา XOR (ตัวอย่างที่ 5.3)

XOR ให้ผล 1 เมื่อสัญญาณเข้าทั้งสองต่างกัน เช่น สวิตช์สองทางของไฟบันได หรือการตรวจความไม่สอดคล้องระหว่างคำสั่งกับสถานะจริงของเบรกเกอร์ หนังสือพิสูจน์ว่าไม่มีค่า $w_1, w_2, b$ ใดทำ XOR ได้ เซลล์ถัดไปลองฝึกจริง 20 รอบ แล้วดูว่าค่าน้ำหนักวนซ้ำไปมาอย่างไร

```python
xor_data = list(zip(map(tuple, INPUTS_2), [0, 1, 1, 0]))
w_xor, b_xor, table_xor, errors_xor = perceptron_train(xor_data, eta=1.0, max_epochs=20)
print("จำนวนที่ทายผิดในแต่ละรอบ:", errors_xor)
end_of_epoch = table_xor.groupby("epoch")[["w1", "w2", "b"]].last()
print("ค่าน้ำหนักเมื่อจบแต่ละรอบ (5 รอบสุดท้าย):")
display(end_of_epoch.tail(5).T)
print("ทุกแถวของรอบที่ 20 (ค่าน้ำหนักถูกปรับไปมาภายในรอบ แล้วกลับมาที่เดิมเมื่อจบรอบ):")
display(table_xor[table_xor["epoch"] == 20])


def plot_boundary(ax, w, b, title, data):
    """วาดสี่กรณีและเส้นแบ่ง w1 x1 + w2 x2 + b = 0"""
    for (x1, x2), y in data:
        ax.scatter(x1, x2, s=160, marker="o", edgecolors=COLORS["blue"] if y == 0 else COLORS["orange"],
                   facecolors="white" if y == 0 else COLORS["orange"], linewidths=2, zorder=3)
    xs = np.linspace(-0.5, 1.5, 50)
    if abs(w[1]) > 1e-12:
        ax.plot(xs, -(w[0] * xs + b) / w[1], color=COLORS["green"])
    else:
        ax.axvline(-b / w[0], color=COLORS["green"])
    ax.set_xlim(-0.5, 1.5)
    ax.set_ylim(-0.5, 1.5)
    ax.set_title(title)
    ax.set_xlabel("x1")
    ax.set_ylabel("x2")


and_data_designed = list(zip(map(tuple, INPUTS_2), [0, 0, 0, 1]))
fig, axes = plt.subplots(1, 3, figsize=(12, 3.8))
plot_boundary(axes[0], [1, 1], -1.5, "AND (Example 5.1): x1 + x2 = 1.5", and_data_designed)
plot_boundary(axes[1], w_or, b_or, "OR learned (Example 5.2): x1 + x2 = 1", or_data)
plot_boundary(axes[2], w_xor, b_xor, f"XOR after 20 epochs: still {errors_xor[-1]} wrong", xor_data)
plt.tight_layout()
plt.show()
```

ไม่มีรอบใดที่ทายถูกครบสี่กรณี ค่าน้ำหนักถูกปรับทุกตัวอย่างภายในรอบ และเมื่อจบรอบก็วนกลับมาที่ค่าเดิม อัลกอริทึมจึงไม่มีวันหยุดเอง ตรงกับทฤษฎีบทการลู่เข้าที่รับประกันเฉพาะข้อมูลที่แยกได้เชิงเส้น

**คำถามชวนคิด** ถ้าเพิ่มคุณลักษณะที่สาม $x_3 = x_1 \cdot x_2$ (ลอจิก AND ของสัญญาณเข้า) Perceptron จะเรียนรู้ XOR ได้หรือไม่ ลองแก้ `xor_data` เป็นสามมิติแล้วรันดู ผลนี้เกี่ยวข้องกับแนวคิด "ชั้นซ่อนสร้างการแทนข้อมูลแบบใหม่" ในส่วนที่ 2 อย่างไร

### 2.1 โครงข่ายสองชั้นแก้ปัญหา XOR (ตัวอย่างที่ 5.4)

### 2.2 การส่งผ่านไปข้างหน้าและการนับพารามิเตอร์ (ตัวอย่างที่ 5.5)

ฟังก์ชัน `dense_forward` คำนวณ $\mathbf{z}^{(l)} = \mathbf{W}^{(l)}\mathbf{a}^{(l-1)} + \mathbf{b}^{(l)}$ และ $\mathbf{a}^{(l)} = f^{(l)}(\mathbf{z}^{(l)})$ ทีละชั้น ใช้กับโครงข่ายประมาณอุณหภูมิน้ำมันหม้อแปลงในตัวอย่างที่ 5.5 ซึ่งรับโหลด (p.u.) และอุณหภูมิแวดล้อม/40 °C ชั้นซ่อนใช้ ReLU และชั้นเอาต์พุตเป็นเชิงเส้น

```python
def identity(z):
    return z


def dense_forward(x, layers):
    """layers คือรายการ (W, b, activation) คืนค่า z และ a ของทุกชั้น"""
    a = np.asarray(x, dtype=float)
    zs, activations = [], [a]
    for W, b, activation in layers:
        z = W @ a + b
        a = activation(z)
        zs.append(z)
        activations.append(a)
    return zs, activations


oil_net = [(np.array([[1.5, 0.5], [-1.0, 0.8]]), np.array([-0.5, 0.1]), relu),
           (np.array([[40.0, 25.0]]), np.array([20.0]), identity)]

for case, (load_pu, ambient_c, book_value) in enumerate([(0.8, 30, 63), (1.2, 30, 87), (0.2, 30, 39.5)], start=1):
    zs, acts = dense_forward([load_pu, ambient_c / 40], oil_net)
    print(f"กรณีที่ {case}: z(1) = {np.round(zs[0], 3)}, a(1) = {np.round(acts[1], 3)}")
    compare(f"         ŷ (°C)", acts[-1][0], book_value, 1e-9)


def count_dense_params(sizes):
    """จำนวนพารามิเตอร์ของ MLP เช่น sizes = [4, 8, 3]: แต่ละชั้นมี (n_in + 1) × n_out"""
    return sum((n_in + 1) * n_out for n_in, n_out in zip(sizes, sizes[1:]))


for sizes, book_value in [([2, 2, 1], 9), ([4, 8, 3], 67), ([4, 8, 8, 3], 139), ([784, 128, 10], 101_770)]:
    print(f"โครงข่าย {'-'.join(map(str, sizes)):>10}: {count_dense_params(sizes):>7,} พารามิเตอร์ (หนังสือ: {book_value:,})")
```

**คำถามชวนคิด** ลองวาด $\hat{y}$ เทียบกับโหลด 0–1.5 p.u. ที่อากาศ 30 °C (ใช้ `dense_forward` ในลูป) กราฟเป็นเส้นตรงหลายช่วงต่อกันหรือไม่ จุดหักอยู่ที่โหลดเท่าใด และช่วงใดที่ความชันขัดกับฟิสิกส์ของหม้อแปลง (แบบฝึกหัดข้อ 5 ในหนังสือ)

### 2.3 ฟังก์ชันกระตุ้น (ตัวอย่างที่ 5.6–5.8 และตารางที่ 5.1)

เริ่มจากตรวจว่าโครงข่ายเชิงเส้นสองชั้นในตัวอย่างที่ 5.6 ยุบเป็นชั้นเดียวได้จริง แล้วสร้างตารางที่ 5.1 และกราฟของฟังก์ชันกระตุ้นกับอนุพันธ์ (รูปที่ 5.5)

```python
# ตัวอย่างที่ 5.6: ไม่มีฟังก์ชันกระตุ้น → W' = W2 W1, b' = W2 b1 + b2
W1, b1 = np.array([[2.0, -1.0], [1.0, 1.0]]), np.array([0.0, -1.0])
W2, b2 = np.array([[1.0, -2.0]]), np.array([0.5])
W_prime, b_prime = W2 @ W1, W2 @ b1 + b2
print("W' =", W_prime, " b' =", b_prime, "(หนังสือ: [0, −3] และ 2.5)")
rng = np.random.default_rng(RANDOM_SEED)
test_points = rng.normal(size=(5, 2))
two_layer = np.array([dense_forward(p, [(W1, b1, identity), (W2, b2, identity)])[1][-1][0] for p in test_points])
print("ผลต่างระหว่างแบบสองชั้นกับแบบยุบรวมที่จุดสุ่ม 5 จุด:", np.abs(two_layer - (test_points @ W_prime.T + b_prime).ravel()).max())

# ตารางที่ 5.1
z_values = np.array([-2, -0.5, 0, 0.5, 2])
table_51 = pd.DataFrame({
    "sigmoid": sigmoid(z_values), "sigmoid'": sigmoid(z_values) * (1 - sigmoid(z_values)),
    "tanh": np.tanh(z_values), "1 - tanh^2": 1 - np.tanh(z_values) ** 2,
    "ReLU": relu(z_values), "ReLU'": relu_grad(z_values), "Leaky ReLU (0.01)": np.maximum(0.01 * z_values, z_values),
}, index=pd.Index(z_values, name="z")).T
display(table_51.round(4))

# รูปที่ 5.5: ฟังก์ชันและอนุพันธ์ (Leaky ReLU ใช้ α = 0.1 เพื่อให้มองเห็นความชันฝั่งลบ)
z = np.linspace(-4, 4, 401)
functions = {
    "Step": (step(z), np.zeros_like(z)),
    "Sigmoid": (sigmoid(z), sigmoid(z) * (1 - sigmoid(z))),
    "Tanh": (np.tanh(z), 1 - np.tanh(z) ** 2),
    "ReLU": (relu(z), relu_grad(z)),
    "Leaky ReLU (0.1)": (np.maximum(0.1 * z, z), np.where(z > 0, 1, 0.1)),
}
fig, axes = plt.subplots(2, 5, figsize=(15, 5), sharex=True)
for col, (name, (f, df)) in enumerate(functions.items()):
    axes[0, col].plot(z, f, color=COLORS["blue"])
    axes[0, col].set_title(name)
    axes[1, col].plot(z, df, color=COLORS["orange"])
    axes[1, col].set_xlabel("z")
axes[1, 1].axhline(0.25, color=COLORS["gray"], linestyle="--")
axes[1, 1].text(-3.9, 0.27, "max 0.25", fontsize=8)
axes[0, 0].set_ylabel("f(z)")
axes[1, 0].set_ylabel("f'(z)")
plt.tight_layout()
plt.show()
```

### 3.1 ครอสเอนโทรปีและเกรเดียนต์ $\mathbf{p} - \mathbf{y}$ (ตัวอย่างที่ 5.9)

ความสูญเสียครอสเอนโทรปีคือ $L = -\ln p_c$ และเมื่อจับคู่กับซอฟต์แมกซ์ เกรเดียนต์เทียบกับ Logit คือ $\mathbf{p} - \mathbf{y}$ เซลล์ถัดไปตรวจทั้งสองข้อ โดยตรวจสูตรเกรเดียนต์ด้วย **การหาอนุพันธ์เชิงตัวเลข** $\big(L(z + \varepsilon) - L(z - \varepsilon)\big)/(2\varepsilon)$ ซึ่งจะใช้ซ้ำในส่วนที่ 3.3

```python
def cross_entropy(logits, true_class):
    return -np.log(softmax(logits)[true_class])


for c, name in enumerate(["normal", "bearing fault", "winding short"]):
    print(f"คลาสจริง = {name:13s}: L = {cross_entropy(logits, c):.3f}")

y_onehot = np.array([0, 1, 0])                       # คลาสจริงคือตลับลูกปืนชำรุด
grad_formula = softmax(logits) - y_onehot
eps = 1e-6
grad_numeric = np.array([(cross_entropy(logits + eps * np.eye(3)[k], 1) - cross_entropy(logits - eps * np.eye(3)[k], 1)) / (2 * eps)
                         for k in range(3)])
print("\n∂L/∂z ด้วยสูตร p − y      :", np.round(grad_formula, 4), "(หนังสือ: 0.659, −0.758, 0.099)")
print("∂L/∂z ด้วยผลต่างสืบเนื่อง  :", np.round(grad_numeric, 4))
```

### 3.2 ผลของอัตราการเรียนรู้ (ตัวอย่างที่ 5.10)

ความสูญเสีย $J(w) = (w - 3)^2$ มีเกรเดียนต์ $J'(w) = 2(w - 3)$ เริ่มจาก $w_0 = 0$ การอัปเดตแต่ละก้าวคูณระยะห่างจากคำตอบด้วย $r = 1 - 2\alpha$ เงื่อนไขลู่เข้าจึงเป็น $0 < \alpha < 1$

```python
def gradient_descent_1d(alpha, w0=0.0, n_steps=3):
    path = [w0]
    for _ in range(n_steps):
        path.append(path[-1] - alpha * 2 * (path[-1] - 3))
    return path


lr_table = pd.DataFrame({alpha: gradient_descent_1d(alpha) for alpha in [0.1, 0.5, 1.0, 1.1]},
                        index=["w0", "w1", "w2", "w3"]).T
lr_table.insert(0, "r = 1 - 2α", [1 - 2 * a for a in lr_table.index])
lr_table.index.name = "α"
display(lr_table.round(4))

fig, axes = plt.subplots(1, 4, figsize=(15, 3.3), sharey=True)
w_grid = np.linspace(-2, 9, 200)
for ax, alpha, label in zip(axes, [0.1, 0.5, 1.0, 1.1], ["slow convergence", "one step", "oscillates", "diverges"]):
    path = np.array(gradient_descent_1d(alpha))
    ax.plot(w_grid, (w_grid - 3) ** 2, color=COLORS["gray"])
    ax.plot(path, (path - 3) ** 2, "o-", color=COLORS["blue"], markerfacecolor=COLORS["orange"])
    for i, w in enumerate(path):
        ax.annotate(str(i), (w, (w - 3) ** 2), textcoords="offset points", xytext=(4, 4), fontsize=8)
    ax.axvline(3, color=COLORS["green"], linestyle=":")
    ax.set_title(f"alpha = {alpha}: {label}")
    ax.set_xlabel("w")
axes[0].set_ylabel("J(w)")
plt.tight_layout()
plt.show()
```

### 3.3 Backpropagation ด้วย NumPy ทุกขั้น (ตัวอย่างที่ 5.12)

โครงข่าย 2-2-1 วินิจฉัยมอเตอร์ ชั้นซ่อนใช้ ReLU ชั้นเอาต์พุตใช้ซิกมอยด์ และใช้ครอสเอนโทรปีแบบสองคลาส ข้อมูลฝึกคือ $\mathbf{x} = (1.0, 0.5)$ ที่ผิดปกติ ($y = 1$) และ $\alpha = 0.5$ พารามิเตอร์เก็บในพจนานุกรม (dict) เพื่อให้อ่านชื่อได้ตรงกับหนังสือ คือ `W` (ค่าน้ำหนักของชั้นซ่อน แถวละหนึ่งนิวรอน) `b` `v` (ค่าน้ำหนักของนิวรอนเอาต์พุต) และ `c`

ฟังก์ชัน `backward_pass` ใช้สมการสี่ข้อของ Backpropagation ในหัวข้อ 5.3.3: $\delta_o = \hat{y} - y$, $\boldsymbol{\delta}_h = \delta_o \mathbf{v} \odot \text{ReLU}'(\mathbf{z})$, เกรเดียนต์ของค่าน้ำหนัก = ความผิดพลาดปลายทาง × สัญญาณเข้า และเกรเดียนต์ของไบแอส = ความผิดพลาด

```python
params_512 = {"W": np.array([[0.5, -0.2], [0.2, 0.6]]), "b": np.array([0.1, -0.1]),
              "v": np.array([0.6, -0.5]), "c": 0.1}
x_512, y_512, alpha_512 = np.array([1.0, 0.5]), 1, 0.5


def forward_pass(params, x):
    """ส่งผ่านไปข้างหน้าและเก็บค่ากลางทางไว้ใช้ตอนย้อนกลับ"""
    z = params["W"] @ x + params["b"]
    a = relu(z)
    z_o = params["v"] @ a + params["c"]
    return {"x": x, "z": z, "a": a, "z_o": z_o, "y_hat": float(sigmoid(z_o))}


def bce_loss(y_hat, y):
    """ครอสเอนโทรปีแบบสองคลาส"""
    return -(y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))


def backward_pass(params, cache, y):
    """คืนค่า (เกรเดียนต์ของทุกพารามิเตอร์, δ_o, δ_h)"""
    delta_o = cache["y_hat"] - y                                   # สมการข้อ 1
    delta_h = delta_o * params["v"] * relu_grad(cache["z"])        # สมการข้อ 2
    grads = {"v": delta_o * cache["a"], "c": delta_o,              # สมการข้อ 3–4 ของชั้นเอาต์พุต
             "W": np.outer(delta_h, cache["x"]), "b": delta_h}     # สมการข้อ 3–4 ของชั้นซ่อน
    return grads, delta_o, delta_h


def update(params, grads, alpha):
    return {name: params[name] - alpha * grads[name] for name in params}


cache = forward_pass(params_512, x_512)
loss_before = bce_loss(cache["y_hat"], y_512)
print("ขั้นที่ 1 ส่งผ่านไปข้างหน้า")
print(f"  z = {cache['z']}, a = {cache['a']}, z_o = {cache['z_o']:.4f}")
compare("  ŷ", cache["y_hat"], 0.5498, 5e-5)
compare("  L", loss_before, 0.5981, 5e-5)

grads, delta_o, delta_h = backward_pass(params_512, cache, y_512)
print("ขั้นที่ 2–4 ค่าความผิดพลาด")
compare("  δ_o", delta_o, -0.4502, 5e-5)
compare("  δ_1", delta_h[0], -0.2701, 5e-5)
compare("  δ_2", delta_h[1], 0.2251, 5e-5)
```

```python
names = ["v1", "v2", "c", "w11", "w12", "b1", "w21", "w22", "b2"]


def flatten(p):
    """เรียงพารามิเตอร์ตามลำดับในตารางของหนังสือ"""
    return np.array([p["v"][0], p["v"][1], p["c"], p["W"][0, 0], p["W"][0, 1], p["b"][0],
                     p["W"][1, 0], p["W"][1, 1], p["b"][1]], dtype=float)


new_params = update(params_512, grads, alpha_512)
steps_table = pd.DataFrame({
    "old": flatten(params_512), "gradient": flatten(grads), "-alpha*gradient": -alpha_512 * flatten(grads),
    "new": flatten(new_params),
    "book new": [0.7125, -0.4100, 0.3251, 0.6350, -0.1325, 0.2350, 0.0875, 0.5437, -0.2125],
}, index=names)
display(steps_table.round(4))

cache_after = forward_pass(new_params, x_512)
print("ขั้นที่ 7 ส่งผ่านไปข้างหน้าอีกครั้งด้วยพารามิเตอร์ใหม่ (ค่าไม่ปัด)")
print(f"  z = {np.round(cache_after['z'], 4)}, z_o = {cache_after['z_o']:.4f}")
compare("  ŷ", cache_after["y_hat"], 0.6980, 5e-5)
compare("  L", bce_loss(cache_after["y_hat"], y_512), 0.3596, 5e-5)
```

**ตรวจเกรเดียนต์ด้วยการหาอนุพันธ์เชิงตัวเลข (Gradient Checking)** ขยับพารามิเตอร์ทีละตัว $\pm\varepsilon$ แล้วดูว่าความสูญเสียเปลี่ยนเท่าใด ถ้าโค้ด Backpropagation ถูก ค่าทั้งสองต้องต่างกันน้อยมาก (ราว $10^{-10}$) วิธีนี้ช้าเพราะต้องส่งผ่านไปข้างหน้าสองครั้งต่อพารามิเตอร์หนึ่งตัว จึงใช้ตรวจโค้ดกับโครงข่ายเล็ก ๆ เท่านั้น

```python
def numerical_gradient(loss_fn, params, eps=1e-5):
    """อนุพันธ์เชิงตัวเลขของ loss_fn(params) เทียบกับทุกพารามิเตอร์ใน dict"""
    numeric = {}
    for name, value in params.items():
        value = np.asarray(value, dtype=float)
        grad = np.zeros_like(value)
        for idx in np.ndindex(value.shape):
            plus, minus = value.copy(), value.copy()
            plus[idx] += eps
            minus[idx] -= eps
            grad[idx] = (loss_fn({**params, name: plus}) - loss_fn({**params, name: minus})) / (2 * eps)
        numeric[name] = grad if grad.shape else float(grad)
    return numeric


loss_512 = lambda p: bce_loss(forward_pass(p, x_512)["y_hat"], y_512)  # noqa: E731
numeric = numerical_gradient(loss_512, params_512)
diff = np.abs(flatten(numeric) - flatten(grads))
display(pd.DataFrame({"backprop": flatten(grads), "numerical": flatten(numeric), "abs diff": diff}, index=names))
print(f"ผลต่างสูงสุด = {diff.max():.1e}")
```

ผลทุกขั้นตรงกับตัวอย่างที่ 5.12 สังเกตว่าค่าน้ำหนักขาเข้าของ $h_1$ (ซึ่งส่งต่อด้วย $v_1 > 0$) ถูกเพิ่มทุกตัว ส่วนของ $h_2$ (ส่งต่อด้วย $v_2 < 0$) ถูกลดทุกตัว คือการแบ่งความรับผิดชอบ (Credit Assignment) และค่าน้ำหนักที่ต่อกับ $x_1 = 1.0$ ถูกปรับมากกว่าที่ต่อกับ $x_2 = 0.5$ สองเท่า (ส่วนที่ 6.1 ตรวจตัวอย่างเดียวกันซ้ำด้วยการหาอนุพันธ์อัตโนมัติของ TensorFlow)

### 3.4 ให้โครงข่ายเรียนรู้ XOR เองจากข้อมูล

ในตัวอย่างที่ 5.4 เราออกแบบค่าน้ำหนักเอง ส่วนนี้ให้โครงข่ายหาค่าน้ำหนักเองด้วย Backpropagation คลาส `NumpyMLP` ขยายโค้ดในส่วนที่ 3.3 ให้มีกี่ชั้นก็ได้ ชั้นซ่อนใช้ ReLU ชั้นเอาต์พุตใช้ซอฟต์แมกซ์กับครอสเอนโทรปี (สำหรับสองคลาสเทียบเท่าซิกมอยด์ ดูตัวอย่างที่ 5.8) ค่าเริ่มต้นสุ่มแบบ He และรองรับโมเมนตัม Dropout และการลดค่าน้ำหนัก ซึ่งจะใช้ในส่วนที่ 4

```python
class NumpyMLP:
    """MLP เชื่อมต่อเต็ม: ชั้นซ่อน ReLU, ชั้นเอาต์พุตซอฟต์แมกซ์, ฝึกด้วย SGD + โมเมนตัม"""

    def __init__(self, sizes, seed=RANDOM_SEED):
        rng = np.random.default_rng(seed)
        self.dropout_rng = np.random.default_rng(seed + 1)
        # He Initialization: ส่วนเบี่ยงเบนมาตรฐาน sqrt(2 / n_in)
        self.W = [rng.normal(0, np.sqrt(2 / n_in), (n_in, n_out)) for n_in, n_out in zip(sizes, sizes[1:])]
        self.b = [np.zeros(n_out) for n_out in sizes[1:]]
        self.velocity = [np.zeros_like(p) for p in self.W + self.b]

    def forward(self, X, dropout=0.0, training=False):
        activations, zs, masks = [X], [], []
        a = X
        for layer, (W, b) in enumerate(zip(self.W, self.b)):
            z = a @ W + b
            zs.append(z)
            if layer < len(self.W) - 1:                       # ชั้นซ่อน
                a = relu(z)
                mask = None
                if training and dropout > 0:                  # Inverted Dropout
                    mask = (self.dropout_rng.random(a.shape) >= dropout) / (1 - dropout)
                    a = a * mask
                masks.append(mask)
            else:                                             # ชั้นเอาต์พุต
                a = softmax(z)
            activations.append(a)
        return activations, zs, masks

    def predict_proba(self, X):
        return self.forward(X)[0][-1]

    def predict(self, X):
        return self.predict_proba(X).argmax(axis=1)

    def loss(self, X, y):
        p = self.predict_proba(X)
        return float(-np.log(p[np.arange(len(y)), y] + 1e-12).mean())

    def train_step(self, X, y, lr, momentum=0.0, dropout=0.0, l2=0.0):
        """Backpropagation หนึ่งกลุ่มข้อมูล แล้วอัปเดตพารามิเตอร์"""
        activations, zs, masks = self.forward(X, dropout, training=True)
        delta = activations[-1].copy()
        delta[np.arange(len(y)), y] -= 1                      # p − y
        delta /= len(y)                                       # เฉลี่ยทั้งกลุ่ม
        grads_W, grads_b = [None] * len(self.W), [None] * len(self.W)
        for layer in range(len(self.W) - 1, -1, -1):
            grads_W[layer] = activations[layer].T @ delta + 2 * l2 * self.W[layer]
            grads_b[layer] = delta.sum(axis=0)
            if layer > 0:                                     # ส่งความผิดพลาดย้อนไปชั้นก่อน (ใช้ W ก่อนอัปเดต)
                delta = (delta @ self.W[layer].T) * relu_grad(zs[layer - 1])
                if masks[layer - 1] is not None:
                    delta = delta * masks[layer - 1]
        for i, (param, grad) in enumerate(zip(self.W + self.b, grads_W + grads_b)):
            self.velocity[i] = momentum * self.velocity[i] + grad
            param -= lr * self.velocity[i]

    def fit(self, X, y, epochs, lr, batch_size=None, momentum=0.0, dropout=0.0, l2=0.0, X_val=None, y_val=None):
        """ฝึกหลายรอบ สุ่มสลับลำดับทุกรอบ คืนประวัติความสูญเสียบนชุดฝึกและชุดตรวจสอบ"""
        rng = np.random.default_rng(RANDOM_SEED)
        batch_size = batch_size or len(X)
        history = {"train": [], "val": []}
        for _ in range(epochs):
            order = rng.permutation(len(X))
            for start in range(0, len(X), batch_size):
                batch = order[start:start + batch_size]
                self.train_step(X[batch], y[batch], lr, momentum, dropout, l2)
            history["train"].append(self.loss(X, y))
            if X_val is not None:
                history["val"].append(self.loss(X_val, y_val))
        return history


X_xor = INPUTS_2.astype(float)
y_xor = np.array([0, 1, 1, 0])
xor_mlp = NumpyMLP([2, 16, 2])
xor_history = xor_mlp.fit(X_xor, y_xor, epochs=500, lr=0.1, momentum=0.9)
print("ความสูญเสียรอบแรก/สุดท้าย:", round(xor_history["train"][0], 4), "/", round(xor_history["train"][-1], 4))
print("ความน่าจะเป็นของคลาส 1:", np.round(xor_mlp.predict_proba(X_xor)[:, 1], 3), "→ ทำนาย", xor_mlp.predict(X_xor))

# scikit-learn ทำแบบเดียวกันได้ในไม่กี่บรรทัด
sk_xor = MLPClassifier(hidden_layer_sizes=(4,), activation="tanh", solver="lbfgs", random_state=RANDOM_SEED).fit(X_xor, y_xor)
print("MLPClassifier ของ scikit-learn ทำนาย:", sk_xor.predict(X_xor))

fig, axes = plt.subplots(1, 2, figsize=(10, 3.8))
axes[0].plot(xor_history["train"], color=COLORS["blue"])
axes[0].set_xlabel("epoch")
axes[0].set_ylabel("cross-entropy loss")
axes[0].set_title("NumpyMLP 2-16-2 learning XOR")
g1, g2 = np.meshgrid(np.linspace(-0.5, 1.5, 200), np.linspace(-0.5, 1.5, 200))
proba = xor_mlp.predict_proba(np.column_stack([g1.ravel(), g2.ravel()]))[:, 1].reshape(g1.shape)
axes[1].contourf(g1, g2, proba, levels=[0, 0.5, 1], colors=["#DCE6F0", "#F6DCC9"])
axes[1].contour(g1, g2, proba, levels=[0.5], colors=[COLORS["green"]])
for (x1, x2), y in xor_data:
    axes[1].scatter(x1, x2, s=160, linewidths=2, zorder=3, edgecolors=COLORS["orange"] if y else COLORS["blue"],
                    facecolors=COLORS["orange"] if y else "white")
axes[1].set_title("Learned decision regions (p = 0.5 boundary)")
plt.tight_layout()
plt.show()
```

โครงข่ายเรียนรู้บริเวณที่ไม่ใช่เส้นตรงเส้นเดียวได้เองจากข้อมูลสี่จุด แต่เส้นแบ่งที่ได้ไม่จำเป็นต้องเหมือนแถบระหว่างเส้นขนานสองเส้นในตัวอย่างที่ 5.4 เพราะคำตอบที่ถูกต้องมีได้หลายชุด

**ชุดข้อมูลจำลอง: วินิจฉัยมอเตอร์สามคลาส** ต่อจากนี้ใช้ข้อมูลจำลองมอเตอร์ 3,000 ตัวที่มีคุณลักษณะสี่ตัวตามรูปที่ 5.3 คือ RMS การสั่น (mm/s) อุณหภูมิตลับลูกปืน (°C) กระแส (A) และอัตราส่วนยอดต่อ RMS (Crest Factor) และมีสามคลาสคือ ปกติ (ร้อยละ 70) ตลับลูกปืนชำรุด (ร้อยละ 20) และขดลวดลัดวงจร (ร้อยละ 10) ค่าเฉลี่ยของแต่ละคลาสตั้งตามแนวโน้มทางกายภาพ (ตลับลูกปืนชำรุดสั่นแรงและมีแรงกระแทก ขดลวดลัดวงจรกินกระแสและร้อน) แต่เป็นค่าสมมติ แบ่งเป็นชุดฝึก/ตรวจสอบ/ทดสอบ 60/20/20 แบบแบ่งชั้น และปรับขนาดด้วยค่าจากชุดฝึกเท่านั้น

```python
MOTOR_CLASSES = ["normal", "bearing fault", "winding short"]
FEATURES = ["vibration_rms", "bearing_temp", "current", "crest_factor"]


def make_motor_dataset(n=3000, seed=RANDOM_SEED):
    """ข้อมูลจำลองมอเตอร์: (ค่าเฉลี่ย, ส่วนเบี่ยงเบนมาตรฐาน) ของแต่ละคุณลักษณะในแต่ละคลาส (ค่าสมมติ)"""
    rng = np.random.default_rng(seed)
    spec = {0: ([2.5, 60, 30, 3.0], [0.7, 6, 2.0, 0.35]),      # ปกติ
            1: ([4.5, 68, 31, 4.6], [1.2, 7, 2.2, 0.80]),      # ตลับลูกปืนชำรุด
            2: ([3.0, 76, 35, 3.2], [0.8, 7, 2.5, 0.40])}      # ขดลวดลัดวงจร
    labels = rng.choice(3, size=n, p=[0.7, 0.2, 0.1])
    X = np.array([rng.normal(*spec[c]) for c in labels])
    return pd.DataFrame(X, columns=FEATURES).assign(condition=labels)


motors = make_motor_dataset()
print(motors.groupby("condition")[FEATURES].mean().round(2).set_axis(MOTOR_CLASSES))
X_all, y_all = motors[FEATURES].to_numpy(), motors["condition"].to_numpy()
X_tmp, X_test_m, y_tmp, y_test_m = train_test_split(X_all, y_all, test_size=0.2, stratify=y_all, random_state=RANDOM_SEED)
X_train_m, X_val_m, y_train_m, y_val_m = train_test_split(X_tmp, y_tmp, test_size=0.25, stratify=y_tmp,
                                                          random_state=RANDOM_SEED)
scaler_m = StandardScaler().fit(X_train_m)                 # ใช้ค่าจากชุดฝึกเท่านั้น
X_train_s, X_val_s, X_test_s = (scaler_m.transform(a) for a in (X_train_m, X_val_m, X_test_m))
print("ขนาด ชุดฝึก/ตรวจสอบ/ทดสอบ:", len(X_train_s), len(X_val_s), len(X_test_s))
```

**เปรียบเทียบอัตราการเรียนรู้และตัวปรับค่าน้ำหนัก** ใช้ `MLPClassifier` ของ scikit-learn ขนาด 4-16-3 ฝึก 60 รอบด้วยขนาดกลุ่ม 32 แล้ววาดความสูญเสียบนชุดฝึกในแต่ละรอบ (`loss_curve_`) ตั้ง `n_iter_no_change` ให้เท่าจำนวนรอบเพื่อไม่ให้หยุดก่อน

```python
def fit_mlp(**kwargs):
    settings = dict(hidden_layer_sizes=(16,), batch_size=32, max_iter=60, n_iter_no_change=60, tol=0,
                    random_state=RANDOM_SEED)
    settings.update(kwargs)
    return MLPClassifier(**settings).fit(X_train_s, y_train_m)


optimizer_runs = {
    "SGD alpha=0.001 (too small)": fit_mlp(solver="sgd", learning_rate_init=0.001, momentum=0.0),
    "SGD alpha=0.05": fit_mlp(solver="sgd", learning_rate_init=0.05, momentum=0.0),
    "SGD alpha=0.05 + momentum 0.9": fit_mlp(solver="sgd", learning_rate_init=0.05, momentum=0.9, nesterovs_momentum=False),
    "SGD alpha=1.0 + momentum 0.9 (too large)": fit_mlp(solver="sgd", learning_rate_init=1.0, momentum=0.9,
                                                       nesterovs_momentum=False),
    "Adam alpha=0.001": fit_mlp(solver="adam", learning_rate_init=0.001),
}
batch_runs = {f"Adam, B = {B}": fit_mlp(solver="adam", learning_rate_init=0.001, batch_size=B) for B in [8, 64, 512]}

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
for (name, model), color in zip(optimizer_runs.items(), ["gray", "blue", "green", "orange", "blue"]):
    axes[0].plot(model.loss_curve_, label=name, color=COLORS[color], linestyle="--" if "Adam" in name else "-")
axes[0].set_ylim(0, 1.2)
axes[0].set_title("Optimizers and learning rates (B = 32)")
for name, model in batch_runs.items():
    axes[1].plot(model.loss_curve_, label=name)
axes[1].set_ylim(0, 1.2)
axes[1].set_title("Batch size (Adam, alpha = 0.001)")
for ax in axes:
    ax.set_xlabel("epoch")
    ax.set_ylabel("training loss")
    ax.legend(fontsize=8)
plt.tight_layout()
plt.show()

summary = pd.DataFrame({name: {"final train loss": m.loss_curve_[-1], "val accuracy": m.score(X_val_s, y_val_m),
                               "updates": math.ceil(len(X_train_s) / m.batch_size) * len(m.loss_curve_)}
                        for name, m in {**optimizer_runs, **batch_runs}.items()}).T
display(summary.round(3))
```

อ่านกราฟตามอาการในหัวข้อ 5.4.1: $\alpha$ เล็กเกินไปทำให้ความสูญเสียลดช้ามาก $\alpha$ ใหญ่เกินไปทำให้ความสูญเสียแกว่งอยู่ที่ระดับสูง โมเมนตัมเร่งการลดลงอย่างชัดเจนเมื่อเทียบกับ SGD ที่ $\alpha$ เท่ากัน ส่วน Adam ใช้ค่าตั้งต้นได้ดีโดยแทบไม่ต้องปรับ ในกราฟขวา กลุ่มเล็กได้ปรับพารามิเตอร์บ่อยกว่ามากต่อรอบการฝึก จึงลดความสูญเสียได้เร็วกว่าเมื่อนับเป็นรอบการฝึก (คอลัมน์ `updates` คือจำนวนการอัปเดตทั้งหมด)

**คำถามชวนคิด** ถ้าวาดแกนนอนเป็น "จำนวนการอัปเดต" แทน "รอบการฝึก" กราฟขวาจะเปลี่ยนไปอย่างไร และในงานจริงที่ใช้ GPU ทำไมเราไม่เลือกขนาดกลุ่ม 1 เสมอ

### 4.2 Overfitting การหยุดฝึกก่อนกำหนด และ Dropout (ตัวอย่างที่ 5.15–5.16)

เริ่มจากเขียนการหยุดฝึกก่อนกำหนดตามกติกาในหนังสือ แล้วตรวจกับค่าความสูญเสียในตัวอย่างที่ 5.15

```python
def early_stopping(val_losses, patience):
    """คืนค่า (รอบที่หยุด, รอบที่ดีที่สุด, ตัวนับในแต่ละรอบ)"""
    best, best_epoch, counter, counters = math.inf, 0, 0, []
    for epoch, loss in enumerate(val_losses, start=1):
        if loss < best:
            best, best_epoch, counter = loss, epoch, 0     # ดีขึ้น → บันทึกพารามิเตอร์ของรอบนี้
        else:
            counter += 1
        counters.append(counter)
        if counter >= patience:
            break
    return epoch, best_epoch, counters


val_515 = [0.90, 0.62, 0.48, 0.41, 0.37, 0.35, 0.36, 0.34, 0.36, 0.38, 0.41, 0.45]
for patience in [1, 3]:
    stop, best_epoch, counters = early_stopping(val_515, patience)
    print(f"ความอดทน {patience}: หยุดหลังรอบ {stop} คืนค่าพารามิเตอร์ของรอบ {best_epoch} "
          f"(ความสูญเสีย {val_515[best_epoch - 1]}) ตัวนับ = {counters}")

# ตัวอย่างที่ 5.16 ก: Inverted Dropout p = 0.5
a_hidden = np.array([0.5, 1.2, 0.8, 0.3])
for mask in [(1, 0, 1, 0), (0, 1, 1, 1)]:
    out = a_hidden * np.array(mask) / (1 - 0.5)
    print(f"หน้ากาก {mask}: {out} ผลรวม {out.sum():.1f}")
rng = np.random.default_rng(RANDOM_SEED)
masks = rng.random((100_000, 4)) >= 0.5
print("ค่าเฉลี่ยจากการสุ่มหน้ากาก 100,000 ครั้ง:", np.round((a_hidden * masks / 0.5).mean(axis=0), 3), "≈ ค่าเดิม", a_hidden)

# ตัวอย่างที่ 5.16 ข: การลดค่าน้ำหนัก α = 0.01, λ = 0.05
print(f"ตัวคูณต่อก้าว = {1 - 2 * 0.01 * 0.05}, หลัง 1,000 ก้าวเหลือ {(1 - 2 * 0.01 * 0.05) ** 1000:.3f} เท่า (หนังสือ ≈ 0.368)")
```

**ทดลองให้เห็น Overfitting จริง** ใช้ข้อมูลจำลองสองคุณลักษณะที่ปรับขนาดแล้ว (เช่น การสั่นกับอุณหภูมิ) ซึ่งขอบเขตระหว่างมอเตอร์ปกติกับผิดปกติเป็นเส้นโค้งและมีสัญญาณรบกวนมาก (`make_moons` ของ scikit-learn) ชุดฝึกมีเพียง 50 ตัว แต่โครงข่าย 2-256-256-2 มีพารามิเตอร์ราว 67,000 ค่า จึง "จำ" ข้อมูลฝึกได้ทั้งหมด เปรียบเทียบการฝึก 1,500 รอบแบบไม่ใส่และใส่ Dropout $p = 0.5$ ในชั้นซ่อน

```python
X_moons, y_moons = make_moons(1100, noise=0.35, random_state=RANDOM_SEED)
X_small, y_small = X_moons[:50], y_moons[:50]          # ชุดฝึกเล็ก
X_check, y_check = X_moons[100:], y_moons[100:]        # ชุดตรวจสอบ 1,000 ตัว
print("จำนวนพารามิเตอร์ของ 2-256-256-2:", count_dense_params([2, 256, 256, 2]))

overfit_runs = {}
for label, p_drop in [("no dropout", 0.0), ("dropout p = 0.5", 0.5)]:
    model = NumpyMLP([2, 256, 256, 2])
    history = model.fit(X_small, y_small, epochs=1500, lr=0.02, momentum=0.9, dropout=p_drop,
                        X_val=X_check, y_val=y_check)
    stop, best_epoch, _ = early_stopping(history["val"], patience=100)
    overfit_runs[label] = (model, history, best_epoch)
    print(f"{label:16s}: ความสูญเสียสุดท้าย ฝึก {history['train'][-1]:.3f} ตรวจสอบ {history['val'][-1]:.3f} | "
          f"ดีที่สุดที่รอบ {best_epoch} ({min(history['val']):.3f}) | ความถูกต้องสุดท้ายบนชุดตรวจสอบ "
          f"{(model.predict(X_check) == y_check).mean():.3f}")

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, (label, (model, history, best_epoch)) in zip(axes[:2], overfit_runs.items()):
    ax.plot(history["train"], color=COLORS["blue"], label="train")
    ax.plot(history["val"], color=COLORS["orange"], label="validation")
    ax.axvline(best_epoch - 1, color=COLORS["green"], linestyle="--", label=f"best epoch {best_epoch}")
    ax.set_ylim(0, 1.5)
    ax.set_xscale("symlog", linthresh=10)          # ขยายช่วงต้นให้เห็นจุดต่ำสุด
    ax.set_title(label)
    ax.set_xlabel("epoch")
    ax.set_ylabel("loss")
    ax.legend()
g1, g2 = np.meshgrid(np.linspace(-2, 3, 200), np.linspace(-1.5, 2, 200))
grid = np.column_stack([g1.ravel(), g2.ravel()])
for (label, (model, _, _)), color in zip(overfit_runs.items(), ["gray", "green"]):
    axes[2].contour(g1, g2, model.predict_proba(grid)[:, 1].reshape(g1.shape), levels=[0.5], colors=[COLORS[color]])
axes[2].scatter(*X_small[y_small == 0].T, facecolors="white", edgecolors=COLORS["blue"], label="normal (train)")
axes[2].scatter(*X_small[y_small == 1].T, color=COLORS["orange"], label="abnormal (train)")
axes[2].set_title("Boundaries: gray = no dropout, green = dropout")
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()
```

โครงข่ายที่ไม่ใส่ Dropout ทำความสูญเสียบนชุดฝึกได้เกือบศูนย์ แต่ความสูญเสียบนชุดตรวจสอบต่ำสุดตั้งแต่ไม่กี่รอบแรก (ข้อมูลง่ายพอที่เส้นแบ่งหยาบ ๆ ทำได้ดีแล้ว แกนนอนจึงใช้สเกลลอการิทึมในช่วงหลัง) แล้วเพิ่มขึ้นต่อเนื่อง ซึ่งเป็นอาการ Overfitting ที่ชัดเจน (รูปที่ 5.10) เส้นแบ่งของมันคดเคี้ยวไปตามจุดฝึกแต่ละจุด ส่วนโครงข่ายที่ใส่ Dropout มีความสูญเสียบนชุดฝึกสูงกว่า (เพราะตอนฝึกขาดนิวรอนบางส่วน) แต่ความสูญเสียบนชุดตรวจสอบตอนจบต่ำกว่ามาก การหยุดฝึกก่อนกำหนดที่รอบที่ดีที่สุดก็ให้ผลใกล้เคียงกันโดยใช้เวลาฝึกน้อยกว่ามาก

**คำถามชวนคิด** ถ้าเพิ่มขนาดชุดฝึกจาก 50 เป็น 500 ตัว (`X_moons[:500]`) ความต่างระหว่างสองกรณีจะเพิ่มขึ้นหรือลดลง เพราะเหตุใด (หัวข้อ 5.4.2 "ข้อมูลมากขึ้นคือ Regularization ที่ดีที่สุด")

### 4.3 การระบุเอกลักษณ์ระบบมอเตอร์ขับพัดลมด้วย NARX (ตัวอย่างที่ 5.17–5.18)

ระบบจริง (ที่ผู้สร้างแบบจำลองไม่รู้) คือ $\omega[k+1] = 0.8\,\omega[k] + 2\,u[k] - 0.05\,\omega[k]^2$ เริ่มจากตรวจตารางในตัวอย่างที่ 5.17–5.18

```python
def fan_system(omega, u):
    """ระบบจริงในตัวอย่างที่ 5.17"""
    return 0.8 * omega + 2 * u - 0.05 * omega ** 2


def fan_model_518(omega, u):
    """แบบจำลองที่ประมาณแรงบิดของพัดลมต่ำไปเล็กน้อย (ตัวอย่างที่ 5.18)"""
    return 0.8 * omega + 2 * u - 0.04 * omega ** 2


def simulate(f, omega0, inputs):
    """จำลองแบบอิสระ: ป้อนค่าที่คำนวณได้กลับเป็นสัญญาณเข้าของก้าวถัดไป"""
    out = [omega0]
    for u in inputs:
        out.append(f(out[-1], u))
    return np.array(out)


true_step = simulate(fan_system, 0.0, np.ones(4))
one_step = np.array([fan_model_518(true_step[k - 1], 1) for k in range(1, 5)])
free_run = simulate(fan_model_518, 0.0, np.ones(4))
display(pd.DataFrame({"true omega[k]": true_step[1:], "one-step": one_step, "one-step error": one_step - true_step[1:],
                      "free-run": free_run[1:], "free-run error": free_run[1:] - true_step[1:]},
                     index=pd.Index(range(1, 5), name="k")).round(4))

steady = {u: simulate(fan_system, 0.0, np.full(300, u))[-1] for u in (1.0, 0.5)}
print(f"สภาวะคงตัวของระบบจริง: u = 1 → {steady[1.0]:.3f}, u = 0.5 → {steady[0.5]:.3f} (หนังสือ 4.633, 2.899)")
print(f"แบบจำลองเชิงเส้นจากจุด u = 1 ทำนาย u = 0.5 ได้ {steady[1.0] * 0.5:.3f} ผิด "
      f"{(steady[0.5] - steady[1.0] * 0.5) / steady[0.5] * 100:.1f}%")
print(f"สภาวะคงตัวของแบบจำลองในตัวอย่างที่ 5.18 (u = 1): {simulate(fan_model_518, 0.0, np.full(300, 1.0))[-1]:.3f}")
```

**ระบุเอกลักษณ์จากข้อมูล** ทำตามขั้นตอนในหัวข้อ 5.4.3

1. ออกแบบสัญญาณกระตุ้นแบบขั้นบันไดหลายระดับที่สุ่มทั้งความสูง (0–1.2) และช่วงเวลาค้าง (5–40 ก้าว) คล้าย APRBS ยาว 3,000 ก้าว
2. จำลองระบบจริงและเพิ่มสัญญาณรบกวนการวัดเล็กน้อย
3. สร้างชุดข้อมูลแบบหน่วงเวลา $(\omega[k], u[k]) \rightarrow \omega[k+1]$ ($n_y = n_u = 1$)
4. แบ่งตามเวลา 2,000 ก้าวแรกเป็นชุดฝึก ที่เหลือเป็นชุดทดสอบ
5. ฝึก MLP (`MLPRegressor` ขนาด 2-32-32-1 ใช้ Adam และหยุดฝึกก่อนกำหนด) เทียบกับแบบจำลองเชิงเส้น ARX (`LinearRegression`)
6. ตรวจด้วยทั้งการทำนายหนึ่งก้าวและการจำลองแบบอิสระบนชุดทดสอบ

```python
def multilevel_excitation(n_steps, rng, low=0.0, high=1.2, min_hold=5, max_hold=40):
    """สัญญาณขั้นบันไดหลายระดับที่สุ่มความสูงและช่วงเวลาค้าง (แนวคิดเดียวกับ APRBS)"""
    u = np.empty(n_steps)
    k = 0
    while k < n_steps:
        hold = rng.integers(min_hold, max_hold + 1)
        u[k:k + hold] = rng.uniform(low, high)
        k += hold
    return u


rng = np.random.default_rng(RANDOM_SEED)
N_STEPS, N_TRAIN = 3000, 2000
u_seq = multilevel_excitation(N_STEPS, rng)
omega_true = simulate(fan_system, 0.0, u_seq)                        # ยาว N_STEPS + 1
omega_meas = omega_true + rng.normal(0, 0.02, size=omega_true.shape)  # ค่าวัดที่มีสัญญาณรบกวน

X_narx = np.column_stack([omega_meas[:-1], u_seq])                    # (ω[k], u[k])
y_narx = omega_meas[1:]                                               # ω[k+1]
narx_mlp = make_pipeline(StandardScaler(), MLPRegressor(hidden_layer_sizes=(32, 32), learning_rate_init=0.003,
                                                        max_iter=2000, early_stopping=True, validation_fraction=0.2,
                                                        n_iter_no_change=30, random_state=RANDOM_SEED))
narx_mlp.fit(X_narx[:N_TRAIN], y_narx[:N_TRAIN])
arx_linear = LinearRegression().fit(X_narx[:N_TRAIN], y_narx[:N_TRAIN])
print(f"ARX เชิงเส้น: ω[k+1] = {arx_linear.coef_[0]:.3f} ω[k] + {arx_linear.coef_[1]:.3f} u[k] + {arx_linear.intercept_:.3f}")


def free_run_model(model, omega0, inputs):
    out = [omega0]
    for u in inputs:
        out.append(model.predict(np.array([[out[-1], u]]))[0])
    return np.array(out[1:])


rows = {}
u_test, omega_test = u_seq[N_TRAIN:], omega_true[N_TRAIN + 1:]
for name, model in [("MLP (NARX)", narx_mlp), ("linear ARX", arx_linear)]:
    one_step_pred = model.predict(X_narx[N_TRAIN:])
    free_pred = free_run_model(model, omega_meas[N_TRAIN], u_test)
    rows[name] = {"one-step RMSE": np.sqrt(np.mean((one_step_pred - omega_test) ** 2)),
                  "free-run RMSE": np.sqrt(np.mean((free_pred - omega_test) ** 2)),
                  "steady state u=1": free_run_model(model, 0.0, np.full(80, 1.0))[-1],
                  "steady state u=0.5": free_run_model(model, 0.0, np.full(80, 0.5))[-1]}
    rows[name]["_free"] = free_pred
rows["true system"] = {"one-step RMSE": 0.0, "free-run RMSE": 0.0, "steady state u=1": steady[1.0],
                       "steady state u=0.5": steady[0.5]}
display(pd.DataFrame(rows).T.drop(columns="_free").astype(float).round(3))

fig, axes = plt.subplots(2, 1, figsize=(12, 5), sharex=True)
window = slice(0, 300)
k_axis = np.arange(N_TRAIN, N_STEPS)[window]
axes[0].plot(k_axis, u_test[window], color=COLORS["gray"], drawstyle="steps-post")
axes[0].set_ylabel("input u")
axes[1].plot(k_axis, omega_test[window], color=COLORS["blue"], label="true system")
axes[1].plot(k_axis, rows["MLP (NARX)"]["_free"][window], color=COLORS["orange"], linestyle="--", label="MLP free-run")
axes[1].plot(k_axis, rows["linear ARX"]["_free"][window], color=COLORS["green"], linestyle=":", label="linear ARX free-run")
axes[1].set_ylabel("speed omega")
axes[1].set_xlabel("k (test segment)")
axes[1].legend()
plt.tight_layout()
plt.show()
```

แบบจำลอง MLP ให้ความเร็วคงตัวใกล้ระบบจริงทั้งที่ $u = 1$ และ $u = 0.5$ เพราะเรียนรู้ได้ว่าอัตราขยายเปลี่ยนตามจุดทำงาน ส่วนแบบจำลองเชิงเส้นซึ่งฝึกจากข้อมูลช่วงเดียวกันต้องเฉลี่ยอัตราขยายทุกจุดทำงานเป็นค่าเดียว จึงผิดที่ปลายทั้งสองข้าง และความคลาดเคลื่อนของการจำลองแบบอิสระสูงกว่าการทำนายหนึ่งก้าวเสมอ เพราะความผิดพลาดถูกป้อนกลับและสะสม (ตัวอย่างที่ 5.18)

**คำถามชวนคิด** ถ้าสัญญาณกระตุ้นในชุดฝึกอยู่ในช่วง 0–0.6 เท่านั้น แล้วนำแบบจำลอง MLP ไปจำลองที่ $u = 1$ คิดว่าผลจะเป็นอย่างไร ลองแก้ `multilevel_excitation(..., high=0.6)` แล้วรันดู ผลนี้เกี่ยวข้องกับข้อควรระวัง "โครงข่ายประมาณค่านอกช่วงได้ไม่ดี" อย่างไร

## ส่วนที่ 5 คอนโวลูชัน การรวมกลุ่ม และ CNN ขนาดเล็ก

### 5.1 คอนโวลูชันตรวจจับขอบของจุดร้อน (ตัวอย่างที่ 5.19)

ไลบรารีการเรียนรู้เชิงลึกใช้สูตรสหสัมพันธ์ไขว้ $F(i, j) = \sum_m \sum_n I(i + m, j + n)\,K(m, n) + b$ (ไม่กลับด้านเคอร์เนล) ฟังก์ชัน `conv2d_simple` ด้านล่างเลื่อนเคอร์เนลทีละตำแหน่งตามสูตรนี้ (ระยะก้าว 1 ไม่เติมขอบ) แบบฝึกปฏิบัติข้อ 3 จะขยายให้รองรับระยะก้าวและการเติมขอบ

```python
def conv2d_simple(image, kernel, bias=0.0):
    """สหสัมพันธ์ไขว้แบบระยะก้าว 1 ไม่เติมขอบ"""
    image, kernel = np.asarray(image, dtype=float), np.asarray(kernel, dtype=float)
    k = kernel.shape[0]
    n_out, m_out = image.shape[0] - k + 1, image.shape[1] - k + 1
    feature_map = np.zeros((n_out, m_out))
    for i in range(n_out):
        for j in range(m_out):
            feature_map[i, j] = np.sum(image[i:i + k, j:j + k] * kernel) + bias
    return feature_map


thermal_patch = np.array([[1, 1, 2, 8, 9], [1, 2, 2, 9, 9], [1, 1, 3, 8, 8], [2, 1, 2, 9, 9], [1, 1, 2, 8, 9]])
K_vertical = np.array([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]])
K_horizontal = np.array([[-1, -1, -1], [0, 0, 0], [1, 1, 1]])

S = conv2d_simple(thermal_patch, K_vertical)
print("แผนที่คุณลักษณะ S (หนังสือ: [[4, 21, 19], [3, 22, 19], [3, 22, 19]]):\n", S)
print("ReLU(F − 10):\n", relu(conv2d_simple(thermal_patch, K_vertical, bias=-10)))
print("เคอร์เนลขอบแนวนอน:\n", conv2d_simple(thermal_patch, K_horizontal), "\n(ตำแหน่งแถว 1–3 คอลัมน์ 2–4 ได้ 1 ตามหนังสือ)")

fig, axes = plt.subplots(1, 3, figsize=(10, 3.2))
for ax, data, title in zip(axes, [thermal_patch, K_vertical, S], ["image I (5x5)", "kernel K (vertical edge)", "feature map S"]):
    ax.imshow(data, cmap="inferno" if data is not K_vertical else "coolwarm")
    for (i, j), value in np.ndenumerate(data):
        ax.text(j, i, f"{value:g}", ha="center", va="center", color="white" if data is not K_vertical else "black")
    ax.set_title(title)
    ax.set_xticks([])
    ax.set_yticks([])
plt.tight_layout()
plt.show()
```

### 5.2 ขนาดผลลัพธ์ การรวมกลุ่ม และจำนวนพารามิเตอร์ (ตัวอย่างที่ 5.20–5.23)

ขนาดผลลัพธ์ของคอนโวลูชันคือ $N_{\text{out}} = \lfloor (N - k + 2P)/S \rfloor + 1$ และจำนวนพารามิเตอร์ของชั้นคอนโวลูชันคือ $(k \times k \times C_{\text{in}} + 1) \times C_{\text{out}}$ ซึ่งไม่ขึ้นกับขนาดภาพ

```python
def conv_output_size(n, k, stride=1, padding=0):
    return (n - k + 2 * padding) // stride + 1


def count_conv_params(k, c_in, c_out, bias=True):
    return (k * k * c_in + int(bias)) * c_out


def pool2d(feature_map, size=2, stride=2, mode="max"):
    """การรวมกลุ่มแบบค่าสูงสุดหรือค่าเฉลี่ย"""
    n = conv_output_size(feature_map.shape[0], size, stride)
    out = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            region = feature_map[i * stride:i * stride + size, j * stride:j * stride + size]
            out[i, j] = region.max() if mode == "max" else region.mean()
    return out


# ตัวอย่างที่ 5.20
F = np.array([[1, 3, 2, 0], [4, 6, 1, 2], [0, 2, 5, 7], [1, 1, 3, 4]], dtype=float)
print("Max Pooling:\n", pool2d(F), "\nAverage Pooling:\n", pool2d(F, mode="avg"))
F_shifted = F.copy()
F_shifted[0, 0], F_shifted[1, 1] = 6, 1                 # เลื่อนค่า 6 ภายในบริเวณซ้ายบน
print("เลื่อน 6 ภายในบริเวณ ผล Max Pooling เท่าเดิม:", np.array_equal(pool2d(F), pool2d(F_shifted)))

# ตัวอย่างที่ 5.21: CNN ขนาดเล็กสำหรับภาพ 28 × 28 × 1, 10 คลาส
n1 = conv_output_size(28, 3)
p1 = conv_output_size(n1, 2, 2)
n2 = conv_output_size(p1, 3)
p2 = conv_output_size(n2, 2, 2)
cnn_table = pd.DataFrame([
    ("input", "28 x 28 x 1", 0),
    ("conv 8 @ 3x3 + ReLU", f"{n1} x {n1} x 8", count_conv_params(3, 1, 8)),
    ("max pool 2x2", f"{p1} x {p1} x 8", 0),
    ("conv 16 @ 3x3 + ReLU", f"{n2} x {n2} x 16", count_conv_params(3, 8, 16)),
    ("max pool 2x2", f"{p2} x {p2} x 16", 0),
    ("flatten", f"{p2 * p2 * 16}", 0),
    ("dense 10 + softmax", "10", count_dense_params([p2 * p2 * 16, 10])),
], columns=["layer", "output shape", "parameters"])
display(cnn_table)
print(f"รวม CNN = {cnn_table['parameters'].sum():,} (หนังสือ 5,258) | MLP 784-128-10 = {count_dense_params([784, 128, 10]):,}"
      f" | อัตราส่วน ≈ {count_dense_params([784, 128, 10]) / cnn_table['parameters'].sum():.1f} เท่า")

# ตัวอย่างที่ 5.22–5.23
print(f"\nAlexNet conv1: ภาพ 227 → {conv_output_size(227, 11, 4)}, พารามิเตอร์ {count_conv_params(11, 3, 96):,}")
print(f"ResNet conv1: ภาพ 224 → {conv_output_size(224, 7, 2, 3)} → max pool → {conv_output_size(112, 3, 2, 1)}, "
      f"ค่าน้ำหนัก {count_conv_params(7, 3, 64, bias=False):,}")
print(f"ResNet-50 transfer: ชั้นจำแนกเดิม {count_dense_params([2048, 1000]):,} → ชั้นใหม่ 3 คลาส {count_dense_params([2048, 3]):,}")
```

### 5.3 ฝึก CNN ขนาดเล็กด้วย NumPy บนภาพตัวเลขขนาด 8 × 8

เพื่อให้รันบน CPU ได้เร็วโดยไม่ต้องดาวน์โหลดข้อมูล ส่วนนี้ใช้ชุดข้อมูล **digits** ที่มากับ scikit-learn (ภาพตัวเลขเขียนมือ 0–9 ขนาด 8 × 8 จุดภาพ 1,797 ภาพ ซึ่งเป็นส่วนหนึ่งของชุดข้อมูล Optical Recognition of Handwritten Digits ของ UCI Machine Learning Repository) ภาพแบบเดียวกับตัวเลขบนป้ายหน้าปัดมิเตอร์ที่ตัดเป็นหลัก ๆ ในตัวอย่างที่ 5.21 แต่เล็กกว่า

โครงสร้างคือ คอนโวลูชัน 8 ตัวกรอง 3 × 3 + ReLU → Max Pooling 2 × 2 → Flatten → เชื่อมต่อเต็ม 10 นิวรอน + ซอฟต์แมกซ์ คลาส `TinyCNN` เขียนการส่งผ่านไปข้างหน้าและย้อนกลับเองทั้งหมด โดยใช้เทคนิค **im2col** ที่ดึงทุกบริเวณ 3 × 3 ของภาพมาเรียงเป็นแถว ทำให้คอนโวลูชันกลายเป็นการคูณเมทริกซ์ครั้งเดียว (วิธีเดียวกับที่ไลบรารีจริงใช้) ผู้อ่านไม่จำเป็นต้องเข้าใจทุกบรรทัดของ `backward` แต่ควรเห็นว่ามันคือกฎลูกโซ่ชุดเดียวกับส่วนที่ 3.3 ไล่ย้อนผ่าน ชั้นเชื่อมต่อเต็ม → Max Pooling (ส่งเกรเดียนต์ไปเฉพาะตำแหน่งที่เป็นค่าสูงสุด) → ReLU → คอนโวลูชัน (เกรเดียนต์ของเคอร์เนล = ความผิดพลาด × บริเวณของภาพที่ทาบ)

```python
digits = load_digits()
X_digits = digits.images / 16.0                              # ปรับค่าให้อยู่ในช่วง 0–1
X_dtrain, X_dtest, y_dtrain, y_dtest = train_test_split(X_digits, digits.target, test_size=0.25,
                                                        stratify=digits.target, random_state=RANDOM_SEED)


def im2col(images, k):
    """ดึงทุกบริเวณ k × k ของภาพ (n, h, w) ออกมาเป็น (n, h-k+1, w-k+1, k*k)"""
    n, h, w = images.shape
    out = h - k + 1
    cols = np.empty((n, out, out, k * k))
    for i in range(k):
        for j in range(k):
            cols[:, :, :, i * k + j] = images[:, i:i + out, j:j + out]
    return cols


class TinyCNN:
    """conv(n_filters @ 3x3) + ReLU → max pool 2x2 → dense(10) + softmax"""

    def __init__(self, n_filters=8, k=3, image_size=8, n_classes=10, seed=RANDOM_SEED):
        rng = np.random.default_rng(seed)
        self.k, self.pool = k, (image_size - k + 1) // 2
        self.K = rng.normal(0, np.sqrt(2 / (k * k)), (k * k, n_filters))   # เคอร์เนลแต่ละตัวเป็นหนึ่งคอลัมน์
        self.bk = np.zeros(n_filters)
        n_flat = self.pool * self.pool * n_filters
        self.W = rng.normal(0, np.sqrt(2 / n_flat), (n_flat, n_classes))
        self.b = np.zeros(n_classes)
        self.velocity = [np.zeros_like(p) for p in (self.K, self.bk, self.W, self.b)]

    def n_params(self):
        return sum(p.size for p in (self.K, self.bk, self.W, self.b))

    def forward(self, images):
        cols = im2col(images, self.k)
        z = cols @ self.K + self.bk                          # คอนโวลูชันด้วยการคูณเมทริกซ์
        a = relu(z)
        n, _, _, f = a.shape
        p = self.pool
        blocks = a[:, :2 * p, :2 * p, :].reshape(n, p, 2, p, 2, f)
        pooled = blocks.max(axis=(2, 4))                     # Max Pooling 2 × 2
        flat = pooled.reshape(n, -1)
        probs = softmax(flat @ self.W + self.b)
        return probs, (cols, z, a, blocks, pooled, flat)

    def predict(self, images):
        return self.forward(images)[0].argmax(axis=1)

    def backward(self, cache, probs, labels):
        cols, z, a, blocks, pooled, flat = cache
        n = len(labels)
        delta = probs.copy()
        delta[np.arange(n), labels] -= 1
        delta /= n                                           # p − y เฉลี่ยทั้งกลุ่ม
        grad_W, grad_b = flat.T @ delta, delta.sum(axis=0)
        d_pooled = (delta @ self.W.T).reshape(pooled.shape)
        is_max = blocks == pooled[:, :, None, :, None, :]    # เกรเดียนต์ไปเฉพาะตำแหน่งค่าสูงสุด
        d_a = np.zeros_like(a)
        p = self.pool
        d_a[:, :2 * p, :2 * p, :] = (is_max * d_pooled[:, :, None, :, None, :]).reshape(n, 2 * p, 2 * p, -1)
        d_z = d_a * relu_grad(z)
        grad_K = cols.reshape(-1, cols.shape[-1]).T @ d_z.reshape(-1, d_z.shape[-1])
        grad_bk = d_z.sum(axis=(0, 1, 2))
        return grad_K, grad_bk, grad_W, grad_b

    def train_step(self, images, labels, lr, momentum=0.9):
        probs, cache = self.forward(images)
        loss = -np.log(probs[np.arange(len(labels)), labels] + 1e-12).mean()
        for i, (param, grad) in enumerate(zip((self.K, self.bk, self.W, self.b), self.backward(cache, probs, labels))):
            self.velocity[i] = momentum * self.velocity[i] + grad
            param -= lr * self.velocity[i]
        return loss


cnn = TinyCNN()
rng = np.random.default_rng(RANDOM_SEED)
cnn_log = []
for epoch in range(1, 31):
    lr = 0.02 * 0.5 ** ((epoch - 1) // 10)                # ลดเป็นขั้น ×0.5 ทุก 10 รอบ (ตัวอย่างที่ 5.14)
    order = rng.permutation(len(X_dtrain))
    losses = [cnn.train_step(X_dtrain[order[s:s + 32]], y_dtrain[order[s:s + 32]], lr=lr)
              for s in range(0, len(order), 32)]
    cnn_log.append({"epoch": epoch, "train loss": np.mean(losses),
                    "test accuracy": (cnn.predict(X_dtest) == y_dtest).mean()})
cnn_log = pd.DataFrame(cnn_log).set_index("epoch")
display(cnn_log.iloc[[0, 4, 9, 14, 19, 24, 29]].round(3))

mlp_digits = MLPClassifier(hidden_layer_sizes=(32,), max_iter=300, random_state=RANDOM_SEED)
mlp_digits.fit(X_dtrain.reshape(len(X_dtrain), -1), y_dtrain)
mlp_params = sum(w.size for w in mlp_digits.coefs_) + sum(b.size for b in mlp_digits.intercepts_)
print(f"TinyCNN: พารามิเตอร์ {cnn.n_params():,} ความถูกต้องบนชุดทดสอบ {cnn_log['test accuracy'].iloc[-1]:.3f}")
print(f"MLP 64-32-10: พารามิเตอร์ {mlp_params:,} ความถูกต้องบนชุดทดสอบ "
      f"{mlp_digits.score(X_dtest.reshape(len(X_dtest), -1), y_dtest):.3f}")
```

```python
# ตรวจ Backpropagation ของ TinyCNN ด้วยการหาอนุพันธ์เชิงตัวเลข (เคอร์เนลทุกค่า)
check_cnn = TinyCNN(seed=1)
x_small_batch, y_small_batch = X_dtrain[:5], y_dtrain[:5]


def cnn_loss(model):
    probs = model.forward(x_small_batch)[0]
    return -np.log(probs[np.arange(5), y_small_batch]).mean()


probs, cache_cnn = check_cnn.forward(x_small_batch)
grad_K = check_cnn.backward(cache_cnn, probs, y_small_batch)[0]
numeric_K = np.zeros_like(grad_K)
for idx in np.ndindex(grad_K.shape):
    original = check_cnn.K[idx]
    check_cnn.K[idx] = original + 1e-5
    loss_plus = cnn_loss(check_cnn)
    check_cnn.K[idx] = original - 1e-5
    loss_minus = cnn_loss(check_cnn)
    check_cnn.K[idx] = original
    numeric_K[idx] = (loss_plus - loss_minus) / 2e-5
print(f"ผลต่างสูงสุดระหว่างเกรเดียนต์ของเคอร์เนลจาก backward กับเชิงตัวเลข: {np.abs(grad_K - numeric_K).max():.1e}")

# แสดงเคอร์เนลที่เรียนรู้ได้และผลลัพธ์ของภาพหนึ่งภาพ
fig, axes = plt.subplots(2, 9, figsize=(13, 3.2))
axes[0, 0].imshow(X_dtest[0], cmap="gray_r")
axes[0, 0].set_title(f"digit {y_dtest[0]}")
axes[1, 0].axis("off")
feature_maps = relu(im2col(X_dtest[:1], 3) @ cnn.K + cnn.bk)[0]
for f in range(8):
    axes[0, f + 1].imshow(cnn.K[:, f].reshape(3, 3), cmap="coolwarm")
    axes[0, f + 1].set_title(f"kernel {f + 1}", fontsize=8)
    axes[1, f + 1].imshow(feature_maps[:, :, f], cmap="inferno")
for ax in axes.ravel():
    ax.set_xticks([])
    ax.set_yticks([])
plt.tight_layout()
plt.show()
```

### 6.3 CNN ตามตัวอย่างที่ 5.21 เทียบกับ MLP บน MNIST

MNIST คือภาพตัวเลขเขียนมือขนาด 28 × 28 จุดภาพ ชุดฝึก 60,000 ภาพ ชุดทดสอบ 10,000 ภาพ (LeCun et al., 1998) `model.summary()` ต้องแสดงจำนวนพารามิเตอร์รวม 5,258 สำหรับ CNN และ 101,770 สำหรับ MLP ตรงกับตัวอย่างที่ 5.21 ฝึกเพียง 3 รอบเพื่อให้เสร็จเร็ว

```python
def build_book_cnn():
    """CNN ในตัวอย่างที่ 5.21"""
    return tf.keras.Sequential([
        tf.keras.Input(shape=(28, 28, 1)),
        tf.keras.layers.Conv2D(8, 3, activation="relu"),
        tf.keras.layers.MaxPooling2D(2),
        tf.keras.layers.Conv2D(16, 3, activation="relu"),
        tf.keras.layers.MaxPooling2D(2),
        tf.keras.layers.Flatten(),
        tf.keras.layers.Dense(10, activation="softmax"),
    ], name="book_cnn")


def build_book_mlp():
    return tf.keras.Sequential([
        tf.keras.Input(shape=(28, 28, 1)),
        tf.keras.layers.Flatten(),
        tf.keras.layers.Dense(128, activation="relu"),
        tf.keras.layers.Dense(10, activation="softmax"),
    ], name="book_mlp")


def mnist_experiment(epochs=3):
    (x_tr, y_tr), (x_te, y_te) = tf.keras.datasets.mnist.load_data()
    x_tr, x_te = x_tr[..., None] / 255.0, x_te[..., None] / 255.0
    results = {}
    for build in (build_book_cnn, build_book_mlp):
        model = build()
        model.summary()
        model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
        model.fit(x_tr, y_tr, epochs=epochs, batch_size=128, validation_split=0.1, verbose=2)
        results[model.name] = {"parameters": model.count_params(),
                               "test accuracy": model.evaluate(x_te, y_te, verbose=0)[1]}
    display(pd.DataFrame(results).T)


if HAS_TF:
    try:
        mnist_experiment()
    except Exception as error:          # เช่น ไม่มีอินเทอร์เน็ตสำหรับดาวน์โหลด MNIST
        print("รัน MNIST ไม่สำเร็จ:", type(error).__name__, error)
else:
    print("ข้าม (ไม่มี TensorFlow)")
```

### 6.4 Transfer Learning: นับพารามิเตอร์ที่ต้องฝึก (ตัวอย่างที่ 5.23)

ส่วนนี้สร้าง MobileNetV2 แบบไม่มีชั้นจำแนกเดิม (`include_top=False`) ซึ่งให้เวกเตอร์คุณลักษณะขนาด 1,280 หลัง Global Average Pooling แล้วต่อชั้นจำแนกใหม่ 4 คลาส (ลูกถ้วยฉนวน ปกติ/สกปรก/แตกร้าว/หลุดหาย ตามแบบฝึกหัดข้อ 17 ในหนังสือ) เพื่อนับว่าแต่ละขั้นของ Transfer Learning ต้องฝึกพารามิเตอร์กี่ค่า

> **หมายเหตุ** โค้ดนี้ใช้ `weights=None` เพื่อไม่ต้องดาวน์โหลดค่าน้ำหนักและรันได้เร็ว จำนวนพารามิเตอร์เท่ากันทุกประการ ในงานจริงให้เปลี่ยนเป็น `weights="imagenet"` เพื่อใช้ค่าน้ำหนักที่ฝึกไว้ก่อน แล้วฝึกด้วยข้อมูลภาพของงานตามขั้นตอน TRANSFER-LEARN ในหัวข้อ 5.5.3

```python
def transfer_learning_counts(n_classes=4, n_unfrozen=20):
    base = tf.keras.applications.MobileNetV2(input_shape=(96, 96, 3), include_top=False, pooling="avg", weights=None)
    model = tf.keras.Sequential([tf.keras.Input(shape=(96, 96, 3)), base,
                                 tf.keras.layers.Dense(n_classes, activation="softmax")])

    def trainable_count():
        return sum(int(np.prod(w.shape)) for w in model.trainable_weights)

    rows = {"total parameters": model.count_params()}
    base.trainable = False                                  # ขั้นที่ 1: สกัดคุณลักษณะ (ตรึงทั้งหมด)
    rows["stage 1: trainable (head only)"] = trainable_count()
    base.trainable = True                                   # ขั้นที่ 2: ปรับละเอียดเฉพาะชั้นท้าย ๆ
    for layer in base.layers[:-n_unfrozen]:
        layer.trainable = False
    rows[f"stage 2: trainable (head + last {n_unfrozen} layers)"] = trainable_count()
    display(pd.Series(rows, name="parameters").to_frame())
    print("ชั้นจำแนกใหม่ 1,280 × 4 + 4 =", 1280 * 4 + 4,
          "| ขั้นที่ 2 ต้องคอมไพล์ใหม่ด้วยอัตราการเรียนรู้เล็กกว่าขั้นที่ 1 ราว 10–100 เท่า")


if HAS_TF:
    transfer_learning_counts()
else:
    print("ข้าม (ไม่มี TensorFlow)")
```
