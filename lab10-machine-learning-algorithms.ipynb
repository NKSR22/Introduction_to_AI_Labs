## ส่วนที่ 1 Linear Regression

### 1.1 วิธีกำลังสองน้อยที่สุดด้วยสูตรปิด (ตัวอย่างที่ 4.14)

$$w_1 = \frac{S_{xy}}{S_{xx}} = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sum (x_i - \bar{x})^2}, \qquad w_0 = \bar{y} - w_1 \bar{x}$$

และในรูปเมทริกซ์ $\mathbf{w} = (\mathbf{X}^\top \mathbf{X})^{-1}\mathbf{X}^\top\mathbf{y}$ ซึ่งในโค้ดควรใช้ `np.linalg.solve` แทนการหาตัวผกผันตรง ๆ เพราะแม่นยำกว่า

```python
def fit_simple_linear(x, y):
    """FIT-SIMPLE-LINEAR ในหนังสือ: คืน (w0, w1) ด้วยวิธีกำลังสองน้อยที่สุด"""
    x_bar, y_bar = x.mean(), y.mean()
    s_xy = np.sum((x - x_bar) * (y - y_bar))
    s_xx = np.sum((x - x_bar) ** 2)
    w1 = s_xy / s_xx
    return y_bar - w1 * x_bar, w1


w0, w1 = fit_simple_linear(x_train, y_train)
print(f"สูตรปิด: ŷ = {w0:.2f} + {w1:.2f}x")

X_design = np.column_stack([np.ones_like(x_train), x_train])          # คอลัมน์แรกเป็น 1
w_matrix = np.linalg.solve(X_design.T @ X_design, X_design.T @ y_train)
print("รูปเมทริกซ์:", w_matrix.round(6))

sk_linear = LinearRegression().fit(x_train.reshape(-1, 1), y_train)
print(f"scikit-learn: w0 = {sk_linear.intercept_:.4f}, w1 = {sk_linear.coef_[0]:.4f}")

residuals = y_train - (w0 + w1 * x_train)
print("ส่วนเหลือ:", residuals.round(6), "| ผลรวม =", round(residuals.sum(), 9),
      "| Σx·e =", round(np.sum(x_train * residuals), 9))
compare("SSE ชุดฝึก", np.sum(residuals ** 2), 368, 1e-9)
compare("R² ชุดฝึก", 1 - np.sum(residuals ** 2) / np.sum((y_train - y_train.mean()) ** 2), 0.985, 5e-4)
```

**ค่าผิดปกติดึงเส้น** หนังสือยกตัวอย่างว่าถ้าวัน D8 บันทึกผิดเป็น 988 kW (ผิด 100 kW) ความชันจะเปลี่ยนจาก 20 เป็นราว 26.7 kW/°C ลองตรวจด้วยโค้ด

```python
y_bad = y_train.copy()
y_bad[-1] = 988
w0_bad, w1_bad = fit_simple_linear(x_train, y_bad)
compare("ความชันเมื่อ D8 = 988", w1_bad, 26.7, 0.05)
compare("ȳ ใหม่", y_bad.mean(), 812.5, 1e-9)
```

### 1.2 Gradient Descent ที่เขียนเอง (ตัวอย่างที่ 4.15)

ใช้ $x' = x - 30$ (จัดค่าเฉลี่ยเป็นศูนย์) โมเดล $\hat{y} = \theta_0 + \theta_1 x'$ เริ่มที่ $\theta_0 = \theta_1 = 0$ และ $\alpha = 0.05$ เกรเดียนต์ของ MSE คือ $\partial J/\partial\theta_0 = -\frac{2}{n}\sum e_i$ และ $\partial J/\partial\theta_1 = -\frac{2}{n}\sum e_i x'_i$ ฟังก์ชันต่อไปนี้คำนวณเกรเดียนต์จากข้อมูลจริงทุกรอบ (ไม่ใช้สูตรลดรูปของหนังสือ) แล้วบันทึกประวัติไว้

```python
def gradient_descent(x, y, alpha, n_iter, theta_init=(0.0, 0.0)):
    """Batch Gradient Descent บน MSE ของโมเดล ŷ = θ0 + θ1·x คืน DataFrame ประวัติทุกรอบ"""
    theta = np.array(theta_init, dtype=float)
    history = []
    for t in range(n_iter + 1):
        e = y - (theta[0] + theta[1] * x)                      # ส่วนเหลือ
        grad = np.array([-2 * e.mean(), -2 * (e * x).mean()])  # เกรเดียนต์ของ MSE
        history.append({"iter": t, "theta0": theta[0], "theta1": theta[1],
                        "grad0": grad[0], "grad1": grad[1], "J": np.mean(e ** 2)})
        theta = theta - alpha * grad                            # ก้าวลงเขา
    return pd.DataFrame(history).set_index("iter")


x_centered = x_train - 30
gd = gradient_descent(x_centered, y_train, alpha=0.05, n_iter=100)
display(gd.loc[[0, 1, 2, 3, 50, 100]].round(4))
compare("J รอบ 0", gd.loc[0, "J"], 643046, 0.5)
compare("J รอบ 3", gd.loc[3, "J"], 340169.0, 0.05)
compare("θ0 รอบ 50", gd.loc[50, "theta0"], 795.88, 0.005)
compare("θ0 รอบ 100", gd.loc[100, "theta0"], 799.98, 0.005)
```

ทุกแถวตรงกับตารางในตัวอย่างที่ 4.15 ความชัน $\theta_1$ ลู่เข้า 20 ภายในไม่กี่รอบ แต่จุดตัดแกน $\theta_0$ ต้องใช้ราว 100 รอบ เพราะหุบเขาในทิศ $\theta_0$ ลาดน้อยกว่า (ความโค้ง 2 เทียบกับ 15)

**ผลของอัตราการเรียนรู้** เงื่อนไขลู่เข้าคือ $\alpha < 2/15 \approx 0.133$ ลองเทียบ $\alpha = 0.001$ (ช้ามาก), $0.05$ และ $0.14$ (ลู่ออก) แล้ววาดเส้นทางบนเส้นชั้นความสูงของ $J$ (รูปที่ 4.7)

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.2))

# (ก) เส้นทางบนเส้นชั้นความสูงของ J(θ0, θ1) ที่คำนวณจากข้อมูลจริง
t0_grid, t1_grid = np.meshgrid(np.linspace(-50, 900, 200), np.linspace(-5, 30, 200))
J_grid = np.mean((y_train[None, None, :] - t0_grid[..., None] - t1_grid[..., None] * x_centered) ** 2, axis=-1)
ax = axes[0]
ax.contour(t0_grid, t1_grid, J_grid, levels=np.geomspace(50, 7e5, 14), colors=COLORS["gray"], linewidths=0.6)
ax.plot(gd["theta0"], gd["theta1"], "o-", ms=3, color=COLORS["orange"], label="α = 0.05 (100 iterations)")
ax.plot(800, 20, "*", ms=14, color=COLORS["green"], label="optimum (800, 20)")
ax.set_xlabel("θ0 (intercept)")
ax.set_ylabel("θ1 (slope)")
ax.set_title("(a) Gradient descent path on MSE contours")
ax.legend(loc="lower right")

# (ข) θ1 ตามรอบเมื่อ α ต่างกัน
ax = axes[1]
for alpha, color in [(0.001, COLORS["gray"]), (0.05, COLORS["blue"]), (0.14, COLORS["orange"])]:
    path = gradient_descent(x_centered, y_train, alpha=alpha, n_iter=10)
    ax.plot(path.index, path["theta1"], "o-", color=color, label=f"α = {alpha}")
ax.axhline(20, ls="--", color="black", lw=0.8)
ax.set_xlabel("Iteration")
ax.set_ylabel("θ1")
ax.set_title("(b) Effect of learning rate on θ1")
ax.legend()
plt.tight_layout()
plt.show()

diverge = gradient_descent(x_centered, y_train, alpha=0.14, n_iter=10)
ratio = (diverge["theta1"].iloc[1:].to_numpy() - 20) / (diverge["theta1"].iloc[:-1].to_numpy() - 20)
compare("ตัวคูณระยะห่างของ θ1 เมื่อ α = 0.14", ratio.mean(), -1.1, 1e-9)
```

**ทำไมต้องจัดค่าเฉลี่ยเป็นศูนย์** ถ้าใช้อุณหภูมิดิบ 26–34 °C เลขเงื่อนไขของหุบเขาจะสูงราว $1.1 \times 10^5$ อัตราการเรียนรู้ที่ใหญ่ที่สุดที่ยังไม่ลู่ออกจึงเล็กมาก และทิศที่ลาดน้อยต้องใช้รอบจำนวนมหาศาล ลองดูด้วยตัวเลข

### 2.1 ฟังก์ชันซิกมอยด์ (ตารางที่ 4.9)

$\sigma(z) = \dfrac{1}{1 + e^{-z}}$ แปลงผลรวมถ่วงน้ำหนัก $z$ ใด ๆ ให้เป็นความน่าจะเป็นระหว่าง 0 ถึง 1 และ $z$ คือลอการิทึมของอัตราต่อรอง (Logit)

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


z_values = np.array([-6, -3, -2, -1, 0, 1, 2, 3, 6])
display(pd.DataFrame({"z": z_values, "sigma(z)": sigmoid(z_values).round(4)}).set_index("z").T)
print("เส้นแบ่งที่เกณฑ์ 0.3 อยู่ที่ z =", round(math.log(0.3 / 0.7), 3))
```

### 2.2 ทำนายสภาพมอเตอร์ด้วยค่าน้ำหนักที่กำหนด (ตัวอย่างที่ 4.16)

หนังสือกำหนด $z = -16 + 2v + 0.1T$ เพื่อให้คำนวณด้วยมือได้ เส้นแบ่ง ($\hat{p} = 0.5$ หรือ $z = 0$) คือ $v = 8 - 0.05T$ และเส้นที่ $\hat{p} = 0.1$, $0.9$ คือ $2v + 0.1T = 16 \pm \ln 9$

```python
def book_logit(v, T):
    return -16 + 2 * v + 0.1 * T


book_pred = motors.assign(z=book_logit(motors["vibration_mm_s"], motors["bearing_temp_c"]))
book_pred["p_hat"] = sigmoid(book_pred["z"])
book_pred["predicted"] = np.where(book_pred["p_hat"] >= 0.5, "A", "N")
display(book_pred.round(3))
print("จำแนกถูกครบ 10 ตัว:", (book_pred["predicted"] == book_pred["condition"]).all())

z_q = book_logit(6.2, 66)
compare("p̂ ของมอเตอร์ใหม่ Q (6.2 mm/s, 66 °C)", sigmoid(z_q), 0.953, 5e-4)
print(f"การสั่นเพิ่ม 1 mm/s คูณอัตราต่อรองด้วย e^2 = {math.exp(2):.2f} เท่า")
```

### 2.3 Log-loss บนมอเตอร์ทดสอบ (ตารางที่ 4.10 และตัวอย่างที่ 4.17)

$J = -\frac{1}{n}\sum \big[y_i \ln \hat{p}_i + (1 - y_i)\ln(1 - \hat{p}_i)\big]$ ลงโทษความมั่นใจที่ผิดอย่างรุนแรง เทียบค่าที่คำนวณเองกับ `sklearn.metrics.log_loss`

```python
print("ตารางที่ 4.10  −ln p̂:", {p: round(-math.log(p), 3) for p in [0.99, 0.9, 0.6, 0.5, 0.1, 0.01]})

test_motors = pd.DataFrame({"v": [6.6, 4.4, 5.2, 6.0], "T": [72, 80, 54, 56], "y": [1, 1, 0, 0]},
                           index=["X1", "X2", "X3", "X4"])
test_motors["p_hat"] = sigmoid(book_logit(test_motors["v"], test_motors["T"]))
test_motors["loss"] = -(test_motors["y"] * np.log(test_motors["p_hat"])
                        + (1 - test_motors["y"]) * np.log(1 - test_motors["p_hat"]))
display(test_motors.round(3))
compare("Log-loss เฉลี่ย (คำนวณเอง)", test_motors["loss"].mean(), 0.691, 5e-4)
compare("Log-loss เฉลี่ย (sklearn)", log_loss(test_motors["y"], test_motors["p_hat"]), 0.691, 5e-4)
```

### 2.4 Gradient Descent ของ Logistic Regression (ตัวอย่างที่ 4.18)

เกรเดียนต์ของ Log-loss คือ $\partial J/\partial w_j = \frac{1}{n}\sum(\hat{p}_i - y_i)x_{ij}$ ใช้มอเตอร์ 4 ตัว (M3, M4, M7, M10) และคุณลักษณะเดียวคือการสั่น เริ่มจาก $w = 0$ และ $\alpha = 0.1$ แล้วไล่ต่อหลายรอบเพื่อดูสิ่งที่หนังสือเตือนว่า "ข้อมูลที่แยกได้สมบูรณ์ทำให้ค่าน้ำหนักโตไม่สิ้นสุด"

```python
v4 = np.array([3.2, 3.8, 6.2, 5.6])
y4 = np.array([0, 0, 1, 1])


def logistic_gd(x, y, alpha, n_iter):
    """Gradient Descent บน Log-loss ของ p̂ = σ(w0 + w1·x) คืนประวัติ"""
    w = np.zeros(2)
    rows = []
    for t in range(n_iter + 1):
        p = sigmoid(w[0] + w[1] * x)
        loss = -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))
        grad = np.array([np.mean(p - y), np.mean((p - y) * x)])
        rows.append({"iter": t, "w0": w[0], "w1": w[1], "grad0": grad[0], "grad1": grad[1], "log_loss": loss})
        w = w - alpha * grad
    return pd.DataFrame(rows).set_index("iter")


lg = logistic_gd(v4, y4, alpha=0.1, n_iter=100_000)
display(lg.loc[[0, 1]].round(4))
compare("Log-loss หลังรอบที่ 1", lg.loc[1, "log_loss"], 0.668, 5e-4)
print("\nไล่ต่อหลายรอบ (ข้อมูลแยกได้สมบูรณ์):")
display(lg.loc[[10, 100, 1_000, 10_000, 100_000], ["w0", "w1", "log_loss"]].round(4))
```

หลังรอบแรกค่าทุกตัวตรงกับตัวอย่างที่ 4.18 เมื่อไล่ต่อ Log-loss ลดลงเข้าใกล้ศูนย์เรื่อย ๆ ขณะที่ $w_1$ โตขึ้นไม่หยุด (เส้นแบ่ง $-w_0/w_1$ ไปอยู่ระหว่าง 3.8 กับ 5.6 แล้ว แต่ซิกมอยด์ยังชันขึ้นต่อไป) นี่คือเหตุผลที่ต้องใช้ Regularization

### 2.5 `LogisticRegression` ของ scikit-learn

`LogisticRegression` ใช้ L2 Regularization โดยปริยาย (`C=1.0` ยิ่งน้อยยิ่งลงโทษแรง) จึงได้ค่าน้ำหนักจำกัดแม้ข้อมูลแยกได้สมบูรณ์ ค่าที่ได้ต่างจากค่าที่หนังสือกำหนด แต่ลองเทียบ **ทิศทางของเส้นแบ่ง** (อัตราส่วน $w_T / w_v$)

```python
sk_model = LogisticRegression(max_iter=10_000).fit(X_motor, y_motor_binary)
T_line = np.linspace(45, 95, 50)
fig, ax = plt.subplots(figsize=(6.5, 4.5))
for cond, marker, color in [("N", "o", COLORS["blue"]), ("A", "^", COLORS["orange"])]:
    part = motors[motors["condition"] == cond]
    ax.scatter(part["bearing_temp_c"], part["vibration_mm_s"], marker=marker, s=70, color=color, label=cond)
ax.plot(T_line, 8 - 0.05 * T_line, color="black", label="book: v = 8 - 0.05T")
for offset, style in [(math.log(9), ":"), (-math.log(9), ":")]:
    ax.plot(T_line, (16 + offset - 0.1 * T_line) / 2, style, color=COLORS["gray"])
(w_v, w_T), b = sk_model.coef_[0], sk_model.intercept_[0]
ax.plot(T_line, -(b + w_T * T_line) / w_v, "--", color=COLORS["green"], label="scikit-learn (C = 1)")
ax.scatter([66], [6.2], marker="*", s=200, color=COLORS["green"], label="new motor Q")
ax.set_xlabel("Bearing temperature T (°C)")
ax.set_ylabel("Vibration v (mm/s)")
ax.set_ylim(1, 9)
ax.set_title("Logistic regression decision boundaries (dotted: book p = 0.1 / 0.9)")
ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

เส้นแบ่งของ scikit-learn เอียงไปทางเดียวกับของหนังสือ (ยิ่งตลับลูกปืนร้อน ยิ่งยอมรับการสั่นได้น้อยลง) และอัตราส่วน $w_T / w_v$ ใกล้ 0.05 แต่ค่าน้ำหนักเล็กกว่าเพราะ Regularization ทำให้ความน่าจะเป็นของ Q ต่ำกว่า เมื่อเพิ่ม `C` (ลงโทษน้อยลง) ค่าน้ำหนักโตขึ้นและความน่าจะเป็นเข้าใกล้ 0 หรือ 1 มากขึ้น

**คำถามชวนคิด** เพราะเหตุใดจึงเปรียบเทียบ $w_v = 2$ กับ $w_T = 0.1$ ตรง ๆ เพื่อสรุปว่า "การสั่นสำคัญกว่าอุณหภูมิ 20 เท่า" ไม่ได้ และต้องทำอย่างไรก่อนจึงเปรียบเทียบได้

## ส่วนที่ 3 k-Nearest Neighbors

### 3.1 ผลของการปรับขนาดต่อเพื่อนบ้านใกล้ที่สุด (ตัวอย่างที่ 4.19)

มอเตอร์ใหม่ Q สั่น 6.2 mm/s อุณหภูมิ 66 °C ใช้ k-NN ที่ $k = 3$ เทียบค่าดิบกับค่าที่ปรับเป็นช่วง 0–1 ตารางระยะทางคำนวณด้วย numpy ส่วนการทำนายใช้ `KNeighborsClassifier` ซึ่งเมื่อใส่ใน `Pipeline` กับ `MinMaxScaler` จะปรับขนาดข้อมูลใหม่ด้วยค่าจากชุดฝึกให้อัตโนมัติ

```python
Q = pd.DataFrame([[6.2, 66]], columns=X_motor.columns)
scaler = MinMaxScaler().fit(X_motor)
dist_raw = np.sqrt(((X_motor.to_numpy() - Q.to_numpy()) ** 2).sum(axis=1))
dist_scaled = np.sqrt(((scaler.transform(X_motor) - scaler.transform(Q)) ** 2).sum(axis=1))
distance_table = pd.DataFrame({"condition": y_motor, "raw distance": dist_raw,
                               "raw rank": pd.Series(dist_raw).rank(method="min").astype(int).to_numpy(),
                               "scaled distance": dist_scaled,
                               "scaled rank": pd.Series(dist_scaled.round(9)).rank(method="min").astype(int).to_numpy()},
                              index=X_motor.index)
display(distance_table.round(3))

# ใช้อาร์เรย์ numpy ทั้งตอน fit และตอนถามเพื่อนบ้าน (kneighbors) เพื่อให้รูปแบบข้อมูลตรงกัน
knn_raw = KNeighborsClassifier(n_neighbors=3).fit(X_motor.to_numpy(), y_motor)
knn_scaled = make_pipeline(MinMaxScaler(), KNeighborsClassifier(n_neighbors=3)).fit(X_motor.to_numpy(), y_motor)
for name, model, query in [("raw", knn_raw, Q.to_numpy()), ("scaled", knn_scaled, Q.to_numpy())]:
    knn_step = model if name == "raw" else model[-1]
    query_for_neighbors = query if name == "raw" else model[0].transform(query)
    _, idx = knn_step.kneighbors(query_for_neighbors)
    print(f"{name:7s}: neighbors = {list(X_motor.index[idx[0]])}, prediction = {model.predict(query)[0]}")
```

ข้อมูลเดียวกัน อัลกอริทึมเดียวกัน แต่คำตอบกลับข้าง: ค่าดิบได้เพื่อนบ้าน M4, M5, M2 (ปกติทั้งหมด) ส่วนค่าปรับขนาดได้ M7, M8, M4 (ผิดปกติ 2 : ปกติ 1) ตรงกับตัวอย่างที่ 4.19

กราฟต่อไปวาดด้วยสเกลแกนเท่ากัน (1 mm/s ยาวเท่า 1 °C) จะเห็นว่าในค่าดิบ ข้อมูลเกือบเป็นเส้นแบนตามแกนอุณหภูมิ ระยะทางจึงถูกอุณหภูมิครอบงำ (รูปที่ 4.9)

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
panels = [("(a) raw values, equal axis scale", X_motor.to_numpy(), Q.to_numpy()[0], knn_raw, " (°C)", " (mm/s)"),
          ("(b) after min-max scaling", scaler.transform(X_motor), scaler.transform(Q)[0], knn_scaled[-1],
           " (scaled)", " (scaled)")]
for ax, (title, pts, q, model, x_unit, y_unit) in zip(axes, panels):
    _, idx = model.kneighbors([q])
    radius = np.sqrt(((pts[idx[0][-1]] - q) ** 2).sum())
    for cond, marker, color in [("N", "o", COLORS["blue"]), ("A", "^", COLORS["orange"])]:
        mask = (y_motor == cond).to_numpy()
        ax.scatter(pts[mask, 1], pts[mask, 0], marker=marker, s=60, color=color, label=cond)
    for i, name in enumerate(X_motor.index):
        ax.annotate(name, (pts[i, 1], pts[i, 0]), xytext=(3, 3), textcoords="offset points", fontsize=8)
    for i in idx[0]:
        ax.plot([q[1], pts[i, 1]], [q[0], pts[i, 0]], color=COLORS["green"], lw=0.8)
    ax.add_patch(plt.Circle((q[1], q[0]), radius, fill=False, ls="--", color=COLORS["gray"]))
    ax.scatter([q[1]], [q[0]], marker="*", s=220, color=COLORS["green"], label="Q", zorder=3)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Bearing temperature" + x_unit)
    ax.set_ylabel("Vibration" + y_unit)
    ax.legend(fontsize=8, loc="upper left")
plt.tight_layout()
plt.show()
```

### 3.2 เลือก $k$ ด้วย Leave-One-Out Cross-Validation (ตัวอย่างที่ 4.20)

หนังสือปรับขนาดด้วยค่าจากมอเตอร์ทั้ง 10 ตัวก่อนทำ LOOCV เพื่อให้คำนวณด้วยมือได้ ซึ่งเป็นการรั่วของข้อมูลเล็กน้อย วิธีที่ถูกคือใส่ `MinMaxScaler` ไว้ใน `Pipeline` แล้วส่งให้ `cross_val_score` ซึ่งจะคำนวณค่าปรับขนาดใหม่จาก 9 ตัวที่เหลือในทุกรอบ ลองทำทั้งสองแบบ

```python
X_motor_scaled_all = MinMaxScaler().fit_transform(X_motor)        # แบบในหนังสือ (รั่วเล็กน้อย)
rows = []
for k in [1, 3, 5, 9]:
    book_way = cross_val_score(KNeighborsClassifier(k), X_motor_scaled_all, y_motor, cv=LeaveOneOut()).mean()
    pipe = make_pipeline(MinMaxScaler(), KNeighborsClassifier(k))
    correct_way = cross_val_score(pipe, X_motor, y_motor, cv=LeaveOneOut()).mean()
    predictions = "".join(cross_val_predict(pipe, X_motor, y_motor, cv=LeaveOneOut()))
    rows.append({"k": k, "LOOCV (scaled on all 10)": book_way, "LOOCV (Pipeline)": correct_way,
                 "predictions M1..M10": predictions})
loocv_table = pd.DataFrame(rows).set_index("k")
display(loocv_table)
for k, expected in zip([1, 3, 5, 9], [0.9, 0.9, 0.8, 0.6]):
    compare(f"LOOCV k = {k} (Pipeline)", loocv_table.loc[k, "LOOCV (Pipeline)"], expected, 1e-9)
```

ในข้อมูลชุดนี้ทั้งสองวิธีให้ผลเท่ากัน (0.9, 0.9, 0.8, 0.6) แต่ในข้อมูลอื่นอาจต่างกันได้ M7 (ตัวที่ 7) ถูกทำนายผิดทุกค่า $k$ และที่ $k = 5$ M10 เริ่มถูกมอเตอร์ปกติ "ข่มเสียง" ที่ $k = 9$ ทุกตัวถูกทำนายเป็นคลาสที่มากกว่า (N)

**คำถามชวนคิด** ทำไม 1-NN จึงถูกต้อง 100% บนชุดฝึกเสมอ และทำไมจึงไม่ควรใช้ความถูกต้องบนชุดฝึกเลือก $k$

### 4.1 เอนโทรปีและ Information Gain (ตารางที่ 4.11 และตัวอย่างที่ 4.21)

$$H(S) = -\sum_c p_c \log_2 p_c, \qquad \text{IG}(S, a) = H(S) - \sum_v \frac{|S_v|}{|S|} H(S_v)$$

ใช้บันทึกการตรวจมอเตอร์ 12 ครั้ง (ตารางที่ 4.12) ซึ่งใน notebook ใช้ค่าภาษาอังกฤษ: `low/medium/high` และ `normal/high/abnormal`

```python
def entropy(labels):
    """เอนโทรปี (บิต) ของชุดป้ายกำกับ"""
    p = pd.Series(labels).value_counts(normalize=True).to_numpy()
    return float(-np.sum(p * np.log2(p)))


def information_gain(df, attribute, target="condition"):
    """IG ของการแบ่ง df ตามค่าทุกค่าของ attribute (แบ่งหลายกิ่งแบบ ID3)"""
    remainder = sum(len(group) / len(df) * entropy(group[target]) for _, group in df.groupby(attribute))
    return entropy(df[target]) - remainder


print("ตารางที่ 4.11:", {f"{n}/{d}": round(entropy(["a"] * n + ["b"] * (d - n)), 3)
                         for n, d in [(1, 6), (1, 5), (1, 4), (2, 7), (1, 3), (2, 5), (1, 2)]})
root_ig = {a: information_gain(inspections, a) for a in ["vibration", "bearing_temp", "sound"]}
print("IG ที่ราก:", {a: round(v, 3) for a, v in root_ig.items()})
compare("IG(การสั่น)", root_ig["vibration"], 0.667, 5e-4)
compare("IG(อุณหภูมิ)", root_ig["bearing_temp"], 0.350, 5e-4)
compare("IG(เสียง)", root_ig["sound"], 0.196, 5e-4)

medium = inspections[inspections["vibration"] == "medium"]
print("IG ในกิ่ง 'การสั่นปานกลาง':", {a: round(information_gain(medium, a), 3) for a in ["bearing_temp", "sound"]})
print("Gain Ratio การสั่น = IG / H(4,4,4) =", round(root_ig["vibration"] / math.log2(3), 3),
      "| หมายเลขการตรวจ = 1 / log2(12) =", round(1 / math.log2(12), 3))
```

### 4.2 สร้างต้นไม้แบบ ID3 ด้วยโค้ดที่เขียนเอง

ฟังก์ชัน `build_tree` ทำตามรหัสเทียม BUILD-TREE ในหนังสือ: ถ้าทุกตัวอย่างเป็นคลาสเดียวกันให้เป็นใบ ไม่เช่นนั้นเลือกคุณลักษณะที่ IG สูงสุด แบ่งตามทุกค่า แล้วทำซ้ำกับแต่ละกิ่ง ผลลัพธ์พิมพ์เป็นกฎ ถ้า...แล้ว...

```python
def build_tree(df, attributes, target="condition", parent_majority=None):
    """ID3: คืนต้นไม้เป็นดิกชันนารีซ้อน {attribute: {value: subtree}} หรือชื่อคลาส (ใบ)"""
    if df.empty:
        return parent_majority
    classes = df[target].unique()
    if len(classes) == 1:
        return classes[0]
    majority = df[target].mode()[0]
    if not attributes:
        return majority
    best = max(attributes, key=lambda a: information_gain(df, a, target))
    remaining = [a for a in attributes if a != best]
    return {best: {value: build_tree(group, remaining, target, majority) for value, group in df.groupby(best)}}


def tree_to_rules(tree, conditions=()):
    """แปลงต้นไม้เป็นรายการกฎ ถ้า...แล้ว..."""
    if not isinstance(tree, dict):
        return [(conditions, tree)]
    (attribute, branches), = tree.items()
    rules = []
    for value, subtree in branches.items():
        rules += tree_to_rules(subtree, conditions + (f"{attribute} = {value}",))
    return rules


id3_tree = build_tree(inspections, ["vibration", "bearing_temp", "sound"])
print(id3_tree)
for i, (conds, label) in enumerate(tree_to_rules(id3_tree), 1):
    print(f"R{i}: IF {' AND '.join(conds)} THEN {label}")
```

ได้กฎ 4 ข้อเดียวกับ R1–R4 ในตัวอย่างที่ 4.21 (ลำดับต่างกันได้) และต้นไม้ไม่ใช้ "เสียง" เลย

## ส่วนที่ 5 k-Means

### 5.1 ไล่ k-Means ทีละรอบ (ตัวอย่างที่ 4.22)

อัลกอริทึมของ Lloyd สลับสองขั้น: **กำหนดกลุ่ม** (ทุกจุดไปอยู่กับเซนทรอยด์ที่ใกล้ที่สุด) และ **ปรับเซนทรอยด์** (เป็นค่าเฉลี่ยของสมาชิก) เริ่มจากเซนทรอยด์ที่อาคาร C และ D ซึ่งจงใจเลือกให้ไม่ดี บันทึก $J$ (WCSS) หลังทุกขั้น

```python
def kmeans_lloyd(points, init_centroids, max_iter=20):
    """k-Means แบบ Lloyd ที่บันทึก J หลังทุกขั้น คืน (labels, centroids, history)"""
    centroids = np.array(init_centroids, dtype=float)
    labels, history = None, []
    for it in range(1, max_iter + 1):
        sq_dist = ((points[:, None, :] - centroids[None, :, :]) ** 2).sum(axis=2)
        new_labels = sq_dist.argmin(axis=1)
        history.append({"step": f"assign {it}", "J": sq_dist.min(axis=1).sum(),
                        "centroids": centroids.round(4).tolist(), "labels": new_labels.tolist()})
        if labels is not None and np.array_equal(new_labels, labels):
            break                                            # กลุ่มไม่เปลี่ยน → หยุด
        labels = new_labels
        centroids = np.array([points[labels == j].mean(axis=0) for j in range(len(centroids))])
        history.append({"step": f"update {it}", "J": ((points - centroids[labels]) ** 2).sum(),
                        "centroids": centroids.round(4).tolist(), "labels": labels.tolist()})
    return labels, centroids, pd.DataFrame(history).set_index("step")


building_points = buildings.to_numpy(float)
init = buildings.loc[["C", "D"]].to_numpy(float)
km_labels, km_centroids, km_history = kmeans_lloyd(building_points, init)
display(km_history)
print("J ตามหนังสือ: 150 → 73 → 37 → 12.5 → 12.5 (หยุด)")

sk_km = KMeans(n_clusters=2, init=init, n_init=1).fit(building_points)
print("scikit-learn:", "inertia =", sk_km.inertia_, "| centroids =", sk_km.cluster_centers_.tolist(),
      "| labels =", sk_km.labels_.tolist())
```

```python
def plot_buildings(ax, labels, centroids, title, previous=None):
    colors = np.array([COLORS["blue"], COLORS["orange"]])
    ax.scatter(buildings["day_load"], buildings["night_load"], c=colors[labels], s=60)
    for name, (x, y) in buildings.iterrows():
        ax.annotate(name, (x, y), xytext=(4, 4), textcoords="offset points")
    ax.scatter(centroids[:, 0], centroids[:, 1], marker="X", s=200, c=colors[:len(centroids)], edgecolor="black")
    if previous is not None:
        for (x0, y0), (x1, y1) in zip(previous, centroids):
            ax.annotate("", (x1, y1), (x0, y0), arrowprops=dict(arrowstyle="->", color=COLORS["gray"], ls="--"))
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 10)
    ax.set_aspect("equal")
    ax.set_xlabel("Daytime load (x10 kW)")
    ax.set_ylabel("Night-time load (x10 kW)")
    ax.set_title(title)


steps = km_history.reset_index()
fig, axes = plt.subplots(1, 3, figsize=(13, 4.3))
plot_buildings(axes[0], np.array(steps.loc[0, "labels"]), np.array(steps.loc[0, "centroids"]), "(a) start: C, D  J = 150")
plot_buildings(axes[1], np.array(steps.loc[2, "labels"]), np.array(steps.loc[1, "centroids"]),
               "(b) after iteration 1  J = 37", previous=np.array(steps.loc[0, "centroids"]))
plot_buildings(axes[2], km_labels, km_centroids, "(c) after iteration 2 (stop)  J = 12.5",
               previous=np.array(steps.loc[1, "centroids"]))
plt.tight_layout()
plt.show()
```

$J$ ลดลงทุกขั้นตามที่หนังสือพิสูจน์ กลุ่ม {A, B, C, D} ใช้ไฟกลางคืนมาก (หอพัก?) และกลุ่ม {E, F, G, H} ใช้ไฟกลางวันมาก (อาคารเรียน?) อัลกอริทึมไม่รู้จักคำเหล่านี้ **มนุษย์เป็นผู้ตีความ** จากเซนทรอยด์

### 5.2 วิธีข้อศอกและค่าซิลูเอตต์ (ตัวอย่างที่ 4.23–4.24)

`KMeans` ของ scikit-learn ใช้ k-means++ และรันหลายจุดเริ่มต้น (`n_init`) แล้วเลือกผลที่ $J$ (`inertia_`) ต่ำที่สุด ค่าซิลูเอตต์คำนวณด้วย `silhouette_samples` (รายจุด) และ `silhouette_score` (ค่าเฉลี่ย)

```python
elbow = []
for k in range(1, 6):
    model = KMeans(n_clusters=k, n_init=50, random_state=RANDOM_SEED).fit(building_points)
    sil = silhouette_score(building_points, model.labels_) if k >= 2 else np.nan
    elbow.append({"k": k, "WCSS (J)": model.inertia_, "silhouette": sil})
elbow = pd.DataFrame(elbow).set_index("k")
elbow["decrease"] = -elbow["WCSS (J)"].diff()
display(elbow.round(3))
for k, expected in zip(range(1, 6), [112.75, 12.5, 8.25, 5.75, 3.5]):
    compare(f"J ต่ำสุด k = {k}", elbow.loc[k, "WCSS (J)"], expected, 1e-9)

s_samples = silhouette_samples(building_points, [0, 0, 0, 0, 1, 1, 1, 1])
compare("ค่าซิลูเอตต์ของอาคาร A (k = 2)", s_samples[0], 0.765, 5e-4)

fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
axes[0].plot(elbow.index, elbow["WCSS (J)"], "o-", color=COLORS["blue"])
axes[0].scatter([2], [elbow.loc[2, "WCSS (J)"]], s=150, color=COLORS["orange"], zorder=3)
axes[0].set_xlabel("Number of clusters k")
axes[0].set_ylabel("WCSS (J)")
axes[0].set_title("Elbow method")
axes[1].plot(elbow.index[1:], elbow["silhouette"].iloc[1:], "o-", color=COLORS["green"])
axes[1].set_xlabel("Number of clusters k")
axes[1].set_ylabel("Mean silhouette")
axes[1].set_title("Silhouette score")
plt.tight_layout()
plt.show()
```

ค่า $J$ ต่ำสุดของทุก $k$ ตรงกับตัวอย่างที่ 4.23 ข้อศอกอยู่ที่ $k = 2$ ชัดเจน และค่าซิลูเอตต์เฉลี่ยก็สูงสุดที่ $k = 2$ เช่นกัน

**คำถามชวนคิด** ถ้าเริ่ม k-Means ด้วย `n_init=1` และจุดเริ่มต้นสุ่ม ผลจะได้ $J$ ต่ำสุดเสมอหรือไม่ (ลองเปลี่ยน `random_state` หลายค่ากับ `k = 3` ดู)

## ส่วนที่ 6 Overfitting, Bias–Variance และ Regularization

### 6.1 โมเดลสามระดับความซับซ้อน (ตัวอย่างที่ 4.25)

- ก ค่าเฉลี่ยชุดฝึก (ง่ายเกินไป) — ข เส้นตรง $200 + 20x$ (พอดี) — ค 1-NN สำหรับการถดถอย (ซับซ้อนเกินไป ถ้าใกล้เท่ากันสองวันให้เฉลี่ย)

> **ข้อควรระวัง** `KNeighborsRegressor(n_neighbors=1)` ของ scikit-learn เลือกเพื่อนบ้าน **ตัวเดียว** เมื่อระยะเท่ากัน (ไม่เฉลี่ย) วัน E2 (30 °C ห่างจาก D4 และ D5 เท่ากัน) จึงให้ค่าต่างจากหนังสือ เราจึงเขียน 1-NN เองตามนิยามในหนังสือ และแสดงผลของ scikit-learn เทียบไว้

```python
def one_nn_regress(x_train, y_train, x_new):
    """1-NN สำหรับการถดถอย ถ้าระยะเท่ากันหลายวันให้เฉลี่ย (นิยามในตัวอย่างที่ 4.25)"""
    out = []
    for x in np.atleast_1d(x_new):
        d = np.abs(x_train - x)
        out.append(y_train[d == d.min()].mean())
    return np.array(out)


def mse(y, y_hat):
    return float(np.mean((np.asarray(y) - np.asarray(y_hat)) ** 2))


models_425 = {
    "A mean (underfit)": (lambda x: np.full(len(np.atleast_1d(x)), y_train.mean())),
    "B line 200 + 20x": (lambda x: 200 + 20 * np.asarray(x)),
    "C 1-NN (overfit)": (lambda x: one_nn_regress(x_train, y_train, x)),
}
table_425 = pd.DataFrame({name: {"train MSE": mse(y_train, f(x_train)), "test MSE": mse(y_test, f(x_test))}
                          for name, f in models_425.items()}).T
display(table_425.round(1))
print("1-NN ทำนายชุดทดสอบ:", one_nn_regress(x_train, y_train, x_test))
sk_1nn = KNeighborsRegressor(n_neighbors=1).fit(x_train.reshape(-1, 1), y_train)
print("KNeighborsRegressor(1):", sk_1nn.predict(x_test.reshape(-1, 1)), "← E2 ต่างจากหนังสือ")
compare("MSE ทดสอบ 1-NN", table_425.loc["C 1-NN (overfit)", "test MSE"], 991.4, 1e-9)
```

### 6.2 เห็น Variance ด้วยข้อมูลฝึกหลายชุด (ตัวอย่างที่ 4.26 และการจำลอง)

หนังสือแบ่งชุดฝึกเป็น S1 = {D1, D3, D5, D7} และ S2 = {D2, D4, D6, D8} แล้วพยากรณ์ที่ 30 °C ผลต่างระหว่างสองชุดคือ 6.2 kW สำหรับเส้นตรง และ 24 kW สำหรับ 1-NN

ต่อจากนั้นขยายเป็นการทดลองทางความคิดในหัวข้อ 4.7.1: สมมติความสัมพันธ์จริงคือ $f(x) = 200 + 20x$ และสัญญาณรบกวนมีส่วนเบี่ยงเบนมาตรฐาน 8 kW (ค่าสมมติ) สุ่มชุดฝึก 8 วันที่อุณหภูมิเดิม 2,000 ชุด ฝึกแต่ละโมเดล แล้วประมาณ Bias² และ Variance ของคำพยากรณ์ที่ 30 °C และ 36 °C

```python
S1, S2 = [0, 2, 4, 6], [1, 3, 5, 7]
for name, idx in [("S1", S1), ("S2", S2)]:
    w0_s, w1_s = fit_simple_linear(x_train[idx], y_train[idx])
    print(f"{name}: line ŷ(30) = {w0_s + w1_s * 30:.1f} kW (slope {w1_s:.2f}) | 1-NN ŷ(30) = "
          f"{one_nn_regress(x_train[idx], y_train[idx], 30)[0]:.0f} kW")

rng = np.random.default_rng(RANDOM_SEED)
true_f = lambda x: 200 + 20 * x
NOISE_SD, N_DATASETS = 8.0, 2000
preds = {"line": {30: [], 36: []}, "1-NN": {30: [], 36: []}}
for _ in range(N_DATASETS):
    y_sim = true_f(x_train) + rng.normal(0, NOISE_SD, len(x_train))
    a0, a1 = fit_simple_linear(x_train, y_sim)
    for x0 in (30, 36):
        preds["line"][x0].append(a0 + a1 * x0)
        preds["1-NN"][x0].append(one_nn_regress(x_train, y_sim, x0)[0])
rows = []
for model, by_x in preds.items():
    for x0, values in by_x.items():
        values = np.array(values)
        bias2 = (values.mean() - true_f(x0)) ** 2
        rows.append({"model": model, "x (°C)": x0, "Bias^2": bias2, "Variance": values.var(),
                     "expected MSE = Bias^2 + Var + noise": bias2 + values.var() + NOISE_SD ** 2})
display(pd.DataFrame(rows).set_index(["model", "x (°C)"]).round(1))
```

เมื่อความสัมพันธ์จริงเป็นเส้นตรง เส้นตรงแทบไม่มี Bias และ Variance ต่ำ ส่วน 1-NN มี Variance สูงกว่ามาก (คำทำนายขึ้นกับวันเดียว) และที่ 36 °C ซึ่งอยู่นอกช่วงข้อมูลฝึก 1-NN ยังมี Bias สูงมาก เพราะทายได้ไม่เกินโหลดของวันที่ร้อนที่สุดในชุดฝึก

### 6.3 ความซับซ้อนของโมเดลกับ Overfitting: การถดถอยพหุนาม

ลองเพิ่มความซับซ้อนทีละขั้นด้วยการถดถอยพหุนาม (หัวข้อ 4.4.1) บนข้อมูลจำลองที่ความสัมพันธ์จริงโค้งขึ้นเมื่อร้อนเกิน 33 °C: $f(x) = 200 + 20x + 4\max(x - 33, 0)^2$ สุ่มวันทำการ 15 วันเป็นชุดฝึก และ 500 วันเป็นชุดทดสอบ (อุณหภูมิ 26–36 °C ทั้งสองชุด) ก่อนสร้างพจน์พหุนาม จัด $x$ ให้อยู่ราวช่วง −1 ถึง 1 ด้วย $(x - 31)/5$ และปรับเป็นค่ามาตรฐาน เพื่อไม่ให้ตัวเลขกำลังสูงใหญ่เกินไป

```python
def true_curve(x):
    return 200 + 20 * x + 4 * np.clip(x - 33, 0, None) ** 2


def center_temperature(x):
    return (x - 31) / 5


rng_poly = np.random.default_rng(RANDOM_SEED)
xp_train = rng_poly.uniform(26, 36, 15)
yp_train = true_curve(xp_train) + rng_poly.normal(0, 12, 15)
xp_test = rng_poly.uniform(26, 36, 500)
yp_test = true_curve(xp_test) + rng_poly.normal(0, 12, 500)


def poly_model(degree, regressor):
    return make_pipeline(FunctionTransformer(center_temperature), PolynomialFeatures(degree, include_bias=False),
                         StandardScaler(), regressor)


degree_rows = []
for degree in [1, 2, 3, 5, 8, 12]:
    m = poly_model(degree, LinearRegression()).fit(xp_train.reshape(-1, 1), yp_train)
    degree_rows.append({"degree": degree, "train MSE": mse(yp_train, m.predict(xp_train.reshape(-1, 1))),
                        "test MSE": mse(yp_test, m.predict(xp_test.reshape(-1, 1))),
                        "max |coef|": np.abs(m[-1].coef_).max()})
degree_table = pd.DataFrame(degree_rows).set_index("degree")
display(degree_table.round(1))
print("สัญญาณรบกวนที่ลดไม่ได้ (σ² = 12²) =", 12 ** 2)
```

MSE ฝึกลดลงทุกครั้งที่เพิ่มกำลัง แต่ MSE ทดสอบลดลงช่วงแรก (กำลัง 2–3 จับความโค้งได้) แล้วพุ่งสูงขึ้นมากเมื่อกำลังสูง นี่คือรูปตัว U ของรูปที่ 4.13 และค่าสัมประสิทธิ์ของโมเดลกำลังสูงใหญ่มาก ซึ่งเป็นสัญญาณของ Overfitting

### 6.4 Regularization

**(ก) Ridge กับพหุนามกำลัง 12** ใช้ `Ridge(alpha=...)` แทน `LinearRegression` ค่า `alpha` คือ $\lambda$ ในหนังสือ (ยิ่งมากยิ่งลงโทษแรง) เลือก `alpha` ด้วย 5-fold CV บนชุดฝึก 15 วันเท่านั้น แล้วดูผลบนชุดทดสอบ

```python
alphas = [1e-4, 1e-3, 1e-2, 1e-1, 1, 10, 100]
ridge_rows = []
for a in alphas:
    m = poly_model(12, Ridge(alpha=a)).fit(xp_train.reshape(-1, 1), yp_train)
    cv_mse = -cross_val_score(poly_model(12, Ridge(alpha=a)), xp_train.reshape(-1, 1), yp_train,
                              cv=KFold(5, shuffle=True, random_state=RANDOM_SEED),
                              scoring="neg_mean_squared_error").mean()
    ridge_rows.append({"alpha": a, "train MSE": mse(yp_train, m.predict(xp_train.reshape(-1, 1))),
                       "CV MSE (train only)": cv_mse, "test MSE": mse(yp_test, m.predict(xp_test.reshape(-1, 1))),
                       "max |coef|": np.abs(m[-1].coef_).max()})
ridge_table = pd.DataFrame(ridge_rows).set_index("alpha")
display(ridge_table.round(1))
best_alpha = ridge_table["CV MSE (train only)"].idxmin()
print(f"alpha ที่ CV เลือก = {best_alpha} → test MSE = {ridge_table.loc[best_alpha, 'test MSE']:.1f} "
      f"(พหุนามกำลัง 12 ไม่มี Regularization: {degree_table.loc[12, 'test MSE']:,.0f})")

grid = np.linspace(26, 36, 300).reshape(-1, 1)
plt.figure(figsize=(7, 4))
plt.scatter(xp_test, yp_test, s=5, color=COLORS["gray"], alpha=0.4, label="test data")
plt.scatter(xp_train, yp_train, color=COLORS["blue"], zorder=3, label="15 training days")
plt.plot(grid, true_curve(grid), "k--", label="true f(x)")
plt.plot(grid, poly_model(12, LinearRegression()).fit(xp_train.reshape(-1, 1), yp_train).predict(grid),
         color=COLORS["orange"], label="degree 12, no regularization")
plt.plot(grid, poly_model(12, Ridge(alpha=best_alpha)).fit(xp_train.reshape(-1, 1), yp_train).predict(grid),
         color=COLORS["green"], lw=2, label=f"degree 12, Ridge alpha = {best_alpha}")
plt.ylim(650, 1050)
plt.xlabel("Daily max temperature (°C)")
plt.ylabel("Peak load (kW)")
plt.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

Ridge หดค่าสัมประสิทธิ์ลงมาก เส้นโค้งเรียบขึ้น และ MSE ทดสอบลดลงจากหลักล้านเหลือหลักร้อย สังเกตว่า CV บนข้อมูลเพียง 15 วันก็ยังแกว่ง ค่า `alpha` ที่ CV เลือกอาจไม่ใช่ค่าที่ดีที่สุดบนชุดทดสอบ ซึ่งเป็นข้อจำกัดของข้อมูลน้อย

**(ข) Ridge และ Lasso บนข้อมูลโหลดของบท (ตัวอย่างที่ 4.27)** สำหรับคุณลักษณะเดียวที่จัดค่าเฉลี่ยศูนย์แล้ว Ridge ได้ $w_1 = S_{xy}/(S_{xx} + \lambda)$ และ Lasso ได้ $w_1 = (S_{xy} - \lambda/2)/S_{xx}$ (เป็น 0 เมื่อ $\lambda \ge 2S_{xy}$)

> **ข้อควรระวัง** `Ridge(alpha)` ของ scikit-learn ลด $\text{SSE} + \alpha\lVert w\rVert^2$ จึงใช้ `alpha = λ` ได้ตรง ๆ แต่ `Lasso(alpha)` ลด $\frac{1}{2n}\text{SSE} + \alpha\lVert w\rVert_1$ ต้องใช้ `alpha = λ / (2n)` = λ/16 สำหรับข้อมูล 8 วัน

```python
X8 = x_train.reshape(-1, 1)
rows = []
for method, lam in [("OLS", 0), ("Ridge", 20), ("Ridge", 60), ("Ridge", 240), ("Lasso", 600), ("Lasso", 1200), ("Lasso", 2400)]:
    if method == "OLS":
        model = LinearRegression()
    elif method == "Ridge":
        model = Ridge(alpha=lam)
    else:
        model = Lasso(alpha=lam / (2 * len(x_train)), max_iter=100_000)
    model.fit(X8, y_train)
    rows.append({"method": method, "lambda": lam, "w1 (kW/°C)": model.coef_[0], "intercept at 30 °C":
                 model.intercept_ + model.coef_[0] * 30, "train MSE": mse(y_train, model.predict(X8)),
                 "test MSE": mse(y_test, model.predict(x_test.reshape(-1, 1)))})
table_427 = pd.DataFrame(rows).set_index(["method", "lambda"])
display(table_427.round(2))
print("หนังสือ: w1 = 20, 15, 10, 4, 15, 10, 0 | test MSE = 177.8, 912.8, 2357.8, 5029, 912.8, 2357.8, 7377.8")
```

ตรงกับตัวอย่างที่ 4.27 ทุกแถว ในกรณีนี้ Regularization ทำให้ผลทดสอบแย่ลงทุกกรณี เพราะเส้นตรงที่มีพารามิเตอร์สองตัวไม่ได้ Overfit อยู่แล้ว ต่างจากพหุนามกำลัง 12 ในข้อ (ก) **Regularization เป็นยาแก้ Overfitting ไม่ใช่ยาบำรุงที่ใช้ได้เสมอ** และต้องเลือกความแรงด้วย Cross-Validation

## ส่วนที่ 7 เปรียบเทียบอัลกอริทึมบนข้อมูลจำลองขนาดใหญ่

### 7.1 จำแนกสภาพมอเตอร์ 1,000 ตัวด้วย `StratifiedKFold`

ใช้ชุดข้อมูลจำลองเดียวกับปฏิบัติการที่ 9 (มอเตอร์ปกติราวร้อยละ 92 ผิดปกติสองแบบ: ตลับลูกปืนชำรุด และเพลาเยื้องศูนย์) เปรียบเทียบ 5 โมเดลด้วย 5-fold CV แบบแบ่งชั้น ทุกโมเดลอยู่ใน `Pipeline` ที่เติมค่าที่หายไปด้วยมัธยฐาน ตัววัดหลักคือความระลึก ความแม่นยำ และ F1 ของคลาสผิดปกติ

```python
def make_motor_fleet(n=1000, seed=RANDOM_SEED):
    """ข้อมูลจำลองสภาพมอเตอร์ n ตัว (เหมือนปฏิบัติการที่ 9)"""
    rng = np.random.default_rng(seed)
    kind = rng.choice(["normal", "bearing", "misalign"], size=n, p=[0.92, 0.05, 0.03])
    vibration = np.select([kind == "normal", kind == "bearing"],
                          [rng.lognormal(np.log(2.9), 0.28, n), rng.normal(5.4, 1.1, n)],
                          rng.normal(6.0, 0.9, n))
    bearing_temp = np.select([kind == "normal", kind == "bearing"],
                             [rng.normal(63, 8, n), rng.normal(80, 7, n)], rng.normal(60, 6, n))
    current = np.where(kind == "normal", rng.normal(78, 9, n), rng.normal(84, 10, n))
    ambient = rng.normal(33, 3, n)
    bearing_temp = bearing_temp + 0.5 * (ambient - 33)
    status = np.where(kind == "normal", "N", "A")
    repair = np.where(status == "A", rng.gamma(4, 3, n),
                      np.where(rng.random(n) < 0.03, rng.gamma(2, 1, n), 0.0))
    df = pd.DataFrame({"motor_id": [f"M{i + 1:04d}" for i in range(n)],
                       "vibration_mm_s": vibration.clip(0.5).round(2), "bearing_temp_c": bearing_temp.round(1),
                       "current_pct": current.round(1), "ambient_temp_c": ambient.round(1),
                       "repair_hours": repair.round(1), "status": status})
    sensor_dropout = rng.random(n) < np.where(df["bearing_temp_c"] > 75, 0.25, 0.02)
    df.loc[sensor_dropout, "bearing_temp_c"] = np.nan
    return df


fleet = make_motor_fleet()
SENSOR_FEATURES = ["vibration_mm_s", "bearing_temp_c", "current_pct", "ambient_temp_c"]   # ไม่ใช้ repair_hours (รั่ว)
X_fleet, y_fleet = fleet[SENSOR_FEATURES], (fleet["status"] == "A").astype(int)
imputer = lambda: SimpleImputer(strategy="median")

fleet_models = {
    "Logistic Regression": make_pipeline(imputer(), StandardScaler(), LogisticRegression()),
    "k-NN (k=5, raw)": make_pipeline(imputer(), KNeighborsClassifier(5)),
    "k-NN (k=5, scaled)": make_pipeline(imputer(), StandardScaler(), KNeighborsClassifier(5)),
    "Decision Tree (full)": make_pipeline(imputer(), DecisionTreeClassifier(random_state=RANDOM_SEED)),
    "Decision Tree (depth 3)": make_pipeline(imputer(), DecisionTreeClassifier(max_depth=3, random_state=RANDOM_SEED)),
}
cv_strat = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_SEED)
rows = []
for name, model in fleet_models.items():
    res = cross_validate(model, X_fleet, y_fleet, cv=cv_strat, scoring=["recall", "precision", "f1"],
                         return_train_score=True)
    rows.append({"model": name, "recall": res["test_recall"].mean(), "precision": res["test_precision"].mean(),
                 "F1 (CV)": res["test_f1"].mean(), "F1 std": res["test_f1"].std(), "F1 (train)": res["train_f1"].mean()})
fleet_cv = pd.DataFrame(rows).set_index("model")
display(fleet_cv.round(3))
```

อ่านผล:

- **การปรับขนาดสำคัญกับ k-NN** บนข้อมูลจริงขนาดใหญ่ก็เช่นเดียวกับตัวอย่างที่ 4.19 k-NN ที่ไม่ปรับขนาดมีความระลึกต่ำมาก
- **ต้นไม้ที่โตเต็มที่ Overfit** F1 บนชุดฝึกเป็น 1.0 แต่ F1 จาก CV ต่ำกว่าต้นไม้ความลึก 3 การจำกัดความลึกคือ Regularization ของต้นไม้
- ส่วนเบี่ยงเบนมาตรฐานของ F1 ระหว่าง fold ไม่เล็ก (คลาสผิดปกติใน fold หนึ่งมีเพียงราว 15 ตัว) ผลต่างระหว่างโมเดลที่เล็กกว่าส่วนเบี่ยงเบนมาตรฐานจึงยังสรุปไม่ได้ (ตัวอย่างที่ 4.7)

**เส้นแบ่งของแต่ละอัลกอริทึม** ใช้เพียงสองคุณลักษณะ (การสั่น อุณหภูมิ) เพื่อวาดเส้นแบ่งบนระนาบ ให้เห็นรูปแบบตามตารางที่ 4.14: เส้นตรง ยืดหยุ่นตามตัวอย่าง และขั้นบันไดขนานแกน

```python
two = ["bearing_temp_c", "vibration_mm_s"]
X_two = fleet[two]
boundary_models = {
    "Logistic Regression": make_pipeline(imputer(), StandardScaler(), LogisticRegression()),
    "k-NN (k=5, scaled)": make_pipeline(imputer(), StandardScaler(), KNeighborsClassifier(5)),
    "Decision Tree (depth 3)": make_pipeline(imputer(), DecisionTreeClassifier(max_depth=3, random_state=RANDOM_SEED)),
}
tt, vv = np.meshgrid(np.linspace(35, 105, 250), np.linspace(0.5, 10, 250))
grid_df = pd.DataFrame({"bearing_temp_c": tt.ravel(), "vibration_mm_s": vv.ravel()})
fig, axes = plt.subplots(1, 3, figsize=(14, 4), sharey=True)
for ax, (name, model) in zip(axes, boundary_models.items()):
    model.fit(X_two, y_fleet)
    zz = model.predict(grid_df).reshape(tt.shape)
    ax.contourf(tt, vv, zz, levels=[-0.5, 0.5, 1.5], colors=["#DCE6F1", "#F6DCC8"])
    for status, marker, color in [(0, "o", COLORS["blue"]), (1, "^", COLORS["orange"])]:
        part = X_two[y_fleet == status]
        ax.scatter(part["bearing_temp_c"], part["vibration_mm_s"], s=8, marker=marker, color=color, alpha=0.6)
    cv_f1 = cross_val_score(model, X_two, y_fleet, cv=cv_strat, scoring="f1").mean()
    ax.set_title(f"{name}\nCV F1 = {cv_f1:.3f}")
    ax.set_xlabel("Bearing temperature (°C)")
axes[0].set_ylabel("Vibration (mm/s)")
plt.tight_layout()
plt.show()
```

**เส้นโค้งการเรียนรู้ (Learning Curve)** วาดความคลาดเคลื่อน (1 − F1) บนชุดฝึกและชุดตรวจสอบเทียบกับจำนวนข้อมูลฝึก ช่องว่างกว้างระหว่างสองเส้นคืออาการของ Overfitting (หัวข้อ 4.7.1)

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 3.8), sharey=True)
for ax, name in zip(axes, ["Decision Tree (full)", "Decision Tree (depth 3)"]):
    sizes, train_scores, val_scores = learning_curve(
        fleet_models[name], X_fleet, y_fleet, cv=cv_strat, scoring="f1",
        train_sizes=np.linspace(0.15, 1.0, 6))
    ax.plot(sizes, 1 - train_scores.mean(axis=1), "o-", color=COLORS["blue"], label="training error (1 - F1)")
    ax.plot(sizes, 1 - val_scores.mean(axis=1), "o-", color=COLORS["orange"], label="validation error (1 - F1)")
    ax.set_title(name)
    ax.set_xlabel("Number of training examples")
axes[0].set_ylabel("Error = 1 - F1")
axes[0].legend()
plt.tight_layout()
plt.show()
```

ต้นไม้ที่โตเต็มที่มีความคลาดเคลื่อนบนชุดฝึกเป็นศูนย์ทุกขนาดข้อมูล แต่บนชุดตรวจสอบสูงกว่ามาก ช่องว่างกว้างนี้คืออาการของ Variance สูง และยังไม่แคบลงชัดเจนแม้ใช้ข้อมูลทั้งหมด ส่วนต้นไม้ความลึก 3 มีสองเส้นเข้าหากันเมื่อข้อมูลเพิ่มขึ้น และความคลาดเคลื่อนบนชุดตรวจสอบต่ำกว่า ในกรณีนี้การจำกัดความซับซ้อนได้ผลกว่าการรอเก็บข้อมูลเพิ่ม

### 7.2 จัดกลุ่มรูปแบบโหลดจากมิเตอร์อัจฉริยะ 200 ราย

ตามกรณีศึกษาท้ายหัวข้อ 4.6.2: สร้างรูปแบบโหลดรายชั่วโมง (24 ค่า) ของผู้ใช้ไฟ 4 แบบ (ค่าสมมติ) คือ สำนักงาน (กลางวัน) ที่พักอาศัย (ยอดตอนเย็น) โรงงานสามกะ (ราบเกือบทั้งวัน) และร้านค้า (10:00–22:00 น.) ขนาดโหลดต่างกันมาก (บ้านไม่กี่ kW โรงงานหลายร้อย kW) แล้วเปรียบเทียบการจัดกลุ่มด้วยค่าดิบ กับค่าที่หารด้วยโหลดสูงสุดของแต่ละราย (ทำให้เป็นบรรทัดฐาน) ประเภทจริงใช้ตรวจผลเท่านั้น ไม่ได้ให้อัลกอริทึมเห็น

```python
def make_load_profiles(seed=RANDOM_SEED):
    """รูปแบบโหลดรายชั่วโมงจำลองของผู้ใช้ไฟ 200 ราย คืน (profiles kW, ประเภทจริง)"""
    rng = np.random.default_rng(seed)
    hours = np.arange(24)
    bump = lambda center, width: np.exp(-0.5 * ((hours - center) / width) ** 2)
    shapes = {"office": 0.25 + 0.75 * ((hours >= 8) & (hours < 18)),
              "residence": 0.35 + 0.55 * bump(20, 2.5) + 0.2 * bump(7, 1.5),
              "factory": 0.85 + 0.1 * np.sin(2 * np.pi * hours / 8),
              "shop": 0.2 + 0.8 * ((hours >= 10) & (hours < 22))}
    typical_kw = {"office": 300, "residence": 5, "factory": 800, "shop": 50}
    profiles, kinds = [], []
    for kind, count in [("office", 60), ("residence", 80), ("factory", 30), ("shop", 30)]:
        for _ in range(count):
            shape = np.clip(shapes[kind] + rng.normal(0, 0.07, 24), 0.02, None)
            profiles.append(shape * rng.lognormal(np.log(typical_kw[kind]), 0.5))
            kinds.append(kind)
    return pd.DataFrame(profiles, columns=[f"h{h:02d}" for h in hours]), np.array(kinds)


profiles, true_kind = make_load_profiles()
normalized = profiles.div(profiles.max(axis=1), axis=0)          # หารด้วยโหลดสูงสุดของแต่ละราย

rows = []
for k in range(2, 8):
    km = KMeans(n_clusters=k, n_init=20, random_state=RANDOM_SEED).fit(normalized)
    rows.append({"k": k, "WCSS": km.inertia_, "silhouette": silhouette_score(normalized, km.labels_),
                 "agreement with true type (ARI)": adjusted_rand_score(true_kind, km.labels_)})
profile_table = pd.DataFrame(rows).set_index("k")
display(profile_table.round(3))

raw_km = KMeans(n_clusters=4, n_init=20, random_state=RANDOM_SEED).fit(profiles)
print("k = 4 บนค่าดิบ (ไม่ทำให้เป็นบรรทัดฐาน): ARI =", round(adjusted_rand_score(true_kind, raw_km.labels_), 3),
      "| ขนาดกลุ่ม =", np.bincount(raw_km.labels_).tolist())
display(pd.crosstab(pd.Series(raw_km.labels_, name="cluster (raw kW)"), pd.Series(true_kind, name="true type")))
```

```python
km4 = KMeans(n_clusters=4, n_init=20, random_state=RANDOM_SEED).fit(normalized)
fig, axes = plt.subplots(1, 3, figsize=(14, 3.8))
axes[0].plot(profile_table.index, profile_table["WCSS"], "o-", color=COLORS["blue"])
axes[0].set_title("Elbow (normalized profiles)")
axes[0].set_xlabel("k")
axes[0].set_ylabel("WCSS")
axes[1].plot(profile_table.index, profile_table["silhouette"], "o-", color=COLORS["green"])
axes[1].set_title("Silhouette")
axes[1].set_xlabel("k")
for j, centroid in enumerate(km4.cluster_centers_):
    axes[2].plot(range(24), centroid, lw=2, label=f"cluster {j} (n = {(km4.labels_ == j).sum()})")
axes[2].set_title("Cluster centroids = representative load shapes")
axes[2].set_xlabel("Hour of day")
axes[2].set_ylabel("Load / daily peak")
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()
display(pd.crosstab(pd.Series(km4.labels_, name="cluster"), pd.Series(true_kind, name="true type")))
```

หลังทำให้เป็นบรรทัดฐาน ทั้งวิธีข้อศอกและค่าซิลูเอตต์ชี้ที่ $k = 4$ และกลุ่มตรงกับประเภทจริงทุกราย (ARI = 1) เซนทรอยด์แต่ละกลุ่มคือ "รูปแบบโหลดตัวแทน" ที่วิศวกรตีความได้ ส่วนการจัดกลุ่มบนค่าดิบได้กลุ่มตาม **ขนาด** ของโหลด (โรงงานถูกแบ่งเป็นหลายกลุ่มตามขนาด ส่วนบ้าน ร้านค้า และสำนักงานขนาดเล็กรวมเป็นกลุ่มเดียว) ตรงกับคำเตือนในกรณีศึกษาของหนังสือ

```python
def forecast_metrics(y, y_hat):
    e = np.asarray(y) - np.asarray(y_hat)
    return {"MAE": np.mean(np.abs(e)), "RMSE": np.sqrt(np.mean(e ** 2)),
            "MAPE (%)": 100 * np.mean(np.abs(e / y)), "error at E4": e[3]}


ridge20 = Ridge(alpha=20).fit(x_train.reshape(-1, 1), y_train)
book_models = {"baseline: train mean (800 kW)": np.full(5, y_train.mean()),
               "1-NN on temperature": one_nn_regress(x_train, y_train, x_test),
               "Ridge (lambda = 20)": ridge20.predict(x_test.reshape(-1, 1)),
               "line 200 + 20x": 200 + 20 * x_test}
table_419 = pd.DataFrame({name: forecast_metrics(y_test, pred) for name, pred in book_models.items()}).T
display(table_419.round(2))
print("หนังสือ: MAE 69.0/23.8/24.0/10.6, RMSE 85.9/31.49/30.21/13.33, MAPE 8.43/2.91/2.91/1.26, E4 +145/+57/+55/+25")

THRESHOLD_KW = 900
for name in ["line 200 + 20x", "1-NN on temperature"]:
    forecast_e4 = book_models[name][3]
    decision = "prepare load shedding" if forecast_e4 > THRESHOLD_KW else "no action"
    correct = (forecast_e4 > THRESHOLD_KW) == (y_test[3] > THRESHOLD_KW)
    print(f"{name:22s}: forecast E4 = {forecast_e4:.0f} kW → {decision:22s} (actual 945 kW) → "
          f"{'correct' if correct else 'WRONG'} decision")
```

ตามขั้นตอนของหัวข้อ 4.7.3: ใช้ ม.ค.–ต.ค. เป็นชุดฝึก และ พ.ย.–ธ.ค. เป็นชุดทดสอบ (แบ่งตามเวลา) เลือกไฮเปอร์พารามิเตอร์ด้วย `TimeSeriesSplit` บนชุดฝึกเท่านั้น และเทียบกับค่าฐานสองแบบ: ค่าเฉลี่ยชุดฝึก และ **Persistence** (โหลดชั่วโมงเดียวกันของสัปดาห์ก่อน)

```python
THAI_HOLIDAYS_2025 = ["2025-01-01", "2025-02-12", "2025-04-07", "2025-04-14", "2025-04-15", "2025-04-16",
                      "2025-05-01", "2025-05-05", "2025-05-12", "2025-06-03", "2025-07-10", "2025-07-28",
                      "2025-08-12", "2025-10-13", "2025-10-23", "2025-12-05", "2025-12-10", "2025-12-31"]


def make_hourly_load(seed=RANDOM_SEED):
    """ข้อมูลจำลองโหลดรายชั่วโมงของอาคารสำนักงาน 1 ปี"""
    rng = np.random.default_rng(seed)
    idx = pd.date_range("2025-01-01 00:00", "2025-12-31 23:00", freq="h")
    day_of_year, hour = idx.dayofyear.to_numpy(), idx.hour.to_numpy()
    daily_mean = 28.5 + 2.5 * np.cos(2 * np.pi * (day_of_year - 110) / 365)
    day_noise = np.repeat(rng.normal(0, 1.0, 366), 24)[:len(idx)]
    temp = daily_mean + day_noise + 3.5 * np.cos(2 * np.pi * (hour - 15) / 24) + rng.normal(0, 0.4, len(idx))
    workday = (idx.dayofweek < 5) & ~idx.normalize().isin(pd.to_datetime(THAI_HOLIDAYS_2025))
    occupied = workday & (hour >= 8) & (hour < 18)
    base = 180 + np.where(workday & (hour >= 7) & (hour < 19), 120, 0)
    cooling = (np.where(occupied, 18, 4) * np.clip(temp - 24, 0, None)
               + np.where(occupied, 3.0, 0.5) * np.clip(temp - 32, 0, None) ** 2)
    load = base + cooling + rng.normal(0, 10, len(idx))
    return pd.DataFrame({"timestamp": idx, "temp_c": temp.round(2), "is_workday": workday.astype(int),
                         "load_kw": load.round(1)})


hourly = make_hourly_load()
hourly["hour"] = hourly["timestamp"].dt.hour
hourly["hour_x_workday"] = hourly["hour"].astype(str) + "_" + hourly["is_workday"].astype(str)   # ปฏิสัมพันธ์
hourly["load_last_week"] = hourly["load_kw"].shift(168)          # ค่าในอดีตที่รู้แล้ว ณ เวลาพยากรณ์
hourly = hourly.dropna().reset_index(drop=True)

is_test_h = hourly["timestamp"] >= "2025-11-01"
h_train, h_test = hourly[~is_test_h], hourly[is_test_h]
print(f"ชุดฝึก {len(h_train):,} ชั่วโมง ({h_train['timestamp'].min():%d %b} – {h_train['timestamp'].max():%d %b}) | "
      f"ชุดทดสอบ {len(h_test):,} ชั่วโมง")

week = hourly[(hourly["timestamp"] >= "2025-04-07") & (hourly["timestamp"] < "2025-04-21")]
fig, ax1 = plt.subplots(figsize=(11, 3.2))
ax1.plot(week["timestamp"], week["load_kw"], color=COLORS["blue"], lw=1)
ax1.set_ylabel("Load (kW)")
ax2 = ax1.twinx()
ax2.plot(week["timestamp"], week["temp_c"], color=COLORS["orange"], lw=0.8, alpha=0.7)
ax2.set_ylabel("Temperature (°C)", color=COLORS["orange"])
ax1.set_title("Two weeks of simulated hourly load (Songkran holidays 14-16 Apr)")
plt.tight_layout()
plt.show()
```

```python
tscv = TimeSeriesSplit(n_splits=5)
CAL = ["temp_c", "hour", "is_workday"]
linear_calendar = make_pipeline(
    ColumnTransformer([("hour_x_workday", OneHotEncoder(handle_unknown="ignore"), ["hour_x_workday"]),
                       ("numeric", "passthrough", ["temp_c"])]),
    LinearRegression())
tree_search = GridSearchCV(DecisionTreeRegressor(random_state=RANDOM_SEED),
                           {"max_depth": [2, 4, 6, 8, 10, 12, None]}, cv=tscv,
                           scoring="neg_root_mean_squared_error")
knn_search = GridSearchCV(make_pipeline(StandardScaler(), KNeighborsRegressor()),
                          {"kneighborsregressor__n_neighbors": [1, 5, 10, 20, 50]}, cv=tscv,
                          scoring="neg_root_mean_squared_error")

candidates = {
    "linear: temperature only": (make_pipeline(LinearRegression()), ["temp_c"]),
    "linear: hour x workday + temperature": (linear_calendar, ["hour_x_workday", "temp_c"]),
    "k-NN (k by TimeSeriesSplit)": (knn_search, CAL),
    "decision tree (depth by TimeSeriesSplit)": (tree_search, CAL),
}
results_h, predictions_h = {}, {}
results_h["baseline: train mean"] = forecast_metrics(h_test["load_kw"].to_numpy(), np.full(len(h_test), h_train["load_kw"].mean()))
results_h["baseline: persistence (last week)"] = forecast_metrics(h_test["load_kw"].to_numpy(), h_test["load_last_week"].to_numpy())
for name, (model, cols) in candidates.items():
    model.fit(h_train[cols], h_train["load_kw"])
    predictions_h[name] = model.predict(h_test[cols])
    results_h[name] = forecast_metrics(h_test["load_kw"].to_numpy(), predictions_h[name])
results_h = pd.DataFrame(results_h).T.drop(columns="error at E4")
display(results_h.round(2))
print("max_depth ที่เลือก:", tree_search.best_params_["max_depth"],
      "| k ที่เลือก:", knn_search.best_params_["kneighborsregressor__n_neighbors"])
display(pd.DataFrame({"max_depth": tree_search.cv_results_["param_max_depth"],
                      "CV RMSE": -tree_search.cv_results_["mean_test_score"]}).round(2).T)
```

- **Overfitting** เห็นได้จาก 1-NN (MSE ฝึก 0 แต่ทดสอบ 991.4) พหุนามกำลังสูง และต้นไม้ที่โตเต็มที่ การจำลองแสดงว่า 1-NN มี Variance สูงกว่าเส้นตรงมาก **Ridge** แก้พหุนามกำลัง 12 ได้ แต่ทำให้เส้นตรงที่ไม่ Overfit แย่ลง (ตัวอย่างที่ 4.27) (ส่วนที่ 6)
