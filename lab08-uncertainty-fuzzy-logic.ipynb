## ส่วนที่ 1 การแจกแจงร่วมและความน่าจะเป็นแบบมีเงื่อนไข (ตัวอย่างที่ 3.13)

บันทึกการตรวจมอเตอร์ 1,000 ครั้งให้การแจกแจงร่วมของ $F$ = สภาพตลับลูกปืน และ $V$ = ระดับการสั่น (ค่าสมมติ) เก็บเป็นตาราง `pandas` แล้วคำนวณ

- ความน่าจะเป็นส่วนเพิ่ม: รวมแถวหรือคอลัมน์ $P(F) = \sum_v P(F, v)$
- ความน่าจะเป็นแบบมีเงื่อนไข: $P(a \mid b) = P(a \land b) / P(b)$
- ความเป็นอิสระ: $F$ กับ $V$ เป็นอิสระต่อกันก็ต่อเมื่อ $P(f \land v) = P(f)\,P(v)$

```python
joint = pd.DataFrame([[0.04, 0.01], [0.10, 0.85]],
                     index=["F=ชำรุด", "F=ปกติ"], columns=["V=สูง", "V=ปกติ"])
p_f = joint.loc["F=ชำรุด"].sum()          # รวมตามแถว
p_high = joint["V=สูง"].sum()             # รวมตามคอลัมน์
display(joint.assign(รวม=joint.sum(axis=1)))

compare("P(f)", p_f, 0.05, 1e-9)
compare("P(V=สูง)", p_high, 0.14, 1e-9)
compare("P(f | V=สูง)", joint.loc["F=ชำรุด", "V=สูง"] / p_high, 0.286, 5e-4)
compare("P(V=สูง | f)", joint.loc["F=ชำรุด", "V=สูง"] / p_f, 0.80, 1e-9)
print(f"P(f)P(V=สูง) = {p_f * p_high:.3f} แต่ P(f ∧ V=สูง) = {joint.loc['F=ชำรุด', 'V=สูง']:.2f} → ไม่เป็นอิสระต่อกัน")
```

**คำถามชวนคิด** ตลับลูกปืนที่ชำรุดทำให้สั่นสูงถึงร้อยละ 80 แต่เมื่อเห็นการสั่นสูง ทำไมโอกาสที่ตลับลูกปืนชำรุดจริงจึงมีเพียงราวร้อยละ 29 (ดูว่าการสั่นสูง 0.14 มาจากแถวใดบ้าง)

## ส่วนที่ 2 ทฤษฎีของเบย์และกับดักอัตราฐาน

### 2.1 คำนวณด้วยสูตร (ตัวอย่างที่ 3.14 และตารางที่ 3.12)

ระบบตรวจจับตลับลูกปืนชำรุดมีความไว $P(\text{เตือน} \mid \text{ชำรุด}) = 0.95$ อัตราการแจ้งเตือนผิด $P(\text{เตือน} \mid \text{ปกติ}) = 0.05$ และอัตราฐาน $P(\text{ชำรุด}) = 0.01$

$$P(\text{ชำรุด} \mid \text{เตือน}) = \frac{P(\text{เตือน} \mid \text{ชำรุด})\,P(\text{ชำรุด})}{P(\text{เตือน} \mid \text{ชำรุด})\,P(\text{ชำรุด}) + P(\text{เตือน} \mid \text{ปกติ})\,P(\text{ปกติ})}$$

```python
def bayes_posterior(prior, sensitivity, false_alarm_rate):
    """คืน (P(ชำรุด | เตือน), P(เตือน)) ด้วยทฤษฎีของเบย์และกฎความน่าจะเป็นรวม"""
    p_alarm = sensitivity * prior + false_alarm_rate * (1 - prior)
    return sensitivity * prior / p_alarm, p_alarm


post, p_alarm = bayes_posterior(0.01, 0.95, 0.05)
compare("P(เตือน)", p_alarm, 0.0590, 5e-5)
compare("P(ชำรุด | เตือน)", post, 0.161, 5e-4)
compare("P(ชำรุด | ไม่เตือน)", 0.05 * 0.01 / (1 - p_alarm), 0.00053, 5e-6)

# ตารางที่ 3.12 เปลี่ยนค่าทีละตัวจากกรณีฐาน
CASES_312 = {"1. กรณีฐาน": (0.01, 0.95, 0.05),
             "2. มอเตอร์เก่าที่ชำรุดบ่อย": (0.10, 0.95, 0.05),
             "3. ลดอัตราเตือนผิดเหลือ 1%": (0.01, 0.95, 0.01),
             "4. เพิ่มความไวเป็น 99%": (0.01, 0.99, 0.05)}
rows = []
for name, (prior, sens, fa) in CASES_312.items():
    post, p_alarm = bayes_posterior(prior, sens, fa)
    rows.append({"กรณี": name, "P(ชำรุด)": prior, "ความไว": sens, "อัตราเตือนผิด": fa,
                 "P(เตือน)": round(p_alarm, 4), "P(ชำรุด|เตือน)": round(post, 3)})
table_312 = pd.DataFrame(rows).set_index("กรณี")
table_312
```

ตารางตรงกับตารางที่ 3.12 ทุกแถว (0.161, 0.679, 0.490, 0.167)

### 2.2 ยืนยันด้วยการจำลองความถี่

สูตรอาจดูเป็นนามธรรม ลองจำลองมอเตอร์จำนวนมากด้วยการสุ่ม: สุ่มว่าตัวใดชำรุดตามอัตราฐาน แล้วสุ่มว่าระบบเตือนหรือไม่ตามความไวหรืออัตราเตือนผิด จากนั้น **นับ** ว่าในมอเตอร์ที่ถูกเตือนทั้งหมด มีกี่ตัวที่ชำรุดจริง ซึ่งก็คือต้นไม้ความถี่ธรรมชาติในรูปที่ 3.5

```python
def simulate_alarms(n_motors, prior, sensitivity, false_alarm_rate, rng):
    """สุ่มสภาพจริงและสัญญาณเตือนของมอเตอร์ n_motors ตัว คืนอาร์เรย์บูลีนสองชุด"""
    faulty = rng.random(n_motors) < prior
    alarm_prob = np.where(faulty, sensitivity, false_alarm_rate)
    alarm = rng.random(n_motors) < alarm_prob
    return faulty, alarm


rng = np.random.default_rng(RANDOM_SEED)
faulty, alarm = simulate_alarms(10_000, 0.01, 0.95, 0.05, rng)
counts = pd.crosstab(pd.Series(faulty, name="ชำรุดจริง"), pd.Series(alarm, name="ระบบเตือน"), margins=True)
display(counts)
print(f"มอเตอร์ 10,000 ตัว: เตือน {alarm.sum()} ครั้ง เป็นของจริง {(faulty & alarm).sum()} ครั้ง"
      f" → สัดส่วน {(faulty & alarm).sum() / alarm.sum():.3f} (สูตร 0.161, หนังสือคาด 95/590)")

# ทำซ้ำกับมอเตอร์ 1,000,000 ตัว สำหรับทุกกรณีของตารางที่ 3.12
for name, (prior, sens, fa) in CASES_312.items():
    faulty, alarm = simulate_alarms(1_000_000, prior, sens, fa, rng)
    estimate = (faulty & alarm).sum() / alarm.sum()
    print(f"{name:30s} จำลองได้ {estimate:.3f}  สูตร {bayes_posterior(prior, sens, fa)[0]:.3f}")
```

ผลการจำลองใกล้กับสูตรมาก และยิ่งจำนวนมอเตอร์มาก ค่าที่ได้ยิ่งใกล้ (กฎของจำนวนมาก) กราฟต่อไปแสดงว่า $P(\text{ชำรุด} \mid \text{เตือน})$ ขึ้นกับอัตราฐานอย่างไร เมื่อใช้ระบบเดียวกัน

```python
priors = np.logspace(-3, 0, 300)
plt.figure(figsize=(6.5, 3.6))
for fa, color in [(0.05, "#1F4E79"), (0.01, "#E07B39")]:
    plt.semilogx(priors, bayes_posterior(priors, 0.95, fa)[0], color=color,
                 label=f"sensitivity 0.95, false alarm {fa:.0%}")
plt.axvline(0.01, color="#6B6B6B", ls="--")
plt.text(0.011, 0.05, "base rate 1%", color="#6B6B6B")
plt.xlabel("base rate P(fault)")
plt.ylabel("P(fault | alarm)")
plt.title("Posterior depends strongly on the base rate")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

### 2.3 หลักฐานชิ้นที่สองและรูปแบบอัตราต่อรอง (ตัวอย่างที่ 3.15)

หลังระบบเตือน ช่างวัดซ้ำด้วยเครื่องวิเคราะห์สเปกตรัมการสั่นที่มีความไว 0.90 และอัตราเตือนผิด 0.02 สมมติว่าผลทั้งสองเป็นอิสระต่อกันเมื่อรู้สภาพตลับลูกปืน ใช้ความน่าจะเป็นภายหลังของขั้นก่อนเป็นความน่าจะเป็นก่อนของขั้นใหม่ หรือใช้รูปแบบอัตราต่อรอง $O(H \mid e) = \text{LR} \times O(H)$ ก็ได้ผลเท่ากัน

```python
def odds(p):
    return p / (1 - p)


def prob_from_odds(o):
    return o / (1 + o)


p1 = bayes_posterior(0.01, 0.95, 0.05)[0]                 # หลังสัญญาณเตือน
p2_positive = bayes_posterior(p1, 0.90, 0.02)[0]           # วัดซ้ำพบความผิดปกติ
p2_negative = (0.10 * p1) / (0.10 * p1 + 0.98 * (1 - p1))  # วัดซ้ำไม่พบ
compare("หลังสัญญาณเตือน", p1, 0.161, 5e-4)
compare("วัดซ้ำพบความผิดปกติ", p2_positive, 0.896, 5e-4)
compare("วัดซ้ำไม่พบความผิดปกติ", p2_negative, 0.019, 5e-4)
compare("รูปแบบอัตราต่อรอง (LR 19 × 45)", prob_from_odds(odds(0.01) * 19 * 45), 0.896, 5e-4)

# ยืนยันด้วยการจำลอง: สุ่มผลการทดสอบสองวิธีแบบอิสระเมื่อรู้สภาพจริง
rng = np.random.default_rng(RANDOM_SEED)
faulty, alarm1 = simulate_alarms(2_000_000, 0.01, 0.95, 0.05, rng)
alarm2 = rng.random(faulty.size) < np.where(faulty, 0.90, 0.02)
both = alarm1 & alarm2
print(f"จำลอง: P(ชำรุด | เตือน และวัดซ้ำพบ) ≈ {(faulty & both).sum() / both.sum():.3f}")
```

**คำถามชวนคิด** ตารางที่ 3.12 บอกว่าการลดอัตราเตือนผิดช่วยได้มากกว่าการเพิ่มความไว ลองอธิบายด้วยจำนวนมอเตอร์ในการจำลอง: เมื่อลดอัตราเตือนผิดจาก 5% เป็น 1% จำนวน "เตือนผิด" ลดลงจากเท่าใดเป็นเท่าใด

## ส่วนที่ 3 Naive Bayes จากการนับ (ตัวอย่างที่ 3.16)

### 3.1 สร้างบันทึกการซ่อมบำรุงและฝึกด้วยการนับ

ตัวอย่างที่ 3.16 ให้จำนวนนับจากบันทึกการซ่อมบำรุง 100 ครั้ง (ค่าสมมติ) เราสร้างตารางบันทึก 100 แถวที่มีจำนวนนับตรงกับหนังสือ (สลับลำดับแถวแบบสุ่มด้วย seed) แล้ว "ฝึก" Naive Bayes ซึ่งเป็นเพียงการนับ

$$P(C \mid x_1, \ldots, x_n) = \alpha\, P(C) \prod_{i} P(x_i \mid C)$$

คุณลักษณะสองค่า 3 ตัว: $V$ = สั่นสูง, $H$ = ตลับลูกปืนร้อน, $S$ = เสียงผิดปกติ คลาส: N = ปกติ, B = ตลับลูกปืนชำรุด, M = เพลาเยื้องศูนย์

```python
COUNTS_316 = {"N": (60, 6, 6, 12), "B": (25, 20, 15, 20), "M": (15, 12, 3, 6)}   # (จำนวน, มี V, มี H, มี S)
FEATURES = ["V", "H", "S"]


def make_maintenance_records(counts, rng):
    """สร้างบันทึกที่จำนวนนับของแต่ละคุณลักษณะในแต่ละคลาสตรงกับ counts"""
    frames = []
    for cls, (n, *positives) in counts.items():
        columns = {}
        for feature, k in zip(FEATURES, positives):
            column = np.zeros(n, dtype=int)
            column[:k] = 1
            columns[feature] = rng.permutation(column)
        frames.append(pd.DataFrame({"คลาส": cls, **columns}))
    return pd.concat(frames, ignore_index=True)


def train_naive_bayes(records, alpha=0):
    """นับ P(C) และ P(x_i = 1 | C) ถ้า alpha = 1 คือการปรับเรียบแบบลาปลาซสำหรับคุณลักษณะสองค่า"""
    priors = records["คลาส"].value_counts(normalize=True, sort=False)
    grouped = records.groupby("คลาส", sort=False)[FEATURES]
    likelihoods = (grouped.sum() + alpha).div(grouped.count() + 2 * alpha)
    return priors, likelihoods


def predict_naive_bayes(evidence, priors, likelihoods):
    """evidence เช่น {"V": 1, "H": 0, "S": 0} คืนตารางตัวเศษและความน่าจะเป็นภายหลัง"""
    scores = priors.copy()
    for feature, value in evidence.items():
        p_one = likelihoods[feature]
        scores = scores * (p_one if value else 1 - p_one)
    return pd.DataFrame({"ตัวเศษ": scores.round(4), "ความน่าจะเป็นภายหลัง": (scores / scores.sum()).round(3)})


records = make_maintenance_records(COUNTS_316, np.random.default_rng(RANDOM_SEED))
print("ตัวอย่างบันทึก 5 แถวแรก:")
display(records.head())
priors, likelihoods = train_naive_bayes(records)
display(likelihoods.assign(**{"P(C)": priors}).round(2))
```

```python
print("กรณีที่ 1: สั่นสูง แต่ตลับลูกปืนไม่ร้อน และไม่มีเสียงผิดปกติ")
case1 = predict_naive_bayes({"V": 1, "H": 0, "S": 0}, priors, likelihoods)
display(case1)
compare("P(M | กรณีที่ 1)", case1.loc["M", "ความน่าจะเป็นภายหลัง"], 0.493, 5e-4)

print("กรณีที่ 2: มีอาการครบทั้งสามอย่าง")
case2 = predict_naive_bayes({"V": 1, "H": 1, "S": 1}, priors, likelihoods)
display(case2)
compare("P(B | กรณีที่ 2)", case2.loc["B", "ความน่าจะเป็นภายหลัง"], 0.899, 5e-4)
```

### 3.2 ความถี่เป็นศูนย์และการปรับเรียบแบบลาปลาซ

สมมติว่าในข้อมูลไม่มีมอเตอร์เพลาเยื้องศูนย์ตัวใดเลยที่ตลับลูกปืนร้อน ($P(H \mid M) = 0/15$) แล้วพบมอเตอร์ที่สั่นสูง ตลับลูกปืนร้อนเล็กน้อย และไม่มีเสียง ลองดูว่าคลาส M ถูกตัดทิ้งทันทีหรือไม่ และการบวก 1 ให้ทุกการนับ (`alpha=1`) เปลี่ยนผลอย่างไร

```python
counts_zero = dict(COUNTS_316, M=(15, 12, 0, 6))       # ไม่มี M ตัวใดที่ H = 1
records_zero = make_maintenance_records(counts_zero, np.random.default_rng(RANDOM_SEED))
evidence = {"V": 1, "H": 1, "S": 0}
for alpha in (0, 1):
    priors_z, likelihoods_z = train_naive_bayes(records_zero, alpha=alpha)
    print(f"alpha = {alpha}: P(H | M) = {likelihoods_z.loc['M', 'H']:.3f}")
    display(predict_naive_bayes(evidence, priors_z, likelihoods_z))
```

เมื่อไม่ปรับเรียบ คลาส M ได้ความน่าจะเป็น 0 ทันทีเพราะมีตัวคูณเป็นศูนย์ ทั้งที่การสั่นสูงและไม่มีเสียงชี้ไปที่ M ค่อนข้างมาก เมื่อใช้การปรับเรียบแบบลาปลาซ $P(H \mid M) = 1/17 \approx 0.059$ ตรงกับหนังสือ คลาส M จึงยังมีโอกาส

**คำถามชวนคิด** ถ้ามีบันทึกการซ่อมบำรุง 10,000 ครั้งแทน 100 ครั้ง ผลของการบวก 1 ต่อความน่าจะเป็นจะมากขึ้นหรือน้อยลง เพราะอะไร

### 4.1 นิยามข่ายงาน (รูปที่ 3.6 และตารางที่ 3.13)

แต่ละโหนดเก็บรายชื่อพ่อแม่และ CPT ซึ่งเป็นดิกชันนารีจากชุดค่าของพ่อแม่ไปยัง $P(\text{โหนด} = \text{จริง} \mid \text{พ่อแม่})$ ค่าเท็จได้จาก 1 ลบ

- $O$ = โหลดเกินพิกัด, $C$ = ระบบระบายความร้อนขัดข้อง (สาเหตุต้นทาง)
- $T$ = อุณหภูมิน้ำมันสูง (พ่อแม่ $O, C$), $A$ = สัญญาณเตือนอุณหภูมิ (พ่อแม่ $T$), $I$ = มิเตอร์แสดงกระแสสูง (พ่อแม่ $O$)

การแจกแจงร่วมคือผลคูณของ CPT: $P(x_1, \ldots, x_n) = \prod_i P(x_i \mid \text{parents}(X_i))$

```python
TRANSFORMER_BN = {
    "O": {"parents": [], "cpt": {(): 0.10}},
    "C": {"parents": [], "cpt": {(): 0.05}},
    "T": {"parents": ["O", "C"], "cpt": {(True, True): 0.95, (True, False): 0.60,
                                         (False, True): 0.70, (False, False): 0.02}},
    "A": {"parents": ["T"], "cpt": {(True,): 0.90, (False,): 0.05}},
    "I": {"parents": ["O"], "cpt": {(True,): 0.90, (False,): 0.10}},
}


def node_probability(bn, var, value, world):
    """P(var = value | ค่าของพ่อแม่ใน world)"""
    parent_values = tuple(world[p] for p in bn[var]["parents"])
    p_true = bn[var]["cpt"][parent_values]
    return p_true if value else 1 - p_true


def joint_probability(bn, world):
    """ผลคูณ CPT ของทุกโหนด (world ต้องกำหนดค่าครบทุกตัวแปร)"""
    result = 1.0
    for var in bn:
        result *= node_probability(bn, var, world[var], world)
    return result


world = {"O": True, "C": False, "T": True, "A": True, "I": True}
compare("ตัวอย่างที่ 3.17: P(o, ¬c, t, a, i)", joint_probability(TRANSFORMER_BN, world), 0.04617, 5e-6)
n_params = sum(len(node["cpt"]) for node in TRANSFORMER_BN.values())
print(f"จำนวนพารามิเตอร์ของข่ายงาน = {n_params} ค่า เทียบกับตารางร่วมเต็ม 2^5 − 1 = {2 ** 5 - 1} ค่า")
```

### 4.2 ENUMERATION-ASK

ฟังก์ชันต่อไปเขียนตามรหัสเทียม ENUMERATION-ASK ในหัวข้อ 3.5.2: สำหรับแต่ละค่าของตัวแปรที่สนใจ รวมการแจกแจงร่วมทุกชุดค่าของตัวแปรซ่อน แล้วทำให้เป็นบรรทัดฐาน ฟังก์ชันนี้ไม่รู้จักหม้อแปลงเลย ใช้กับข่ายงานใดก็ได้ที่เขียนในรูปแบบเดียวกัน

```python
def enumeration_ask(bn, query, evidence):
    """คืน (P(query = จริง | evidence), P(evidence))"""
    hidden = [v for v in bn if v != query and v not in evidence]
    totals = {}
    for query_value in (True, False):
        total = 0.0
        for values in itertools.product((True, False), repeat=len(hidden)):
            world = {**evidence, query: query_value, **dict(zip(hidden, values))}
            total += joint_probability(bn, world)
        totals[query_value] = total
    p_evidence = totals[True] + totals[False]
    return totals[True] / p_evidence, p_evidence


p_t, _ = enumeration_ask(TRANSFORMER_BN, "T", {})
p_a, _ = enumeration_ask(TRANSFORMER_BN, "A", {})
compare("P(t)", p_t, 0.11035, 5e-6)
compare("P(a)", p_a, 0.1438, 5e-5)
compare("ตัวอย่างที่ 3.18: P(c | a)", enumeration_ask(TRANSFORMER_BN, "C", {"A": True})[0], 0.232, 5e-4)
compare("ตัวอย่างที่ 3.18: P(o | a)", enumeration_ask(TRANSFORMER_BN, "O", {"A": True})[0], 0.400, 5e-4)
compare("คำถามเชิงทำนาย: P(a | o)", enumeration_ask(TRANSFORMER_BN, "A", {"O": True})[0], 0.575, 5e-4)
```

### 4.3 การอธิบายหักล้าง (ตัวอย่างที่ 3.19 และตารางที่ 3.15)

หลังสัญญาณเตือนดัง ผู้ควบคุมดูค่าจากมิเตอร์กระแสเพิ่ม ลองคำนวณความเชื่อเรื่องสาเหตุทั้งสองเมื่อได้หลักฐานเพิ่มทีละชิ้น

```python
EVIDENCE_315 = {"ยังไม่มีหลักฐาน": {},
                "สัญญาณเตือนดัง": {"A": True},
                "สัญญาณเตือนดัง + กระแสสูง": {"A": True, "I": True},
                "สัญญาณเตือนดัง + กระแสปกติ": {"A": True, "I": False}}
BOOK_315 = {"ยังไม่มีหลักฐาน": (0.100, 0.050), "สัญญาณเตือนดัง": (0.400, 0.232),
            "สัญญาณเตือนดัง + กระแสสูง": (0.857, 0.112), "สัญญาณเตือนดัง + กระแสปกติ": (0.069, 0.318)}


def belief_table(bn):
    rows = []
    for name, evidence in EVIDENCE_315.items():
        rows.append({"หลักฐานที่รู้": name,
                     "P(โหลดเกิน | ·)": round(enumeration_ask(bn, "O", evidence)[0], 3),
                     "P(ระบายความร้อนขัดข้อง | ·)": round(enumeration_ask(bn, "C", evidence)[0], 3)})
    return pd.DataFrame(rows).set_index("หลักฐานที่รู้")


table_315 = belief_table(TRANSFORMER_BN)
display(table_315)
matches = all(tuple(table_315.loc[k]) == v for k, v in BOOK_315.items())
print("ตรงกับตารางที่ 3.15 ทุกแถว:", matches)

table_315.plot.barh(figsize=(7, 3.2), color=["#1F4E79", "#E07B39"])
plt.yticks(range(4), ["no evidence", "alarm", "alarm + high current", "alarm + normal current"])
plt.xlabel("posterior probability")
plt.legend(["P(overload | .)", "P(cooling failure | .)"])
plt.ylabel("")
plt.gca().invert_yaxis()
plt.title("Explaining away in the transformer network")
plt.show()
```

### 4.4 การอนุมานโดยประมาณด้วยการสุ่มตัวอย่าง

ข่ายงานขนาดใหญ่ไล่ทุกกรณีไม่ไหว วิธีหนึ่งคือ **สุ่มตัวอย่างแล้วปฏิเสธ (Rejection Sampling)**: สุ่มโลกจำนวนมากตามลำดับพ่อแม่ → ลูก ทิ้งโลกที่ไม่ตรงกับหลักฐาน แล้วนับสัดส่วนในโลกที่เหลือ

```python
def prior_sample(bn, rng):
    """สุ่มโลกหนึ่งชุดตามลำดับโหนด (ดิกชันนารีเรียงพ่อแม่ก่อนลูกไว้แล้ว)"""
    world = {}
    for var in bn:
        world[var] = rng.random() < node_probability(bn, var, True, world)
    return world


def rejection_sampling(bn, query, evidence, n_samples, rng):
    kept = hits = 0
    for _ in range(n_samples):
        world = prior_sample(bn, rng)
        if all(world[k] == v for k, v in evidence.items()):
            kept += 1
            hits += world[query]
    return hits / kept, kept


rng = np.random.default_rng(RANDOM_SEED)
estimate, kept = rejection_sampling(TRANSFORMER_BN, "C", {"A": True, "I": False}, 100_000, rng)
print(f"P(c | a, ¬i) ≈ {estimate:.3f} จากโลกที่ตรงหลักฐาน {kept:,} จาก 100,000 ชุด (ค่าแม่นตรง 0.318)")
```

**คำถามชวนคิด** ทำไมโลกที่ถูกเก็บไว้จึงมีเพียงราวร้อยละ 8 ของทั้งหมด ถ้าหลักฐานเป็นเหตุการณ์ที่เกิดยากมาก (เช่น ความน่าจะเป็น 0.001) วิธีนี้จะมีปัญหาอะไร

### 5.1 ฟังก์ชันความเป็นสมาชิก (ตารางที่ 3.16 และรูปที่ 3.8)

เขียนฟังก์ชันสี่เหลี่ยมคางหมู `trapmf` ตามสูตรในหัวข้อ 3.6.1 และตีความพจน์ที่ตัวส่วนเป็นศูนย์ว่าเป็น 1 (รูปไหล่) สามเหลี่ยม `trimf(a, b, c)` คือคางหมูที่ $b = c$ ตัวแปรภาษาแต่ละตัวเก็บเป็นดิกชันนารี ชื่อค่าทางภาษา → พารามิเตอร์คางหมู 4 ค่า

```python
def trapmf(x, a, b, c, d):
    """สี่เหลี่ยมคางหมู (a, b, c, d) ถ้า a = b หรือ c = d เป็นรูปไหล่"""
    x = np.asarray(x, dtype=float)
    rise = np.ones_like(x) if b == a else (x - a) / (b - a)
    fall = np.ones_like(x) if d == c else (d - x) / (d - c)
    return np.clip(np.minimum(np.minimum(rise, 1.0), fall), 0.0, 1.0)


def trimf(x, a, b, c):
    return trapmf(x, a, b, b, c)


def tri(a, b, c):
    """เขียนสามเหลี่ยมในรูปพารามิเตอร์คางหมู เพื่อเก็บทุกเซตในรูปแบบเดียวกัน"""
    return (a, b, b, c)


TEMP_SETS = {"cold": (15, 15, 20, 25), "medium": tri(20, 25, 30), "hot": (25, 30, 40, 40)}   # °C
HUM_SETS = {"low": (0, 0, 40, 70), "high": (40, 70, 100, 100)}                             # %RH
SPEED_SETS = {"slow": tri(0, 0, 50), "medium": tri(25, 50, 75), "fast": tri(50, 100, 100)}  # %
SPEED_UNIVERSE = np.linspace(0, 100, 10_001)          # ความละเอียด 0.01% ใกล้เคียงค่าแม่นตรงมาก


def fuzzify(value, sets):
    """ระดับความเป็นสมาชิกของค่าชัดในทุกค่าทางภาษา"""
    return {label: float(trapmf(value, *params)) for label, params in sets.items()}


fig, axes = plt.subplots(3, 1, figsize=(7, 7))
panels = [(TEMP_SETS, np.linspace(15, 40, 501), "temperature T (°C)", 27),
          (HUM_SETS, np.linspace(0, 100, 501), "relative humidity H (%RH)", 64),
          (SPEED_SETS, SPEED_UNIVERSE, "fan speed S (%)", None)]
colors = ["#1F4E79", "#3A7D44", "#E07B39"]
for ax, (sets, x, xlabel, marker) in zip(axes, panels):
    for (label, params), color in zip(sets.items(), colors if len(sets) == 3 else colors[::2]):
        ax.plot(x, trapmf(x, *params), color=color, label=label)
    if marker is not None:
        ax.axvline(marker, color="#6B6B6B", ls="--")
        for label, mu in fuzzify(marker, sets).items():
            if mu > 0:
                ax.annotate(f"{mu:.1f}", (marker, mu), xytext=(5, 0), textcoords="offset points")
    ax.set_xlabel(xlabel)
    ax.set_ylabel("μ")
    ax.legend(loc="center right")
plt.tight_layout()
plt.show()
```

### 5.2 การทำให้เป็นฟัซซี (ตัวอย่างที่ 3.20) และตัวดำเนินการ (ตัวอย่างที่ 3.21)

```python
rows = []
for t in [18, 22, 27, 29.9, 30.1, 35]:
    mu = fuzzify(t, TEMP_SETS)
    rows.append({"T (°C)": t, **{k: round(v, 2) for k, v in mu.items()}, "ร้อนแบบคมชัด (T ≥ 30)": int(t >= 30)})
display(pd.DataFrame(rows).set_index("T (°C)"))

hot = fuzzify(27, TEMP_SETS)["hot"]
humid = fuzzify(64, HUM_SETS)["high"]
print(f"ร้อน = {hot:.1f}, ชื้นสูง = {humid:.1f}")
print(f"ร้อน และ ชื้นสูง: min = {min(hot, humid):.2f}, ผลคูณ = {hot * humid:.2f}")
print(f"ร้อน หรือ ชื้นสูง: max = {max(hot, humid):.2f}, ผลบวกเชิงพีชคณิต = {hot + humid - hot * humid:.2f}")
print(f"ร้อน และ ไม่ร้อน = {min(hot, 1 - hot):.2f} (บูลีนจะได้ 0 เสมอ)")
print(f"ร้อนมาก = {hot ** 2:.2f}, ค่อนข้างร้อน = {hot ** 0.5:.2f}")
```

### 5.3 การอนุมานแบบ Mamdani อินพุตเดียว (ตัวอย่างที่ 3.22)

ขั้นตอน: ทำให้เป็นฟัซซี → ระดับการยิงของกฎ → **ตัดยอด** เซตผลของแต่ละกฎที่ระดับการยิง (min) → **รวมผล** ทีละจุดด้วย max ฟังก์ชัน `aggregate` รับระดับการยิงของแต่ละค่าทางภาษาของเอาต์พุต และเลือกได้ระหว่างการตัดยอด (`"clip"`) กับการย่อขนาด (`"scale"`)

```python
def aggregate(strengths, universe, output_sets=SPEED_SETS, implication="clip"):
    """รวมเซตผลที่ถูกตัดยอด (หรือย่อขนาด) ของทุกค่าทางภาษาด้วย max ทีละจุด"""
    clipped = []
    for label, strength in strengths.items():
        mf = trapmf(universe, *output_sets[label])
        clipped.append(np.minimum(strength, mf) if implication == "clip" else strength * mf)
    return np.max(clipped, axis=0)


mu_t27 = fuzzify(27, TEMP_SETS)
strengths_322 = {"slow": mu_t27["cold"], "medium": mu_t27["medium"], "fast": mu_t27["hot"]}   # R1, R2, R3
print("ระดับการยิง:", strengths_322)

grid10 = np.arange(0, 101, 10)
table_322 = pd.DataFrame({
    "μปานกลาง ดิบ": trapmf(grid10, *SPEED_SETS["medium"]),
    "ตัดยอดที่ 0.6": np.minimum(0.6, trapmf(grid10, *SPEED_SETS["medium"])),
    "μเร็ว ดิบ": trapmf(grid10, *SPEED_SETS["fast"]),
    "ตัดยอดที่ 0.4": np.minimum(0.4, trapmf(grid10, *SPEED_SETS["fast"])),
    "μagg = max": aggregate(strengths_322, grid10),
}, index=grid10).T.round(2)
table_322.columns.name = "s (%)"
display(table_322)

mu_clip = aggregate(strengths_322, SPEED_UNIVERSE)
mu_scale = aggregate(strengths_322, SPEED_UNIVERSE, implication="scale")
plt.figure(figsize=(7, 3))
plt.fill_between(SPEED_UNIVERSE, mu_clip, color="#1F4E79", alpha=0.3, label="clipping (min)")
plt.plot(SPEED_UNIVERSE, mu_scale, color="#E07B39", label="scaling (product)")
plt.xlabel("fan speed S (%)")
plt.ylabel("μ_agg")
plt.title("Aggregated output set at T = 27 °C")
plt.legend()
plt.show()
```

### 5.4 การแปลงกลับเป็นค่าชัด (ตัวอย่างที่ 3.23)

เขียนวิธีแปลงกลับเป็นค่าชัดทั้งหมดในหัวข้อ 3.6.3 แล้วใช้กับเซตรวมเดียวกัน

- **จุดศูนย์ถ่วง** $\int z\,\mu(z)\,dz / \int \mu(z)\,dz$ อินทิเกรตเชิงตัวเลขบนจุดห่าง 0.01% (ใกล้ค่าแม่นตรง) และแบบไม่ต่อเนื่องทุก 10%
- **แบ่งครึ่งพื้นที่** หาจุดที่พื้นที่สะสมเท่ากับครึ่งหนึ่ง
- **MOM / SOM / LOM** ค่าเฉลี่ย ค่าน้อยสุด ค่ามากสุด ของจุดที่ $\mu$ สูงสุด
- **ค่าเฉลี่ยถ่วงน้ำหนัก** $\sum w_i c_i / \sum w_i$ ใช้ตำแหน่งยอด $c_i$ ของเซตผล

```python
def centroid(universe, mu, fallback=100.0):
    """จุดศูนย์ถ่วง ถ้าเซตรวมว่าง (ไม่มีกฎยิง) คืนค่าปริยายที่ปลอดภัย = พัดลมเต็มกำลัง"""
    area = integrate(mu, universe)
    return float(integrate(universe * mu, universe) / area) if area > 0 else fallback


def discrete_centroid(universe, mu):
    return float(np.sum(universe * mu) / np.sum(mu))


def bisector(universe, mu):
    cumulative = np.concatenate([[0.0], np.cumsum((mu[1:] + mu[:-1]) / 2 * np.diff(universe))])
    return float(np.interp(cumulative[-1] / 2, cumulative, universe))


def max_group(universe, mu):
    top = universe[np.isclose(mu, mu.max())]
    return float(top.mean()), float(top.min()), float(top.max())


def weighted_average(strengths, output_sets=SPEED_SETS):
    peaks = {label: params[1] for label, params in output_sets.items()}   # ยอดของสามเหลี่ยม
    total = sum(strengths.values())
    return sum(w * peaks[label] for label, w in strengths.items()) / total


mom, som, lom = max_group(SPEED_UNIVERSE, mu_clip)
methods_323 = {
    "จุดศูนย์ถ่วง (ต่อเนื่อง)": (centroid(SPEED_UNIVERSE, mu_clip), 62.7),
    "จุดศูนย์ถ่วง (ทุก 10%)": (discrete_centroid(grid10, aggregate(strengths_322, grid10)), 64.4),
    "แบ่งครึ่งพื้นที่": (bisector(SPEED_UNIVERSE, mu_clip), 59.9),
    "MOM": (mom, 50.0), "SOM": (som, 40.0), "LOM": (lom, 60.0),
    "ค่าเฉลี่ยถ่วงน้ำหนัก": (weighted_average(strengths_322), 70.0),
}
table_323 = pd.DataFrame([{"วิธี": k, "คำนวณได้ (%)": round(v, 2), "หนังสือ (%)": b,
                           "ตรง": abs(v - b) <= 0.05} for k, (v, b) in methods_323.items()]).set_index("วิธี")
display(table_323)

plt.figure(figsize=(7.5, 3.2))
plt.fill_between(SPEED_UNIVERSE, mu_clip, color="#1F4E79", alpha=0.25)
styles = {"จุดศูนย์ถ่วง (ต่อเนื่อง)": ("centroid", "#E07B39"), "แบ่งครึ่งพื้นที่": ("bisector", "#1F4E79"),
          "MOM": ("MOM", "#3A7D44"), "SOM": ("SOM", "#6B6B6B"), "LOM": ("LOM", "#6B6B6B"),
          "ค่าเฉลี่ยถ่วงน้ำหนัก": ("weighted avg", "#8B4513")}
for i, (key, (name, color)) in enumerate(styles.items()):
    x = methods_323[key][0]
    plt.axvline(x, color=color, ls="-" if name == "centroid" else "--")
    right = name == "bisector"            # วางป้ายด้านซ้ายของเส้นเพื่อไม่ทับเส้น LOM
    plt.text(x - 0.5 if right else x + 0.5, 0.95 - 0.1 * i, f"{name} {x:.1f}", color=color,
             ha="right" if right else "left")
plt.ylim(0, 1.05)
plt.xlabel("fan speed S (%)")
plt.ylabel("μ_agg")
plt.title("Defuzzification methods on the same aggregated set")
plt.show()
```

อินพุตเดียวกันให้คำสั่ง 40–70% ขึ้นกับวิธีที่เลือก และจุดศูนย์ถ่วงทุก 10% คลาดจากค่าแม่นตรงราว 1.8 จุดร้อยละ ซึ่งเป็นเหตุผลที่ควรใช้จุดตัวอย่างละเอียด

### 5.5 ตัวควบคุมสองอินพุตครบทุกขั้น (ตารางที่ 3.18 และตัวอย่างที่ 3.24)

ฐานกฎ 6 ข้อเก็บเป็นดิกชันนารี (ค่าทางภาษาของอุณหภูมิ, ค่าทางภาษาของความชื้น) → ค่าทางภาษาของความเร็ว เรียงตาม R1–R6 ของหนังสือ ฟังก์ชัน `fan_controller` รับฐานกฎและเซตทุกชุดเป็นพารามิเตอร์ เพื่อให้ทดลองเปลี่ยนได้ในแบบฝึกปฏิบัติ

```python
FAN_RULES = {("cold", "low"): "slow",      # R1
             ("cold", "high"): "slow",     # R2
             ("medium", "low"): "medium",  # R3
             ("medium", "high"): "fast",   # R4
             ("hot", "low"): "fast",       # R5
             ("hot", "high"): "fast"}      # R6


def rule_strengths(temperature, humidity, rules=FAN_RULES, temp_sets=TEMP_SETS, hum_sets=HUM_SETS,
                   speed_sets=SPEED_SETS):
    """คืน (ระดับการยิงของแต่ละกฎ, ระดับการยิงรวมของแต่ละค่าทางภาษาของเอาต์พุตด้วย max)"""
    mu_t, mu_h = fuzzify(temperature, temp_sets), fuzzify(humidity, hum_sets)
    per_rule = {(t, h): min(mu_t[t], mu_h[h]) for (t, h) in rules}       # AND = min
    per_output = {label: 0.0 for label in speed_sets}
    for antecedent, label in rules.items():
        per_output[label] = max(per_output[label], per_rule[antecedent])
    return per_rule, per_output


def fan_controller(temperature, humidity, rules=FAN_RULES, temp_sets=TEMP_SETS, hum_sets=HUM_SETS,
                   speed_sets=SPEED_SETS, universe=SPEED_UNIVERSE):
    """Mamdani: min (AND), ตัดยอด, max (รวมผล), จุดศูนย์ถ่วง คืนคำสั่งความเร็วพัดลม (%)"""
    _, per_output = rule_strengths(temperature, humidity, rules, temp_sets, hum_sets, speed_sets)
    return centroid(universe, aggregate(per_output, universe, speed_sets))


per_rule, per_output = rule_strengths(27, 64)
display(pd.DataFrame([{"กฎ": f"R{i}", "IF T เป็น": t, "AND H เป็น": h, "THEN S เป็น": FAN_RULES[(t, h)],
                       "ระดับการยิง": round(w, 2)} for i, ((t, h), w) in enumerate(per_rule.items(), 1)])
        .set_index("กฎ"))
print("กรณี ก (27 °C, 64 %RH) ระดับการยิงรวม:", {k: round(v, 2) for k, v in per_output.items()})
```

```python
CASES_324 = {"ข": (22, 46), "ง": (27, 30), "ก": (27, 64), "ค": (32, 80)}
BOOK_319 = {"ข": (39.5, 37.6), "ง": (62.7, 64.4), "ก": (73.1, 75.3), "ค": (83.3, 86.7)}

rows = []
for case, (t, h) in CASES_324.items():
    _, per_output = rule_strengths(t, h)
    exact = fan_controller(t, h)
    coarse = discrete_centroid(grid10, aggregate(per_output, grid10))
    rows.append({"กรณี": case, "T (°C)": t, "H (%RH)": h,
                 "ระดับการยิง ช้า/ปานกลาง/เร็ว": " / ".join(f"{per_output[k]:.1f}" for k in SPEED_SETS),
                 "จุดศูนย์ถ่วงต่อเนื่อง (%)": round(exact, 1), "จุดศูนย์ถ่วงทุก 10% (%)": round(coarse, 1),
                 "ตรงกับตารางที่ 3.19": (round(exact, 1), round(coarse, 1)) == BOOK_319[case]})
table_319 = pd.DataFrame(rows).set_index("กรณี")
display(table_319)

fig, axes = plt.subplots(1, 4, figsize=(13, 2.8), sharey=True)
for ax, (case, (t, h)) in zip(axes, CASES_324.items()):
    _, per_output = rule_strengths(t, h)
    mu = aggregate(per_output, SPEED_UNIVERSE)
    ax.fill_between(SPEED_UNIVERSE, mu, color="#1F4E79", alpha=0.3)
    s_star = centroid(SPEED_UNIVERSE, mu)
    ax.axvline(s_star, color="#E07B39")
    ax.set_title(f"T={t} °C, H={h}% → {s_star:.1f}%", fontsize=10)
    ax.set_xlabel("fan speed S (%)")
axes[0].set_ylabel("μ_agg")
plt.tight_layout()
plt.show()
```

ทั้งสี่กรณีตรงกับตารางที่ 3.19 สังเกตว่า

- กรณี ง และ ก มีอุณหภูมิเท่ากัน แต่ความชื้นที่สูงขึ้นทำให้ความเร็วเพิ่มจาก 62.7% เป็น 73.1% เพราะกฎ R3 กับ R4 ต่างกันเท่านั้น
- กรณี ค ร้อนและชื้นที่สุด แต่คำสั่งสูงสุดเพียง 83.3% เพราะจุดศูนย์ถ่วงของเซต "เร็ว" tri(50, 100, 100) คือ 83.3 (ข้อสังเกตที่ 3 ในหนังสือ) แบบฝึกปฏิบัติข้อ 3 จะแก้ปัญหานี้

### 5.6 พื้นผิวควบคุม (รูปที่ 3.12)

พื้นผิวควบคุมแสดงคำสั่งความเร็วสำหรับทุกคู่ของอุณหภูมิและความชื้น เป็นเครื่องมือตรวจการออกแบบที่สำคัญที่สุดอย่างหนึ่ง ฟังก์ชัน `control_surface` คำนวณบนกริด (ใช้จุดตัวอย่างเอาต์พุตทุก 0.1% เพื่อให้เร็วขึ้น)

```python
SURFACE_UNIVERSE = np.linspace(0, 100, 1001)    # ทุก 0.1%


def control_surface(controller, temps=np.linspace(15, 40, 51), hums=np.linspace(0, 100, 51), **kwargs):
    """คืนกริด (T, H, S) สำหรับวาดพื้นผิว controller ต้องรับ (temperature, humidity, **kwargs)"""
    T, H = np.meshgrid(temps, hums)
    S = np.vectorize(lambda t, h: controller(t, h, **kwargs))(T, H)
    return T, H, S


def plot_surface(T, H, S, title, marks=CASES_324):
    fig = plt.figure(figsize=(12, 4.5))
    ax3d = fig.add_subplot(1, 2, 1, projection="3d")
    ax3d.plot_surface(T, H, S, cmap="coolwarm", edgecolor="none")
    ax3d.set_xlabel("T (°C)")
    ax3d.set_ylabel("H (%RH)")
    ax3d.set_zlabel("S (%)")
    ax3d.set_zlim(0, 100)
    ax3d.set_title(title)
    ax2d = fig.add_subplot(1, 2, 2)
    image = ax2d.pcolormesh(T, H, S, cmap="coolwarm", vmin=0, vmax=100, shading="auto")
    contours = ax2d.contour(T, H, S, levels=range(10, 100, 10), colors="k", linewidths=0.6)
    ax2d.clabel(contours, fmt="%d", fontsize=7)
    for case, (t, h) in marks.items():
        ax2d.plot(t, h, "o", color="#3A7D44")
        ax2d.annotate(f"{fan_controller(t, h):.1f}", (t, h), xytext=(4, 4), textcoords="offset points",
                      color="#3A7D44", fontsize=8)
    ax2d.set_xlabel("T (°C)")
    ax2d.set_ylabel("H (%RH)")
    fig.colorbar(image, ax=ax2d, label="S (%)")
    plt.tight_layout()
    plt.show()


T_grid, H_grid, S_grid = control_surface(fan_controller, universe=SURFACE_UNIVERSE)
plot_surface(T_grid, H_grid, S_grid, "Fan controller (book design)")
print(f"ช่วงคำสั่งบนพื้นผิว: {S_grid.min():.1f}% ถึง {S_grid.max():.1f}%")
print(f"T = 18 °C, H = 55%: {fan_controller(18, 55):.1f}%  |  T = 35 °C, H = 55%: {fan_controller(35, 55):.1f}%")
```

พื้นผิวราบเรียบ ไม่มีหลุมหรือหน้าผา คำสั่งต่ำสุดราว 16.7% (พัดลมไม่มีวันหยุด) และสูงสุดราว 83.3% ตรงกับข้อสังเกตที่ 3 และคำบรรยายรูปที่ 3.12 และที่ความชื้น 55% ช่วงคำสั่งแคบลงอีก (ราว 19.4% ถึง 80.6%) เพราะกฎ "ช้า" หรือ "เร็ว" ยิงไม่เต็มที่

**คำถามชวนคิด** ตามแนวยาว T จาก 20 ถึง 30 °C ที่ความชื้นต่ำ (H ≤ 40) เส้นชั้นความสูงห่างเท่ากันหรือไม่ เทียบกับการวิเคราะห์ในหัวข้อ 3.6.2 ที่บอกว่าแบบ Sugeno ให้ความสัมพันธ์เชิงเส้น $S = 10(T - 20)$

## ส่วนที่ 6 เทียบผลกับ scikit-fuzzy
