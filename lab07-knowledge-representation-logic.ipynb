## ส่วนที่ 1 ตารางค่าความจริงและการตรวจการตามมา

### 1.1 ประโยคคือฟังก์ชันของแบบจำลอง

ในตรรกศาสตร์เชิงประพจน์ **แบบจำลอง (Model)** คือการกำหนดค่าจริงหรือเท็จให้สัญลักษณ์ทุกตัว (หัวข้อ 3.2.1) เราแทนแบบจำลองด้วยดิกชันนารี เช่น `{"O": True, "T": False}` และแทน **ประโยค** ด้วยฟังก์ชันที่รับแบบจำลองแล้วคืนค่าจริงหรือเท็จ เช่น ประโยค $O \Rightarrow T$ เขียนเป็น `lambda m: implies(m["O"], m["T"])`

ตัวเชื่อม $\lnot, \land, \lor$ ใช้ `not`, `and`, `or` ของ Python ได้ทันที ส่วน $\Rightarrow$ และ $\Leftrightarrow$ ต้องเขียนฟังก์ชันเอง

```python
def implies(p, q):
    """P ⇒ Q เท็จกรณีเดียว คือเหตุจริงแต่ผลเท็จ"""
    return (not p) or q


def iff(p, q):
    """P ⇔ Q จริงเมื่อทั้งสองมีค่าเหมือนกัน"""
    return p == q


def all_models(symbols):
    """สร้างแบบจำลองทั้ง 2^n แบบ เรียงแบบเดียวกับตารางในหนังสือ (0 ก่อน 1)"""
    for values in itertools.product([False, True], repeat=len(symbols)):
        yield dict(zip(symbols, values))


def truth_table(symbols, sentences):
    """symbols: ลิสต์ชื่อสัญลักษณ์, sentences: ดิกชันนารี ชื่อคอลัมน์ -> ฟังก์ชันของแบบจำลอง"""
    rows = []
    for model in all_models(symbols):
        row = {s: int(model[s]) for s in symbols}
        for name, sentence in sentences.items():
            row[name] = int(sentence(model))
        rows.append(row)
    table = pd.DataFrame(rows)
    table.index = range(1, len(table) + 1)
    table.index.name = "แถว"
    return table


# ตารางที่ 3.4 ตารางค่าความจริงของตัวเชื่อมทั้ง 5
truth_table(["P", "Q"], {
    "¬P": lambda m: not m["P"],
    "P ∧ Q": lambda m: m["P"] and m["Q"],
    "P ∨ Q": lambda m: m["P"] or m["Q"],
    "P ⇒ Q": lambda m: implies(m["P"], m["Q"]),
    "P ⇔ Q": lambda m: iff(m["P"], m["Q"]),
})
```

ผลตรงกับตารางที่ 3.4 สังเกตคอลัมน์ $P \Rightarrow Q$ ซึ่งเท็จเฉพาะแถวที่ 3 ($P = 1, Q = 0$) เท่านั้น

### 1.2 ตรวจการตามมาด้วยการตรวจแบบจำลอง (ตัวอย่างที่ 3.5)

$KB \models \alpha$ ก็ต่อเมื่อ **ในทุกแบบจำลองที่ $KB$ จริง $\alpha$ ต้องจริงด้วย** ฟังก์ชัน `tt_entails` ไล่ทุกแถว เก็บแถวที่ $KB$ จริง และเก็บ **แถวตัวอย่างค้าน (Counterexample)** คือแถวที่ $KB$ จริงแต่ $\alpha$ เท็จ ถ้าไม่มีแถวตัวอย่างค้านเลย แปลว่า $KB \models \alpha$

ตัวอย่างที่ 3.5 ใช้สัญลักษณ์ $O$ = รีเลย์โอเวอร์โหลดทำงาน, $T$ = เบรกเกอร์ตัด, $R$ = มอเตอร์หมุน และ $KB$: (1) $O \Rightarrow T$ (2) $T \Rightarrow \lnot R$ (3) $O$ ถามว่า $KB \models \lnot R$ หรือไม่

```python
def tt_entails(symbols, kb, alpha):
    """ตรวจ KB ⊨ alpha ด้วยการไล่ทุกแบบจำลอง คืน (ผล, แถวที่ KB จริง, แถวตัวอย่างค้าน)"""
    kb_rows, counterexamples = [], []
    for row_number, model in enumerate(all_models(symbols), start=1):
        if all(sentence(model) for sentence in kb):
            kb_rows.append(row_number)
            if not alpha(model):
                counterexamples.append(row_number)
    return len(counterexamples) == 0, kb_rows, counterexamples


SYMBOLS_35 = ["O", "T", "R"]
S1 = lambda m: implies(m["O"], m["T"])        # (1) O ⇒ T
S2 = lambda m: implies(m["T"], not m["R"])    # (2) T ⇒ ¬R
S3 = lambda m: m["O"]                         # (3) O
NOT_R = lambda m: not m["R"]

display(truth_table(SYMBOLS_35, {"(1) O⇒T": S1, "(2) T⇒¬R": S2, "(3) O": S3,
                                 "KB": lambda m: S1(m) and S2(m) and S3(m), "¬R": NOT_R}))

result, kb_rows, counter = tt_entails(SYMBOLS_35, [S1, S2, S3], NOT_R)
compare("ขั้นที่ 1 แถวที่ KB จริง", kb_rows, [7])
compare("ขั้นที่ 2 KB ⊨ ¬R", result, True)

result2, kb2_rows, counter2 = tt_entails(SYMBOLS_35, [S1, S2], NOT_R)
compare("ขั้นที่ 3 แถวที่ KB' = (1)∧(2) จริง", kb2_rows, [1, 2, 3, 7])
compare("ขั้นที่ 3 แถวตัวอย่างค้านของ KB' ⊨ ¬R", counter2, [2])
compare("ขั้นที่ 3 KB' ⊨ ¬R", result2, False)
```

### 1.3 การอนุมานที่ถูกและผิด (ตารางที่ 3.6 และกับดักในตัวอย่างที่ 3.6)

กฎการอนุมานที่ **ถูกต้อง (Sound)** คือกฎที่ประโยคใต้เส้นตามมาจากประโยคเหนือเส้นเสมอ ตรวจได้ด้วย `tt_entails` เดียวกัน เราตรวจ Modus Ponens, Modus Tollens และการแก้ปม (Resolution) เทียบกับ **การยืนยันผล (Affirming the Consequent)** ซึ่งเป็นการอนุมานผิดที่พบบ่อย ("เบรกเกอร์ตัด จึงโอเวอร์โหลด")

```python
A = lambda m: m["A"]
B = lambda m: m["B"]
C = lambda m: m["C"]
inference_rules = {
    "Modus Ponens: A⇒B, A ∴ B": ([lambda m: implies(A(m), B(m)), A], B),
    "Modus Tollens: A⇒B, ¬B ∴ ¬A": ([lambda m: implies(A(m), B(m)), lambda m: not B(m)], lambda m: not A(m)),
    "Resolution: A∨B, ¬B∨C ∴ A∨C": ([lambda m: A(m) or B(m), lambda m: (not B(m)) or C(m)],
                                    lambda m: A(m) or C(m)),
    "ยืนยันผล (ผิด): A⇒B, B ∴ A": ([lambda m: implies(A(m), B(m)), B], A),
}
rows = []
for name, (premises, conclusion) in inference_rules.items():
    ok, _, counter = tt_entails(["A", "B", "C"], premises, conclusion)
    example = "—"
    if counter:  # แสดงแบบจำลองแรกที่เป็นตัวอย่างค้าน
        model = list(all_models(["A", "B", "C"]))[counter[0] - 1]
        example = ", ".join(f"{k}={int(v)}" for k, v in model.items())
    rows.append({"กฎ": name, "ถูกต้อง (sound)": ok, "ตัวอย่างค้าน": example})
pd.DataFrame(rows)
```

ตัวอย่างค้านของการยืนยันผลคือ $A = 0, B = 1$ ซึ่งตรงกับแถวที่ 3 ของตัวอย่างที่ 3.5 ($O = 0, T = 1$): เบรกเกอร์ตัดทั้งที่ไม่ได้โอเวอร์โหลด เช่น ตัดเพราะลัดวงจร

### 1.4 ความสมมูล และราคาของการตรวจแบบจำลอง

ประโยคสองประโยค **สมมูลกัน** ถ้าจริงในแบบจำลองชุดเดียวกันพอดี เซลล์ต่อไปตรวจความสมมูลในตารางที่ 3.5 บางข้อ แล้วตรวจตัวอย่างที่ 3.6 ($A \land C \Rightarrow L$, $L \Rightarrow S$, $A$, $C$ ⊨ $S$) ซึ่งต้องไล่ $2^4 = 16$ แถว

```python
def equivalent(symbols, alpha, beta):
    """α ≡ β ก็ต่อเมื่อมีค่าความจริงเท่ากันในทุกแบบจำลอง"""
    return all(bool(alpha(m)) == bool(beta(m)) for m in all_models(symbols))


P = lambda m: m["P"]
Q = lambda m: m["Q"]
equivalences = {
    "P⇒Q ≡ ¬P∨Q": (lambda m: implies(P(m), Q(m)), lambda m: (not P(m)) or Q(m)),
    "P⇒Q ≡ ¬Q⇒¬P (แย้งสลับที่)": (lambda m: implies(P(m), Q(m)), lambda m: implies(not Q(m), not P(m))),
    "¬(P∧Q) ≡ ¬P∨¬Q (เดอมอร์แกน)": (lambda m: not (P(m) and Q(m)), lambda m: (not P(m)) or (not Q(m))),
    "P⇒Q ≡ Q⇒P (ผิด: กลับทิศ)": (lambda m: implies(P(m), Q(m)), lambda m: implies(Q(m), P(m))),
}
for name, (alpha, beta) in equivalences.items():
    print(f"{name:35s} → {equivalent(['P', 'Q'], alpha, beta)}")

kb_36 = [lambda m: implies(m["A"] and m["C"], m["L"]), lambda m: implies(m["L"], m["S"]),
         lambda m: m["A"], lambda m: m["C"]]
ok, kb_rows, _ = tt_entails(["A", "C", "L", "S"], kb_36, lambda m: m["S"])
print(f"\nตัวอย่างที่ 3.6: KB ⊨ S = {ok} (ไล่ {2 ** 4} แถว KB จริงเพียงแถว {kb_rows})")
```

ตารางค่าความจริงโตแบบ $2^n$ ลองวัดเวลาตรวจฐานความรู้แบบลูกโซ่ $P_1,\ P_1 \Rightarrow P_2,\ \ldots,\ P_{n-1} \Rightarrow P_n$ ⊨ $P_n$ เมื่อ $n$ เพิ่มขึ้น

```python
def chain_kb(n):
    """KB: P1, P1⇒P2, ..., P(n-1)⇒Pn และคำถาม Pn"""
    symbols = [f"P{i}" for i in range(1, n + 1)]
    kb = [lambda m: m["P1"]]
    for a, b in zip(symbols, symbols[1:]):
        kb.append(lambda m, a=a, b=b: implies(m[a], m[b]))
    return symbols, kb, lambda m: m[symbols[-1]]


sizes, seconds = list(range(8, 17)), []
for n in sizes:
    symbols, kb, query = chain_kb(n)
    start = time.perf_counter()
    assert tt_entails(symbols, kb, query)[0]
    seconds.append(time.perf_counter() - start)

plt.figure(figsize=(6, 3.2))
plt.semilogy(sizes, seconds, "o-", color="#1F4E79")
plt.xlabel("number of symbols n")
plt.ylabel("time (s, log scale)")
plt.title("Model checking time grows like 2^n")
plt.grid(alpha=0.3)
plt.show()
print(f"n = 16 ใช้ {seconds[-1]:.2f} วินาที ถ้า n = 30 จะช้ากว่าราว 2^14 = {2 ** 14:,} เท่า")
```

## ส่วนที่ 2 ระบบผู้เชี่ยวชาญแบบกฎ: Forward Chaining และคำอธิบาย HOW

### 2.1 ฐานกฎ หน่วยความจำใช้งาน และกลไกอนุมาน

ระบบผู้เชี่ยวชาญแบบกฎ (หัวข้อ 3.3.1) แยกเป็นส่วนชัดเจน

- **ฐานกฎ** — ลิสต์ของ `Rule` แต่ละกฎมีชื่อ เงื่อนไข และผลสรุป เงื่อนไขเป็นได้สองแบบ คือ **ข้อเท็จจริง** (สตริง เช่น `"สั่นสูง"`) หรือ **การทดสอบค่าจากเซนเซอร์** (`Test` เช่น การสั่น > 7 mm/s)
- **หน่วยความจำใช้งาน (Working Memory: WM)** — ข้อเท็จจริงที่รู้แล้วของกรณีที่กำลังพิจารณา และค่าที่อ่านได้จากเซนเซอร์ (`readings`)
- **กลไกอนุมาน** — ฟังก์ชัน `forward_chaining` ทำงานตามวงรอบจดจำ-ทำงาน: จับคู่ → แก้ข้อขัดแย้ง → ยิงกฎ จนไม่มีกฎใดยิงได้

กลไกอนุมานไม่รู้อะไรเกี่ยวกับมอเตอร์เลย ฐานกฎเท่านั้นที่เปลี่ยนเมื่อเปลี่ยนงาน

```python
@dataclass(frozen=True)
class Test:
    """เงื่อนไขเชิงตัวเลขของค่าจากเซนเซอร์ เช่น Test("vibration", ">", 7)"""
    sensor: str
    op: str          # ">" หรือ "<="
    threshold: float

    def holds(self, value):
        return value > self.threshold if self.op == ">" else value <= self.threshold

    def __str__(self):
        return f"{self.sensor} {self.op} {self.threshold:g}"


@dataclass
class Rule:
    name: str
    conditions: list     # สตริง (ข้อเท็จจริง) หรือ Test
    conclusion: str

    def __str__(self):
        return f"{self.name}: IF {' AND '.join(map(str, self.conditions))} THEN {self.conclusion}"


def condition_holds(condition, wm, readings):
    """เงื่อนไขเป็นจริงถ้าเป็นข้อเท็จจริงที่อยู่ใน WM หรือเป็น Test ที่ค่าจากเซนเซอร์ผ่านเกณฑ์"""
    if isinstance(condition, Test):
        return condition.sensor in readings and condition.holds(readings[condition.sensor])
    return condition in wm


def select_lowest_number(conflict_set, fact_time):
    """กลยุทธ์แก้ข้อขัดแย้งที่บทใช้: เลือกกฎหมายเลขน้อยที่สุด (ลำดับในฐานกฎ)"""
    return conflict_set[0]


def forward_chaining(rules, readings=None, initial_facts=None, select=select_lowest_number):
    """คืน (WM, how, ตารางไล่รอบ) โดย how บันทึกว่าข้อเท็จจริงแต่ละข้อได้มาจากกฎใด"""
    readings = readings or {}
    fact_time = dict(initial_facts or {})      # ข้อเท็จจริง -> เวลาที่เข้ามา (ใช้กับกลยุทธ์ความใหม่)
    how, trace = {}, []
    while True:
        # จับคู่: กฎที่เงื่อนไขครบ และผลสรุปยังไม่อยู่ใน WM (ไม่ยิงซ้ำ)
        conflict = [r for r in rules if r.conclusion not in fact_time
                    and all(condition_holds(c, fact_time, readings) for c in r.conditions)]
        row = {"เซตข้อขัดแย้ง": "{" + ", ".join(r.name for r in conflict) + "}"}
        if not conflict:
            trace.append({**row, "กฎที่ยิง": "—", "ข้อเท็จจริงที่เพิ่มใน WM": "หยุด"})
            break
        rule = select(conflict, fact_time)                  # แก้ข้อขัดแย้ง
        fact_time[rule.conclusion] = max(fact_time.values(), default=0) + 1   # ยิงกฎ
        how[rule.conclusion] = rule
        trace.append({**row, "กฎที่ยิง": rule.name, "ข้อเท็จจริงที่เพิ่มใน WM": rule.conclusion})
    table = pd.DataFrame(trace)
    table.index = range(1, len(table) + 1)
    table.index.name = "รอบ"
    return list(fact_time), how, table
```

### 2.4 กรณี ก และกรณี ข ของตัวอย่างที่ 3.11

ฟังก์ชัน `diagnose` เรียก `forward_chaining` แล้วแสดงตารางไล่รอบ ผลวินิจฉัย และคำแนะนำ เปรียบเทียบตารางกับตัวอย่างที่ 3.11 ในหนังสือทีละรอบ

```python
CASE_A = {"winding_temp": 128, "vibration": 8.2, "bearing_temp": 96, "current_unbalance": 4, "current": 112}
CASE_B = {"winding_temp": 125, "vibration": 3.0, "bearing_temp": 70, "current_unbalance": 14, "current": 95}


def diagnose(readings, rules=MOTOR_RULES, show=True):
    """รัน Forward Chaining แล้วแยกผลวินิจฉัย (ผลสรุปของกฎที่มีเงื่อนไขเป็นข้อเท็จจริง) ออกจากอาการ"""
    wm, how, table = forward_chaining(rules, readings)
    diagnosis_facts = {r.conclusion for r in rules if any(not isinstance(c, Test) for c in r.conditions)}
    diagnoses = [f for f in wm if f in diagnosis_facts]
    if show:
        display(table)
        for d in diagnoses:
            print(f"ผลวินิจฉัย: {d}\n    คำแนะนำ: {RECOMMENDATIONS.get(d, '-')}")
    return wm, how, table, diagnoses


print("กรณี ก")
wm_a, how_a, table_a, diag_a = diagnose(CASE_A)
compare("กฎที่ยิง กรณี ก", list(table_a["กฎที่ยิง"]), ["R1", "R2", "R3", "R6", "R7", "R9", "—"])
compare("เซตข้อขัดแย้งรอบที่ 4", table_a.loc[4, "เซตข้อขัดแย้ง"], "{R6, R7}")
```

### 2.5 ส่วนอธิบายเหตุผล: คำถาม HOW

เพราะ `forward_chaining` บันทึกไว้ใน `how` ว่าข้อเท็จจริงแต่ละข้อมาจากกฎใด เราจึงตอบคำถาม "**สรุปได้อย่างไร (HOW)**" ได้ด้วยการไล่ย้อนตามกฎที่ใช้จนถึงค่าจากเซนเซอร์

```python
def explain_how(fact, how, readings, indent=0):
    """พิมพ์ต้นไม้เหตุผลของ fact ย้อนไปถึงค่าจากเซนเซอร์"""
    pad = "    " * indent
    rule = how.get(fact)
    if rule is None:
        print(f"{pad}{fact} (ข้อเท็จจริงตั้งต้น)")
        return
    print(f"{pad}{fact}  ← ตาม {rule.name}")
    for condition in rule.conditions:
        if isinstance(condition, Test):
            print(f"{pad}    {condition.sensor} = {readings[condition.sensor]:g}  ผ่านเกณฑ์ {condition}")
        else:
            explain_how(condition, how, readings, indent + 1)


for d in diag_a:
    explain_how(d, how_a, CASE_A)
    print()
```

### 2.6 เมื่อเซนเซอร์เสีย: ทำไมจึงมีกฎ R4

ถ้าเขียน R8 ว่า "สั่นสูง AND NOT ตลับลูกปืนร้อน" ระบบจะต้องใช้สมมติฐานโลกปิด (ตัวอย่างที่ 3.8) คือถือว่าสิ่งที่ไม่รู้เป็นเท็จ ลองจำลองกรณีที่เซนเซอร์อุณหภูมิตลับลูกปืนเสีย (ไม่มีค่า `bearing_temp`) แล้วเทียบฐานกฎของหนังสือกับฐานกฎที่ใช้ NOT

```python
CASE_BROKEN_SENSOR = {"winding_temp": 80, "vibration": 9.0, "current_unbalance": 3, "current": 90}  # ไม่มี bearing_temp

print("ฐานกฎของหนังสือ (มี R4):")
_ = diagnose(CASE_BROKEN_SENSOR)


@dataclass(frozen=True)
class NotFact:
    """เงื่อนไข NOT: จริงเมื่อข้อเท็จจริงไม่อยู่ใน WM (สมมติฐานโลกปิด)"""
    fact: str

    def __str__(self):
        return f"NOT {self.fact}"


def condition_holds(condition, wm, readings):
    """ขยายฟังก์ชันจับคู่เดิมให้รู้จักเงื่อนไข NOT (forward_chaining เรียกชื่อนี้ จึงใช้ฉบับใหม่ทันที)"""
    if isinstance(condition, NotFact):
        return condition.fact not in wm
    if isinstance(condition, Test):
        return condition.sensor in readings and condition.holds(readings[condition.sensor])
    return condition in wm


CWA_RULES = [r for r in MOTOR_RULES if r.name != "R8"] + [
    Rule("R8*", ["สั่นสูง", NotFact("ตลับลูกปืนร้อน")], "สงสัยเพลาเยื้องศูนย์หรือไม่สมดุลทางกล")]
print("\nฐานกฎที่ใช้ NOT (สมมติฐานโลกปิด):", CWA_RULES[-1])
_ = diagnose(CASE_BROKEN_SENSOR, rules=CWA_RULES)
```

ฐานกฎของหนังสือไม่สรุปอะไรเกี่ยวกับตลับลูกปืนเลย เพราะทั้ง R3 และ R4 ไม่ยิง ส่วนฐานกฎที่ใช้ NOT สรุปผิดทันทีว่า "สงสัยเพลาเยื้องศูนย์" เพียงเพราะ **ไม่มีข้อมูล** ว่าตลับลูกปืนร้อน

**คำถามชวนคิด** ในงานความปลอดภัย ระบบควรแจ้งผู้ใช้อย่างไรเมื่อพบว่าเซนเซอร์บางตัวไม่มีค่า ลองเสนอกฎเพิ่ม 1 ข้อที่ทำให้ระบบเตือนว่า "ข้อมูลไม่พอสำหรับวินิจฉัยตลับลูกปืน"

## ส่วนที่ 3 Backward Chaining และคำอธิบาย WHY

### 3.1 กลไกอนุมานแบบย้อนกลับ

Backward Chaining (หัวข้อ 3.3.2) เริ่มจาก **เป้าหมาย** หากฎที่มีผลสรุปตรงกับเป้าหมาย แล้วตั้งเงื่อนไขของกฎนั้นเป็นเป้าหมายย่อย ถ้าเงื่อนไขเป็นค่าจากเซนเซอร์ ระบบ **ขอให้ช่างวัด** เฉพาะค่าที่จำเป็น และจำค่าที่วัดแล้วไว้ไม่ให้วัดซ้ำ

คลาส `BackwardChainer` เก็บสถานะไว้ในตัว ได้แก่ WM, ค่าที่วัดแล้ว, บันทึกการทำงานทีละขั้น และ **กองเป้าหมาย (Goal Stack)** ที่กำลังพิสูจน์อยู่ ซึ่งใช้ทั้งป้องกันการวนซ้ำ และตอบคำถาม "**ทำไม (WHY)** ต้องวัดค่านี้"

```python
class BackwardChainer:
    def __init__(self, rules, measure, initial_facts=(), explain_why=False):
        self.rules = rules
        self.measure = measure             # ฟังก์ชันขอค่าจากช่างหรือเซนเซอร์: measure(sensor) -> ค่า
        self.wm = set(initial_facts)
        self.measured = {}                 # ค่าที่วัดแล้ว (ไม่วัดซ้ำ)
        self.how = {}
        self.log = []
        self.explain_why = explain_why

    def _record(self, goal, level, action, result=""):
        self.log.append({"เป้าหมาย (ระดับ)": f"{goal} ({level})", "การทำงาน": action, "ผล": result})

    def _why(self, stack):
        """ตอบ WHY จากกองเป้าหมาย: สมมติฐานบนสุด + สิ่งที่รู้แล้ว + เป้าหมายย่อยที่กำลังตรวจ"""
        top_goal, top_rule = stack[0]
        known = [c for c in top_rule.conditions if isinstance(c, str) and c in self.wm]
        goal, rule = stack[-1]
        text = f"เพราะกำลังตรวจสมมติฐาน '{top_goal}' ({top_rule.name})"
        if known:
            text += f" ซึ่งทราบแล้วว่า {', '.join(known)}"
        if len(stack) > 1:
            text += f" และต้องรู้ว่า '{goal}' หรือไม่ ({rule.name})"
        return text

    def get_value(self, sensor, stack):
        if sensor in self.measured:
            return self.measured[sensor], "ใช้ค่าที่วัดแล้ว"
        if self.explain_why:
            print(f"ระบบขอค่า {sensor} → WHY? {self._why(stack)}")
        self.measured[sensor] = self.measure(sensor)
        return self.measured[sensor], "ขอให้ช่างวัด"

    def prove(self, goal, level=0, stack=()):
        if goal in self.wm:
            self._record(goal, level, "อยู่ใน WM แล้ว", "✓")
            return True
        if any(goal == g for g, _ in stack):             # ป้องกันการวนซ้ำ เช่น A⇒B และ B⇒A
            return False
        candidate_rules = [r for r in self.rules if r.conclusion == goal]
        for rule in candidate_rules:
            new_stack = stack + ((goal, rule),)
            facts = [c for c in rule.conditions if not isinstance(c, Test)]
            if facts:
                self._record(goal, level, f"ไม่อยู่ใน WM → {rule.name} → เป้าหมายย่อย {', '.join(facts)}", "—")
            satisfied = True
            for condition in rule.conditions:
                if isinstance(condition, Test):
                    value, how_obtained = self.get_value(condition.sensor, new_stack)
                    satisfied = condition.holds(value)
                    self._record(goal, level, f"{rule.name} ต้องการ {condition} → {how_obtained} ได้ {value:g}",
                                 f"✓ เพิ่ม '{goal}'" if satisfied else "✗")
                else:
                    satisfied = self.prove(condition, level + 1, new_stack)
                if not satisfied:
                    break
            if satisfied:
                self.wm.add(goal)
                self.how[goal] = rule
                if facts:
                    self._record(goal, level, f"เงื่อนไขของ {rule.name} ครบ", "✓ พิสูจน์ได้")
                return True
        if candidate_rules and any(not isinstance(c, Test) for r in candidate_rules for c in r.conditions):
            names = ", ".join(r.name for r in candidate_rules)
            self._record(goal, level, f"{names} ล้มเหลว และไม่มีกฎอื่นสรุปเป้าหมายนี้", "✗ พิสูจน์ไม่ได้")
        return False

    def log_table(self):
        table = pd.DataFrame(self.log)
        table.index = range(1, len(table) + 1)
        table.index.name = "ขั้น"
        return table
```

## ส่วนที่ 4 กลยุทธ์การแก้ข้อขัดแย้ง (ตัวอย่างที่ 3.9)

ฐานความรู้ของระบบเฝ้าระวังปั๊มมีกฎ 3 ข้อ ข้อเท็จจริง "ตลับลูกปืนร้อน" เข้ามาที่เวลา $t = 1$ และ "สั่นสูง" เข้ามาที่ $t = 2$ ฟังก์ชัน `forward_chaining` รับกลยุทธ์ผ่านพารามิเตอร์ `select` เราจึงเขียนกลยุทธ์ใหม่เป็นฟังก์ชันสั้น ๆ ได้โดยไม่ต้องแก้กลไกอนุมาน

- **ลำดับของกฎ** — เลือกกฎแรกในเซตข้อขัดแย้ง (`select_lowest_number`)
- **ความเฉพาะเจาะจง** — เลือกกฎที่มีเงื่อนไขมากที่สุด ถ้าเสมอกันเลือกกฎที่อยู่ก่อน
- **ความใหม่** — เลือกกฎที่ใช้ข้อเท็จจริงที่เข้ามาล่าสุด ถ้าเสมอกันตัดสินด้วยความเฉพาะเจาะจง

```python
PUMP_RULES = [
    Rule("RA", ["ตลับลูกปืนร้อน"], "ลดความเร็วรอบลง 20%"),
    Rule("RB", ["ตลับลูกปืนร้อน", "สั่นสูง"], "หยุดเครื่องทันที"),
    Rule("RC", ["สั่นสูง"], "แจ้งเตือนช่าง"),
]
PUMP_FACTS = {"ตลับลูกปืนร้อน": 1, "สั่นสูง": 2}


def select_by_specificity(conflict_set, fact_time):
    return max(conflict_set, key=lambda r: len(r.conditions))          # max คืนตัวแรกเมื่อเสมอ


def select_by_recency(conflict_set, fact_time):
    def newest(rule):
        return max(fact_time.get(c, 0) for c in rule.conditions)
    return max(conflict_set, key=lambda r: (newest(r), len(r.conditions)))


strategies = {"ลำดับของกฎ": select_lowest_number, "ความเฉพาะเจาะจง": select_by_specificity,
              "ความใหม่": select_by_recency}
rows = []
for name, select in strategies.items():
    _, _, table = forward_chaining(PUMP_RULES, initial_facts=PUMP_FACTS, select=select)
    fired = [f for f in table["กฎที่ยิง"] if f != "—"]
    first_action = table.loc[1, "ข้อเท็จจริงที่เพิ่มใน WM"]
    rows.append({"กลยุทธ์": name, "กฎที่ยิงก่อน": fired[0], "การกระทำแรก": first_action,
                 "ลำดับการยิงทั้งหมด": " → ".join(fired)})
strategy_table = pd.DataFrame(rows)
display(strategy_table)
compare("กฎที่ยิงก่อน", list(strategy_table["กฎที่ยิงก่อน"]), ["RA", "RB", "RB"])
```

ผลตรงกับตัวอย่างที่ 3.9 กลยุทธ์ "ลำดับของกฎ" สั่งลดความเร็วรอบก่อน ทั้งที่อาการรุนแรงพอจะต้องหยุดเครื่อง ส่วนอีกสองกลยุทธ์สั่งหยุดเครื่องทันที ในระบบนี้ทุกกฎยิงครบในที่สุดเพราะไม่มีกฎใดลบข้อเท็จจริง แต่ในระบบจริง เมื่อหยุดเครื่องแล้ว การกระทำอื่นอาจไม่มีความหมายหรือขัดกัน ลำดับจึงสำคัญ

**คำถามชวนคิด** ถ้าต้องการให้กฎหยุดเครื่องยิงก่อนเสมอโดยไม่ขึ้นกับลำดับการเขียนกฎ ควรเพิ่มอะไรให้ `Rule` และเขียนฟังก์ชัน `select` แบบใด (แนวคิด Salience ในหัวข้อ 3.3.1)

### ข้อ 3 ตรวจลอจิกอินเทอร์ล็อกด้วยการตามมา

ตู้ควบคุมมีอินเทอร์ล็อก "ถ้าประตูตู้เปิด ห้ามจ่ายไฟ" กำหนด $D$ = ประตูตู้เปิด, $P$ = มีการจ่ายไฟเข้าบัส, $L$ = หลอดแสดง "มีไฟ" ติด และฐานความรู้

- (1) $D \Rightarrow \lnot P$ (อินเทอร์ล็อก)
